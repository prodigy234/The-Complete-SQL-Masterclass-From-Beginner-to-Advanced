# PART 1 - DATABASE DESIGN FUNDAMENTALS

**Database design** is the process of deciding how data will be structured, stored, related, and constrained so that it stays correct, is easy to query, and performs well at scale. It sits above SQL syntax: SQL is *how* you implement a design; design is deciding *what* to implement.

**Related terms, disambiguated**

| Term | Meaning |
|---|---|
| Data modeling | The activity of defining entities, attributes, and relationships - the blueprint |
| Data architecture | The overall strategy for how data flows and is organized across an entire organization |
| Database architecture | How a *specific* database system is structured (schemas, storage, replication) |
| Schema design | The concrete set of tables/columns/constraints implementing a data model |
| Conceptual design | Business-level model - entities and relationships, no technology (e.g. an ER diagram with no data types) |
| Logical design | Technology-independent but fully detailed - attributes, keys, normalization, still no vendor-specific types |
| Physical design | Vendor-specific implementation - actual data types, indexes, partitioning, storage |

**Requirements types**

- **Business requirements**: what the business needs (e.g. "track every customer order and its fulfillment status").
- **Technical requirements**: constraints the solution must satisfy (e.g. "must support 10,000 transactions/second").
- **Functional requirements**: what the system must *do* (e.g. "generate a monthly revenue report by region").
- **Non-functional requirements**: quality attributes - performance, availability, security, scalability - *how well* it must do it.

**The database design lifecycle**

```
Requirements
   -> Conceptual Model      -- entities & relationships, business language, no data types
      -> Logical Model        -- fully attributed, normalized, keys defined, still platform-agnostic
         -> Physical Model      -- actual DDL: data types, indexes, partitioning, chosen for one platform
            -> Implementation     -- CREATE TABLE scripts run against a real database
               -> Testing            -- constraint tests, load tests, data-quality tests
                  -> Deployment         -- migrated into production, versioned
                     -> Monitoring        -- performance, data quality, growth tracked over time
                        -> Evolution        -- schema changes as the business changes, done safely (Part 48-49)
```

**Why poor database design compounds into real damage**

| Poor design choice | Consequence |
|---|---|
| No normalization | Data duplication -> inconsistency when only one copy gets updated |
| Missing constraints | Insert/update/delete anomalies (Part 4) - invalid or orphaned data enters silently |
| Wrong grain/keys | Poor performance, broken joins, double-counted metrics |
| No naming standard | Difficult maintenance - every new engineer re-learns table meaning by guessing |
| No governance | Security problems - nobody owns deciding who should see what |
| Ignoring analytical needs | Difficult reporting - every report re-implements the same joins/logic differently, producing conflicting numbers |
| No plan for scale | Scaling problems discovered in production, under load, when it's most expensive to fix |

A design mistake made in Part 1 gets baked into every layer built on top of it - this is why database design is worth deliberate, upfront investment rather than "we'll fix it later."

# PART 2 - DATA MODELING

**Three levels of modeling, matched to Part 1's lifecycle**

- **Conceptual model**: entities and relationships only, in business language, for stakeholders who don't write SQL.
- **Logical model**: fully attributed, keys and relationships defined, normalized, but no vendor-specific data types.
- **Physical model**: the actual implementation - real data types, indexes, constraints, for one specific database engine.

**Core building blocks**

- **Entity**: a thing the business cares about and tracks data on (Customer, Order, Product).
- **Attribute**: a property of an entity (Customer.email, Order.order_date).
- **Relationship**: how two entities are connected (a Customer *places* an Order).
- **Key**: an attribute (or set of attributes) that uniquely identifies an entity instance (Part 3).
- **Constraint**: a rule the data must obey (Part 7).
- **Cardinality**: how many instances of one entity relate to how many of another (one-to-one, one-to-many, many-to-many).
- **Optionality**: whether the relationship is mandatory or optional (must every Order have a Customer? Usually yes. Must every Customer have an Order? Usually no.).
- **Business rule**: a constraint or behavior specific to the business, not inherent to the data structure (e.g. "a loan cannot be approved if the customer's credit score is below 600").

**ER modeling vocabulary**

| Term | Meaning |
|---|---|
| Strong entity | Exists independently, has its own primary key (Customer) |
| Weak entity | Cannot exist without a related strong entity; its key includes the strong entity's key (OrderLine depends on Order) |
| Associative entity | Resolves a many-to-many relationship into two one-to-many relationships (Enrollment between Student and Course) |
| Composite attribute | An attribute made of sub-parts (Address = Street + City + PostalCode) |
| Derived attribute | Computed from other attributes, not stored independently (Age derived from DateOfBirth) |
| Multivalued attribute | An attribute that can hold multiple values for one entity instance (a Customer's multiple PhoneNumbers) - modeled as a separate table in a relational design |
| Relationship degree | How many entities participate (binary = 2, ternary = 3) |
| Relationship cardinality | 1:1, 1:N, or M:N |
| Relationship optionality | Mandatory ("must have") vs optional ("may have") participation |

**Cardinality, with realistic examples**

| Cardinality | Example | Meaning |
|---|---|---|
| One-to-one | Employee <-> EmployeeBadge | Each employee has exactly one badge, each badge belongs to exactly one employee |
| One-to-many | Customer -> Orders | One customer places many orders; each order belongs to one customer |
| Many-to-one | Employees -> Department | Many employees belong to one department (same relationship as above, viewed from the other side) |
| Many-to-many | Students <-> Courses | A student takes many courses; a course has many students - resolved with an associative entity `Enrollment` |

Further realistic examples used throughout this handbook: `Department -> Employees` (1:N), `Doctors <-> Patients` (M:N, resolved via `Appointment`/`Visit`), `Products -> Categories` (N:1, a product belongs to one category; or M:N if products can belong to multiple categories), `Users <-> Roles` (M:N, resolved via `UserRole`).

```sql
-- Associative entity resolving Students <-> Courses (M:N)
CREATE TABLE enrollment (
    student_id  INT REFERENCES students(student_id),
    course_id   INT REFERENCES courses(course_id),
    enrolled_at DATE,
    grade       CHAR(2),
    PRIMARY KEY (student_id, course_id)   -- composite key enforces one enrollment per student/course pair
);
```

# PART 3 - KEYS

| Key type | Definition |
|---|---|
| Primary key | The chosen unique identifier for a row, enforced by the database, cannot be NULL |
| Candidate key | Any column/combination that *could* serve as the primary key (uniquely identifies a row) |
| Alternate key | A candidate key not chosen as the primary key, but still worth a UNIQUE constraint |
| Foreign key | A column referencing another table's primary key, enforcing a relationship |
| Natural key | A real-world attribute used as identifier (email, national ID, SKU) |
| Surrogate key | A system-generated identifier with no business meaning (auto-increment integer, UUID) |
| Composite key | A primary key made of two or more columns together |
| Business key | Synonym for natural key, emphasizing it comes from business processes |
| Unique key | Any column/set enforced unique, whether or not it's the primary key |

**Natural key vs surrogate key - when to use each**

| | Natural key | Surrogate key |
|---|---|---|
| Example | `email`, `ISBN`, `national_id` | Auto-increment `customer_id`, UUID |
| Stability | Can change (email changes, SKUs get reissued) | Never changes once assigned |
| Meaning | Self-documenting - readable in a join result | Meaningless outside the database |
| Warehouse fit | Poor - breaks SCD Type 2 (Part 19), breaks across source-system merges | Excellent - required for SCD Type 2, safe across source systems |
| Join performance | Often wider/slower (strings, composites) | Fast (small integers) |
| Use when | Small reference/lookup tables with genuinely immutable identifiers (ISO country codes) | Almost everything else, especially any dimension/entity that will ever be updated, merged, or historized |

```sql
-- Natural key as primary key - fine here because ISO codes are stable and standardized
CREATE TABLE country (
    country_code CHAR(2) PRIMARY KEY,   -- 'NG', 'US' - natural key, unlikely to ever change
    country_name TEXT
);

-- Surrogate key preferred - customer_id from a source system could be reissued or merged
CREATE TABLE dim_customer (
    customer_key   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,   -- [PostgreSQL/SQL Server-style identity]
    customer_id    INT UNIQUE,        -- natural/business key kept as an attribute, not the PK
    customer_name  TEXT
);
```

**Identifier generation strategies**

| Strategy | Description | Trade-offs |
|---|---|---|
| Integer identity/auto-increment | Database auto-generates a sequential integer | Fast, small, but predictable and centrally generated (poor fit for distributed writes) |
| Sequence | Explicit, named, database-managed counter object; can be shared across tables | More control than identity (can pre-fetch, skip, reset) |
| UUID / GUID | 128-bit globally unique value, generated anywhere (app, multiple DB nodes) without coordination | No central bottleneck, good for distributed systems; larger (16 bytes), can fragment indexes if used as a clustering key |
| Snowflake ID | Time-ordered 64-bit ID (timestamp + machine ID + sequence), used by distributed systems (originated at Twitter) | Roughly sortable by creation time unlike UUID, compact, needs a coordinated ID-generation service |
| Hash-based identifier | Deterministic hash of business attributes (Part 39 of the SQL-for-DE handbook) | Reproducible without a lookup, good for deduplication/reconciliation, but not a good primary key (collisions possible, not sequential) |

**How key design affects the rest of the system**: small sequential integer keys make joins and indexes fast and compact; UUIDs as primary/clustering keys can cause index fragmentation on engines that physically order rows by the clustering key (random UUID insert order scatters writes across the whole index); natural keys complicate data integration (the same "customer" might have a different natural key per source system, but should map to one surrogate key in a warehouse); and in distributed systems, surrogate keys generated by a single auto-increment sequence become a write bottleneck, which is exactly why UUIDs/Snowflake IDs exist.

# PART 4 - NORMALIZATION

Normalization exists to eliminate **redundancy** and the **anomalies** it causes, by ensuring each fact is stored in exactly one place.

**The three anomalies (why normalization exists)**

- **Insert anomaly**: you cannot add a fact without also being forced to add unrelated data (can't add a new course unless a student is already enrolled in it, if course and enrollment are merged into one table).
- **Update anomaly**: the same fact is stored in multiple rows, so updating it requires finding and changing every copy - miss one and the data becomes inconsistent.
- **Delete anomaly**: deleting one fact unintentionally deletes another (deleting the only enrollment in a course also deletes all knowledge that the course exists).

**Normal forms**

| Form | Requirement | Problem it solves |
|---|---|---|
| 1NF | Atomic values, no repeating groups | Eliminates multi-valued cells (e.g. one column holding "phone1, phone2, phone3") |
| 2NF | 1NF + no partial dependency on a composite key | Eliminates attributes that depend on only *part* of a composite primary key |
| 3NF | 2NF + no transitive dependency | Eliminates attributes that depend on a *non-key* attribute rather than the key directly |
| BCNF | Every determinant is a candidate key | Closes edge cases 3NF misses when a table has multiple overlapping candidate keys |
| 4NF | No multivalued dependency | Eliminates independent multi-valued facts incorrectly combined in one table |
| 5NF | No join dependency not implied by candidate keys | Eliminates redundancy only visible when the table is decomposed into 3+ smaller tables and rejoined |

**Functional dependency vocabulary**: a **functional dependency** `A -> B` means the value of A determines the value of B. A **partial dependency** happens when a non-key attribute depends on only part of a composite key. A **transitive dependency** happens when a non-key attribute depends on another non-key attribute rather than on the key itself. A **multivalued dependency** happens when one key determines a *set* of values for an attribute, independently of another such set in the same table. A **join dependency** is the general case 4NF/5NF address - the table can be losslessly decomposed and must be, to remove all redundancy.

**Worked example: unnormalized -> 3NF**

```
Unnormalized order table (repeating groups, mixed grain):

order_id | customer_name | customer_email | product1  | product2 | salesperson | dept
1001     | John Doe      | john@test.com  | Laptop    | Mouse    | Amaka       | Sales
```

```sql
-- Step 1 -> 1NF: eliminate repeating groups (product1, product2), one row per order line
CREATE TABLE order_line_1nf (
    order_id INT, customer_name TEXT, customer_email TEXT,
    product TEXT, salesperson TEXT, dept TEXT
);
-- Now (1001, John Doe, john@test.com, Laptop, Amaka, Sales) and
--     (1001, John Doe, john@test.com, Mouse,  Amaka, Sales) are separate rows.

-- Step 2 -> 2NF: assume composite key (order_id, product); customer_name/email/salesperson/dept
-- depend only on order_id (a partial dependency) - split them out
CREATE TABLE orders_2nf (order_id INT PRIMARY KEY, customer_name TEXT, customer_email TEXT, salesperson TEXT, dept TEXT);
CREATE TABLE order_lines_2nf (order_id INT REFERENCES orders_2nf(order_id), product TEXT, PRIMARY KEY (order_id, product));

-- Step 3 -> 3NF: dept depends on salesperson, not on order_id directly (a transitive dependency) - split again
CREATE TABLE salespeople_3nf (salesperson TEXT PRIMARY KEY, dept TEXT);
CREATE TABLE orders_3nf (
    order_id INT PRIMARY KEY, customer_name TEXT, customer_email TEXT,
    salesperson TEXT REFERENCES salespeople_3nf(salesperson)
);
CREATE TABLE order_lines_3nf (order_id INT REFERENCES orders_3nf(order_id), product TEXT, PRIMARY KEY (order_id, product));
```

Now: a salesperson's department is stored exactly once (fixing update anomalies), a new salesperson can be added before they have any orders (fixing insert anomalies), and deleting an order no longer risks deleting knowledge of a salesperson or their department (fixing delete anomalies). Real-world use: this is exactly the discipline behind a normalized OLTP schema (Part 6).

# PART 5 - DENORMALIZATION

**Denormalization** deliberately reintroduces redundancy to reduce the number of joins a query needs, trading write-side consistency risk for read-side speed and simplicity.

**When to denormalize**: read-heavy analytical workloads where join cost dominates query time; reporting tables that are rebuilt wholesale on a schedule (so the "risk" of stale redundant copies is bounded and predictable); pre-computed aggregates that are expensive to recompute per query. **When NOT to denormalize**: OLTP systems with frequent concurrent writes (redundant copies multiply the chance of inconsistency); any table where the business genuinely needs the write-time guarantee that a fact exists in exactly one place (financial ledgers, regulatory records).

```sql
-- NORMALIZED (OLTP): customer's region looked up via a join every time
SELECT c.customer_name, r.region_name, SUM(o.amount)
FROM orders o
JOIN customers c ON c.customer_id = o.customer_id
JOIN regions r ON r.region_id = c.region_id
GROUP BY c.customer_name, r.region_name;

-- DENORMALIZED (reporting table): region_name copied directly onto the reporting row
CREATE TABLE reporting.sales_flat AS
SELECT o.order_id, c.customer_name, r.region_name, o.amount
FROM orders o
JOIN customers c ON c.customer_id = o.customer_id
JOIN regions r ON r.region_id = c.region_id;
-- Every downstream report queries sales_flat directly - one table, no joins, faster and simpler.
```

**Denormalization techniques**: precomputed columns (store a calculated value instead of recomputing it every query), summary/aggregation tables (pre-aggregated rollups, e.g. `daily_sales_summary`), redundant attributes (copy a frequently-joined attribute directly onto the fact row, as above), and materialized views (a query result physically stored and refreshed on a schedule - the database-native version of a denormalized summary table).

**The guideline, and its limit**: "normalize OLTP, denormalize analytics" is a useful default, not a law - some OLTP systems denormalize specific hot-read paths for performance (e.g. caching an order's current total on the order row instead of always summing line items), and some analytical models stay substantially normalized (Data Vault, Part 34) when auditability matters more than query simplicity. Always denormalize deliberately, for a stated reason, not by default.

# PART 6 - OLTP DATABASE DESIGN

**OLTP** (Online Transaction Processing) systems run the day-to-day business: many short, concurrent transactions, each touching a small number of rows, requiring strict correctness. **ACID** properties are the contract that makes this safe: **Atomicity** (a transaction fully happens or not at all), **Consistency** (a transaction moves the database from one valid state to another, per its constraints), **Isolation** (concurrent transactions don't see each other's uncommitted changes), **Durability** (once committed, a transaction survives a crash). OLTP schemas are normalized (Part 4) because the priority is write correctness and storage efficiency, not read simplicity.

**Ten OLTP designs (condensed - full DDL for two, entity/relationship summary for the rest to keep this section navigable; each follows the identical design discipline)**

**1. E-commerce** - core entities: `customers`, `products`, `categories`, `orders`, `order_items`, `payments`, `shipments`.

```sql
CREATE TABLE customers (
    customer_id   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    full_name     TEXT NOT NULL,
    email         TEXT UNIQUE NOT NULL,
    created_at    TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
CREATE TABLE categories (
    category_id   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    category_name TEXT NOT NULL
);
CREATE TABLE products (
    product_id    INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    category_id   INT REFERENCES categories(category_id),
    product_name  TEXT NOT NULL,
    unit_price    NUMERIC(10,2) CHECK (unit_price >= 0)
);
CREATE TABLE orders (
    order_id      INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_id   INT NOT NULL REFERENCES customers(customer_id),
    order_date    DATE NOT NULL DEFAULT CURRENT_DATE,
    status        TEXT CHECK (status IN ('pending','paid','shipped','cancelled'))
);
CREATE TABLE order_items (
    order_id      INT REFERENCES orders(order_id),
    product_id    INT REFERENCES products(product_id),
    quantity      INT CHECK (quantity > 0),
    unit_price    NUMERIC(10,2),
    PRIMARY KEY (order_id, product_id)
);
CREATE TABLE payments (
    payment_id    INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    order_id      INT REFERENCES orders(order_id),
    amount        NUMERIC(10,2) CHECK (amount >= 0),
    paid_at       TIMESTAMP
);

-- Sample transaction: place an order atomically
BEGIN;
INSERT INTO orders (customer_id, status) VALUES (101, 'pending') RETURNING order_id;
INSERT INTO order_items (order_id, product_id, quantity, unit_price) VALUES (5001, 42, 2, 25.00);
COMMIT;   -- either both inserts succeed, or neither does (Atomicity)
```

**2. Banking** - entities: `customers`, `accounts`, `transactions`, `branches`, `employees`. Key relationship: one customer can hold many accounts (1:N); each transaction belongs to exactly one account (N:1). Critical constraint: `transactions.amount` combined with a `transaction_type` CHECK, and account balance updates wrapped in a transaction with `SELECT ... FOR UPDATE` row locking to prevent race conditions on concurrent withdrawals.

```sql
CREATE TABLE accounts (
    account_id   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_id  INT REFERENCES customers(customer_id),
    account_type TEXT CHECK (account_type IN ('checking','savings')),
    balance      NUMERIC(14,2) NOT NULL DEFAULT 0
);
CREATE TABLE transactions (
    transaction_id INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    account_id     INT REFERENCES accounts(account_id),
    txn_type       TEXT CHECK (txn_type IN ('deposit','withdrawal')),
    amount         NUMERIC(14,2) CHECK (amount > 0),
    txn_time       TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

Hospital, University, Telecom, Payroll, Logistics, Inventory, Airline booking, Hotel booking** - each follows the identical discipline demonstrated in full above: identify entities -> define cardinality -> choose surrogate keys -> normalize to 3NF -> add constraints -> index foreign keys. Core entity sets: **Hospital** (`patients`, `doctors`, `appointments`, `diagnoses`, `prescriptions`); **University** (`students`, `courses`, `enrollments`, `instructors`, `departments`); **Telecom** (`subscribers`, `plans`, `calls`, `sms`, `data_usage`); **Payroll** (`employees`, `salaries`, `payslips`, `deductions`, `tax_bands`); **Logistics** (`shipments`, `vehicles`, `drivers`, `routes`, `warehouses`); **Inventory** (`products`, `warehouses`, `stock_levels`, `stock_movements`, `suppliers`); **Airline booking** (`passengers`, `flights`, `bookings`, `seats`, `fares`); **Hotel booking** (`guests`, `rooms`, `reservations`, `room_types`, `rates`). # PART 7 - DATABASE CONSTRAINTS

```sql
CREATE TABLE employees (
    employee_id   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,   -- entity integrity: unique, non-null row identifier
    department_id INT REFERENCES departments(department_id),       -- referential integrity: must exist in departments
    email         TEXT UNIQUE NOT NULL,                             -- domain integrity: no two employees share an email
    salary        NUMERIC(10,2) CHECK (salary >= 0),                 -- domain integrity: business rule on value range
    status        TEXT DEFAULT 'active'                               -- default: sensible value when not specified
);
```

**Integrity categories**: **entity integrity** (every table has a primary key, so every row is uniquely identifiable), **domain integrity** (every column's values conform to its allowed type/range/format - `CHECK`, `NOT NULL`, data type itself), **referential integrity** (foreign keys always point to an existing row, enforced by the database rather than trusted to application code).

**Cascading actions**

```sql
ALTER TABLE order_items
    ADD CONSTRAINT fk_order
    FOREIGN KEY (order_id) REFERENCES orders(order_id)
    ON DELETE CASCADE;      -- deleting an order deletes its line items automatically

ALTER TABLE employees
    ADD CONSTRAINT fk_manager
    FOREIGN KEY (manager_id) REFERENCES employees(employee_id)
    ON DELETE SET NULL;     -- deleting a manager leaves their reports' manager_id as NULL, not deleted

ALTER TABLE payments
    ADD CONSTRAINT fk_order_payment
    FOREIGN KEY (order_id) REFERENCES orders(order_id)
    ON DELETE RESTRICT;     -- deleting an order with existing payments is blocked outright

ALTER TABLE order_items
    ADD CONSTRAINT fk_product
    FOREIGN KEY (product_id) REFERENCES products(product_id)
    ON UPDATE CASCADE;      -- if a product_id ever changes, dependent rows update automatically
```

**When cascading is dangerous**: `ON DELETE CASCADE` on a deeply nested hierarchy can silently wipe out far more data than intended (deleting one customer cascades through orders -> order_items -> shipments -> ...), and in a live production system a single mistaken `DELETE` on a parent row can be catastrophic and hard to undo. Prefer `RESTRICT` (forces an explicit, deliberate decision) or soft deletes (Part 42 of the SQL-for-DE handbook) for any entity with financial, legal, or audit significance, and reserve `CASCADE` for genuinely dependent, low-stakes child data (e.g. deleting a shopping cart deletes its cart items).

# PART 8 - DATABASE SCHEMA DESIGN

**Naming conventions**: pick `snake_case` (most portable across case-sensitive Linux-hosted databases) or a single consistent casing and never mix them; decide singular vs plural table names once (`customer` vs `customers`) and apply it everywhere - plural is the more common modern convention because a table is a *collection* of rows; prefix or suffix foreign keys to make relationships obvious at a glance (`customer_id` on the `orders` table, not just `id`); avoid reserved words as column names (`order`, `user`, `group`).

**Standard audit columns** - add these to almost every operational table:

```sql
CREATE TABLE customers (
    customer_id  INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    full_name    TEXT NOT NULL,
    created_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP,   -- when the row was first inserted
    updated_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP,   -- when the row was last modified
    created_by   TEXT,                                    -- who/what inserted it (user or service account)
    updated_by   TEXT,                                    -- who/what last modified it
    is_deleted   BOOLEAN DEFAULT FALSE,                    -- soft-delete flag
    status       TEXT DEFAULT 'active',                     -- business-state column
    version      INT DEFAULT 1                               -- optimistic concurrency control counter
);
```

`version` supports **optimistic locking**: an `UPDATE ... WHERE customer_id = ? AND version = ?` that also does `SET version = version + 1` fails (updates zero rows) if someone else modified the row in between - the application detects this and can retry or alert, without needing a database-level lock held for the whole read-then-write cycle.

**Data dictionary template** - the single most valuable, most-skipped piece of schema documentation:

| Column | Data Type | Description | Nullable | Default | Key | Business Meaning | Source System |
|---|---|---|---|---|---|---|---|
| `customer_id` | INT | Surrogate key | No | identity | PK | Unique internal identifier | Generated |
| `email` | TEXT | Customer's email | No | - | Alternate (UNIQUE) | Primary contact + login | CRM |
| `status` | TEXT | Account status | No | 'active' | - | 'active','suspended','closed' | CRM |
| `created_at` | TIMESTAMP | Row insertion time | No | now() | - | Audit trail | System-generated |

Maintaining this table (even as a living spreadsheet or a dbt `schema.yml`, Part 43) is what separates a schema a *new* engineer can safely work in from one only its original author understands.

# PART 9 - TEMPORAL DATABASE DESIGN

Business facts change over time, and a design must decide explicitly whether it tracks only the *current* state or the *full history*.

**Vocabulary**: **effective date** / **valid-from** - when a fact became true; **expiration date** / **valid-to** - when it stopped being true (or a sentinel like `9999-12-31` while still current); **transaction time** - when the database recorded the fact (may lag behind when it actually became true - Part 26); **event time** - when the fact actually happened in the real world; **processing time** - when your system processed it.

```sql
-- CURRENT-STATE table: only the latest value, history is lost
CREATE TABLE customer_current (
    customer_id INT PRIMARY KEY,
    city        TEXT
);
UPDATE customer_current SET city = 'Abuja' WHERE customer_id = 101;   -- 'Lagos' is now gone forever

-- HISTORICAL / TEMPORAL table: every version preserved with a validity window
CREATE TABLE customer_history (
    customer_id  INT,
    city         TEXT,
    valid_from   DATE,
    valid_to     DATE,          -- NULL or '9999-12-31' while current
    is_current   BOOLEAN
);
-- Lagos: valid_from=2020-01-01, valid_to=2026-02-06, is_current=FALSE
-- Abuja: valid_from=2026-02-07, valid_to=NULL,       is_current=TRUE

-- AUDIT table: every change event logged, independent of the "current" table's shape
CREATE TABLE customer_audit_log (
    audit_id     BIGSERIAL PRIMARY KEY,
    customer_id  INT,
    field_changed TEXT,
    old_value     TEXT,
    new_value      TEXT,
    changed_at      TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    changed_by       TEXT
);
```

Applied examples: **customer address changes** and **customer status changes** (active/suspended/closed) are natural fits for `customer_history`-style validity windows; **employee salary changes** typically need both a history table (for "what was their salary on this date" reporting) *and* an audit log (for "who approved this raise and when," a compliance question a validity-window table alone doesn't answer); **product price changes** need history primarily so that past orders can still report the price that was actually charged at the time, rather than today's price. This temporal-table pattern is the OLTP-side foundation that SCD Type 2 (Part 19) builds on in the warehouse.

# PART 11 - OLAP

**OLAP** (Online Analytical Processing) is the class of workload data warehouses are built for: aggregating large historical volumes to answer "how much / how many / how has this changed" questions, rather than "what is the current value of this one record."

| OLAP variant | Storage | Description |
|---|---|---|
| MOLAP (Multidimensional OLAP) | Proprietary multidimensional cube | Pre-aggregated, very fast for known query patterns, less flexible, cube must be rebuilt on schema change |
| ROLAP (Relational OLAP) | Standard relational tables (star/snowflake schema) | Flexible, scales with modern warehouses, aggregation computed at query time (or via materialized views) |
| HOLAP (Hybrid OLAP) | Mix - detail in relational tables, summaries in a cube | Balances flexibility and speed; less common now that columnar warehouses make ROLAP fast enough |

Modern cloud warehouses (Part 36) are overwhelmingly ROLAP - a well-designed star schema on a columnar engine now performs competitively with traditional MOLAP cubes, without the rigidity.

**Core OLAP operations**

| Operation | Meaning | Example |
|---|---|---|
| Aggregation | Summarizing many rows into one number | `SUM(sales_amount)` |
| Drill-down | Moving from summary to more detail | Year -> Quarter -> Month -> Day |
| Roll-up | Moving from detail to more summary | Day -> Month -> Year |
| Slice | Fixing one dimension to a single value | "Show sales for Region = 'West' only" |
| Dice | Fixing multiple dimensions to sub-ranges | "Show sales for Region IN ('West','East') AND Quarter = 'Q1'" |
| Pivot | Rotating dimensions between rows and columns | Turning `month` values into columns, one per month |

```sql
-- Drill-down: from yearly to monthly, same measure, finer date grain
SELECT d.year, SUM(f.sales_amount) FROM fact_sales f JOIN dim_date d ON d.date_key=f.date_key GROUP BY d.year;
SELECT d.year, d.month_name, SUM(f.sales_amount) FROM fact_sales f JOIN dim_date d ON d.date_key=f.date_key GROUP BY d.year, d.month_name;

-- Slice: fix Region = 'West'
SELECT d.month_name, SUM(f.sales_amount)
FROM fact_sales f JOIN dim_date d ON d.date_key=f.date_key JOIN dim_store s ON s.store_key=f.store_key
WHERE s.region = 'West'
GROUP BY d.month_name;

-- Pivot: months as columns (portable CASE-based approach)
SELECT s.region,
    SUM(CASE WHEN d.month_name='January' THEN f.sales_amount END) AS jan,
    SUM(CASE WHEN d.month_name='February' THEN f.sales_amount END) AS feb
FROM fact_sales f JOIN dim_date d ON d.date_key=f.date_key JOIN dim_store s ON s.store_key=f.store_key
GROUP BY s.region;
```

# PART 12 - DATA WAREHOUSE ARCHITECTURE

```
Source Systems (OLTP DBs, APIs, files, SaaS apps, ERP, CRM, event streams)
   -> Ingestion            -- bulk load, API pull, CDC capture, streaming consumer
      -> Landing Zone         -- raw files/events exactly as received, no transformation
         -> Raw Layer            -- structured copy of landing data, typed but untransformed
            -> Staging Layer        -- cleaned, standardized, still 1:1 with source rows
               -> Integration Layer    -- multiple sources conformed into unified entities
                  -> Core Warehouse       -- dimensional model: conformed facts and dimensions
                     -> Data Marts           -- domain-specific subsets/aggregates
                        -> Semantic Layer        -- business-friendly metric/dimension definitions
                           -> BI / Analytics / ML / AI   -- dashboards, notebooks, models, applications
```

Each layer exists to isolate a specific kind of change: a source-system schema change only affects Raw/Staging; a new integration rule only affects the Integration Layer; a new metric definition only affects the Semantic Layer - without this separation, every change ripples unpredictably through the whole stack.

**Source system categories**: **operational databases** (the OLTP systems from Part 6), **APIs** (SaaS platforms exposing REST/GraphQL endpoints - Stripe, Salesforce), **files** (CSV/Excel exports, partner data drops), **SaaS applications** (HR systems, marketing platforms, support desks), **ERP** (enterprise resource planning - SAP, Oracle ERP - usually large, complex, many interrelated tables), **CRM** (customer relationship management - Salesforce, HubSpot), **databases** (direct DB connections to other systems), and **event streams** (Kafka, Kinesis - continuous behavioral/telemetry data).

# PART 13 - ETL AND ELT

**ETL**: Extract -> Transform -> Load. Transformation happens in a separate processing engine *before* data reaches its final destination.
**ELT**: Extract -> Load -> Transform. Raw data is loaded first; transformation happens *inside* the target warehouse using its own compute (SQL).

| | ETL | ELT |
|---|---|---|
| Advantage | Target only ever holds clean data; works with limited-compute targets | Raw history always retrievable; leverages warehouse's elastic compute; simpler pipeline (fewer moving parts) |
| Disadvantage | Requires a separate transformation engine/server; raw history often not retained | Warehouse holds (and is paying to store) raw, possibly messy data |
| Best fit | Traditional on-prem warehouses, compute-constrained targets, strict pre-load data governance requirements | Modern cloud warehouses (Part 36) with decoupled, elastic, pay-per-use compute |

```sql
-- ETL: transform happens before load (conceptually, in an external Spark/Informatica job) -
-- the warehouse only ever receives this already-clean statement's *result*
-- (shown here as the transform step that would run externally, then loaded as a finished table)

-- ELT: raw JSON loaded as-is, transformed afterward with plain SQL inside the warehouse
COPY raw.orders_raw FROM 's3://bucket/orders.json' FORMAT AS JSON 'auto';   -- [Redshift] load raw, untransformed

CREATE TABLE warehouse.fact_orders AS               -- transform AFTER load, using warehouse compute
SELECT (payload:order_id)::INT AS order_id, (payload:amount)::NUMERIC AS amount
FROM raw.orders_raw;                                 -- [Snowflake]-style JSON dot notation
```

Modern data platforms default to ELT because storage and compute are decoupled and elastically priced (Part 36) - it is cheap to keep everything raw and only pay for transformation compute when you actually run it, and any transformation bug can be fixed and cheaply re-run against untouched raw history.

# PART 14 - DIMENSIONAL MODELING

**Facts**: numeric, measurable events (a sale, a payment, a shipment). **Dimensions**: the descriptive context around a fact (who, what, where, when). **Measures**: the specific numeric columns on a fact table (`sales_amount`, `quantity`). **Attributes**: descriptive columns on a dimension (`customer_name`, `product_category`). **Hierarchies**: natural drill-down paths within a dimension (Year -> Quarter -> Month -> Day within `dim_date`; Country -> Region -> City within `dim_geography`).

**Grain is the single most important decision in dimensional modeling** - it is what one row in the fact table represents. Every fact table design in this handbook, and every one you build professionally, must start with an explicit sentence:

> "The grain of this fact table is **one row per [X]**."

For example: "The grain of `FactSales` is one row per order line item." Once grain is fixed, every dimension the fact table joins to, and every measure it carries, must be consistent with that grain - a measure that doesn't naturally exist at that grain (e.g. an order-level shipping fee, if the grain is order line item) must either be allocated down to the grain or left out and handled at a different grain in a separate fact table. Getting grain wrong is the root cause of the majority of real-world warehouse bugs: double-counted revenue, incorrect averages, and joins that silently multiply rows.

# PART 15 - STAR SCHEMA

```
                     DimCustomer
                          |
     DimDate  --------  FactSales  --------  DimProduct
                          |
                      DimStore
```

A **star schema** has one central fact table directly connected to fully denormalized dimension tables - no dimension joins to another dimension. This shape minimizes the number of joins any analytical query needs and is the default, recommended shape for the vast majority of warehouse models.

```sql
-- The grain of FactSales is one row per order line item.

CREATE TABLE dim_customer (
    customer_key    INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,   -- surrogate key
    customer_id     INT UNIQUE,                                      -- natural/business key
    customer_name   TEXT,
    segment         TEXT,
    region          TEXT
);

CREATE TABLE dim_product (
    product_key     INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    product_id      INT UNIQUE,
    product_name    TEXT,
    category        TEXT,
    unit_cost       NUMERIC(10,2)
);

CREATE TABLE dim_date (
    date_key        INT PRIMARY KEY,             -- e.g. 20260207, see Part 20
    full_date       DATE,
    month_name      TEXT,
    quarter         INT,
    year            INT
);

CREATE TABLE dim_store (
    store_key       INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    store_id        INT UNIQUE,
    store_name      TEXT,
    city            TEXT,
    region          TEXT
);

CREATE TABLE fact_sales (
    sale_key        BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_key    INT REFERENCES dim_customer(customer_key),
    product_key     INT REFERENCES dim_product(product_key),
    date_key        INT REFERENCES dim_date(date_key),
    store_key       INT REFERENCES dim_store(store_key),
    quantity        INT,
    sales_amount    NUMERIC(12,2),
    cost_amount     NUMERIC(12,2)
);
```

Every column ending in `_key` is a surrogate key (Part 3, Part 23); `sales_amount`/`cost_amount`/`quantity` are the measures - the only columns a BI tool will typically `SUM()`/`AVG()`; everything else in `fact_sales` is a foreign key. This is the shape a BI tool's semantic layer (Part 42) is built to consume directly, with minimal or no further joins needed.

# PART 16 - SNOWFLAKE SCHEMA

A **snowflake schema** takes a star schema's dimensions and further normalizes them into sub-dimensions - e.g. splitting `dim_product`'s `category` column out into a separate `dim_category` table that `dim_product` references.

```sql
-- Snowflaked: dim_product normalized further into dim_category
CREATE TABLE dim_category (
    category_key  INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    category_name TEXT
);
CREATE TABLE dim_product_snowflaked (
    product_key   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    product_id    INT UNIQUE,
    product_name  TEXT,
    category_key  INT REFERENCES dim_category(category_key)   -- one extra join vs the star version
);
```

| | Star Schema | Snowflake Schema |
|---|---|---|
| Dimension storage | Denormalized (some redundancy) | Normalized (sub-dimensions) |
| Storage used | Slightly more | Slightly less |
| Query complexity | Fewer joins, simpler SQL | More joins, more complex SQL |
| Query performance | Generally faster on modern columnar engines | Can be slower due to extra joins, though negligible on some engines |
| Maintainability | A category rename touches denormalized rows in `dim_product` | A category rename touches exactly one row in `dim_category` |

**When to use each**: default to star schema for BI-facing marts - the storage saved by snowflaking is trivial on modern columnar/compressed storage, and the extra joins cost real query complexity and (on some engines) real performance for no meaningful benefit. Consider snowflaking only when a sub-dimension is large, changes independently and frequently, and is shared across many parent dimensions in a way that denormalized redundancy would make genuinely expensive or error-prone to maintain.

# PART 17 - FACT TABLE TYPES

| Type | Grain | Update pattern | Example |
|---|---|---|---|
| Transaction fact | One row per business event | Insert-only, immutable once written | Sales transactions - one row per sale |
| Periodic snapshot fact | One row per entity per fixed time interval | Insert-only, new snapshot row each period | Daily account balance - one row per account per day |
| Accumulating snapshot fact | One row per process instance | Updated in place as the process progresses | Order fulfillment lifecycle - one row per order, columns for each stage's date |
| Factless fact table | One row per co-occurrence, no numeric measure | Insert-only | Student attendance - one row per student per class per day, the "measure" is just that it happened |

```sql
-- Transaction fact
CREATE TABLE fact_sales_txn (sale_key BIGSERIAL PRIMARY KEY, date_key INT, customer_key INT, amount NUMERIC(12,2));

-- Periodic snapshot fact
CREATE TABLE fact_account_balance_daily (
    date_key INT, account_key INT, ending_balance NUMERIC(14,2),
    PRIMARY KEY (date_key, account_key)
);

-- Accumulating snapshot fact - one row per order, updated as it moves through stages
CREATE TABLE fact_order_fulfillment (
    order_key            BIGINT PRIMARY KEY,
    ordered_date_key      INT,
    paid_date_key          INT,
    shipped_date_key         INT,
    delivered_date_key         INT,
    days_order_to_delivery       INT   -- recomputed each time a new stage completes
);

-- Factless fact table
CREATE TABLE fact_attendance (
    date_key INT, student_key INT, class_key INT
    -- no measure column - COUNT(*) is the metric ("how many attendance events")
);
```

# PART 18 - DIMENSION TYPES

| Dimension type | Definition | Example |
|---|---|---|
| Conformed dimension | Shared, identically defined, across multiple fact tables/marts | `dim_customer` used by both `fact_sales` and `fact_support_tickets` |
| Role-playing dimension | One physical dimension used multiple times with different meanings | `dim_date` joined as both `order_date` and `ship_date` |
| Degenerate dimension | A dimension-like attribute with no separate table, stored directly on the fact | `order_number` on `fact_sales` - has no other attributes worth a separate table |
| Junk dimension | Bundles several low-cardinality flags into one small dimension | A dimension combining `is_gift_wrapped`, `payment_method`, `is_expedited` |
| Mini-dimension | A frequently-changing subset of a large dimension's attributes, split out to avoid excessive SCD Type 2 row growth | Splitting `customer_income_band`/`customer_age_band` out of `dim_customer` |
| Outrigger dimension | A dimension that itself references another dimension (a controlled exception to the pure star shape) | `dim_product` referencing `dim_supplier` |
| Slowly Changing Dimension | A dimension whose attribute values change over time, handled per Part 19 | `dim_customer.city` changing |
| Inferred dimension | A placeholder dimension row created because a fact arrived before its dimension record | An order for a customer not yet in `dim_customer` gets a stub row inserted immediately |
| Late-arriving dimension | The eventual "real" dimension record that arrives after an inferred placeholder was already used | The real `dim_customer` details arriving the next day, updating the stub |

```sql
-- Junk dimension example
CREATE TABLE dim_order_flags (
    flag_key           INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    is_gift_wrapped     BOOLEAN,
    payment_method       TEXT,
    is_expedited            BOOLEAN
);
-- fact_sales references flag_key instead of carrying 3 separate low-cardinality columns directly.

-- Inferred/late-arriving dimension pattern
INSERT INTO dim_customer (customer_id, customer_name, is_inferred)
SELECT 9999, 'UNKNOWN', TRUE
WHERE NOT EXISTS (SELECT 1 FROM dim_customer WHERE customer_id = 9999);
-- fact_sales can now safely reference customer_key for customer_id 9999 immediately.
-- When the real customer record arrives later, UPDATE the stub row in place (Type 1 style)
-- rather than inserting a second row, so existing fact references remain valid.
```

# PART 19 - SLOWLY CHANGING DIMENSIONS

| Type | Behavior | History kept? |
|---|---|---|
| Type 0 | Never changes - fixed attribute | N/A (immutable) |
| Type 1 | Overwrite the old value | No |
| Type 2 | Insert a new versioned row, keep the old one | Yes, full history |
| Type 3 | Add a "previous value" column, limited history | Partial (usually just one prior value) |
| Type 4 | Split into a current table and a separate full-history table | Yes, in a separate table |
| Type 6 | Hybrid: combine Types 1 + 2 + 3 in one dimension | Yes, plus an easy "current value" column even on historical rows |

**Example: customer changes city, Lagos -> Abuja**

```sql
-- TYPE 1: overwrite - simple, but "customer was in Lagos" is now unrecoverable
UPDATE dim_customer SET city = 'Abuja' WHERE customer_id = 101;

-- TYPE 2: full history via new row + surrogate key + validity window
CREATE TABLE dim_customer_scd2 (
    customer_key    INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,  -- unique per VERSION
    customer_id     INT,                                            -- business key, repeats across versions
    city            TEXT,
    effective_date  DATE,
    expiry_date     DATE,          -- '9999-12-31' or NULL while current
    is_current      BOOLEAN
);

-- Step 1: close the old version
UPDATE dim_customer_scd2
SET expiry_date = CURRENT_DATE - 1, is_current = FALSE
WHERE customer_id = 101 AND is_current = TRUE;

-- Step 2: insert the new version
INSERT INTO dim_customer_scd2 (customer_id, city, effective_date, expiry_date, is_current)
VALUES (101, 'Abuja', CURRENT_DATE, DATE '9999-12-31', TRUE);

-- TYPE 3: one extra column for the immediately-prior value only
ALTER TABLE dim_customer ADD COLUMN previous_city TEXT;
UPDATE dim_customer SET previous_city = city, city = 'Abuja' WHERE customer_id = 101;
```

`customer_key` is what every `fact_sales` row references - it pins a fact to the *exact version* of the customer that was true when the sale happened, which is precisely why Type 2 preserves historical accuracy that Type 1 permanently destroys (a sale made while the customer lived in Lagos stays correctly attributed to "Lagos" forever, even after the customer moves). Use Type 1 for corrections/attributes with no analytical value in tracking history (fixing a typo in a name); use Type 2 for attributes analysts genuinely need to slice historical facts by (region, segment, tier); use Type 3 only for the rare case where exactly one prior value matters and full history doesn't (e.g. "previous vs current sales rep" for hand-off reporting).

# PART 20 - DATE DIMENSION

```sql
CREATE TABLE dim_date (
    date_key       INT PRIMARY KEY,        -- e.g. 20260207
    full_date      DATE,
    day            INT,
    day_name       TEXT,
    day_of_week    INT,
    week           INT,
    month          INT,
    month_name     TEXT,
    quarter        INT,
    year           INT,
    fiscal_year    INT,
    fiscal_quarter INT,
    is_weekend     BOOLEAN,
    is_holiday     BOOLEAN
);

-- [PostgreSQL] generate a full calendar year
INSERT INTO dim_date
SELECT
    TO_CHAR(d,'YYYYMMDD')::INT, d,
    EXTRACT(DAY FROM d)::INT, TO_CHAR(d,'Day'), EXTRACT(ISODOW FROM d)::INT,
    EXTRACT(WEEK FROM d)::INT, EXTRACT(MONTH FROM d)::INT, TO_CHAR(d,'Month'),
    EXTRACT(QUARTER FROM d)::INT, EXTRACT(YEAR FROM d)::INT,
    CASE WHEN EXTRACT(MONTH FROM d) >= 4 THEN EXTRACT(YEAR FROM d)::INT ELSE EXTRACT(YEAR FROM d)::INT - 1 END,  -- example fiscal year starting April
    EXTRACT(QUARTER FROM d + INTERVAL '9 months')::INT,
    EXTRACT(ISODOW FROM d) IN (6,7),
    FALSE   -- is_holiday populated separately from a holiday reference list
FROM generate_series('2026-01-01'::DATE, '2026-12-31'::DATE, '1 day') AS d;
```

**Why a pre-built date dimension beats computing date attributes live in every report**: business calendars (fiscal periods, holidays, "is this a trading day") are genuinely complex and organization-specific - encoding that logic once in a table beats re-deriving it with SQL date functions in every single report, which inevitably drifts inconsistent across dashboards; joining on an integer `date_key` is also faster and more uniform than calling date functions per row; and BI tools build cleaner date hierarchies (Year > Quarter > Month > Day) from an explicit dimension than from a raw `DATE` column.

# PART 21 - TIME DIMENSION

A **time dimension** models intra-day time separately from the date dimension, needed whenever analysis operates below daily granularity.

```sql
CREATE TABLE dim_time (
    time_key      INT PRIMARY KEY,     -- e.g. 143000 for 14:30:00
    hour          INT,
    minute        INT,
    second        INT,
    shift         TEXT,                -- 'Morning','Afternoon','Night'
    is_business_hours BOOLEAN,
    is_peak       BOOLEAN
);

INSERT INTO dim_time
SELECT
    EXTRACT(HOUR FROM t)::INT*10000 + EXTRACT(MINUTE FROM t)::INT*100 + EXTRACT(SECOND FROM t)::INT,
    EXTRACT(HOUR FROM t)::INT, EXTRACT(MINUTE FROM t)::INT, EXTRACT(SECOND FROM t)::INT,
    CASE WHEN EXTRACT(HOUR FROM t) < 12 THEN 'Morning'
         WHEN EXTRACT(HOUR FROM t) < 18 THEN 'Afternoon' ELSE 'Night' END,
    EXTRACT(HOUR FROM t) BETWEEN 9 AND 17,
    EXTRACT(HOUR FROM t) BETWEEN 17 AND 20        -- example peak window
FROM generate_series('2026-01-01 00:00:00'::TIMESTAMP, '2026-01-01 23:59:00'::TIMESTAMP, '1 minute') AS t;
```

**Use cases**: **telecom** (call volume by hour to plan network capacity), **call centers** (staffing by shift and peak/off-peak demand), **retail** (footfall and sales by hour to plan staffing), **manufacturing** (production-line throughput by shift). A fact table joins both `dim_date` (for the day) and `dim_time` (for the time-of-day) via two separate foreign keys when both granularities matter - keeping them as separate dimensions (rather than one giant per-second date-time dimension) keeps `dim_date` reusable at its natural daily grain while still supporting intra-day analysis where needed.

# PART 22 - BRIDGE TABLES

A **bridge table** resolves a many-to-many relationship between a fact and a dimension - something a pure star schema (one FK per dimension) cannot represent directly.

```sql
-- Customer <-> Multiple Accounts (a fact naturally at "transaction" grain needs to
-- attribute one transaction to potentially several account-holders)
CREATE TABLE bridge_customer_account (
    customer_key INT REFERENCES dim_customer(customer_key),
    account_key  INT REFERENCES dim_account(account_key),
    weighting_factor NUMERIC(4,3)   -- e.g. 0.5 if two customers jointly own one account
);

-- Product <-> Multiple Categories
CREATE TABLE bridge_product_category (
    product_key  INT REFERENCES dim_product(product_key),
    category_key INT REFERENCES dim_category(category_key)
);
```

**Weighting factors** matter when a measure must be *allocated* rather than duplicated across the bridge - e.g. if a joint bank account's balance is reported once per co-owner without weighting, `SUM(balance)` across customers double-counts the actual money; multiplying by `weighting_factor` (0.5 for two equal co-owners) keeps the aggregate correct. Other realistic examples: **Patient <-> Multiple Doctors** (a patient under several specialists - a bridge avoids forcing one "primary doctor" FK that loses information), **Customer <-> Multiple Segments** (marketing segments overlap - a customer can be both "high value" and "at risk" simultaneously).

# PART 23 - DATA WAREHOUSE KEYS

| Key type in a warehouse | Definition |
|---|---|
| Surrogate key | Warehouse-generated integer identity, unique per dimension row *version* (Part 3, Part 19) |
| Natural key | The source system's original identifier, retained as an attribute for traceability |
| Business key | Synonym for natural key in warehouse contexts |
| Foreign key | Links a fact row to a dimension row via the dimension's surrogate key |
| Durable key | A key that stays stable across dimension *versions* for the same real-world entity - typically the natural/business key, used to trace "all versions of this customer" across SCD Type 2 rows |

Warehouse dimensions overwhelmingly use surrogate keys because: (1) SCD Type 2 requires multiple rows for the same real-world entity, which is impossible if the natural key is the primary key (it would violate uniqueness); (2) surrogate keys decouple the warehouse from any single source system's ID scheme, which matters the moment two source systems need to be merged into one conformed dimension; (3) small integer surrogate keys join and index faster than wide natural keys. The `durable key` (natural/business key kept as a plain attribute) is what lets you write `SELECT * FROM dim_customer_scd2 WHERE customer_id = 101 ORDER BY effective_date` to see a customer's full version history despite each version having a different surrogate `customer_key`.

# PART 24 - DATA WAREHOUSE LOADING

| Strategy | Description | When to use |
|---|---|---|
| Full load | Reprocess the entire source every run | Small tables, or the first-ever load of any table (**initial load**) |
| Incremental load | Only new/changed rows since the last run | Large tables with a reliable change indicator (timestamp, sequence) |
| Delta load | Synonym for incremental, emphasizing "just the difference" | Same as incremental |
| Append-only | Never update, only insert new rows | Immutable event/log data |
| Upsert | Insert new, update existing, in one operation | Slowly changing reference/dimension data (Type 1) |
| Merge | The SQL `MERGE` statement implementing upsert (and optionally delete) logic in one statement | Cross-platform upsert implementation (see SQL-for-DE handbook Part 13) |
| CDC | Capture inserts/updates/deletes directly from the source's change stream | High-volume OLTP sources needing near-real-time freshness (Part 25) |
| Snapshot | Copy the full state of a table at a point in time | Sources with no reliable change indicator, or when point-in-time history itself is the requirement |
| SCD loading | The specific close-then-insert (Type 2) or overwrite (Type 1) pattern for dimensions | Any dimension attribute that changes over time (Part 19) |

**Initial load vs incremental load**: the initial load moves the *entire* historical dataset once, typically as a full load (often run once, off-hours, potentially in large batches to manage load on the source); every load after that is incremental, processing only what changed since the last successful run, tracked via a stored watermark (see SQL-for-DE handbook Part 12 for the full `>` vs `>=` watermark boundary treatment). Getting the initial load right (validated, reconciled against the source) matters disproportionately, because every subsequent incremental load builds on top of it - an error in the initial load silently persists forever unless separately caught.

# PART 25 - CHANGE DATA CAPTURE

| CDC approach | Mechanism | Trade-off |
|---|---|---|
| Log-based CDC | Reads the database's internal transaction log (WAL/binlog/redo log) directly | No extra load on source tables, captures deletes, requires log access/tooling (e.g. Debezium) |
| Trigger-based CDC | A database trigger writes every change into a separate change table | Simple to understand, but adds write overhead to every source transaction |
| Timestamp-based extraction | Filter source rows by an `updated_at` watermark | Simple, no source-side setup, but misses deletes and depends on a trustworthy timestamp column |
| Version columns | A monotonically increasing version/sequence number per row, used as the change indicator instead of a timestamp | More robust than timestamps under clock skew, still misses deletes unless paired with a tombstone flag |
| Change tables | The structured output of trigger- or log-based CDC - one row per change event with an operation code | The consumable artifact downstream pipelines read from |
| Event-based CDC | Change events published onto a message stream (Kafka) for multiple downstream consumers | Decouples producers from consumers, scales to many consuming pipelines |

```sql
-- Simplified change table, as trigger- or log-based CDC would populate it
CREATE TABLE cdc.customer_changes (
    change_id    BIGSERIAL PRIMARY KEY,
    operation    CHAR(1),          -- 'I','U','D'
    customer_id  INT,
    before_city  TEXT,
    after_city   TEXT,
    changed_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Downstream: apply changes into the warehouse's SCD Type 2 dimension
-- (a 'D' operation typically expires the current version rather than deleting warehouse history)
UPDATE dim_customer_scd2 SET expiry_date = CURRENT_DATE, is_current = FALSE
WHERE customer_id = (SELECT customer_id FROM cdc.customer_changes WHERE operation='D' ORDER BY changed_at DESC LIMIT 1)
  AND is_current = TRUE;
```

**CDC architecture** (conceptual): source database -> log reader (e.g. Debezium tailing the WAL/binlog) -> message stream (Kafka topic per table) -> stream consumer -> warehouse-loading job applying inserts/updates/deletes as upserts/expirations. This architecture decouples "detecting a change happened" from "applying that change to every downstream system," letting one CDC capture feed a warehouse, a search index, and a cache simultaneously.

# PART 26 - LATE-ARRIVING DATA

| Term | Meaning |
|---|---|
| Late-arriving fact | A fact whose event time falls in an already-processed period, but which physically arrives later (a January sale recorded in the source in February) |
| Late-arriving dimension | A dimension record (e.g. a new customer) that arrives after facts referencing it have already loaded |
| Early-arriving fact | A fact that references a dimension member that doesn't exist yet at all - the inverse problem, requiring an inferred/stub dimension row (Part 18) |

**Handling late facts**: if the fact table is partitioned by date (recommended - see SQL-for-DE handbook Part 22), reload only the affected historical partition so previously-published period totals get corrected rather than the late row being silently misattributed to "today." **Handling late dimensions**: use a placeholder/inferred dimension row (Part 18) at fact-load time, then update that row in place (not a new SCD version) once the real dimension data arrives, so existing fact references remain valid without needing to be rewired. **Early-arriving facts** are handled the same way as late dimensions from the fact's perspective - an inferred/stub `dim_customer` row is created just-in-time so the fact load never fails on a missing foreign key.

```sql
-- Detect late-arriving facts: event date is well before the load date
SELECT * FROM staging.stg_sales
WHERE sale_date < DATE_TRUNC('month', loaded_at)
  AND loaded_at >= DATE_TRUNC('month', CURRENT_DATE);
```

# PART 27 - DATA QUALITY IN WAREHOUSES

The seven dimensions of data quality, applied specifically to a warehouse context: **completeness** (are all expected source rows present after load), **accuracy** (do warehouse values reflect the true source values), **consistency** (does the same fact agree across marts/reports), **timeliness** (is the warehouse fresh enough for its stated SLA), **validity** (do values conform to expected formats/domains), **uniqueness** (no unintended duplicate fact/dimension rows), **referential integrity** (every fact FK resolves to a real dimension row).

```sql
-- Duplicate records at the fact grain
SELECT order_id, product_id, COUNT(*) FROM fact_sales GROUP BY order_id, product_id HAVING COUNT(*) > 1;

-- Missing keys / null values in required columns
SELECT COUNT(*) FROM fact_sales WHERE customer_key IS NULL OR product_key IS NULL;

-- Invalid dates (future-dated sales)
SELECT COUNT(*) FROM fact_sales f JOIN dim_date d ON d.date_key=f.date_key WHERE d.full_date > CURRENT_DATE;

-- Negative amounts where they shouldn't occur
SELECT COUNT(*) FROM fact_sales WHERE sales_amount < 0;

-- Invalid categories (values outside the expected domain)
SELECT DISTINCT category FROM dim_product WHERE category NOT IN ('Electronics','Grocery','Apparel','Home');

-- Orphan records: fact rows whose dimension key does not exist
SELECT f.* FROM fact_sales f
LEFT JOIN dim_customer c ON c.customer_key = f.customer_key
WHERE c.customer_key IS NULL;
```

Each of these is a single, automatable check that should run after every load, feeding a warehouse-wide `dq_results` table (see SQL-for-DE handbook Part 10) so quality is monitored continuously rather than discovered by an analyst noticing a wrong number on a dashboard.

# PART 28 - DATA WAREHOUSE TESTING

| Test type | What it verifies |
|---|---|
| Source-to-target testing | Data made it from source to warehouse correctly, value for value |
| Row-count testing | The number of rows loaded matches the expected/source count |
| Aggregate reconciliation | Sums/totals in the warehouse match the source system's totals |
| Null testing | Required columns contain no unexpected NULLs |
| Uniqueness testing | Keys are actually unique at their stated grain |
| Referential integrity testing | Every FK resolves (Part 27) |
| Transformation testing | A specific transformation rule produces the expected output for known input |
| Regression testing | A previously-fixed bug has not reappeared after a code change |
| Freshness testing | The warehouse was updated within its expected SLA window |

```sql
-- Row-count testing
SELECT (SELECT COUNT(*) FROM source.orders) AS source_count,
       (SELECT COUNT(*) FROM warehouse.fact_sales) AS warehouse_count;

-- Aggregate reconciliation
SELECT (SELECT SUM(amount) FROM source.orders) AS source_total,
       (SELECT SUM(sales_amount) FROM warehouse.fact_sales) AS warehouse_total;

-- Transformation testing: a known input row must produce a known output row
SELECT * FROM warehouse.fact_sales WHERE order_id = 90210;  -- compare manually/in a test framework against expected values

-- Freshness testing
SELECT CURRENT_TIMESTAMP - MAX(loaded_at) AS staleness FROM staging.stg_sales;
```

A mature warehouse runs the first six of these as automated, scheduled checks on every load (hard-failing or soft-warning per severity), and the last three (transformation, regression, freshness) as part of both CI (before a code change ships) and ongoing production monitoring.

# PART 29 - DATA MARTS

A **data mart** is a subject-specific, usually smaller, slice of the warehouse serving one team or domain - `sales_mart`, `finance_mart`, `hr_mart`, `marketing_mart`, `customer_mart`, `supply_chain_mart`.

| | Dependent data mart | Independent data mart |
|---|---|---|
| Source | Built *from* the central/enterprise warehouse | Built directly from source systems, bypassing the EDW |
| Consistency | High - inherits conformed dimensions and single source of truth | Lower - risk of diverging definitions/metrics from other marts |
| Governance | Easier - one integration point to govern | Harder - each mart can drift independently |
| Speed to build | Slower initially (EDW must exist first) | Faster initially, but accrues technical debt |
| Recommended default | Yes - this is the Kimball "bus architecture" pattern (Part 32) | Only as a deliberate, temporary/tactical exception |

Independent data marts are the classic anti-pattern behind "why do Finance and Sales report different revenue numbers" - each mart re-derives its own version of shared concepts (customer, revenue, region) with subtly different logic. Dependent marts, built from the same conformed dimensions and fact tables (Part 30), structurally prevent this by construction.

# PART 30 - CONFORMED DIMENSIONS

A **conformed dimension** is a dimension shared, with identical structure and meaning, across multiple fact tables and data marts.

```
DimCustomer (conformed)
   |-- used by FactSales        (Sales mart)
   |-- used by FactSupportTickets (Customer Service mart)
   |-- used by FactMarketingResponses (Marketing mart)
   |-- used by FactInvoices     (Finance mart)
```

Because every mart joins to the *same* `DimCustomer` (same surrogate keys, same attribute definitions, same SCD history), a customer's segment, region, or tier means exactly the same thing whether you're looking at a sales report or a support-ticket report - cross-domain questions ("do high-value customers file more support tickets?") become simple joins across marts instead of requiring a bespoke reconciliation project. This is the single most important enabler of trustworthy enterprise-wide reporting, and is the technical foundation of the Kimball "bus architecture" (Part 32): design conformed dimensions once, centrally, and let every subsequent fact table/mart reuse them.

*Intermediate*: 1) Build a bridge table with a weighting factor and prove `SUM()` stays correct with it. 2) Design a CDC-based loading strategy for a high-volume OLTP source. 3) Detect and correctly re-process a late-arriving fact in a partitioned fact table. 4) Write the full source-to-target reconciliation suite (row count + aggregate + freshness) for one pipeline. 5) Identify a conformed dimension opportunity across two marts you know and describe how to unify them.

# PART 31 - ENTERPRISE DATA WAREHOUSE

An **Enterprise Data Warehouse (EDW)** is the single, organization-wide, integrated warehouse serving every business domain, built on conformed dimensions (Part 30) so every downstream mart and report shares one consistent view of the business.

| Architecture style | Description |
|---|---|
| Centralized | One physical EDW, all marts built dependently from it (Kimball bus architecture) |
| Federated | Multiple domain-owned warehouses/marts, integrated logically (via a semantic layer or virtualization) rather than physically merged |

**Foundational concepts an EDW depends on**: **governance** (who owns what data, who can change a shared dimension's definition - Part 44), **metadata** (data about the data - table descriptions, column meanings, lineage - Part 44/47), **master data** (the single trusted version of core business entities - customer, product, employee - Part 45). Without governance and master data discipline, even a technically well-modeled centralized EDW degrades back into inconsistent, mart-by-mart reinvention over time.

**Kimball vs Inmon** (previewed here, detailed in Parts 32-33): Kimball builds the EDW bottom-up as an integrated collection of conformed dimensional data marts; Inmon builds a normalized enterprise core first, then derives dimensional marts from it. Both aim at the same end state - a governed, conformed, enterprise-wide analytical resource - via different sequencing and different intermediate data shapes.

# PART 32 - KIMBALL METHODOLOGY

The **Kimball approach** (dimensional modeling, bottom-up) builds the warehouse as a series of business-process-oriented dimensional models (star schemas), unified by shared, conformed dimensions - the **bus architecture**.

**Bus architecture**: a matrix with business processes as rows and conformed dimensions as columns; each fact table (one per business process) uses a subset of the conformed dimensions consistently.

| Business Process | DimCustomer | DimProduct | DimDate | DimStore |
|---|---|---|---|---|
| Sales | X | X | X | X |
| Returns | X | X | X | X |
| Inventory | - | X | X | X |
| Support Tickets | X | - | X | - |

Because `DimCustomer` and `DimDate` are conformed (identical structure/meaning) across Sales, Returns, and Support Tickets, you can join `FactSales` and `FactSupportTickets` through the shared `DimCustomer` to answer cross-process questions directly.

**Kimball's core principles**: dimensional modeling for every mart (never normalized reporting schemas), conformed dimensions built and governed centrally, fact tables at a clearly stated grain, and **incremental development** - deliver one business process's star schema at a time (Sales first, then Returns, then Inventory), each immediately useful on its own, rather than requiring the entire EDW to be finished before anything ships. This incremental delivery is Kimball's most practically important advantage: faster time-to-value, with the bus matrix ensuring today's fast-delivered mart still integrates cleanly with tomorrow's.

# PART 33 - INMON METHODOLOGY

The **Inmon approach** (top-down, "Corporate Information Factory") builds a **normalized** enterprise data warehouse core first - a subject-oriented, integrated, non-volatile, time-variant single source of truth - and derives dimensional data marts *from* that normalized core for specific departments.

```
Source Systems -> ETL -> Normalized Enterprise Data Warehouse (3NF) -> Dimensional Data Marts (per department)
```

| | Kimball | Inmon |
|---|---|---|
| Sequencing | Bottom-up: build dimensional marts first, conform dimensions across them | Top-down: build a normalized enterprise core first, then derive marts |
| Core schema | Dimensional (star schemas) throughout | Normalized (3NF) at the center, dimensional only at the mart layer |
| Time-to-value | Faster - first mart can ship in weeks | Slower - enterprise core must be substantially built first |
| Redundancy | Some, by design (denormalized dimensions) | Minimal at the core; redundancy pushed to the derived marts |
| Best fit | Most organizations, especially those needing fast, iterative delivery | Very large, highly regulated enterprises prioritizing a single normalized system of record over speed |
| Risk | Marts built before governance matures can drift non-conformed if discipline slips | Long time-to-first-value; risk of the "big upfront design" project never finishing |

Neither is strictly "correct" - most real-world modern platforms are pragmatic hybrids: a moderately normalized integration layer (a lightweight Inmon-style core) feeding conformed, Kimball-style dimensional marts, which is exactly the shape of the layered architecture in Part 12.

# PART 34 - DATA VAULT

**Data Vault** models the enterprise core around three table types, optimized for auditability, historical tracking, and scalability to many changing sources, at the cost of query simplicity.

| Component | Purpose |
|---|---|
| Hub | Stores only the unique list of business keys for an entity (one row per distinct customer, ever) |
| Link | Stores relationships between hubs (a customer placed an order - just the keys, no descriptive attributes) |
| Satellite | Stores the descriptive, time-variant attributes for a hub or link, fully historized |

```sql
CREATE TABLE hub_customer (
    customer_hash_key CHAR(32) PRIMARY KEY,   -- hash of the business key, for join performance
    customer_id        TEXT,                    -- the actual business key
    load_date            TIMESTAMP,
    record_source          TEXT                    -- which source system this business key came from
);

CREATE TABLE link_customer_order (
    link_hash_key       CHAR(32) PRIMARY KEY,
    customer_hash_key   CHAR(32) REFERENCES hub_customer(customer_hash_key),
    order_hash_key       CHAR(32),
    load_date              TIMESTAMP,
    record_source            TEXT
);

CREATE TABLE sat_customer_details (
    customer_hash_key CHAR(32) REFERENCES hub_customer(customer_hash_key),
    load_date          TIMESTAMP,
    city               TEXT,
    segment            TEXT,
    hash_diff          CHAR(32),      -- hash of all tracked attributes, for cheap change detection
    PRIMARY KEY (customer_hash_key, load_date)
);
```

**Why Data Vault**: business keys and relationships (hubs/links) are completely separated from descriptive attributes (satellites), which means adding a new source system or a new attribute never requires restructuring existing tables - you only add new satellites or hubs, an append-only, low-risk change pattern. Every row is fully historized and tagged with its source and load time, giving strong auditability. **Trade-off**: querying a Data Vault directly is verbose (many joins across hubs/links/satellites), so Data Vault is almost always paired with a Kimball-style dimensional layer built *on top* of it for actual BI consumption - Data Vault as the auditable, source-of-truth integration layer; star schemas as the consumption layer.

**Kimball vs Inmon vs Data Vault**

| | Kimball | Inmon | Data Vault |
|---|---|---|---|
| Primary strength | Fast time-to-value, BI-friendly | Single normalized enterprise truth | Auditability, scalable to many changing sources |
| Primary weakness | Discipline required to keep dimensions conformed | Slow initial delivery | Verbose to query directly, needs a consumption layer on top |
| Typical use | Most organizations | Very large, stable, highly regulated enterprises | Environments with many volatile sources, heavy compliance/audit needs |

# PART 35 - MODERN DATA ARCHITECTURE

```
Raw (Bronze)  ->  Cleaned (Silver)  ->  Curated (Gold)
```

| Term | Definition |
|---|---|
| Modern Data Warehouse | Cloud-native, elastically-scaled warehouse (Part 36), usually ELT-based |
| Data Lake | Cheap, schema-on-read object storage for raw data of any format |
| Data Lakehouse | Lake storage plus warehouse-style ACID transactions and query performance |
| Medallion Architecture | The Bronze/Silver/Gold layering pattern, most associated with Databricks/Delta Lake but applicable anywhere |
| Bronze | Raw, exactly as ingested, append-only, minimal/no transformation |
| Silver | Cleaned, typed, deduplicated, validated, conformed |
| Gold | Business-ready: aggregated, dimensional, ready for BI/ML consumption |

```
Source Systems
   -> Bronze (Delta/Iceberg tables over object storage - raw ingested data)
      -> Silver (typed, deduplicated, validated - "trusted" data other teams build on)
         -> Gold (star-schema dimensional model - what BI tools and analysts actually query)
            -> BI / ML / AI
```

This end-to-end architecture unifies the classic layered-warehouse thinking (Part 12) with lakehouse technology: Bronze/Silver/Gold map directly onto Raw/Staging-Integration/Core-Marts, but physically live as open-format tables (Delta, Iceberg) on cheap object storage rather than proprietary warehouse storage - giving one copy of data that both SQL analytics *and* ML/AI workloads (which typically want direct file access, not just SQL) can consume without duplicating data into two separate systems.

# PART 36 - CLOUD DATA WAREHOUSING

| Platform | Architecture | Storage/Compute | SQL Dialect |
|---|---|---|---|
| Google BigQuery | Serverless, fully managed | Fully decoupled, auto-scaled | GoogleSQL (Standard SQL) |
| Snowflake | Multi-cluster shared data | Fully decoupled, virtual warehouses | Snowflake SQL (ANSI-based) |
| Amazon Redshift | MPP clusters | Coupled by default (RA3 nodes decouple) | PostgreSQL-derived SQL |
| Azure Synapse Analytics | MPP (dedicated pools) + serverless (on-demand) | Both coupled (dedicated) and decoupled (serverless) options | T-SQL |
| Databricks SQL | Lakehouse (Delta Lake on object storage) | Fully decoupled, SQL Warehouses | Spark SQL |

**Shared architectural concepts, implemented differently per platform**: **storage/compute separation** lets you scale query power up or down (or to zero) independently of how much data you store, avoiding the old model of buying one fixed-size cluster for both; **elastic scaling** means compute capacity can grow (or shrink) automatically or on-demand as workload changes; **columnar storage** stores each column contiguously, so scanning 3 of 50 columns only reads those 3 columns' data, dramatically speeding up typical analytical queries; **distributed query execution** splits a query's work across many machines/nodes in parallel; **partitioning** and **clustering** (Parts 37-38) physically organize data so queries skip irrelevant data; **compression** exploits columnar storage's repeated values to shrink storage footprint and I/O; **parallel processing** is the underlying mechanism distributed execution relies on.

Each platform implements every one of these differently - BigQuery and Snowflake fully automate storage layout decisions behind simple `PARTITION BY`/clustering-key declarations; Redshift requires explicit distribution-style and sort-key choices with real performance consequences if chosen wrong; Azure Synapse gives you the choice between a dedicated MPP pool (Redshift-like, explicit distribution) and serverless (BigQuery-like, pay-per-query); Databricks SQL operates on open Delta/Parquet files with explicit `OPTIMIZE`/`ZORDER` maintenance commands rather than fully automatic optimization. None of these are interchangeable at the syntax or architecture level - always consult the target platform's own documentation for the authoritative implementation details.

# PART 37 - PARTITIONING

| Partitioning type | Splits data by | Typical key |
|---|---|---|
| Range partitioning | A continuous value range | Date ranges (most common) |
| List partitioning | Discrete enumerated values | Country code, region |
| Hash partitioning | Hash of a key, for even distribution | A high-cardinality ID with no natural range/list |
| Time-based partitioning | A special case of range partitioning by date/timestamp | Daily/monthly partitions on `order_date` |

```sql
-- [PostgreSQL] range partitioning by month
CREATE TABLE fact_sales (sale_id INT, order_date DATE, amount NUMERIC) PARTITION BY RANGE (order_date);
CREATE TABLE fact_sales_2026_01 PARTITION OF fact_sales FOR VALUES FROM ('2026-01-01') TO ('2026-02-01');

-- Partition pruning in action - the planner only touches the January partition
EXPLAIN SELECT SUM(amount) FROM fact_sales WHERE order_date >= '2026-01-01' AND order_date < '2026-02-01';
```

**Partition pruning** is the query planner's ability to skip entire partitions that cannot contain matching rows, based on the filter predicate - the single biggest performance lever partitioning provides. **Query performance**: queries that filter on the partition key touch dramatically less data. **Data lifecycle management**: dropping an entire old partition (a metadata operation) is vastly cheaper than a row-by-row `DELETE` for archiving/retention (see SQL-for-DE handbook Part 43). **Archiving**: old partitions can be detached and moved to cheaper storage tiers without touching the live table.

# PART 38 - CLUSTERING AND SORTING

| Concept | Definition |
|---|---|
| Clustering | Physically co-locating rows with similar values in specified columns, so filtered scans read fewer blocks |
| Sort key | [Redshift] defines on-disk row order, enabling zone-map-style pruning |
| Distribution key | [Redshift] controls which compute node stores each row, minimizing cross-node data movement during joins |
| Data locality | The general principle that data physically stored near other data it's queried/joined with performs better |
| File organization | How rows are laid out within files/pages on disk - row-oriented vs columnar, sorted vs unsorted |

| Platform | How it implements clustering/sorting |
|---|---|
| BigQuery | Partitioning (coarse, usually by date) + clustering (fine, up to 4 columns, auto-maintained) |
| Snowflake | Automatic micro-partitioning; optional explicit clustering keys for very large tables that need manual tuning |
| Redshift | Explicit distribution style (KEY/ALL/EVEN) + explicit sort keys (compound or interleaved) |
| Azure Synapse (dedicated pool) | Explicit distribution (hash/round-robin/replicate) + clustered columnstore indexes |
| Databricks | Partitioning + `OPTIMIZE ... ZORDER BY` (or newer liquid clustering) for physical file layout |

The general pattern across all platforms: fully-managed serverless platforms (BigQuery, Snowflake) try to make good storage-layout decisions automatically from simple declarative hints; classic MPP platforms (Redshift, Synapse dedicated pools) require the designer to make explicit, consequential choices about data distribution across nodes; lakehouse platforms (Databricks) expose file-layout optimization as an explicit, periodically-run maintenance operation rather than either extreme.

# PART 39 - PHYSICAL DATABASE DESIGN

Physical design is where a platform-agnostic logical model becomes concrete, vendor-specific implementation decisions.

| Concern | Decisions involved |
|---|---|
| Index design | Which columns to index, index type (B-tree, hash, columnstore, bitmap), covering indexes |
| Storage | Row-oriented vs columnar, compression settings, tablespace/storage-tier placement |
| Data types | Choosing the narrowest correct type (affects storage size, index size, comparison speed) |
| Page/block concepts | The fixed-size unit (e.g. 8KB in PostgreSQL) the database reads/writes at a time - more rows per page means fewer I/O operations per query |
| Partitioning | Physical splitting for pruning and lifecycle management (Part 37) |
| Compression | Reducing physical storage footprint, often also reducing I/O (fewer bytes to read) |
| Clustering | Physical row ordering for locality (Part 38) |
| Materialized views | A query result physically persisted and refreshed on a schedule, trading storage and staleness for query speed |

```sql
-- Materialized view: precompute an expensive aggregation, refreshed on a schedule
CREATE MATERIALIZED VIEW mart.daily_sales_summary AS
SELECT date_key, SUM(sales_amount) AS total_sales
FROM fact_sales GROUP BY date_key;

REFRESH MATERIALIZED VIEW mart.daily_sales_summary;   -- [PostgreSQL] re-run on a schedule
```

**How physical design affects performance**: a poorly chosen data type (e.g. storing a date as `TEXT`) prevents efficient range comparisons and date functions; a missing index on a frequently-filtered column forces full scans; over-wide rows reduce how many rows fit per page, increasing I/O per query; the right partitioning/clustering choices (Parts 37-38) are also physical design decisions with first-order performance impact. Physical design is the layer where the same logical model can perform radically differently depending on execution.

# PART 40 - PERFORMANCE DESIGN

Schema design itself - not just query-writing - determines how fast queries *can* be, independent of how well any individual query is written.

| Lever | How schema design affects it |
|---|---|
| Index strategy | Which columns are indexed determines which filters/joins can avoid full scans |
| Join strategy | Surrogate integer keys (Part 23) join faster than wide natural/composite keys |
| Data types | Narrow, fixed-size types compare and index faster than wide/variable types |
| Cardinality | High-cardinality columns (many distinct values) are better index/partition candidates than low-cardinality ones |
| Selectivity | How much a predicate narrows the result set - the planner favors indexes on highly selective predicates |
| Statistics | The optimizer's cardinality estimates (row-count guesses) come from table statistics - stale statistics after schema/data changes cause bad query plans |
| Partition pruning | A partitioning scheme aligned with real query filters (Part 37) is a schema-level performance decision |
| Predicate pushdown | A schema that lets filters apply at the storage/scan layer (rather than after a join) reduces data movement |
| Column pruning | Columnar storage lets a query read only the columns it needs - schema width and column ordering can matter for this |
| Query planning | The optimizer's plan choice is directly shaped by all of the above - schema design is upstream of every plan the optimizer can even consider |

The core insight: query optimization (rewriting a slow `SELECT`) has a ceiling set by the underlying schema design - the best-written query against a badly-designed schema (wrong keys, no partitioning, wrong data types) will still be slow, while a well-designed schema makes even a mediocre query reasonably fast. Invest in schema-level performance design before micro-optimizing individual queries.

# PART 41 - ANALYTICAL QUERY DESIGN

Warehouse schemas exist to make common analytical questions cheap and simple to answer. The following patterns all assume the star schema from Part 15.

```sql
-- Revenue, profit, gross margin
SELECT d.year, SUM(f.sales_amount) AS revenue, SUM(f.sales_amount - f.cost_amount) AS profit,
       ROUND(100.0 * SUM(f.sales_amount - f.cost_amount) / NULLIF(SUM(f.sales_amount),0), 2) AS gross_margin_pct
FROM fact_sales f JOIN dim_date d ON d.date_key = f.date_key
GROUP BY d.year;

-- Customer lifetime value (simple historical CLV)
SELECT c.customer_id, SUM(f.sales_amount) AS lifetime_value
FROM fact_sales f JOIN dim_customer c ON c.customer_key = f.customer_key
GROUP BY c.customer_id;

-- Retention / churn (customers active in month N who were also active in month N-1)
WITH monthly_customers AS (
    SELECT DISTINCT d.year, d.month, f.customer_key
    FROM fact_sales f JOIN dim_date d ON d.date_key = f.date_key
)
SELECT curr.year, curr.month,
       COUNT(DISTINCT curr.customer_key) AS active_customers,
       COUNT(DISTINCT prev.customer_key) AS retained_from_prev_month
FROM monthly_customers curr
LEFT JOIN monthly_customers prev
    ON prev.customer_key = curr.customer_key AND prev.month = curr.month - 1 AND prev.year = curr.year
GROUP BY curr.year, curr.month;

-- Cohort analysis (customers grouped by first purchase month, tracked over subsequent months)
WITH first_purchase AS (
    SELECT customer_key, MIN(date_key) AS cohort_date_key FROM fact_sales GROUP BY customer_key
)
SELECT fp.cohort_date_key, d.date_key AS activity_date_key, COUNT(DISTINCT f.customer_key) AS active_customers
FROM fact_sales f
JOIN first_purchase fp ON fp.customer_key = f.customer_key
JOIN dim_date d ON d.date_key = f.date_key
GROUP BY fp.cohort_date_key, d.date_key;

-- Market basket analysis (which products are commonly bought together)
SELECT a.product_key AS product_a, b.product_key AS product_b, COUNT(*) AS times_bought_together
FROM fact_sales a
JOIN fact_sales b ON a.order_id = b.order_id AND a.product_key < b.product_key
GROUP BY a.product_key, b.product_key
ORDER BY times_bought_together DESC;

-- Inventory turnover
SELECT p.category, SUM(f.quantity_sold) / NULLIF(AVG(i.on_hand_quantity), 0) AS turnover_ratio
FROM fact_sales f JOIN dim_product p ON p.product_key = f.product_key
JOIN fact_inventory_snapshot i ON i.product_key = f.product_key
GROUP BY p.category;
```

Each of these patterns follows directly from a correctly-grained star schema (Part 14-15) - the analytical value of dimensional modeling is precisely that these business questions become straightforward joins and aggregations rather than bespoke, hard-to-verify logic re-derived per report.

# PART 42 - BI AND SEMANTIC MODELING

BI tools (Power BI, Tableau, Looker, Excel) connect to the warehouse and let business users explore data without writing SQL - but only if the warehouse exposes a clean, well-modeled **semantic layer** for them to connect to.

| Concept | Definition |
|---|---|
| Semantic layer | The business-friendly abstraction between raw warehouse tables and BI tools - defines metrics, dimensions, and relationships once, centrally |
| Metric / Measure | A named, precisely-defined calculation (e.g. "Net Revenue" = `sales_amount - returns_amount - discounts`) |
| Dimension (BI context) | A named attribute/hierarchy business users filter and group by (Region, Product Category, Month) |
| Hierarchy | A drill-path within a dimension (Year > Quarter > Month > Day) |
| Relationship | How BI-tool tables relate (usually mirroring the star schema's fact-to-dimension foreign keys) |
| KPI definition | A specific, governed metric with a stated formula, owner, and target used company-wide |

**Why business definitions should not be recreated independently in every dashboard**: if "Active Customer" is defined one way in the Sales dashboard (purchased in the last 30 days) and another way in the Marketing dashboard (any account not marked churned), executives get two different, both "correct," conflicting numbers for the same-sounding metric - eroding trust in data generally, not just in one dashboard. A governed semantic layer (built once, e.g. as a Power BI dataset, a Looker LookML model, or a dbt semantic layer / metrics layer) defines "Active Customer" exactly once, and every dashboard built on top of it inherits the same definition automatically. Warehouse design directly enables this: a clean, conformed star schema (Parts 15, 30) is what makes a single semantic layer *possible* to build in the first place - a messy, non-conformed warehouse forces every BI tool/team to reinvent business logic independently, which is precisely the failure mode a good semantic layer exists to prevent.

# PART 43 - ANALYTICS ENGINEERING

**Analytics Engineering** is the discipline of applying software-engineering practices - modularity, testing, version control, CI/CD - to the SQL transformation layer that turns raw warehouse data into trusted, documented models.

```
Source -> Staging -> Intermediate -> Marts
```

- **Source**: raw tables as loaded, referenced (never modified) as the starting point of every model's lineage.
- **Staging**: one model per source table, typed/renamed/lightly cleaned, 1:1 with the source.
- **Intermediate**: reusable joined/reshaped building blocks not yet exposed to BI.
- **Marts**: final, business-facing, documented, tested models - what BI tools actually query.

```yaml
# dbt-style schema.yml - documentation + tests colocated with the model definition
models:
  - name: dim_customer
    description: "Conformed customer dimension, SCD Type 2"
    columns:
      - name: customer_key
        tests: [unique, not_null]
      - name: customer_id
        tests: [not_null]
  - name: fact_sales
    columns:
      - name: customer_key
        tests:
          - relationships: {to: ref('dim_customer'), field: customer_key}
```

**Core practices**: **modular SQL** (small, single-purpose models chained via references, not one giant query); **reusable models** (a staging model used by many downstream marts instead of each mart re-deriving it); **testing** (declarative not-null/unique/relationships tests, run on every build); **documentation** (descriptions colocated with the model definition, not a separate wiki that goes stale); **lineage** (auto-derived from model references - which models feed which); **version control** (every model is a file in git, reviewed via pull request like application code); **CI/CD** (models are tested and built automatically on every change before merging to production). This is the same discipline covered in depth in the SQL-for-Data-Engineering handbook (Part 34) - here it is the *design methodology* that keeps a warehouse's transformation layer maintainable as it grows to hundreds of models.

# PART 45 - MASTER DATA MANAGEMENT

**Master Data Management (MDM)** creates and maintains one trusted, deduplicated "golden record" for core business entities - customer, product, supplier, employee - that are referenced by many systems and must mean the same thing everywhere.

| Concept | Definition |
|---|---|
| Golden record | The single, trusted, merged representation of an entity, resolved from potentially many conflicting source records |
| Duplicate resolution | Identifying that multiple source records refer to the same real-world entity |
| Survivorship | The rule set deciding which source's value "wins" for each attribute when merging duplicates (e.g. "most recently updated source wins," or "CRM wins for contact info, ERP wins for billing address") |
| Matching | The technique (exact key match, fuzzy/probabilistic match) used to detect that two records are duplicates |
| Reference data | Small, stable, shared lookup data (country codes, currency codes, unit-of-measure codes) governed similarly to master data but simpler |

```sql
-- Simplified survivorship: prefer the most recently updated source's value per attribute
SELECT customer_natural_key,
       (ARRAY_AGG(email ORDER BY source_updated_at DESC))[1] AS golden_email,
       (ARRAY_AGG(phone ORDER BY source_updated_at DESC))[1] AS golden_phone
FROM staging.customer_records_all_sources
GROUP BY customer_natural_key;
```

MDM and warehouse dimensional modeling are complementary, not competing: MDM's golden `Customer` record is exactly the trusted, deduplicated feed a well-designed `dim_customer` (Parts 15, 19, 30) should be built from - a warehouse without MDM discipline upstream inherits every source system's duplicate/conflicting customer records directly into its "conformed" dimension, undermining conformance from the start.

# PART 46 - DATABASE SECURITY

| Concept | Definition |
|---|---|
| Authentication | Verifying who is connecting (password, SSO, certificate) |
| Authorization | Determining what an authenticated identity is allowed to do |
| Roles | Named bundles of permissions, assigned to users/service accounts rather than granting permissions individually |
| Permissions | Specific allowed actions (SELECT, INSERT, UPDATE on specific objects) |
| Least privilege | Granting only the minimum access needed to perform a job, nothing more |
| Encryption | Protecting data at rest (storage) and in transit (network) from unauthorized reading |
| Row-level security | Restricting which *rows* of a table a given role/user can see |
| Column-level security | Restricting which *columns* of a table a given role/user can see |
| Data masking | Obscuring sensitive values (e.g. showing only the last 4 digits of an account number) for roles that shouldn't see the full value |
| Sensitive data | PII, financial, health, or otherwise regulated data requiring special handling |
| Auditing | Recording who accessed or changed what, and when, for accountability and compliance |

```sql
CREATE ROLE bi_analyst_role;
GRANT SELECT ON mart.sales_mart TO bi_analyst_role;                -- least privilege: read-only on one mart

CREATE VIEW mart.customers_masked AS
SELECT customer_id, customer_name, CONCAT(LEFT(email,2),'***@***') AS email_masked
FROM dim_customer;
GRANT SELECT ON mart.customers_masked TO bi_analyst_role;           -- column-level protection via a masked view

CREATE POLICY region_policy ON fact_sales
    USING (region = current_setting('app.user_region'));            -- [PostgreSQL] row-level security
```

**Warehouse security for BI users specifically**: BI users typically need broad *read* access to marts to be useful, which is in direct tension with protecting sensitive columns - the standard resolution is masked/aggregated views (as above) granted to broad BI roles, with raw/sensitive columns restricted to a narrow, audited set of roles that genuinely need them (e.g. Finance seeing unmasked payment details). Row-level security additionally scopes *which rows* a given business unit or region's analysts can see within an otherwise shared mart.

# PART 47 - DATA LINEAGE

```
CRM
   -> Raw Customer            -- landed exactly as received
      -> Staging Customer        -- typed, cleaned, still 1:1 with source
         -> DimCustomer              -- conformed, SCD Type 2, surrogate-keyed
            -> FactSales                 -- references DimCustomer via customer_key
               -> Sales Mart                 -- aggregated, business-facing
                  -> Power BI Dashboard          -- what an executive actually looks at
```

**Why lineage matters**: when a dashboard number looks wrong, lineage lets you trace backward through this exact chain to find precisely which upstream table/transformation is responsible; when a source system changes (a CRM field is renamed), lineage lets you trace forward to see everything - reports, models, dashboards - that will break, *before* it breaks in production. Lineage is also what makes governance (Part 44) and impact analysis for schema evolution (Part 48) tractable at scale - without it, every source change requires manually asking every team "does this affect you," which does not scale past a handful of pipelines.

# PART 48 - SCHEMA EVOLUTION

| Change type | Backward compatible? |
|---|---|
| Adding a nullable column | Yes - existing consumers ignore it |
| Removing a column | No - anything selecting it explicitly breaks |
| Renaming a column | No - equivalent to a drop + add for consumers |
| Changing a data type (widening, e.g. INT to BIGINT) | Usually yes | 
| Changing a data type (narrowing, e.g. BIGINT to INT, or changing precision) | Usually no - risk of overflow/truncation |

```sql
ALTER TABLE dim_customer ADD COLUMN loyalty_tier TEXT;      -- backward-compatible
ALTER TABLE dim_customer DROP COLUMN old_unused_flag;       -- breaking
ALTER TABLE dim_customer RENAME COLUMN cust_nm TO customer_name;  -- breaking, treat as drop+add
```

**What schema changes can break**: **ETL/ELT pipelines** that reference a dropped/renamed column fail outright; **APIs** built on a table's shape return errors or silently wrong data; **dashboards** referencing a removed field break or silently show blanks; **ML pipelines** expecting a specific feature column fail training/inference; **reports** with hardcoded column references break at generation time. **Backward compatibility** means old consumers keep working unmodified after the change; **forward compatibility** means a system built *before* a change can still function reasonably once the change happens (rarer to guarantee, usually only relevant for genuinely additive changes). **Versioning** and staged **migration strategies** (introduce new alongside old, dual-write, migrate consumers, deprecate old) turn an unavoidable breaking change into a managed, low-risk process instead of a surprise outage - this is the same discipline detailed with full SQL examples in the SQL-for-Data-Engineering handbook, Part 32.

# PART 49 - DATABASE MIGRATIONS

Migration tools (**Flyway**, **Liquibase**, **Alembic**) all implement the same core discipline: every schema change is a small, numbered, version-controlled script; a metadata table tracks which migrations have already run against each environment; running the tool applies only unapplied migrations, in order, keeping dev/staging/production schemas provably in sync.

```sql
-- 001_create_dim_customer.sql
CREATE TABLE dim_customer (customer_key SERIAL PRIMARY KEY, customer_id INT, customer_name TEXT);

-- 002_add_loyalty_tier.sql
ALTER TABLE dim_customer ADD COLUMN loyalty_tier TEXT DEFAULT 'standard';
```

**Version-controlled schema changes**: every migration lives in git, reviewed like application code, giving a complete auditable history of how the schema evolved. **Rollbacks**: rather than physically reversing a migration against a table that may already hold real data in the new shape (risky), most mature teams practice **forward-only migrations** - if migration 002 is wrong, write migration 003 to fix it forward. **Backward-compatible changes** (Part 48) can typically be deployed and rolled out safely without coordinating a simultaneous consumer update; breaking changes require a coordinated, staged **deployment strategy** across both the schema change and every dependent consumer.

```python
from sqlalchemy import create_engine, text
import pandas as pd

engine = create_engine("postgresql+psycopg://etl:pw@host/warehouse")

# Extraction: pull from an OLTP source into a DataFrame
df = pd.read_sql("SELECT * FROM source.orders WHERE order_date = CURRENT_DATE - 1", engine)

# Validation: Python's richer assertion/branching logic complements set-based SQL
assert df['amount'].min() >= 0, "Negative order amount detected - halting load"
assert df['customer_id'].notnull().all(), "Null customer_id detected - halting load"

# Transformation (light, row-level logic easier in pandas than SQL, e.g. custom business rules)
df['order_month'] = pd.to_datetime(df['order_date']).dt.to_period('M')

# Loading: write the validated, transformed DataFrame into the warehouse
df.to_sql('stg_orders', engine, if_exists='append', index=False, method='multi')

# Downstream: the actual dimensional transformation still happens as SQL, inside the warehouse
with engine.begin() as conn:
    conn.execute(text("""
        INSERT INTO warehouse.fact_sales (customer_key, amount, date_key)
        SELECT c.customer_key, s.amount, d.date_key
        FROM stg_orders s
        JOIN dim_customer c ON c.customer_id = s.customer_id
        JOIN dim_date d ON d.full_date = s.order_date
    """))
```

Python and SQL are complementary in warehouse work exactly as they are in general data engineering (see SQL-for-Data-Engineering handbook, Parts 52-53): Python excels at orchestration, validation, and calling non-database systems (APIs, files); SQL excels at set-based transformation and aggregation once data is inside the warehouse. Database connectors (`psycopg`, `pyodbc`, platform-specific drivers) and SQLAlchemy provide the connection layer; pandas provides in-memory manipulation for anything more naturally row-wise than set-based.

# PART 51 - DATA WAREHOUSE AND MACHINE LEARNING

Warehouse design directly determines how easy (or leakage-prone) it is to build correct ML training datasets.

```sql
-- Point-in-time-correct feature extraction for a churn model
-- CRITICAL: every feature must only use data available strictly BEFORE the label's cutoff date,
-- or the model "cheats" during training in a way it cannot in live production (data leakage).

SELECT
    c.customer_id,
    COUNT(f.sale_key)                                    AS historical_transaction_count,
    SUM(f.sales_amount)                                     AS historical_total_spend,
    MAX(d.full_date)                                           AS last_purchase_date,
    label_cutoff.churned_within_90_days                          AS label
FROM dim_customer c
LEFT JOIN fact_sales f ON f.customer_key = c.customer_key
LEFT JOIN dim_date d ON d.date_key = f.date_key AND d.full_date < '2026-01-01'   -- as-of cutoff
JOIN ml.churn_labels label_cutoff ON label_cutoff.customer_id = c.customer_id
GROUP BY c.customer_id, label_cutoff.churned_within_90_days;
```

**Feature tables** built from a well-designed star schema are straightforward: one row per entity, one column per feature, computed as-of a stated cutoff. **Historical features** (rolling windows - "spend in the last 90 days") rely directly on the fact table's grain and date dimension being correctly designed. **Training/validation/test datasets** are typically split by *time* (train on data before date X, validate/test after) specifically to simulate the point-in-time constraint realistically - a warehouse with a clean, well-partitioned date dimension makes this split trivial; a poorly time-modeled warehouse makes it error-prone. Worked examples: a **customer churn model** and a **fraud detection model** both need point-in-time rolling behavioral features exactly as shown above; a **credit risk model** needs historical account/payment-behavior features with the same leakage discipline; a **sales forecasting model** needs time-series features (lagged sales, moving averages) that depend entirely on the fact table's date grain being correct and complete (no silent gaps from missed incremental loads).

# PART 53 - DATA ARCHITECTURE FOR E-COMMERCE

**Sources**: Website, Mobile App, Payment Gateway, CRM, Inventory System, Marketing Platform.

```
Website/App events, Payment Gateway, CRM, Inventory, Marketing
   -> Raw Layer (one raw table per source, untransformed)
      -> Staging Layer (typed, deduplicated, 1:1 with source)
         -> Warehouse (conformed dimensions + fact tables below)
            -> Data Marts (sales_mart, marketing_mart, inventory_mart)
               -> Semantic Layer -> BI (Power BI / Tableau / Looker)
```

**Dimensions**: `DimCustomer` (conformed across every mart), `DimProduct`, `DimDate`, `DimStore` (or DimChannel for online-only), `DimPromotion`, `DimPayment`, `DimGeography`.

# PART 61 - DATABASE DESIGN INTERVIEW MASTERY (curated core across all 23 categories, extensible pattern)

Each entry: Question -> Correct answer -> Explanation -> Example -> Common wrong answer -> Interview tip. One representative question per category is given in full below; use the identical five-part format to extend each category to its full requested count using your own study list.

**A. Database Design Fundamentals** - *Q: What is the difference between logical and physical data models?* **A**: Logical is technology-agnostic and fully attributed; physical is the vendor-specific implementation. **Explanation**: logical design decisions (entities, keys, normalization) should survive a database migration; physical decisions (data types, indexes, partitioning) are chosen per-platform. **Example**: a logical "Customer" entity with a "Total Spend" attribute becomes, physically, a `NUMERIC(12,2)` column with a specific index strategy on PostgreSQL vs BigQuery. **Common wrong answer**: treating them as the same step, or thinking "physical" just means "the ER diagram with data types added" without considering platform-specific storage/index choices. **Interview tip**: always mention that logical design should be reusable across a platform migration - it signals you understand the separation is deliberate.

**B. ER Modeling** - *Q: How do you resolve a many-to-many relationship in a relational model?* **A**: With an associative entity (junction/bridge table) holding foreign keys to both sides. **Explanation**: relational databases cannot natively store M:N without an intermediate table (Part 2). **Example**: `Students <-> Courses` resolved via `Enrollment(student_id, course_id)`. **Common wrong answer**: storing a comma-separated list of IDs in one column (violates 1NF, Part 4). **Interview tip**: mention that the junction table itself can carry relationship attributes (enrollment date, grade) - shows you understand it's a first-class entity, not just a technical workaround.

**C. Keys** - *Q: Why do warehouse dimensions use surrogate keys instead of the source system's natural key?* **A**: To support SCD Type 2 history and to decouple the warehouse from any single source's ID scheme. **Explanation**: (Part 3, Part 23) a natural key can't have multiple "current" rows for the same entity, which Type 2 requires. **Example**: `dim_customer_scd2.customer_key` differs per version; `customer_id` (natural key) repeats across those versions. **Common wrong answer**: "surrogate keys are just faster for joins" - true but incomplete, misses the SCD/integration rationale which is the primary reason. **Interview tip**: always connect surrogate keys to SCD Type 2 specifically - it's the answer interviewers are listening for.

**D. Normalization** - *Q: What anomaly does 2NF specifically address that 1NF does not?* **A**: Partial dependency - a non-key attribute depending on only part of a composite primary key. **Explanation**: (Part 4) 1NF only requires atomic values; it says nothing about dependency structure. **Example**: in `(order_id, product_id) -> product_name`, `product_name` depends only on `product_id`, a partial dependency, violating 2NF. **Common wrong answer**: confusing 2NF with removing all redundancy (that's the general goal of normalization, not what specifically distinguishes 2NF from 1NF). **Interview tip**: be ready to state, precisely, what's newly enforced at each normal form - vague "reduces redundancy" answers for every level signal shallow understanding.

**E. Denormalization** - *Q: When is denormalization an acceptable trade-off?* **A**: Read-heavy analytical workloads where join cost dominates and the redundant copy is refreshed on a controlled schedule. **Explanation**: (Part 5) the risk (stale/inconsistent redundant data) is bounded and predictable for a scheduled-rebuild reporting table, unlike a live OLTP table with concurrent writes. **Example**: a `sales_flat` reporting table joining customer/region once, at build time, instead of per query. **Common wrong answer**: "denormalize whenever it makes queries faster," with no consideration of write-consistency risk. **Interview tip**: always name the specific trade-off being accepted (write-consistency risk) and the mitigation (scheduled, controlled rebuild) - shows engineering judgment, not just a rule memorized.

**F. OLTP** - *Q: Why are OLTP schemas normalized while OLAP schemas are typically dimensional/denormalized?* **A**: OLTP prioritizes write correctness and storage efficiency for many small transactions; OLAP prioritizes read simplicity/speed for large aggregating scans. **Explanation/Example/Wrong answer/Tip**: see Part 10's OLTP vs OLAP comparison table - reproduce it, then explain the *why* behind each row, not just the row itself.

**G. OLAP** - *Q: What is the difference between roll-up and drill-down?* **A**: Roll-up moves from detail to summary (Day -> Month); drill-down moves from summary to detail (Year -> Quarter). **Example**: see Part 11's worked SQL. **Common wrong answer**: reversing the two definitions - a frequent, easy-to-catch mistake, so state the direction explicitly and confidently.

**H. Dimensional Modeling** - *Q: Why must grain be defined before choosing a fact table's dimensions and measures?* **A**: Every dimension and measure must be consistent with what one row represents - defining grain first prevents mixing multiple grains in one fact table (Part 14, Part 64). **Interview tip**: always be ready with the exact sentence "the grain of this fact table is..." - interviewers specifically probe for this framing.

**I. Star Schema** - *Q: Why does a star schema minimize joins compared to a snowflake schema?* **A**: Dimensions are fully denormalized, so a fact table joins directly to each dimension with no intermediate sub-dimension tables (Part 15-16).

**J. Snowflake Schema** - *Q: What real cost does snowflaking introduce, and is it usually worth it?* **A**: Extra joins per query, for a usually-marginal storage saving on modern compressed columnar storage - rarely worth it except for large, independently-changing, widely-shared sub-dimensions (Part 16).

**K. Fact Tables** - *Q: Give an example of a factless fact table and explain why it has no measures.* **A**: `fact_attendance` - one row per student per class per day; the fact being recorded is that the event *occurred*, not a numeric quantity, so `COUNT(*)` itself is the metric (Part 17).

**L. Dimension Tables** - *Q: What is a role-playing dimension and why not just create two separate date tables?* **A**: One physical `dim_date` used multiple times via different foreign keys/aliases (`order_date`, `ship_date`) - creating two separate physical tables would duplicate maintenance (holiday lists, fiscal periods) and risk them drifting inconsistent (Part 18, Part 20).

**M. SCD** - *Q: Walk through what goes wrong if you insert a new SCD Type 2 row without first closing the old one.* **A**: Two rows end up with `is_current = TRUE` for the same business key, breaking any query that assumes exactly one current row per entity, and any new fact load resolving "current customer_key" gets a non-deterministic/duplicated result (Part 19, Part 65 debugging example).

**N. Data Warehousing** - *Q: What's the difference between an EDW and a data mart?* **A**: See Part 10's comparison table - scope (organization-wide vs domain-specific) and typical relationship (marts are usually built *from* the EDW).

**O. ETL/ELT** - *Q: Why do modern cloud warehouses favor ELT?* **A**: Storage/compute decoupling makes it cheap to store raw data and pay for transform compute only when run, keeping full raw history re-derivable (Part 13, Part 36).

**P. CDC** - *Q: Why is log-based CDC preferred over trigger-based CDC in high-volume OLTP systems?* **A**: Log-based CDC adds no write overhead to the source transactions (it reads the transaction log after the fact); triggers add overhead to every single write (Part 25).

**Q. Data Quality** - *Q: Name the seven data-quality dimensions and give one SQL check for each.* **A**: See Part 27's full table and SQL examples - be ready to write at least the orphan-record and duplicate-record checks from memory.

**R. Data Governance** - *Q: What's the difference between data ownership and data stewardship?* **A**: Ownership is accountability for a dataset's correctness/definition; stewardship is the operational, day-to-day maintenance role - they're often different people (Part 44).

**S. Cloud Warehousing** - *Q: Contrast how Redshift and BigQuery each handle scaling.* **A**: Redshift's classic model couples storage and compute in a fixed-size cluster you must resize manually (RA3 nodes decouple this); BigQuery is fully serverless, scaling compute automatically per query with storage fully decoupled (Part 36).

**T. Data Architecture** - *Q: What is the medallion architecture and how does it map onto the classic Raw/Staging/Core layering?* **A**: Bronze=Raw, Silver=Staging/Integration, Gold=Core/Marts - same discipline, different naming, typically implemented on open lakehouse table formats (Part 35).

**U. Senior Data Engineer** - *Q: How would you redesign a fact table discovered to be mixing two grains?* **A**: Identify the finer of the two intended grains, split the table into two fact tables (one per grain), and re-derive any existing report logic that assumed the incorrect single table - reference the grain-mismatch mechanism from Part 2/14 explicitly.

**V. Data Architect** - *Q: How do you decide between Kimball, Inmon, and Data Vault for a new enterprise platform?* **A**: See Part 34's comparison table - the decision hinges on time-to-value needs, source volatility/auditability requirements, and organizational scale/regulation.

**W. Analytics Engineer** - *Q: How does a dbt-style staging/intermediate/marts layering prevent metric drift across dashboards?* **A**: Centralizing transformation logic and metric definitions in tested, documented, version-controlled models means every downstream consumer inherits the same logic, rather than re-deriving it independently per dashboard (Part 43, Part 42).

# PART 62 - DATABASE DESIGN CODING CHALLENGES (representative progression across all listed categories, extensible pattern)

**1. Schema design** - *Problem*: design a normalized schema for a library system (books, members, loans, reservations). *Solution*: `books(book_id PK, title, isbn UNIQUE)`, `members(member_id PK, name, email UNIQUE)`, `loans(loan_id PK, book_id FK, member_id FK, loan_date, due_date, return_date)`, `reservations(reservation_id PK, book_id FK, member_id FK, reserved_at)`.

**2. Normalization** - *Problem*: normalize a flat `student_courses(student_name, student_email, course1, course2, instructor1_dept)` table to 3NF. *Solution*: follow Part 4's exact three-step decomposition (1NF removes repeating course columns, 2NF splits attributes depending on only part of a composite key, 3NF splits instructor->department as a transitive dependency).

**3. Constraints** - *Problem*: add constraints to prevent a negative salary, an invalid department reference, and a duplicate email. *Solution*: `CHECK (salary >= 0)`, `FOREIGN KEY (department_id) REFERENCES departments`, `UNIQUE (email)`.

**4. Keys** - *Problem*: decide and justify natural vs surrogate key for a `country_lookup` table and a `customer` table. *Solution*: natural key (`country_code`) for the stable, standardized lookup table; surrogate key for `customer` since customer attributes and identity can change/merge over time (Part 3).

**5. Fact tables** - *Problem*: design a fact table for "website page views," stating its grain. *Solution*: "the grain of `fact_page_views` is one row per page-view event" - columns: `event_key, customer_key (nullable for anonymous), date_key, time_key, page_url, session_id`.

**6. Dimension tables** - *Problem*: design `dim_product` supporting both current and historical category assignments. *Solution*: SCD Type 2 with `effective_date`/`expiry_date`/`is_current` on the `category` attribute specifically (Part 19).

**7. SCD Type 2** - *Problem*: write the full close-then-insert SQL for a `dim_employee` job-title change. *Solution*: directly apply Part 19's two-statement pattern, substituting `job_title` for `city`.

**8. Date dimensions** - *Problem*: extend Part 20's `dim_date` to include a `is_holiday` flag populated from a holiday reference list. *Solution*: `UPDATE dim_date SET is_holiday = TRUE WHERE full_date IN (SELECT holiday_date FROM ref_holidays);`

**9. ETL transformations** - *Problem*: write a transformation cleaning and standardizing a raw `country` column with mixed values ('NG','Nigeria','nigeria'). *Solution*: `CASE WHEN UPPER(country) IN ('NG','NIGERIA') THEN 'NG' ELSE UPPER(LEFT(country,2)) END` (SQL-for-DE handbook Part 8 pattern).

**10. Data quality** - *Problem*: write a single query returning all seven data-quality violation counts for `fact_sales` in one result row. *Solution*: one `SELECT` with seven `SUM(CASE WHEN ... THEN 1 ELSE 0 END)` expressions, one per Part 27 check.

**11. Deduplication** - *Problem*: deduplicate a staging table on `(order_id, product_id)` keeping the latest `updated_at`. *Solution*: `ROW_NUMBER() OVER (PARTITION BY order_id, product_id ORDER BY updated_at DESC)`, filter `rn = 1` (SQL-for-DE handbook Part 11).

**12. Incremental loading** - *Problem*: design and write a watermark-based incremental load for a `customers` staging table. *Solution*: apply the watermark table pattern from the SQL-for-DE handbook Part 12, including correct `>=`/downstream-dedup boundary handling.

**13. CDC** - *Problem*: design a change table capturing inserts/updates/deletes for an `accounts` table and write the downstream MERGE applying it. *Solution*: directly apply Part 25's change-table and MERGE pattern.

**14. Warehouse queries** - *Problem*: write a query computing month-over-month revenue growth percentage from `fact_sales`. *Solution*:
```sql
WITH monthly AS (
    SELECT d.year, d.month, SUM(f.sales_amount) AS revenue
    FROM fact_sales f JOIN dim_date d ON d.date_key = f.date_key
    GROUP BY d.year, d.month
)
SELECT year, month, revenue,
       ROUND(100.0 * (revenue - LAG(revenue) OVER (ORDER BY year, month)) / NULLIF(LAG(revenue) OVER (ORDER BY year, month),0), 2) AS mom_growth_pct
FROM monthly;
```

# PART 64 - COMMON DATABASE DESIGN MISTAKES (curated 25, extensible pattern)

Each follows: why it's wrong -> what happens -> correct approach.

1. **No primary key** - Why wrong: rows aren't uniquely identifiable. What happens: duplicate rows accumulate silently, updates/deletes can't target a single row reliably. Fix: every table gets an explicit primary key, surrogate if no good natural key exists (Part 3).
2. **Wrong data types** - Why wrong: e.g. storing dates as TEXT. What happens: no native date arithmetic/validation, poor index usage, silent format inconsistency. Fix: always use the narrowest, most specific correct type (Part 39).
3. **Over-normalization** - Why wrong: normalizing analytical/reporting schemas as if they were OLTP. What happens: excessive joins, slow, hard-to-write reports. Fix: denormalize deliberately for analytical workloads (Part 5).
4. **Under-normalization** - Why wrong: OLTP schemas with repeating groups/redundant data. What happens: update anomalies, inconsistent data (Part 4). Fix: normalize to at least 3NF for transactional systems.
5. **Missing indexes** - Why wrong: frequently-filtered/joined columns have no index. What happens: full table scans, slow queries at scale. Fix: index foreign keys and frequently-filtered columns; verify with `EXPLAIN`.
6. **Too many indexes** - Why wrong: every column indexed "just in case." What happens: slow writes (every index must be maintained on every INSERT/UPDATE), wasted storage. Fix: index deliberately, based on actual query patterns, not preemptively.
7. **Poor foreign keys** - Why wrong: relationships enforced only in application code, not the database. What happens: orphaned rows, silent referential-integrity violations. Fix: always declare FK constraints (Part 7).
8. **Bad naming** - Why wrong: inconsistent casing, ambiguous abbreviations, reserved words as identifiers. What happens: constant confusion, slower onboarding, avoidable bugs. Fix: agree and document a naming standard (Part 8).
9. **Ignoring grain** - Why wrong: building a fact table without first stating what one row represents. What happens: inconsistent joins, ambiguous measures. Fix: always state grain explicitly before designing a fact table (Part 14).
10. **Mixing grains in fact tables** - Why wrong: some rows represent orders, others represent line items, in the same table. What happens: aggregations silently double-count or under-count depending on which rows are included. Fix: split into separate fact tables, one per grain (Part 17, Part 61-U).
11. **Using natural keys incorrectly** - Why wrong: using a mutable natural key as a dimension's primary key. What happens: breaks the moment the natural key changes or SCD Type 2 history is needed. Fix: use surrogate keys for dimensions (Part 3, Part 23).
12. **Incorrect SCD implementation** - Why wrong: inserting a new Type 2 row without closing the old one. What happens: two "current" rows for one entity (Part 61-M). Fix: always close-then-insert within one transaction.
13. **Duplicate dimensions** - Why wrong: two teams independently build their own version of `dim_customer`. What happens: numbers don't reconcile across reports (Part 30). Fix: govern and enforce conformed dimensions centrally.
14. **Non-conformed dimensions** - Why wrong: the "same" dimension has different keys/attributes/history in different marts. What happens: cross-mart analysis becomes impossible or wrong. Fix: same as above - central governance of shared dimensions.
15. **Poor date modeling** - Why wrong: computing date attributes live in every report instead of using a date dimension. What happens: inconsistent fiscal-period/holiday logic across reports (Part 20). Fix: build and use one governed `dim_date`.
16. **Mixing transactional and analytical workloads** - Why wrong: running heavy analytical queries directly against the OLTP production database. What happens: analytical scans lock/slow down the transactional workload the business depends on. Fix: separate OLTP and OLAP systems (Part 10).
17. **Building dashboards directly on OLTP systems** - A specific, very common instance of #16. What happens: dashboard refreshes degrade checkout/application performance. Fix: build a warehouse/mart layer; never point BI tools at production OLTP.
18. **No data-quality checks** - Why wrong: assuming loaded data is correct. What happens: silent corruption compounds undetected for months (Part 27). Fix: automated DQ checks on every load.
19. **No lineage** - Why wrong: nobody can trace where a number came from. What happens: debugging a wrong dashboard number becomes a multi-day investigation (Part 47). Fix: document/automate lineage from the start.
20. **No governance** - Why wrong: no owner for shared data assets. What happens: definitions drift, duplicate/non-conformed dimensions proliferate (Part 44). Fix: assign explicit ownership and stewardship.
21. **No incremental loading strategy** - Why wrong: full-reloading a huge table every run "because it's simpler." What happens: load times grow unsustainably as data grows, source systems get hammered repeatedly. Fix: design a watermark/CDC incremental strategy early (Part 24-25).
22. **Ignoring cardinality/selectivity when indexing** - Why wrong: indexing a low-cardinality column (e.g. a boolean flag) expecting it to speed up filters. What happens: the index is barely used because it doesn't narrow the result set enough to beat a scan. Fix: index high-selectivity columns; use low-cardinality columns as partition/cluster keys instead (Part 37-38, Part 40).
23. **No documented data dictionary** - Why wrong: column meanings live only in the original author's memory. What happens: new engineers guess, misinterpret, and build wrong logic on misunderstood columns. Fix: maintain a living data dictionary (Part 8).
24. **Uncontrolled schema changes** - Why wrong: altering a shared production table without checking downstream consumers. What happens: ETL/API/dashboard/ML breakage discovered only after it happens (Part 48). Fix: version-controlled migrations (Part 49) plus lineage-informed impact analysis.
25. **Treating denormalization as the default rather than the exception** - Why wrong: denormalizing OLTP tables "for performance" without need. What happens: update anomalies creep back into transactional systems that need write correctness most (Part 5). Fix: normalize OLTP by default; denormalize only for stated, deliberate analytical reasons.

# PART 65 - DATABASE DESIGN CODE REVIEW

**Sample bad schema under review**

```sql
CREATE TABLE order (
    id INT,
    cust_name VARCHAR(50),
    cust_email VARCHAR(50),
    product1 VARCHAR(50),
    product2 VARCHAR(50),
    amount FLOAT,
    order_date VARCHAR(20),
    region VARCHAR(50)
);
```

**Identify the flaws before reading on**: (pause and list them yourself, then compare)

| Issue category | Finding |
|---|---|
| Design flaw | `order` is a reserved word in most SQL dialects - will cause quoting headaches everywhere |
| Design flaw | No primary key at all - rows can't be uniquely identified or safely updated/deleted |
| Normalization issue | `product1`/`product2` are a repeating group - violates 1NF (Part 4) |
| Normalization issue | `cust_name`/`cust_email`/`region` repeat per order instead of living in a separate `customers` table - update anomaly risk |
| Key problem | No foreign keys at all - `region` is a bare string with no referential integrity to a `regions` table |
| Constraint issue | No `NOT NULL` on clearly-required columns (`cust_email`, `order_date`) |
| Constraint issue | No `CHECK` on `amount` to prevent negative values |
| Performance issue | `amount FLOAT` - floating point is inappropriate for money (rounding errors); should be `NUMERIC(12,2)` |
| Performance issue | `order_date VARCHAR(20)` - storing a date as text prevents date arithmetic and range-based indexing |
| Warehouse modeling issue | If this table is meant to feed analytics, it has no stated grain and mixes customer/order/product attributes at inconsistent grains in one flat table |

**Corrected design**

```sql
CREATE TABLE customers (
    customer_id   INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_name TEXT NOT NULL,
    email         TEXT UNIQUE NOT NULL,
    region_id     INT REFERENCES regions(region_id)
);
CREATE TABLE orders (
    order_id      INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_id   INT NOT NULL REFERENCES customers(customer_id),
    order_date    DATE NOT NULL,
    amount        NUMERIC(12,2) CHECK (amount >= 0)
);
CREATE TABLE order_items (
    order_id      INT REFERENCES orders(order_id),
    product_id    INT REFERENCES products(product_id),
    PRIMARY KEY (order_id, product_id)
);
```

Apply this same review lens - design flaws, normalization, keys, constraints, performance, security, warehouse modeling - to every schema before it ships; Part 63's coding-challenge and mistake catalogs (Part 64) are both good sources of additional bad-schema samples to practice this review process against.

# PART 66 - STANDARD VS PLATFORM-SPECIFIC DESIGN

**Table 66a - Identity/sequence generation, UUIDs, transactions, constraints (Standard / PostgreSQL / MySQL / SQL Server / Oracle / SQLite)**

| Feature | Standard | PostgreSQL | MySQL | SQL Server | Oracle | SQLite |
|---|---|---|---|---|---|---|
| Identity column | `GENERATED ALWAYS AS IDENTITY` | `GENERATED ALWAYS AS IDENTITY` / `SERIAL` | `AUTO_INCREMENT` | `IDENTITY(1,1)` | `GENERATED ALWAYS AS IDENTITY` | `AUTOINCREMENT` |
| Sequence object | `CREATE SEQUENCE` | `CREATE SEQUENCE` | not native (emulated) | `CREATE SEQUENCE` | `CREATE SEQUENCE` | not native |
| UUID generation | - | `gen_random_uuid()` (pgcrypto/core 13+) | `UUID()` | `NEWID()` | `SYS_GUID()` | via extension |
| Transactions | `BEGIN`/`COMMIT` | full ACID, MVCC | InnoDB: full ACID; MyISAM: no transactions | full ACID | full ACID | full ACID (single-writer) |
| CHECK constraints | Yes | Yes | Yes (8.0.16+) | Yes | Yes | Yes |
| Stored procedures | `CREATE PROCEDURE` | PL/pgSQL | Yes | T-SQL | PL/SQL | not supported natively |
| Materialized views | - | `CREATE MATERIALIZED VIEW` | not native (emulated with tables) | Indexed views (limited) | `CREATE MATERIALIZED VIEW` | not supported |
| JSON support | `JSON` type (SQL:2016) | `JSONB` (indexed, binary) | `JSON` type | `NVARCHAR` + `JSON_VALUE`/`OPENJSON` | `JSON` (21c+) / CLOB-based earlier | `JSON1` extension |

**Table 66b - Warehouse-oriented platforms (BigQuery / Snowflake / Redshift / Databricks)**

| Feature | BigQuery | Snowflake | Redshift | Databricks SQL |
|---|---|---|---|---|
| Identity/surrogate generation | `GENERATE_UUID()`, monotonic IDs not native | `AUTOINCREMENT` / sequence | `IDENTITY(1,1)` | UUID or surrogate via `ROW_NUMBER()`/hash |
| Partitioning | `PARTITION BY` (date/timestamp/int) | automatic micro-partitions | manual via sort/dist keys, not true partitioning | `PARTITIONED BY` (Delta) |
| Clustering/indexing | `CLUSTER BY` (up to 4 cols) | optional explicit clustering keys | `SORTKEY`/`DISTKEY` | `ZORDER BY` / liquid clustering |
| Materialized views | Yes, native | Yes, native | Yes, native | Yes, native (Delta) |
| JSON | native `STRUCT`/`ARRAY`, `JSON` type | `VARIANT` type, `:` notation | `SUPER` type | `from_json`, `:` notation on VARIANT-like columns |
| Transactions | limited (table-level, no multi-statement ACID historically) | full ACID (multi-statement) | full ACID | ACID via Delta Lake transaction log |
| Stored procedures | BigQuery scripting (`BEGIN...END` blocks) | JavaScript/SQL/Python stored procedures | limited (via PL/pgSQL-derived stored procedures) | limited (via notebooks/jobs more than in-warehouse procs) |
| Warehouse architecture | Fully serverless, no cluster management | Virtual warehouses, independently sized per workload | MPP clusters (or Serverless RPU-based) | Lakehouse: SQL Warehouses over Delta Lake |

Never assume a syntax or architectural detail from one platform transfers to another - always verify against the specific target platform's current documentation before implementing, especially for identity generation, JSON handling, and transaction semantics, which vary the most across this table.

# PART 67 - PROFESSIONAL DATABASE DOCUMENTATION

**1. ERD documentation** - entities, attributes, relationships, cardinality/optionality notation, generated from or kept in sync with the actual schema (many teams auto-generate ERDs from the live database to prevent drift).

**2. Data dictionary** - see Part 8's full template (column, type, description, nullable, default, key, business meaning, source system).

**3. Table specification** - table name, purpose (one sentence), grain (for fact tables, Part 14), owner, refresh cadence, row-count estimate, retention policy.

**4. Column specification** - name, type, nullable, default, business definition, valid value domain/CHECK constraints, PII/sensitivity classification (Part 46).

**5. Data-quality specification** - which of the seven dimensions (Part 27) apply to this table, the specific check for each, severity (hard-fail vs soft-warn), and current pass-rate history.

**6. ETL specification** - source, target, transformation logic (in plain language plus the actual SQL), load strategy (full/incremental/CDC - Part 24), schedule, dependencies (what must complete before this runs).

**7. Warehouse specification** - the star/snowflake schema diagram, fact table grains, dimension SCD types, conformance bus matrix (Part 32).

**8. Business glossary** - term, definition, owner, related metrics, example usage - the controlled vocabulary underlying the semantic layer (Part 42, Part 44).

**9. Metric dictionary** - metric name, exact formula, grain it's computed at, owner, which dashboards/reports use it, last-changed date.

**10. Architecture document** - the end-to-end diagram (Part 12-style layering), technology choices and why, data flow, security model, governance model, known limitations/trade-offs - the document a new senior hire should be able to read to understand the whole platform in one sitting.

Each of these templates exists to answer one recurring, expensive question ("what does this mean," "where did this come from," "is this trustworthy," "who do I ask") *before* someone has to interrupt a colleague or reverse-engineer the answer from SQL - the ROI on maintaining these compounds directly with organization size and platform age.

## Part 1 - Theory (sample of 10, extend to 50)

1. Define database design and distinguish it from data modeling. **A**: (Part 1-2) design is the overall process; modeling is the specific activity of defining entities/attributes/relationships within it.
2. What is a functional dependency? **A**: (Part 4) `A -> B` means the value of A determines the value of B.
3. Explain the difference between a candidate key and an alternate key. **A**: (Part 3) every candidate key could be the primary key; alternate keys are the candidate keys *not* chosen.
4. What is the difference between OLTP and OLAP schema design philosophy? **A**: (Part 10) normalized for write correctness vs dimensional/denormalized for read simplicity.
5. Define grain and explain why it must be decided first. **A**: (Part 14) what one fact-table row represents; every dimension/measure must be consistent with it.
6. What is a conformed dimension? **A**: (Part 30) a dimension shared, identically defined, across multiple facts/marts.
7. Contrast Kimball and Inmon. **A**: (Part 32-33) bottom-up dimensional-first vs top-down normalized-core-first.
8. What are the three Data Vault table types? **A**: (Part 34) Hub (business keys), Link (relationships), Satellite (descriptive history).
9. What is the medallion architecture? **A**: (Part 35) Bronze (raw) -> Silver (clean) -> Gold (business-ready).
10. Why do warehouse dimensions use surrogate keys? **A**: (Part 3, 23) to support SCD Type 2 history and decouple from source-system ID schemes.

## Part 2 - SQL / Database-Design Questions (sample of 8, extend to 50)

1. Write a CREATE TABLE for a normalized `orders`/`order_items` pair with correct keys and constraints.
```sql
CREATE TABLE orders (order_id INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY, customer_id INT NOT NULL, order_date DATE NOT NULL);
CREATE TABLE order_items (order_id INT REFERENCES orders(order_id), product_id INT, quantity INT CHECK (quantity > 0), PRIMARY KEY (order_id, product_id));
```
2. Write the SQL to add a CHECK constraint preventing negative account balances.
```sql
ALTER TABLE accounts ADD CONSTRAINT chk_balance CHECK (balance >= 0);
```
3. Write a full SCD Type 2 close-then-insert for a `dim_product` price-tier change.
```sql
UPDATE dim_product_scd2 SET expiry_date = CURRENT_DATE - 1, is_current = FALSE WHERE product_id = 55 AND is_current = TRUE;
INSERT INTO dim_product_scd2 (product_id, price_tier, effective_date, expiry_date, is_current)
VALUES (55, 'premium', CURRENT_DATE, DATE '9999-12-31', TRUE);
```
4. Write a query detecting fact rows with an unresolved dimension foreign key.
```sql
SELECT f.* FROM fact_sales f LEFT JOIN dim_product p ON p.product_key = f.product_key WHERE p.product_key IS NULL;
```
5. Write a query generating one calendar year into a `dim_date` table (any dialect, labeled).
```sql
-- [PostgreSQL]
INSERT INTO dim_date SELECT TO_CHAR(d,'YYYYMMDD')::INT, d FROM generate_series('2026-01-01'::DATE,'2026-12-31'::DATE,'1 day') AS d;
```
6. Write a query computing gross margin percentage by product category.
```sql
SELECT p.category, ROUND(100.0*SUM(f.sales_amount-f.cost_amount)/NULLIF(SUM(f.sales_amount),0),2) AS margin_pct
FROM fact_sales f JOIN dim_product p ON p.product_key=f.product_key GROUP BY p.category;
```
7. Normalize a flat `employee(name, dept_name, dept_manager)` table to 3NF and write the resulting CREATE TABLE statements.
```sql
CREATE TABLE departments (dept_id INT PRIMARY KEY, dept_name TEXT, dept_manager TEXT);
CREATE TABLE employees (employee_id INT PRIMARY KEY, name TEXT, dept_id INT REFERENCES departments(dept_id));
```
8. Write a masked view exposing only the last 4 digits of an account number.
```sql
CREATE VIEW accounts_masked AS SELECT account_id, CONCAT('****', RIGHT(account_number,4)) AS account_number_masked FROM accounts;
```

## Part 5 - Debugging Questions (sample of 5, extend to 25)

1. A star schema's `fact_sales` sum doesn't match finance's total by exactly the number of multi-item orders - diagnose. **A**: grain mismatch (Part 2, Part 64-#10) - an order-level amount was joined against line-item-grain rows without aggregating first.
2. Two rows in `dim_customer_scd2` both have `is_current = TRUE` for the same customer - diagnose. **A**: the close-step of Type 2 loading failed or was skipped (Part 19, Part 61-M).
3. A `dim_date` join returns no rows for a valid date - diagnose. **A**: likely a `date_key` format mismatch (e.g. `INT` `YYYYMMDD` vs a `DATE` type) between the fact and dimension.
4. A conformed dimension gives different results in two different marts - diagnose. **A**: the dimension has silently diverged (non-conformed duplicate, Part 64-#14) - trace lineage (Part 47) to find where the copies split.
5. A newly added NOT NULL column broke an existing ETL job - diagnose. **A**: a backward-incompatible schema change (Part 48) was deployed without checking/updating the dependent pipeline first.

## Part 6 - Scenario Questions (sample of 5, extend to 25)

1. The business wants to track "as of last Friday" reporting alongside "as of today" - what design change is needed? **A**: introduce or extend a periodic snapshot fact (Part 17) rather than relying solely on a current-state/transaction fact.
2. Marketing and Sales disagree on the definition of "Active Customer" - what process fixes this long-term? **A**: a governed semantic layer plus business glossary entry with one owner (Part 42, 44).
3. A source system will stop supporting timestamp-based CDC and only offers full daily extracts - how does the loading strategy change? **A**: fall back to full/snapshot loading (Part 24) with a hash-based change-detection layer (SQL-for-DE handbook Part 39) to still support the warehouse's SCD Type 2 logic.
4. A new regulation requires proving what a customer's risk rating was on any given past date - what design guarantees this? **A**: SCD Type 2 on `dim_customer.risk_rating` (Part 19), joined via `fact_*` rows that reference the correct historical `customer_key`.
5. Two acquired companies' customer databases must be merged into one warehouse - what process resolves conflicting records? **A**: MDM matching and survivorship (Part 45) feeding a newly conformed `dim_customer`.
