# PART 1 - DATA ENGINEERING AND SQL

**Data Engineering** is the discipline of designing, building, and operating the systems that move data from source systems into forms other people (analysts, scientists, ML models, executives) can reliably use. A **Data Engineer** builds pipelines, models data, guarantees quality, and keeps everything running - the plumbing beneath analytics and AI.

SQL is the most important Data Engineering skill because almost every warehouse, lake engine, and orchestration tool ultimately executes SQL as the transformation language. Python builds the pipeline scaffolding; SQL does the heavy data manipulation inside the database, where it is fastest.

**Role comparison**

| Role | Primary question | Primary tool |
|---|---|---|
| SQL Developer | How do I query/write this database correctly? | SQL |
| Data Analyst | What does the data say happened? | SQL + BI tools |
| Data Scientist | What will happen / why? | SQL + Python/R + ML |
| Data Engineer | How do I move, model, and guarantee this data reliably at scale? | SQL + Python + orchestration + cloud platforms |

**OLTP vs OLAP**

| | OLTP (operational) | OLAP (analytical) |
|---|---|---|
| Purpose | Run the business (transactions) | Understand the business (analysis) |
| Schema | Highly normalized (3NF) | Denormalized / dimensional (star schema) |
| Query pattern | Short, many, row-level | Long, few, scan-heavy, aggregating |
| Example | Postgres backing a checkout API | Snowflake warehouse behind BI dashboards |

**Core storage concepts**

- **Data warehouse**: curated, modeled, query-optimized store of historical, integrated data for analytics (Snowflake, BigQuery, Redshift).
- **Data lake**: cheap, schema-on-read storage for raw data in any format (S3, GCS, ADLS).
- **Data lakehouse**: lake storage + warehouse-style transactional/query guarantees (Delta Lake, Iceberg, Databricks).
- **Data mart**: a subject-specific slice of the warehouse (e.g. `sales_mart`) for a single team/domain.
- **Data platform**: the whole stack - ingestion, storage, transformation, orchestration, governance, serving.
- **Data pipeline / workflow**: the ordered set of steps that move and transform data, usually orchestrated as a DAG.

**ETL vs ELT vs Reverse ETL**

- **ETL** (Extract, Transform, Load): transform *before* loading into the warehouse - traditional, used when the target is limited or compute is external.
- **ELT** (Extract, Load, Transform): load raw data first, transform *inside* the warehouse using SQL - dominant pattern with modern cloud warehouses because compute is cheap and elastic there.
- **Reverse ETL**: push modeled warehouse data back out into operational tools (CRM, ad platforms, Slack).

**Batch vs streaming vs micro-batch**

- **Batch**: process data in scheduled chunks (hourly/daily). Simple, cheap, latency of hours.
- **Streaming**: process events continuously as they arrive (Kafka, Flink). Complex, low latency (seconds).
- **Micro-batching**: streaming implemented as very frequent small batches (e.g. every 1-5 minutes) - a practical middle ground (Spark Structured Streaming default mode).

**SQL's role at every stage**

```
Source System            -- OLTP DB, SaaS API, files, events
   -> Ingestion           -- COPY / bulk load / CDC captures rows AS SQL rows
   -> Raw / Staging       -- SQL CREATE TABLE AS SELECT (CTAS), preserves source shape
   -> Cleaning            -- SQL TRIM/CAST/COALESCE/dedup logic
   -> Transformation      -- SQL joins, aggregations, business logic
   -> Data Warehouse      -- SQL dimensional model (facts/dimensions)
   -> Data Mart           -- SQL views/tables scoped to one domain
   -> BI Dashboard        -- SQL queries issued live by the BI tool
```

At every arrow, the artifact that moves is (almost always) the *result of a SQL statement*.

# PART 2 - DATA ENGINEERING SQL MENTAL MODEL

Professional Data Engineers do not think "what SELECT gets the answer" - they think "what is the shape, source of truth, and lifecycle of this data."

**Data as a pipeline vs data as a product**
- *Pipeline thinking*: data is a byproduct of moving bytes from A to B.
- *Product thinking*: a table is a product with an owner, a schema contract, an SLA, and consumers who depend on it. Modern data engineering (data mesh, dbt culture) treats tables as products.

**Data contracts, lineage, dependencies, source of truth**
- A **data contract** is an agreement (schema + semantics + freshness + quality guarantees) between a data producer and its consumers.
- **Lineage** traces where a column's value came from, table by table, transformation by transformation.
- **Dependencies** are the DAG edges: table B depends on table A, so A must be fresh and correct before B runs.
- **Source of truth**: the one system/table that is authoritative for a fact - critical when the same entity appears in multiple systems (CRM vs billing vs product DB).

**Data grain - the single most important modeling idea**

*Grain* = what one row represents. Every table must have an explicit, documented, enforced grain.

| Grain | Meaning |
|---|---|
| One row per customer | `customers` dimension |
| One row per order | `orders` fact (order header) |
| One row per order item | `order_items` fact (line-item grain) |
| One row per transaction | `transactions` fact |
| One row per customer per day | daily activity snapshot |
| One row per product per day | daily inventory snapshot |
| One row per event | raw event/clickstream table |

**Why wrong grain breaks everything**

```sql
-- WRONG: joining orders (order grain) to order_items (line-item grain)
-- without aggregating first silently duplicates the order-level amount
-- once per line item.
SELECT o.order_id, o.order_total, oi.product_id
FROM orders o
JOIN order_items oi ON oi.order_id = o.order_id;
-- SUM(order_total) here is now inflated by the number of items per order.
```

```sql
-- CORRECT: aggregate to the same grain before joining, or sum the
-- line-item amount instead of re-using the header total.
SELECT o.order_id, SUM(oi.line_amount) AS computed_total
FROM orders o
JOIN order_items oi ON oi.order_id = o.order_id
GROUP BY o.order_id;
```

Grain mistakes are the #1 cause of "the dashboard numbers don't match finance" incidents in real companies. Immutable (append-only event/log) data and mutable (updatable state) data must also be modeled differently - mutable data needs SCD logic (Part 16); immutable data can simply be appended and deduplicated by natural key + timestamp.

# PART 3 - DATABASE LAYERS FOR DATA ENGINEERING

A professional warehouse is organized in explicit layers so that each transformation step is testable, re-runnable, and owned independently.

```
1. Source        -- the operational system itself (not touched by analytics)
2. Raw            -- exact copy of source, untransformed, append-only
3. Staging        -- typed, renamed, lightly cleaned, still 1:1 with source
4. Intermediate   -- joined/reshaped building blocks, not yet business-facing
5. Core           -- clean, deduplicated, conformed entities (the "warehouse")
6. Dimension      -- descriptive entities (who/what/where)
7. Fact           -- measurable events (how much/how many/when)
8. Data Mart      -- domain-specific, business-friendly aggregates
9. Presentation   -- final BI-ready views/tables, one query away from a chart
```

```sql
-- 2. RAW: land exactly what arrived, add ingestion metadata, never transform
CREATE TABLE raw.customers_raw (
    payload         JSONB,             -- [PostgreSQL] JSON kept as-is
    source_system   TEXT,
    ingestion_ts    TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    batch_id        TEXT
);

-- 3. STAGING: typed, renamed, 1:1 with source rows
CREATE TABLE staging.stg_customers AS
SELECT
    (payload ->> 'customer_id')::INT   AS customer_id,
    TRIM(payload ->> 'name')           AS customer_name,
    LOWER(TRIM(payload ->> 'email'))   AS email,
    (payload ->> 'created_at')::TIMESTAMP AS created_at,
    ingestion_ts,
    batch_id
FROM raw.customers_raw;

-- 4. INTERMEDIATE: reusable joined building block, not yet exposed to BI
CREATE TABLE intermediate.int_customer_orders AS
SELECT c.customer_id, c.customer_name, o.order_id, o.order_total
FROM staging.stg_customers c
JOIN staging.stg_orders o ON o.customer_id = c.customer_id;

-- 6/7. DIMENSION and FACT (core warehouse)
CREATE TABLE core.dim_customer (
    customer_key    SERIAL PRIMARY KEY,   -- surrogate key
    customer_id     INT UNIQUE,           -- business/natural key
    customer_name   TEXT,
    email           TEXT
);

CREATE TABLE core.fact_orders (
    order_key       SERIAL PRIMARY KEY,
    customer_key    INT REFERENCES core.dim_customer(customer_key),
    order_date_key  INT,
    order_amount    NUMERIC(12,2)
);

-- 8. MART: business-domain aggregate for one team
CREATE TABLE mart.sales_mart AS
SELECT c.customer_name, SUM(f.order_amount) AS total_sales
FROM core.fact_orders f
JOIN core.dim_customer c ON c.customer_key = f.customer_key
GROUP BY c.customer_name;
```

Separating layers means: a bug in transformation logic is isolated to one layer, raw data is always recoverable (nothing destructive ever touches `raw`), and each layer can be tested independently (Part 35).

# PART 4 - ETL WITH SQL

**Extract -> Transform -> Load**: transform happens *before* the data lands in its final home, typically in a staging area or external compute engine.

Realistic raw customer feed:

```
customer_id | name          | email            | phone         | country | created_at
1           |  John doe     | JOHN@Test.com    | 08031234567   | NG      | 2026-01-05
2           | Mary O'Neil   | mary@test.com    |  +234-803-000 | Nigeria | 2026-01-06
2           | Mary O'Neil   | mary@test.com    |  +234-803-000 | Nigeria | 2026-01-06   -- duplicate
```

```sql
-- Full clean/standardize/validate/dedupe/transform pipeline, staged in one CTE chain
WITH extracted AS (
    SELECT * FROM raw.customers_raw
),
cleaned AS (
    SELECT
        customer_id,
        TRIM(INITCAP(name))                              AS customer_name,
        LOWER(TRIM(email))                                AS email,
        REGEXP_REPLACE(phone, '[^0-9]', '', 'g')          AS phone_digits,   -- [PostgreSQL]
        CASE
            WHEN UPPER(country) IN ('NG','NIGERIA') THEN 'NG'
            ELSE UPPER(LEFT(country,2))
        END                                                AS country_code,
        created_at
    FROM extracted
),
validated AS (
    SELECT *,
        (email ~ '^[^@]+@[^@]+\.[^@]+$')                  AS is_valid_email  -- [PostgreSQL] regex match
    FROM cleaned
    WHERE customer_id IS NOT NULL
),
deduplicated AS (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id ORDER BY created_at DESC
        ) AS rn
    FROM validated
)
-- LOAD: only the final, clean, deduplicated, valid rows reach the target
INSERT INTO staging.stg_customers
    (customer_id, customer_name, email, phone_digits, country_code, created_at)
SELECT customer_id, customer_name, email, phone_digits, country_code, created_at
FROM deduplicated
WHERE rn = 1 AND is_valid_email;
```

**Staging table types**: temporary staging (session-scoped, used mid-pipeline and discarded), permanent staging (retained for reprocessing/audit), transformation tables (intermediate outputs of a specific step). Choose permanent staging whenever you need to re-run downstream steps without re-extracting from the source - extraction is usually the most expensive/fragile step.

**Common ETL failure scenarios**: source schema changes mid-extract, transformation step throws on an unexpected NULL/type, partial load (job dies halfway), duplicate load (job re-run without a checkpoint), and downstream tables built on a transform that silently produced wrong values. All are addressed later (Parts 28-30).

# PART 5 - ELT WITH SQL

**Extract -> Load -> Transform**: raw data is loaded into the warehouse *first*, and all transformation happens afterward as SQL running inside the warehouse.

**ETL vs ELT**

| | ETL | ELT |
|---|---|---|
| Transform location | External engine (Spark, Informatica) | Inside the warehouse |
| Warehouse load | Only clean data | All raw data, then cleaned in place |
| Flexibility | Must know transformations up front | Raw history always available to re-transform |
| Cost model | Fixed ETL server cost | Pay for warehouse compute on demand |
| Modern fit | Legacy / on-prem | Cloud warehouses (Snowflake, BigQuery, Redshift, Databricks) |

Cloud warehouses favor ELT because storage and compute are decoupled and elastically scalable - it is cheap to store *everything* raw and expensive compute is only spent when you actually run a transform, and you can always re-derive a new model from the same raw history without re-extracting from the source system.

```sql
-- EXTRACT + LOAD: raw JSON landed as-is (fast bulk load, no transform)
COPY raw.orders_raw FROM 's3://bucket/orders/2026-01-31.json'
IAM_ROLE 'arn:aws:iam::123456789012:role/redshift-role'
FORMAT AS JSON 'auto';                                    -- [Redshift]

-- TRANSFORM: happens entirely inside the warehouse, in SQL, after load
CREATE TABLE core.fact_orders AS
SELECT
    (payload:order_id)::INT           AS order_id,        -- [Snowflake] JSON dot-notation
    (payload:customer_id)::INT        AS customer_id,
    (payload:amount)::NUMERIC(12,2)   AS order_amount,
    (payload:order_date)::DATE        AS order_date
FROM raw.orders_raw;
```

Because storage is decoupled from compute, ELT lets you reprocess history cheaply: if a transformation bug is found, fix the SQL and re-run it against the untouched raw layer - no need to re-pull from the source system.

# PART 6 - STAGING TABLES

Staging is the contract between "whatever the source sent" and "what the warehouse understands."

**Staging variants**
- **Raw staging**: byte-for-byte copy of the source payload (JSON/CSV as-is).
- **Clean staging**: typed and lightly normalized, still one row per source row.
- **Temporary staging**: exists only for the duration of a run (session temp table / CTE).
- **Persistent staging**: retained across runs for audit and reprocessing.
- **Landing tables**: the very first table data touches after ingestion, before any staging logic runs.

```sql
CREATE TABLE staging.stg_orders (
    order_id            INT,
    customer_id         INT,
    order_amount         NUMERIC(12,2),
    order_date           DATE,
    -- metadata columns - always add these in staging
    source_system        TEXT,           -- which system this row came from
    ingestion_timestamp   TIMESTAMP,       -- when the pipeline pulled it
    batch_id              TEXT,            -- which run loaded this row
    pipeline_run_id       TEXT,            -- links row to the orchestration run
    record_hash           TEXT,            -- hash of business columns (Part 39)
    loaded_at              TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

These metadata columns matter because: `source_system` lets you merge multiple upstream systems safely; `ingestion_timestamp`/`batch_id`/`pipeline_run_id` make every row traceable back to the exact run that produced it (critical for debugging and reconciliation); `record_hash` makes change detection and deduplication O(1) comparisons instead of comparing every column by hand.

# PART 7 - DATA INGESTION

SQL's ingestion primitives, from slowest to fastest:

```sql
-- Row-by-row INSERT: correct but slow at scale - one round trip per row
INSERT INTO staging.stg_customers VALUES (1, 'John', 'john@test.com');

-- Bulk loading: single operation, minimal logging, dramatically faster
COPY staging.stg_customers FROM '/data/customers.csv' WITH (FORMAT csv, HEADER true);   -- [PostgreSQL]
LOAD DATA INFILE '/data/customers.csv' INTO TABLE stg_customers
    FIELDS TERMINATED BY ',' LINES TERMINATED BY '\n' IGNORE 1 ROWS;                     -- [MySQL]
BULK INSERT stg_customers FROM 'C:\data\customers.csv'
    WITH (FIRSTROW = 2, FIELDTERMINATOR = ',', ROWTERMINATOR = '\n');                    -- [SQL Server]
-- Oracle: SQL*Loader / external tables are the bulk-load mechanism
-- BigQuery/Snowflake/Redshift: LOAD/COPY INTO from cloud storage (Parts 46-49)
```

**Row-by-row vs bulk**: row-by-row INSERT pays per-statement overhead (parsing, logging, index maintenance, network round trip) for every single row; bulk loading amortizes that overhead across the whole file, often minimally logs the operation, and can load millions of rows per minute instead of per hour. Always prefer bulk loading for ingestion; reserve row-level INSERT for small, transactional, application-driven writes.

**External tables**: define a table schema over files sitting in object storage (S3/GCS/ADLS) without physically loading them - useful for exploring a lake before committing to a warehouse load, or for infrequently queried archives.

# PART 8 - DATA CLEANING FOR DATA ENGINEERING

```sql
SELECT
    customer_id,
    TRIM(name)                                             AS name_trimmed,
    LOWER(email)                                           AS email_lower,
    REPLACE(phone, ' ', '')                                AS phone_no_spaces,
    REGEXP_REPLACE(phone, '[^0-9+]', '', 'g')              AS phone_digits_only,       -- [PostgreSQL]
    CAST(created_at AS DATE)                                AS created_date,
    COALESCE(country, 'UNKNOWN')                            AS country_filled,
    NULLIF(TRIM(middle_name), '')                            AS middle_name_or_null,   -- '' becomes NULL
    CASE
        WHEN amount ~ '^[0-9]+(\.[0-9]+)?$' THEN CAST(amount AS NUMERIC)                -- [PostgreSQL]
        ELSE NULL
    END                                                       AS amount_safe
FROM raw.customers_stage;
```

**Safe casting across dialects** (avoid a whole batch failing because one row has bad data):

```sql
SELECT TRY_CAST(amount AS NUMERIC(12,2))   FROM raw.orders;    -- [SQL Server] / [Snowflake]
SELECT TRY_CONVERT(NUMERIC(12,2), amount)  FROM raw.orders;    -- [SQL Server] alternative
SELECT SAFE_CAST(amount AS NUMERIC)        FROM raw.orders;    -- [BigQuery]
-- PostgreSQL/MySQL/Oracle have no built-in TRY_CAST before very recent versions;
-- use a CASE + regex guard, or wrap in a PL/pgSQL function that catches the exception.
```

**Common cleaning categories**: NULL/missing values (`COALESCE`, explicit defaults, or leave NULL and document why), invalid values (regex/range checks), duplicate records and duplicate *business keys* (different from exact duplicates - Part 11), invalid dates/numbers (validate before cast), whitespace and case normalization, Unicode normalization (accents, full-width characters), phone/email/currency normalization to one canonical format, and country standardization (always to ISO codes, never free text).

# PART 9 - DATA VALIDATION

```sql
-- Schema / null validation: customer_id must never be NULL
SELECT COUNT(*) AS violations FROM orders WHERE customer_id IS NULL;

-- Range validation: amount must be non-negative
SELECT COUNT(*) AS violations FROM orders WHERE order_amount < 0;

-- Format validation: date cannot be in the future
SELECT COUNT(*) AS violations FROM orders WHERE order_date > CURRENT_DATE;

-- Format validation: email pattern
SELECT COUNT(*) AS violations
FROM customers
WHERE email !~ '^[^@\s]+@[^@\s]+\.[^@\s]+$';                 -- [PostgreSQL]

-- Referential integrity: every order's customer_id must exist in customers
SELECT COUNT(*) AS orphaned_orders
FROM orders o
LEFT JOIN customers c ON c.customer_id = o.customer_id
WHERE c.customer_id IS NULL;

-- Uniqueness validation: no duplicate business key
SELECT customer_id, COUNT(*)
FROM customers
GROUP BY customer_id
HAVING COUNT(*) > 1;

-- Freshness validation: latest row must be within expected SLA
SELECT MAX(loaded_at) AS latest_load,
       CURRENT_TIMESTAMP - MAX(loaded_at) AS staleness
FROM staging.stg_orders;

-- Completeness validation: expected row count roughly matches source
SELECT COUNT(*) FROM staging.stg_orders WHERE batch_id = 'run_2026_02_01';
```

Each of these checks is a single testable SQL statement returning a violation count - `0` means pass. Part 10 shows how to turn this pattern into a repeatable framework, and Part 35 shows how to run it as automated pipeline tests.

# PART 10 - DATA QUALITY ENGINEERING

The seven classic data-quality dimensions:

1. **Accuracy** - does the value reflect reality?
2. **Completeness** - are required fields populated?
3. **Consistency** - does the same fact agree across tables/systems?
4. **Validity** - does the value conform to its expected format/domain?
5. **Uniqueness** - is each entity represented exactly once at its grain?
6. **Timeliness** - is the data fresh enough to be useful?
7. **Integrity** - do relationships between tables hold (FKs, business rules)?

```sql
-- Build a reusable, auditable data-quality framework in plain SQL

CREATE TABLE dq.dq_results (
    check_name       TEXT,
    table_name       TEXT,
    run_ts           TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    violations       BIGINT,
    passed           BOOLEAN
);

-- Each check is inserted as one row - here, "orders.customer_id not null"
INSERT INTO dq.dq_results (check_name, table_name, violations, passed)
SELECT
    'orders_customer_id_not_null'          AS check_name,
    'orders'                                AS table_name,
    COUNT(*)                                AS violations,
    COUNT(*) = 0                            AS passed
FROM orders
WHERE customer_id IS NULL;

-- Quarantine failing rows instead of silently dropping or silently loading them
CREATE TABLE dq.quarantine_orders AS
SELECT *, 'customer_id_null' AS failure_reason, CURRENT_TIMESTAMP AS quarantined_at
FROM orders
WHERE customer_id IS NULL;

-- A quality score per table: percentage of checks passed
SELECT table_name,
       ROUND(100.0 * SUM(CASE WHEN passed THEN 1 ELSE 0 END) / COUNT(*), 1) AS quality_score_pct
FROM dq.dq_results
WHERE run_ts::DATE = CURRENT_DATE
GROUP BY table_name;
```

**Thresholds and gating**: a pipeline can be configured to hard-fail (stop the DAG) if `quality_score_pct` drops below a threshold (e.g. 98%), or soft-warn (load anyway but alert) for lower-severity checks. Quarantine/error/audit tables let you keep bad rows for investigation without letting them corrupt downstream models.

# PART 11 - DEDUPLICATION

**Exact duplicates** = every column identical. **Business-key duplicates** = same natural key (e.g. `customer_email`) but different metadata (e.g. `updated_at`) - these are the dangerous ones because a naive `DISTINCT` won't catch them.

```sql
-- Identify duplicates by business key
SELECT customer_email, COUNT(*) AS n
FROM staging.stg_customers
GROUP BY customer_email
HAVING COUNT(*) > 1;

-- Keep the latest record per business key, drop the rest
WITH ranked AS (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_email
            ORDER BY updated_at DESC
        ) AS rn
    FROM staging.stg_customers
)
SELECT * FROM ranked WHERE rn = 1;
```

Line by line: `PARTITION BY customer_email` groups rows sharing the same natural key into independent windows; `ORDER BY updated_at DESC` ranks the newest row first inside each group; `ROW_NUMBER()` assigns 1, 2, 3... within each partition; filtering `rn = 1` keeps only the single most recent row per email, discarding every earlier version.

**Source-priority logic**: when duplicates come from *different* source systems (e.g. CRM vs billing both have a customer record), rank by a source-priority column instead of/in addition to recency:

```sql
ROW_NUMBER() OVER (
    PARTITION BY customer_email
    ORDER BY CASE source_system WHEN 'crm' THEN 1 WHEN 'billing' THEN 2 ELSE 3 END,
             updated_at DESC
)
```

**Fuzzy matching** (names/addresses that are "the same" but not byte-identical, e.g. "Jon Doe" vs "John Doe") is handled conceptually with similarity functions - `SOUNDEX`/`DIFFERENCE` [SQL Server], `SIMILARITY` with `pg_trgm` [PostgreSQL], or, more robustly, an external entity-resolution/ML step - plain SQL equality is not enough for this class of problem.

# PART 12 - INCREMENTAL DATA LOADING

**Full load** reprocesses the entire source every run - simple, safe, but expensive and slow as data grows. **Incremental load** processes only new/changed rows since the last successful run.

Key vocabulary: **watermark** / **high-water mark** = the last value (usually a timestamp or sequence number) successfully processed; **last processed timestamp** = the watermark's value; **batch ID** = identifier for a specific run; **sequence-based loading** = using a monotonic ID instead of a timestamp as the watermark (safer when clocks can be skewed).

```sql
-- Store the watermark somewhere durable between runs
CREATE TABLE etl.watermarks (
    pipeline_name TEXT PRIMARY KEY,
    last_watermark TIMESTAMP
);

-- First load: 2026-01-01 to 2026-01-31
INSERT INTO staging.stg_orders
SELECT * FROM source.orders
WHERE updated_at >= '2026-01-01' AND updated_at < '2026-02-01';

UPDATE etl.watermarks SET last_watermark = '2026-01-31 23:59:59'
WHERE pipeline_name = 'orders_incremental';

-- Next load: only pulls rows changed since the stored watermark
INSERT INTO staging.stg_orders
SELECT o.*
FROM source.orders o, etl.watermarks w
WHERE w.pipeline_name = 'orders_incremental'
  AND o.updated_at > w.last_watermark;    -- strictly greater than, see boundary note below

UPDATE etl.watermarks SET last_watermark = (SELECT MAX(updated_at) FROM staging.stg_orders)
WHERE pipeline_name = 'orders_incremental';
```

**The `>` vs `>=` boundary problem**: if you use `>=` against the *exact same* watermark value on the next run, the row that produced that watermark gets reprocessed (duplicate). If you use `>` you avoid that duplicate, but if two rows share the *exact same* `updated_at` timestamp and only one was captured before the watermark was recorded, the second is silently skipped. The robust fix is either (a) use `>=` plus deduplicate downstream by business key + `ROW_NUMBER()`, or (b) use a strictly increasing sequence/ID as the watermark instead of a timestamp, or (c) always set the new watermark to `MAX(updated_at) - small_safety_buffer` and use `>=`, accepting a few reprocessed rows that your idempotent MERGE step will simply overwrite harmlessly.

# PART 13 - UPSERT PATTERNS

"Upsert" = insert new rows, update existing ones, in one logical operation, keyed by business key.

```sql
-- [PostgreSQL] / [SQLite] - ON CONFLICT
INSERT INTO dim_customer (customer_id, name, email)
VALUES (101, 'John Doe', 'john@test.com')
ON CONFLICT (customer_id)
DO UPDATE SET name = EXCLUDED.name, email = EXCLUDED.email;

-- [MySQL] - ON DUPLICATE KEY UPDATE
INSERT INTO dim_customer (customer_id, name, email)
VALUES (101, 'John Doe', 'john@test.com')
ON DUPLICATE KEY UPDATE name = VALUES(name), email = VALUES(email);

-- [SQL Server] / [Oracle] / [Snowflake] / [Databricks SQL] - MERGE
MERGE INTO dim_customer AS t
USING staging.stg_customers AS s
ON t.customer_id = s.customer_id
WHEN MATCHED THEN
    UPDATE SET name = s.name, email = s.email
WHEN NOT MATCHED THEN
    INSERT (customer_id, name, email) VALUES (s.customer_id, s.name, s.email);
```

**MERGE caveats**: MERGE is powerful but has historically had subtle bugs (e.g. SQL Server MERGE under concurrent writes, and it can silently do more than intended if the `ON` clause matches multiple source rows to one target row). For high-stakes production upserts, many teams prefer an explicit, auditable two-step pattern:

```sql
-- Safer explicit pattern: stage, then UPDATE existing + INSERT new separately
UPDATE dim_customer t
SET name = s.name, email = s.email
FROM staging.stg_customers s
WHERE t.customer_id = s.customer_id;                       -- [PostgreSQL] UPDATE...FROM

INSERT INTO dim_customer (customer_id, name, email)
SELECT s.customer_id, s.name, s.email
FROM staging.stg_customers s
LEFT JOIN dim_customer t ON t.customer_id = s.customer_id
WHERE t.customer_id IS NULL;
```

This two-step form is more verbose but easier to reason about, test independently, and log row counts for (`rows_updated` vs `rows_inserted` becomes two separate, auditable statements - directly feeding Part 30's audit table).

# PART 14 - CHANGE DATA CAPTURE (CDC)

**CDC** captures every `INSERT`/`UPDATE`/`DELETE` that happens to a source table, as it happens, instead of periodically re-scanning the whole table.

**Why it matters**: full/incremental SQL scans still cost read overhead on the source system and can miss deletes entirely (a deleted row just disappears - a timestamp-based incremental load never sees it). CDC solves both.

**Types of CDC**
- **Log-based CDC**: reads the database's internal transaction log directly (no extra load on the source tables). This is the gold standard.
- **Trigger-based CDC**: a database trigger writes every change into a separate "change table" - adds write overhead to every transaction.
- **Timestamp-based change detection**: what Part 12 does - simple but misses deletes and depends on a reliable `updated_at` column.
- **Query-based CDC**: periodically diff two snapshots to infer changes - expensive, only for small tables.

**Underlying mechanisms** (conceptually): every relational database keeps an internal log of every write for crash recovery and replication - PostgreSQL's **WAL** (Write-Ahead Log), MySQL's **binlog**, Oracle's **redo log**. Log-based CDC tools (e.g. **Debezium**) tail this log and turn each entry into a structured change event (operation code `I`/`U`/`D`, before-value, after-value), typically publishing it onto a stream like **Kafka** for downstream consumers.

```sql
-- A simplified "change table" as trigger-based CDC would populate it
CREATE TABLE cdc.orders_changes (
    change_id      BIGSERIAL PRIMARY KEY,
    operation      CHAR(1),          -- 'I', 'U', 'D'
    order_id       INT,
    before_amount  NUMERIC(12,2),
    after_amount   NUMERIC(12,2),
    changed_at     TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Downstream SQL consuming a CDC stream applies changes like an upsert/delete
MERGE INTO warehouse.orders AS t
USING cdc.orders_changes AS c
ON t.order_id = c.order_id
WHEN MATCHED AND c.operation = 'D' THEN DELETE
WHEN MATCHED AND c.operation = 'U' THEN UPDATE SET order_amount = c.after_amount
WHEN NOT MATCHED AND c.operation = 'I' THEN INSERT (order_id, order_amount) VALUES (c.order_id, c.after_amount);
```

You do not need to know Kafka in depth to work with CDC as a Data Engineer: conceptually, it is a durable, ordered, replayable message log that CDC tools publish change events onto, and warehouse-loading jobs consume from.

# PART 15 - SNAPSHOTS

A **snapshot** captures the full state of a table (or a defined subset) at a point in time, rather than tracking individual changes.

```sql
-- Daily full snapshot: one partition per day, entire table copied
CREATE TABLE snapshots.inventory_20260207 AS
SELECT *, DATE '2026-02-07' AS snapshot_date
FROM source.inventory;

-- Incremental snapshot: only rows that changed since yesterday's snapshot
INSERT INTO snapshots.inventory_incremental
SELECT i.*, CURRENT_DATE AS snapshot_date
FROM source.inventory i
JOIN snapshots.inventory_20260206 y ON y.product_id = i.product_id
WHERE i.updated_at > '2026-02-06';
```

**When snapshots beat CDC**: snapshots are simpler to build/reason about, work even when you cannot access the source's transaction log (e.g. a third-party SaaS API with no CDC support), and are the natural fit for point-in-time reporting (e.g. "what was our inventory at end of each month"). CDC is preferable when you need every intermediate change (e.g. audit trails, fraud detection) or the table is too large to snapshot fully on every run.

# PART 16 - SLOWLY CHANGING DIMENSIONS (SCD)

| Type | Behavior |
|---|---|
| Type 0 | Never changes - fixed attribute (e.g. date of birth) |
| Type 1 | Overwrite old value - no history kept |
| Type 2 | Insert new row, keep full history, flag current |
| Type 3 | Keep limited history in extra columns ("previous_value") |
| Type 4 | Separate current table + separate history table |
| Type 6 | Hybrid of 1 + 2 + 3 |

**Example change**: customer city changes from Lagos to Abuja.

```sql
-- SCD TYPE 1: overwrite - simple, but the fact "customer was in Lagos" is gone
UPDATE dim_customer SET city = 'Abuja' WHERE customer_id = 101;
```

**SCD Type 2 - full implementation**

```sql
CREATE TABLE dim_customer_scd2 (
    customer_key    SERIAL PRIMARY KEY,   -- surrogate key, unique per version
    customer_id     INT,                   -- business key, repeats across versions
    city            TEXT,
    effective_from  DATE,
    effective_to    DATE,                  -- '9999-12-31' or NULL while current
    is_current      BOOLEAN
);

-- Step 1: close out the old version
UPDATE dim_customer_scd2
SET effective_to = CURRENT_DATE - 1, is_current = FALSE
WHERE customer_id = 101 AND is_current = TRUE;

-- Step 2: insert the new version
INSERT INTO dim_customer_scd2 (customer_id, city, effective_from, effective_to, is_current)
VALUES (101, 'Abuja', CURRENT_DATE, DATE '9999-12-31', TRUE);
```

`customer_key` (surrogate key) is what fact tables join to - it uniquely identifies *one version* of a customer. `customer_id` (business key) identifies the *real-world entity* across all its versions. `effective_from`/`effective_to` bound the validity window of each version; `is_current` is a convenience flag for "give me the customer as they are today." Any fact row loaded while the customer lived in Lagos correctly joins to the Lagos version of the dimension forever - this is exactly why Type 2 preserves historical accuracy that Type 1 destroys.

# PART 17 - DATA WAREHOUSE DESIGN

**Dimensional modeling** organizes a warehouse around business processes (facts) and the descriptive context around them (dimensions), optimized for query simplicity and performance over write-normalization.

- **Star schema**: one central fact table joined directly to denormalized dimension tables - fast, simple, most common.
- **Snowflake schema**: dimensions are further normalized into sub-dimensions - saves storage, costs extra joins; rarely worth it on modern column stores.
- **Bridge table**: resolves many-to-many relationships between a fact and a dimension (e.g. one order has many sales reps).
- **Junk dimension**: bundles several low-cardinality flags/indicators into one small dimension table to avoid cluttering the fact table with many tiny FK columns.
- **Degenerate dimension**: a dimension-like attribute (e.g. `order_number`) stored directly in the fact table with no separate dimension table because it has no other attributes.
- **Role-playing dimension**: one physical dimension used multiple times with different meanings (e.g. `dim_date` joined as both `order_date` and `ship_date`).
- **Conformed dimension**: the same dimension (e.g. `dim_customer`) shared consistently across multiple fact tables/marts so metrics are comparable across domains.

**Fact table grain, worked example**: for a `fact_sales` table, deciding the grain is the single most important design decision:

| Candidate grain | Rows for 1 order with 3 items | Use case |
|---|---|---|
| One row per order | 1 | Order-level revenue, fast, loses item detail |
| One row per order item | 3 | Product-level analysis, most flexible |
| One row per order item per day (snapshot) | 3 x days held | Inventory-style tracking, rarely needed for sales |

Always pick the *finest* grain your business questions require and aggregate upward in marts - you cannot recover detail you never captured, but you can always roll finer grain up to coarser grain.

# PART 18 - FACT TABLES

| Fact type | Description | Example |
|---|---|---|
| Transaction fact | One row per business event, immutable once written | `fact_sales`: one row per sale |
| Periodic snapshot fact | One row per entity per fixed time interval | `fact_inventory_daily`: one row per product per day |
| Accumulating snapshot fact | One row per process instance, updated as it progresses through stages | `fact_order_fulfillment`: one row per order, columns for each stage's timestamp |
| Factless fact | Records that an event happened, with no numeric measure | `fact_attendance`: one row per student per class per day (measure is just "occurred") |

```sql
-- Transaction fact
CREATE TABLE fact_sales (
    sale_key      BIGSERIAL PRIMARY KEY,
    date_key      INT REFERENCES dim_date(date_key),
    customer_key  INT REFERENCES dim_customer(customer_key),
    product_key   INT REFERENCES dim_product(product_key),
    quantity      INT,
    sales_amount  NUMERIC(12,2)
);

-- Accumulating snapshot fact - updated in place as the order progresses
CREATE TABLE fact_order_fulfillment (
    order_key            BIGINT PRIMARY KEY,
    ordered_date_key      INT,
    paid_date_key          INT,
    shipped_date_key       INT,
    delivered_date_key     INT,
    days_order_to_ship      INT
);

-- Factless fact
CREATE TABLE fact_website_visit (
    date_key     INT,
    customer_key INT,
    page_key     INT
    -- no numeric measure - COUNT(*) itself is the metric
);
```

**Measures vs dimensions**: measures are the numeric facts you aggregate (`quantity`, `sales_amount`) - additive (SUM works cleanly), semi-additive (e.g. account balance - SUMs across accounts but not across time), or non-additive (e.g. a ratio). Dimensions are the descriptive attributes you group/filter by (`customer_name`, `product_category`) - never aggregated, always joined.

# PART 19 - DIMENSION TABLES

```sql
CREATE TABLE dim_customer (
    customer_key  SERIAL PRIMARY KEY,     -- surrogate key, used by fact tables
    customer_id   INT UNIQUE,              -- business key from the source system
    customer_name TEXT,
    segment       TEXT,
    signup_date   DATE
);

CREATE TABLE dim_product (
    product_key   SERIAL PRIMARY KEY,
    product_id    INT UNIQUE,
    product_name  TEXT,
    category      TEXT,
    unit_price    NUMERIC(10,2)
);

CREATE TABLE dim_location (
    location_key  SERIAL PRIMARY KEY,
    city          TEXT,
    state         TEXT,
    country       TEXT,
    region        TEXT
);

CREATE TABLE dim_employee (
    employee_key  SERIAL PRIMARY KEY,
    employee_id   INT UNIQUE,
    full_name     TEXT,
    department    TEXT,
    hire_date     DATE
);
```

**Why surrogate keys** (`*_key`, auto-generated integers) instead of reusing the business key directly: they are stable even if the business key changes format, they let SCD Type 2 store multiple versions of the same business entity, they join faster (small integers) than wide natural keys (strings/composite keys), and they decouple the warehouse's internal identity system from any single source system's ID scheme.

# PART 20 - DATE DIMENSIONS

Warehouses use a pre-built date dimension instead of relying purely on live date functions because: it lets business calendars (fiscal periods, holidays, "is a trading day") be modeled as plain lookup data rather than complex logic re-derived in every query, it makes joins uniform (`join on date_key`, an integer, is faster than function calls per row), and it gives BI tools a single consistent table to build calendar hierarchies/filters from.

```sql
CREATE TABLE dim_date (
    date_key        INT PRIMARY KEY,     -- e.g. 20260207
    full_date        DATE,
    day_of_month      INT,
    month_num          INT,
    month_name          TEXT,
    quarter              INT,
    year                  INT,
    week_of_year          INT,
    day_of_week            INT,
    day_name                TEXT,
    is_weekend                BOOLEAN,
    fiscal_period               TEXT
);

-- [PostgreSQL] generate a full calendar year with generate_series
INSERT INTO dim_date
SELECT
    TO_CHAR(d, 'YYYYMMDD')::INT,
    d,
    EXTRACT(DAY FROM d)::INT,
    EXTRACT(MONTH FROM d)::INT,
    TO_CHAR(d, 'Month'),
    EXTRACT(QUARTER FROM d)::INT,
    EXTRACT(YEAR FROM d)::INT,
    EXTRACT(WEEK FROM d)::INT,
    EXTRACT(ISODOW FROM d)::INT,
    TO_CHAR(d, 'Day'),
    EXTRACT(ISODOW FROM d) IN (6,7),
    'FY' || EXTRACT(YEAR FROM d)::TEXT
FROM generate_series('2026-01-01'::DATE, '2026-12-31'::DATE, '1 day') AS d;

-- [SQL Server] recursive CTE date generator
;WITH dates AS (
    SELECT CAST('2026-01-01' AS DATE) AS d
    UNION ALL
    SELECT DATEADD(DAY, 1, d) FROM dates WHERE d < '2026-12-31'
)
SELECT d FROM dates OPTION (MAXRECURSION 366);

-- [Oracle] CONNECT BY row generator
SELECT DATE '2026-01-01' + LEVEL - 1 AS d
FROM dual
CONNECT BY LEVEL <= 365;

-- [MySQL 8+] recursive CTE, same shape as SQL Server's
WITH RECURSIVE dates AS (
    SELECT DATE('2026-01-01') AS d
    UNION ALL
    SELECT d + INTERVAL 1 DAY FROM dates WHERE d < '2026-12-31'
)
SELECT d FROM dates;
```

# PART 21 - DATA MODELING FOR PIPELINES

| Approach | When to use |
|---|---|
| Normalization (3NF) | Source/OLTP systems - minimizes update anomalies, not analytics-friendly |
| Denormalization | Warehouse/mart layers - fewer joins, faster analytical scans |
| Wide tables | BI/ML feature consumption - one row has everything a query needs, no joins |
| Narrow tables | Event/log storage - flexible, storage-efficient, needs joins/pivots to consume |
| Dimensional models | Star schema facts/dimensions - the default for BI warehouses |
| Event models | Append-only, one row per event, timestamp-keyed - source of truth for behavior analytics |

Rule of thumb: normalize close to the source (raw/staging), denormalize as you move toward consumption (marts, ML feature tables, BI). Every join you remove downstream is compute you don't pay for on every dashboard refresh.

# PART 22 - DATA PARTITIONING

- **Horizontal partitioning**: split rows across partitions (most common - by date range, list, or hash).
- **Vertical partitioning**: split columns across separate physical stores (rare, used for very wide tables with hot/cold columns).
- **Range partitioning**: partition by a value range - e.g. one partition per month of `order_date`.
- **List partitioning**: partition by discrete values - e.g. one partition per `country_code`.
- **Hash partitioning**: partition by hash of a key - even distribution when there's no natural range/list.

`date` is the most common partition key because most analytical queries filter by a date range, and new data naturally lands in new partitions (no rewriting old ones).

```sql
-- [PostgreSQL] declarative range partitioning by month
CREATE TABLE fact_sales (
    sale_id INT, order_date DATE, amount NUMERIC
) PARTITION BY RANGE (order_date);

CREATE TABLE fact_sales_2026_01 PARTITION OF fact_sales
    FOR VALUES FROM ('2026-01-01') TO ('2026-02-01');

-- Partition pruning: the planner only scans the January partition
EXPLAIN SELECT SUM(amount) FROM fact_sales WHERE order_date >= '2026-01-01' AND order_date < '2026-02-01';
```

**Good partition key**: high cardinality relative to query patterns, matches how queries actually filter (usually date), roughly even data distribution across partitions, and not overly granular (millions of tiny partitions add metadata overhead). **Bad partition key**: low-cardinality columns with extreme skew (e.g. `status` where 95% of rows are `'completed'`), or a key queries rarely filter on (defeats pruning entirely).

Dialect differences: PostgreSQL/MySQL/SQL Server/Oracle implement partitioning as a physical table feature you declare explicitly; **BigQuery** partitions natively by ingestion time or a date/timestamp column with a simple `PARTITION BY` clause and enforces partition filters on large tables; **Snowflake** does not require manual partitioning at all - it automatically maintains micro-partitions and relies on clustering keys (Part 23) instead; **Redshift** combines distribution style with sort keys rather than traditional partitioning; **Databricks**/Delta Lake supports both directory-style partitioning and, increasingly, Z-ordering/liquid clustering as the modern alternative.

---

# PART 23 - CLUSTERING AND SORTING

- **Clustering / clustering keys**: physically co-locate rows with similar values in specified columns so scans that filter on them read fewer blocks.
- **Sort keys** [Redshift]: define the on-disk row order, enabling zone-map pruning similar to partition pruning.
- **Distribution keys** [Redshift]: control which compute node stores each row, minimizing data shuffling during joins.

| Platform | Optimization mechanism |
|---|---|
| BigQuery | Partitioning (coarse) + clustering (fine, up to 4 columns) |
| Snowflake | Automatic micro-partitions + optional explicit clustering keys for very large tables |
| Redshift | Distribution style (KEY/ALL/EVEN) + sort keys (compound/interleaved) |
| Databricks | Partitioning + Z-ORDER / liquid clustering on Delta tables |

Cloud warehouses differ because their underlying storage engines differ: BigQuery and Snowflake are fully managed, columnar, and auto-optimize storage layout behind the scenes (you mostly declare *intent* via clustering keys); Redshift is a classic MPP (massively parallel processing) architecture where *you* must explicitly choose how data is distributed across nodes, which has a much bigger performance impact if chosen wrong; Databricks sits on open Delta/Parquet files in a lake, so its optimizations (`OPTIMIZE`, `ZORDER`) are explicit maintenance commands you run periodically rather than fully automatic.

# PART 24 - SQL PERFORMANCE FOR DATA ENGINEERS

```sql
-- SLOW: scans the whole table, computes on every row, no early filtering
SELECT *
FROM fact_sales f
JOIN dim_customer c ON c.customer_key = f.customer_key
WHERE EXTRACT(YEAR FROM f.order_date) = 2026;   -- function on column disables index/partition use

-- OPTIMIZED: SARGable predicate (Search ARGument ABLE) - column stays untouched
SELECT f.sale_key, f.amount, c.customer_name       -- only needed columns, never SELECT *
FROM fact_sales f
JOIN dim_customer c ON c.customer_key = f.customer_key
WHERE f.order_date >= '2026-01-01' AND f.order_date < '2027-01-01';   -- range enables pruning/index use
```

**SARGability**: a predicate is SARGable when the database can use an index or partition to answer it directly. Wrapping the *column* in a function (`EXTRACT(YEAR FROM f.order_date) = 2026`) forces a full scan because the engine must compute the function for every row before it can compare; rewriting as a range on the raw column lets the engine use an index range scan or partition pruning.

```sql
-- Reading an execution plan
EXPLAIN ANALYZE
SELECT c.customer_name, SUM(f.amount)
FROM fact_sales f JOIN dim_customer c ON c.customer_key = f.customer_key
WHERE f.order_date >= '2026-01-01'
GROUP BY c.customer_name;
```

`EXPLAIN` shows the *planned* execution strategy (which scans, which joins, estimated costs) without running the query; `EXPLAIN ANALYZE` actually executes it and shows *real* row counts and timings alongside the estimates - the gap between estimated and actual rows is usually where performance problems hide (stale statistics, bad cardinality estimation).

**Other levers**: join optimization (join smallest result sets first, ensure join columns are indexed and same type), reduce scans (filter as early as possible, push predicates before joins), aggregation strategy (pre-aggregate in a materialized layer instead of aggregating raw facts on every dashboard query), materialization (persist expensive intermediate results as tables instead of recomputing CTEs repeatedly - CTEs are not automatically cached in every engine), temporary tables (break a huge query into staged, indexable steps), and statistics/cardinality estimation (run `ANALYZE`/`UPDATE STATISTICS` regularly so the optimizer's row-count guesses stay accurate - stale statistics are a very common cause of a query that "used to be fast").

# PART 25 - LARGE DATASETS

A query pattern that is fine at 100,000 rows can fail or become unusably slow at 100 million+ because: nested-loop joins that were negligible at small scale become quadratic, sorts that fit in memory start spilling to disk, an unindexed scan that took milliseconds now takes minutes, and result sets that could be materialized in application memory no longer fit.

Strategies at scale:

| Row count | Typical strategy |
|---|---|
| ~10M | Proper indexing usually enough |
| ~100M | Add partitioning, avoid full scans, pre-aggregate hot queries |
| ~1B | Partitioning + clustering + batching writes + materialized rollups mandatory |
| 10B+ | Distributed engine (BigQuery/Snowflake/Redshift/Databricks) with partition + cluster tuning, incremental-only processing, heavy pre-aggregation |

```sql
-- Batch a large backfill instead of one giant transaction/scan
DO $$
DECLARE batch_start DATE := '2020-01-01';
BEGIN
  WHILE batch_start < '2026-01-01' LOOP
    INSERT INTO warehouse.fact_sales_backfill
    SELECT * FROM source.sales WHERE order_date >= batch_start AND order_date < batch_start + INTERVAL '1 month';
    batch_start := batch_start + INTERVAL '1 month';
  END LOOP;
END $$;                                                       -- [PostgreSQL] PL/pgSQL
```

Batching a huge backfill into monthly (or daily) chunks bounds memory usage, lets you checkpoint progress (resume from a failed batch instead of restarting from zero), and avoids holding one enormous, lock-heavy transaction open for hours.

# PART 26 - BATCH PROCESSING

```
source.sales (OLTP)
   -> staging.stg_sales           -- extract + type + light clean
   -> validated.val_sales          -- run DQ checks, quarantine failures
   -> warehouse.fact_sales          -- transform, join to dimensions, load
   -> mart.daily_sales_summary       -- pre-aggregated reporting table
```

```sql
-- Daily batch pipeline, one SQL step per stage, each idempotent for its batch_id

-- 1. Staging (incremental by batch date)
INSERT INTO staging.stg_sales
SELECT *, CURRENT_DATE AS batch_date FROM source.sales WHERE sale_date = CURRENT_DATE - 1;

-- 2. Validation (quarantine bad rows, keep good ones)
INSERT INTO validated.val_sales
SELECT * FROM staging.stg_sales
WHERE batch_date = CURRENT_DATE AND customer_id IS NOT NULL AND amount >= 0;

-- 3. Transformation + load into the fact table
INSERT INTO warehouse.fact_sales (customer_key, product_key, date_key, amount)
SELECT c.customer_key, p.product_key, d.date_key, v.amount
FROM validated.val_sales v
JOIN dim_customer c ON c.customer_id = v.customer_id
JOIN dim_product p ON p.product_id = v.product_id
JOIN dim_date d ON d.full_date = v.sale_date
WHERE v.batch_date = CURRENT_DATE;

-- 4. Reporting rollup
INSERT INTO mart.daily_sales_summary
SELECT CURRENT_DATE - 1 AS report_date, SUM(amount) AS total_sales, COUNT(*) AS n_sales
FROM warehouse.fact_sales f JOIN dim_date d ON d.date_key = f.date_key
WHERE d.full_date = CURRENT_DATE - 1;
```

Cadences (hourly/daily/weekly/monthly) are chosen based on how fresh consumers actually need the data to be, balanced against source system load and compute cost - defaulting to the *least frequent* cadence that meets the real business SLA keeps pipelines cheap and simple.

# PART 27 - STREAMING DATA

Core streaming vocabulary:
- **Event time** vs **processing time**: event time is when something actually happened at the source; processing time is when your system saw it. They diverge whenever there's network delay, batching, or retries.
- **Late-arriving data**: an event whose event time is well before the current processing time (e.g. a mobile app buffers events offline and sends them hours later).
- **Idempotency**: reprocessing the same event twice must not double-count it (Part 28).
- **Ordering**: events may arrive out of event-time order; systems either tolerate this or explicitly reorder within a window.
- **Windows**: streaming aggregations are computed over bounded time windows (tumbling, sliding, session) because "sum of all events ever" is not a meaningful continuously-updating answer.
- **Watermarking**: a streaming engine's mechanism for deciding "how late can an event be before I stop waiting for it and finalize the window."

SQL's role in streaming systems is usually as the *query language* layered on top of a streaming engine, not the engine itself - e.g. **Kafka** (durable event log), **Flink**/**Spark Structured Streaming** (processing engines), **Kafka Streams** (library for stream processing on Kafka), and various cloud streaming platforms (Kinesis, Pub/Sub) all offer a SQL-like interface for defining continuous queries:

```sql
-- Conceptual streaming SQL (Flink SQL / Spark Structured Streaming style)
SELECT
    TUMBLE_START(event_time, INTERVAL '1' HOUR) AS window_start,
    product_id,
    SUM(amount) AS hourly_sales
FROM sales_stream
GROUP BY TUMBLE(event_time, INTERVAL '1' HOUR), product_id;
```

This reads like ordinary SQL but runs continuously: as new events arrive on the stream, the engine incrementally updates the aggregate for the currently-open window and emits a result when the window closes (subject to the watermark's late-data tolerance).

# PART 28 - IDEMPOTENT PIPELINES

**Idempotency**: running the same pipeline step twice (with the same input) produces the same end state as running it once - critical because retries, backfills, and orchestrator-triggered re-runs are a normal part of production operations, not an edge case.

```sql
-- BAD: not idempotent - re-running this after a partial failure double-counts
INSERT INTO fact_sales (customer_id, amount, sale_date)
SELECT customer_id, amount, sale_date FROM staging.stg_sales WHERE batch_date = CURRENT_DATE;
-- If this runs twice for the same batch_date, every row is duplicated.
```

```sql
-- GOOD: idempotent - delete-then-insert scoped to the exact batch being (re)loaded
DELETE FROM fact_sales WHERE batch_id = 'run_20260207';

INSERT INTO fact_sales (batch_id, customer_id, amount, sale_date)
SELECT 'run_20260207', customer_id, amount, sale_date
FROM staging.stg_sales WHERE batch_id = 'run_20260207';
-- Re-running this any number of times leaves the same end state.
```

```sql
-- ALTERNATIVE: idempotent via upsert keyed on a natural/event ID instead of delete+insert
MERGE INTO fact_sales AS t
USING staging.stg_sales AS s
ON t.event_id = s.event_id                 -- event_id is a stable, unique identifier per source event
WHEN MATCHED THEN UPDATE SET amount = s.amount
WHEN NOT MATCHED THEN INSERT (event_id, customer_id, amount) VALUES (s.event_id, s.customer_id, s.amount);
```

Safely rerunning daily/hourly jobs, failed jobs, and historical backfills all reduce to the same discipline: scope every write to an explicit, reproducible key (`batch_id`, `run_id`, or a stable `event_id`), and make the write pattern delete+insert or upsert - never a blind append.

# PART 29 - PIPELINE FAILURE AND RECOVERY

| Failure | Typical cause | Mitigation |
|---|---|---|
| Source unavailable | Network/outage/credentials | Retry with backoff, alert if retries exhausted |
| Partial load | Job killed mid-run | Idempotent design (Part 28) + checkpointing |
| Duplicate load | Job re-run without dedup | Idempotent design + unique constraints |
| Corrupted records | Malformed source data | Validation layer + quarantine table (Part 10) |
| Schema change | Upstream added/renamed/dropped a column | Schema validation step before load, alert on mismatch |
| Timeout | Query too slow / resource contention | Query optimization (Part 24), smaller batches |
| Deadlock | Concurrent writes to same rows | Consistent lock ordering, shorter transactions, retry logic |
| Constraint violation | Bad data hits a NOT NULL/FK/UNIQUE constraint | Validate before load, or route violations to quarantine |
| Warehouse unavailable | Platform outage / maintenance | Retry with backoff, circuit breaker, alerting |

Recovery mechanisms: **retries** (with exponential backoff, capped attempts), **rollback** (wrap a multi-statement load in a transaction so a mid-failure leaves no partial state), **checkpoints** (record progress after each safely-completed sub-step so a restart resumes rather than reprocessing from zero), **restartability** (a design property - every step must be safely re-runnable, which is exactly idempotency from Part 28), and **dead-letter / quarantine tables** (route unprocessable records aside instead of failing the whole batch, so 1 bad row doesn't block 999,999 good ones).

# PART 31 - DATA LINEAGE

**Lineage** answers "where did this value come from, and what does it feed into." It matters because when a number looks wrong, lineage lets you trace backward to the exact upstream table/transformation responsible, and when a source table changes, lineage lets you trace forward to everything that will be impacted.

```
source.orders
   -> staging.stg_orders          (table lineage: one hop)
   -> warehouse.fact_sales         (table lineage: two hops; column lineage: order_amount -> sales_amount)
   -> mart.sales_mart                (table lineage: three hops)
   -> "Revenue" dashboard              (business lineage: the metric a VP looks at)
```

- **Column lineage**: tracks a specific column's transformations end to end (e.g. `orders.amount` -> `CAST(amount AS NUMERIC)` -> `fact_sales.sales_amount` -> `SUM(sales_amount)` -> `mart.total_revenue`).
- **Table lineage**: coarser - which tables feed which tables, without tracking individual columns.
- **Pipeline lineage**: which orchestrated jobs/tasks produce which tables.
- **Business lineage**: connects technical lineage to business artifacts (reports, dashboards, KPIs) so a non-technical stakeholder can see what feeds their number.

Every SQL transformation you write *is* a lineage edge - modern catalogs (dbt docs, OpenLineage, Alation, Atlan) parse your SQL's `FROM`/`JOIN` clauses to build the lineage graph automatically, which is one more reason to write explicit, parseable SQL rather than dynamically generated strings the parser can't follow.

# PART 32 - SCHEMA EVOLUTION

```sql
-- Adding a column: safe, backward-compatible (existing readers ignore the new column)
ALTER TABLE customers ADD COLUMN loyalty_tier TEXT;

-- Removing a column: breaking - anything that SELECTs it explicitly now fails
ALTER TABLE customers DROP COLUMN old_unused_flag;

-- Renaming a column: breaking - treat as drop+add for consumers
ALTER TABLE customers RENAME COLUMN cust_nm TO customer_name;

-- Changing a data type: potentially breaking if precision/format changes
ALTER TABLE customers ALTER COLUMN phone TYPE TEXT;                 -- [PostgreSQL]
ALTER TABLE customers ALTER COLUMN phone VARCHAR(20);               -- [SQL Server]/[MySQL] style
```

**Backward compatibility** means old consumers keep working after a schema change without modification. Additive changes (new nullable column) are backward-compatible by default; removals, renames, and type-narrowing are not. **Schema contracts** (Part 36) formalize this: a producer commits to only making backward-compatible changes without a version bump and advance notice, so consumers can trust the pipeline won't silently break.

**Migration strategy for a breaking change**: (1) add the new column/table alongside the old one, (2) dual-write both for a transition period, (3) migrate all consumers to the new shape, (4) deprecate and finally drop the old shape - never make a breaking change in a single atomic step in a shared production table.

# PART 33 - DATABASE MIGRATIONS

```sql
-- Migration file: 001_create_customers.sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT
);

-- Migration file: 002_add_loyalty_tier.sql
ALTER TABLE customers ADD COLUMN loyalty_tier TEXT DEFAULT 'standard';

-- Rollback for 002 (forward-only teams instead write a *new* forward migration
-- that reverses the change, rather than physically rolling back)
ALTER TABLE customers DROP COLUMN loyalty_tier;
```

Migration tools (**Flyway**, **Liquibase**, **Alembic**) all follow the same core idea: every schema change is a small, numbered, version-controlled file; the tool keeps a metadata table (e.g. `schema_history`) recording which migrations have already run against each environment; running the tool applies only the migrations not yet recorded as applied, in order, guaranteeing dev/staging/production schemas stay in sync. **Forward-only migrations** avoid ever running a destructive rollback against production - if migration 002 turns out wrong, you write migration 003 to fix it forward, rather than reversing 002, because a rollback executed against a table that already has real data in the new column can itself be destructive.

# PART 34 - DBT / ANALYTICS ENGINEERING

dbt (data build tool) turns a folder of `SELECT` statements into a managed, tested, documented transformation pipeline, running entirely inside your warehouse (pure ELT).

Core concepts:
- **models**: a `.sql` file containing a `SELECT` - dbt wraps it in `CREATE TABLE/VIEW AS` for you.
- **sources**: declared references to raw tables dbt did not create, so lineage starts from the true origin.
- **seeds**: small static CSV files loaded as tables (e.g. a country-code lookup).
- **tests**: declarative data-quality assertions (`not_null`, `unique`, `accepted_values`, `relationships`) run with one command.
- **snapshots**: dbt's built-in mechanism for building SCD Type 2 tables automatically from a mutable source.
- **macros**: reusable Jinja+SQL functions (like SQL "subroutines").
- **refs**: `{{ ref('model_name') }}` - how models reference each other, which is how dbt auto-builds the dependency DAG.
- **materializations**: how a model is physically built - `view`, `table`, `incremental`, `ephemeral` (CTE-inlined, never persisted).
- **incremental models**: dbt only processes new/changed rows on subsequent runs (the dbt-native version of Part 12's watermark pattern).

```sql
-- models/staging/stg_customers.sql
SELECT
    customer_id,
    LOWER(TRIM(email)) AS email,
    created_at
FROM {{ source('raw', 'customers') }}

-- models/marts/fact_sales.sql  (incremental materialization)
{{ config(materialized='incremental', unique_key='order_id') }}

SELECT order_id, customer_id, amount, order_date
FROM {{ ref('stg_orders') }}
{% if is_incremental() %}
WHERE order_date > (SELECT MAX(order_date) FROM {{ this }})
{% endif %}
```

`{{ ref('stg_orders') }}` both resolves to the correctly-environment-scoped table name *and* registers a lineage edge; `{% if is_incremental() %}` makes the same model file behave as a full build on the first run and an incremental watermark-filtered build on every run after - collapsing Part 12's manual watermark table into a framework-managed pattern. The natural layering is `raw` (sources) -> `staging` (1:1 typed/cleaned models) -> `intermediate` (reusable joins) -> `marts` (business-facing final models) - directly mirroring Part 3's layered architecture, implemented as dbt models instead of hand-run SQL scripts.

# PART 35 - SQL TESTING

```yaml
# schema.yml - dbt-style declarative tests (the modern standard for SQL testing)
models:
  - name: dim_customer
    columns:
      - name: customer_id
        tests: [not_null, unique]
      - name: email
        tests: [not_null]
  - name: fact_sales
    columns:
      - name: customer_key
        tests:
          - relationships:
              to: ref('dim_customer')
              field: customer_key
```

```sql
-- The same tests expressed as raw SQL (what the framework runs under the hood)

-- uniqueness
SELECT customer_id, COUNT(*) FROM dim_customer GROUP BY customer_id HAVING COUNT(*) > 1;

-- not-null
SELECT * FROM dim_customer WHERE customer_id IS NULL;

-- accepted values
SELECT * FROM dim_customer WHERE segment NOT IN ('retail','enterprise','sme');

-- relationships (referential integrity)
SELECT f.* FROM fact_sales f
LEFT JOIN dim_customer c ON c.customer_key = f.customer_key
WHERE c.customer_key IS NULL;
```

**Testing levels**: unit tests check one transformation's logic against known fixed input/output (e.g. "given these 3 raw rows, does the cleaning model produce exactly these 3 clean rows"); data tests (the queries above) check properties of the *actual* data continuously; integration tests verify a full pipeline run end-to-end produces correct downstream tables; regression tests re-run after every code change to catch a previously-fixed bug reappearing. A mature pipeline runs data tests on every scheduled run (failing the DAG on violation) and unit + regression tests in CI before code merges.

# PART 36 - DATA CONTRACTS

A **data contract** is an explicit agreement between a producer team and consumer teams covering:

- **Schema**: exact columns, types, nullability - changes require a version bump.
- **Ownership**: who is accountable when the contract breaks.
- **Freshness**: how current the data must be (e.g. "updated within 1 hour of the source event").
- **Quality**: which DQ checks (Part 10) must pass before data is considered "published."
- **Semantics**: what a column *means* (e.g. is `amount` gross or net of tax - documented, not assumed).
- **SLA** (Service Level Agreement): the overall commitment (e.g. "99.5% of daily loads complete by 6am").
- **SLI** (Service Level Indicator): the actual measured metric (e.g. "load completed at 5:52am").
- **SLO** (Service Level Objective): the target threshold for the SLI (e.g. "complete by 6am on 99% of days").

Data Engineers work with upstream producers by getting contracts agreed *before* building a pipeline against their table, subscribing to schema-change notifications, adding automated contract-validation checks at ingestion (Part 9) that fail fast and alert the producer rather than silently loading violating data, and treating a contract breach as an incident with the producer team, not something the pipeline should silently absorb.

# PART 37 - DATA FRESHNESS

```sql
-- Expected: new data every hour. Detect staleness explicitly.
SELECT
    MAX(loaded_at)                                          AS latest_record,
    CURRENT_TIMESTAMP                                          AS now,
    CURRENT_TIMESTAMP - MAX(loaded_at)                            AS staleness,
    CASE WHEN CURRENT_TIMESTAMP - MAX(loaded_at) > INTERVAL '2 hours'
         THEN 'STALE' ELSE 'FRESH' END                              AS freshness_status
FROM staging.stg_events;
```

With `latest_record = 2026-09-07 08:00` and `now = 2026-09-07 11:00`, `staleness` evaluates to `3 hours`, which exceeds the `2 hour` SLO, so `freshness_status = 'STALE'` - this single query is exactly what a monitoring/alerting job runs on a schedule to page someone when a pipeline has silently stopped updating (one of the most common real-world production incidents, and one that a naive "did the job return success" check completely misses if the job *ran* successfully but processed zero new rows).

# PART 38 - DATA RECONCILIATION

**Scenario**: source database reports 1,000,000 transactions; the warehouse only has 999,850 - 150 rows are missing.

```sql
-- Step 1: row-count reconciliation
SELECT
  (SELECT COUNT(*) FROM source.transactions)   AS source_count,
  (SELECT COUNT(*) FROM warehouse.fact_transactions) AS warehouse_count;

-- Step 2: find exactly which keys are missing
SELECT s.transaction_id
FROM source.transactions s
LEFT JOIN warehouse.fact_transactions w ON w.transaction_id = s.transaction_id
WHERE w.transaction_id IS NULL;

-- Step 3: check whether the missing rows share a pattern (e.g. all from one batch/date)
SELECT DATE(s.created_at) AS created_date, COUNT(*) AS missing_count
FROM source.transactions s
LEFT JOIN warehouse.fact_transactions w ON w.transaction_id = s.transaction_id
WHERE w.transaction_id IS NULL
GROUP BY DATE(s.created_at)
ORDER BY missing_count DESC;

-- Step 4: sum/control-total reconciliation (catches value errors, not just missing rows)
SELECT
  (SELECT SUM(amount) FROM source.transactions) AS source_total,
  (SELECT SUM(amount) FROM warehouse.fact_transactions) AS warehouse_total;
```

The standard reconciliation toolkit: **row counts** (do the totals match), **sums/control totals** (do the numeric totals match - catches value corruption even when counts match), **hashes** (a hash of all business columns per row, compared between source and target, pinpoints exactly which rows differ - Part 39), **min/max dates** (did the load cover the expected time range completely), and **distinct key counts** (catches duplication as easily as it catches loss). Once step 2 or 3 narrows down *which* rows/batch is missing, you go back to the pipeline logs/audit table (Part 30) for that specific run to find the root cause (a failed batch, a watermark boundary bug from Part 12, a validation rule silently rejecting valid rows).

# PART 39 - HASHING FOR DATA ENGINEERING

```sql
-- Row hash for change detection / deduplication / reconciliation
SELECT
    customer_id,
    MD5(name || '|' || email || '|' || COALESCE(phone,''))   AS row_hash    -- [PostgreSQL]/[MySQL]
FROM staging.stg_customers;

SELECT HASHBYTES('SHA2_256',
    CONCAT(name, '|', email, '|', ISNULL(phone,'')))          AS row_hash    -- [SQL Server]
FROM staging.stg_customers;

SELECT TO_HEX(SHA256(CONCAT(name, '|', email)))                AS row_hash    -- [BigQuery]
FROM staging.stg_customers;
```

**Change detection**: store `row_hash` alongside each dimension row; on the next load, compute the hash of the incoming record and compare - if it matches the stored hash, nothing changed (skip); if it differs, something changed (trigger SCD Type 2 logic in Part 16) - this is far cheaper than comparing every individual column. **Deduplication**: two rows with the same hash are byte-identical on their business columns, so exact duplicates collapse to a single `GROUP BY row_hash`. **Reconciliation**: comparing source-side and warehouse-side hashes for the same key pinpoints exactly which rows differ, without transferring/comparing every column value across systems.

# PART 40 - HANDLING LATE-ARRIVING DATA

**Late-arriving data**: a record whose real-world (event) time falls in an already-processed period, but which physically arrives afterward - e.g. a January transaction that reaches the warehouse in February due to a delayed batch from an offline point-of-sale device.

```sql
-- Detect late arrivals: event date is earlier than the load date by more than expected
SELECT *
FROM staging.stg_transactions
WHERE transaction_date < DATE_TRUNC('month', loaded_at)   -- event happened before this month
  AND loaded_at >= DATE_TRUNC('month', CURRENT_DATE);      -- but only just arrived this month
```

**Late facts**: reprocess (or incrementally re-aggregate) the specific historical partition/period the late row belongs to, rather than only appending it to the current period - otherwise January's numbers stay permanently wrong. Partitioned fact tables (Part 22) make this cheap: reload just the affected partition. **Late dimensions**: if a dimension record (e.g. a new customer) arrives after facts referencing it have already loaded, use a placeholder/"unknown member" surrogate key at load time, then backfill the correct surrogate key once the dimension record catches up. **Historical corrections**: always correct via a new, audited write (Part 41) into the correct historical partition, never by silently overwriting an already-published reporting number without a documented correction trail.

# PART 41 - DATA CORRECTIONS

```sql
-- Direct UPDATE: fast, but destroys the prior (possibly wrong-but-published) value with no trail
UPDATE fact_sales SET amount = 4500.00 WHERE sale_id = 90210;

-- Safer: soft-delete the old row and append the corrected one, preserving history
UPDATE fact_sales SET is_deleted = TRUE, deleted_at = CURRENT_TIMESTAMP WHERE sale_id = 90210;
INSERT INTO fact_sales (sale_id, amount, corrected_from_sale_id)
VALUES (90211, 4500.00, 90210);

-- Safest for regulated/audited domains: append-only correction table, never touch the original
INSERT INTO fact_sales_corrections (original_sale_id, corrected_amount, corrected_by, corrected_at, reason)
VALUES (90210, 4500.00, 'data_eng_kajola', CURRENT_TIMESTAMP, 'source system rounding bug');
-- Downstream reporting views JOIN against this table and prefer the correction when present.
```

Blindly `UPDATE`/`DELETE`-ing production facts is dangerous because it destroys the audit trail (you can no longer answer "what did this number look like when the VP saw it last Tuesday"), it can silently break any downstream table that was built assuming immutability (e.g. an accumulating snapshot or a cached rollup), and in regulated domains (finance, healthcare) it may be a compliance violation outright. Prefer append-only corrections with full audit metadata (`corrected_by`, `corrected_at`, `reason`) wherever the business/regulatory context allows it.

# PART 42 - SOFT DELETES

```sql
ALTER TABLE customers ADD COLUMN deleted_at TIMESTAMP;
ALTER TABLE customers ADD COLUMN is_deleted BOOLEAN DEFAULT FALSE;
ALTER TABLE customers ADD COLUMN active_flag BOOLEAN DEFAULT TRUE;

-- "Delete" without physically removing the row
UPDATE customers SET is_deleted = TRUE, deleted_at = CURRENT_TIMESTAMP, active_flag = FALSE
WHERE customer_id = 101;

-- Every downstream query must remember to filter these out
SELECT * FROM customers WHERE is_deleted = FALSE;
```

**Advantages**: preserves history/audit trail, supports "undelete," lets fact tables that already reference the row keep working, and is safer under GDPR-style "right to be forgotten" workflows when combined with a later hard-delete/anonymization pass. **Disadvantages**: every query everywhere must remember to filter soft-deleted rows (easy to forget, causing silent bugs), the table grows forever unless periodically archived (Part 43), and "true" deletion for compliance (e.g. actually erasing PII) still eventually requires a real, separate hard-delete/anonymization process - soft delete alone does not satisfy strict data-erasure requirements.

# PART 43 - DATA RETENTION

```sql
-- Archive rows older than the retention window into cold storage, then purge from hot table
INSERT INTO archive.fact_sales_archive
SELECT * FROM fact_sales WHERE order_date < CURRENT_DATE - INTERVAL '7 years';

DELETE FROM fact_sales WHERE order_date < CURRENT_DATE - INTERVAL '7 years';

-- Cheaper at scale: drop an entire old partition instead of row-by-row DELETE
ALTER TABLE fact_sales DETACH PARTITION fact_sales_2018_01;               -- [PostgreSQL]
DROP TABLE fact_sales_2018_01;
```

Retention policy design balances **regulatory considerations** (many financial/healthcare/tax regimes mandate a minimum retention period - conceptually, "you must be *able* to produce records for N years," which is a business/legal question outside SQL's scope but directly shapes the schema) against **business considerations** (storage cost, query performance on ever-growing hot tables, and genuine business need for old detail vs. a pre-aggregated summary being sufficient). Partition dropping (rather than row-level `DELETE`) is dramatically cheaper at scale because it is a metadata operation, not a scan-and-delete over billions of rows.

# PART 44 - SECURITY FOR DATA ENGINEERING

```sql
-- Principle of least privilege: a pipeline service account gets only what it needs
CREATE ROLE etl_service_role;
GRANT SELECT ON source.orders TO etl_service_role;
GRANT INSERT, UPDATE ON staging.stg_orders TO etl_service_role;
GRANT SELECT ON staging.stg_orders TO analytics_reader_role;   -- readers cannot write

-- Masking a sensitive column for a broad reporting role
CREATE VIEW mart.customers_masked AS
SELECT customer_id,
       name,
       CONCAT(LEFT(email, 2), '***@***') AS email_masked,     -- PII masked
       phone
FROM dim_customer;
GRANT SELECT ON mart.customers_masked TO bi_analyst_role;       -- analysts get the masked view, not raw PII

-- Row-level security: a user only sees rows for their own region [PostgreSQL]
CREATE POLICY region_policy ON fact_sales
    USING (region = current_setting('app.user_region'));
```

Production pipelines should not run as a database administrator account because a DBA account can do far more damage than a pipeline ever needs to (drop tables, alter permissions, read every schema) - if that credential leaks or a pipeline bug executes unintended SQL, blast radius is the whole database instead of the few tables the pipeline actually touches. Standard practice: dedicated service accounts per pipeline with least-privilege grants, secrets stored in a secrets manager (never hardcoded in SQL/scripts), sensitive columns (PII - names, emails, national IDs, health data) identified and masked/encrypted for broad-access roles, and row-level security applied wherever different consumers should see different subsets of the same table.

# PART 46 - BIGQUERY SQL

```sql
-- [BigQuery] partition + cluster a table
CREATE TABLE analytics.fact_sales
PARTITION BY DATE(order_date)
CLUSTER BY customer_id, product_id
AS SELECT * FROM staging.stg_sales;

-- [BigQuery] arrays and STRUCT - native nested/repeated data, no join needed
SELECT
    customer_id,
    ARRAY_AGG(STRUCT(order_id, amount)) AS orders    -- nested repeated STRUCT column
FROM fact_sales
GROUP BY customer_id;

-- [BigQuery] UNNEST - flatten an array back into rows
SELECT customer_id, o.order_id, o.amount
FROM analytics.customer_orders_nested, UNNEST(orders) AS o;

-- [BigQuery] QUALIFY - filter on a window function without a wrapping subquery
SELECT customer_id, order_id, amount,
       ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY amount DESC) AS rn
FROM fact_sales
QUALIFY rn = 1;

-- [BigQuery] external table over a bucket, queryable without loading
CREATE EXTERNAL TABLE raw.events_ext
OPTIONS (format = 'PARQUET', uris = ['gs://my-bucket/events/*.parquet']);
```

`ARRAY`/`STRUCT`/`UNNEST` let BigQuery natively model one-to-many relationships inside a single row (e.g. one customer row holding an array of all their orders) - useful for denormalized, join-free analytical tables; `QUALIFY` removes the need to wrap a `ROW_NUMBER()` query in an outer `SELECT ... WHERE rn = 1`, which every other dialect here requires. Scheduled queries let you run a `.sql` file on a cron-like schedule natively inside BigQuery, a lightweight alternative to a full orchestrator for simple pipelines.

# PART 47 - SNOWFLAKE SQL

```sql
-- [Snowflake] object hierarchy: database -> schema -> table; warehouses are separate compute
CREATE WAREHOUSE etl_wh WITH WAREHOUSE_SIZE = 'MEDIUM' AUTO_SUSPEND = 60 AUTO_RESUME = TRUE;
USE WAREHOUSE etl_wh;

-- [Snowflake] stage: a named pointer to a cloud storage location
CREATE STAGE raw_stage URL = 's3://my-bucket/raw/' CREDENTIALS = (AWS_KEY_ID='...' AWS_SECRET_KEY='...');

-- [Snowflake] file format + bulk load
CREATE FILE FORMAT csv_fmt TYPE = 'CSV' FIELD_DELIMITER = ',' SKIP_HEADER = 1;
COPY INTO staging.stg_customers FROM @raw_stage/customers/ FILE_FORMAT = (FORMAT_NAME = csv_fmt);

-- [Snowflake] streams: built-in CDC on a table (tracks INSERT/UPDATE/DELETE since last consumed)
CREATE STREAM orders_stream ON TABLE source.orders;
SELECT * FROM orders_stream;   -- shows only changed rows since the stream was last read

-- [Snowflake] tasks: native scheduled SQL execution (a lightweight built-in orchestrator)
CREATE TASK refresh_fact_sales
  WAREHOUSE = etl_wh
  SCHEDULE = 'USING CRON 0 * * * * UTC'
AS
  MERGE INTO fact_sales t USING orders_stream s ON t.order_id = s.order_id
  WHEN MATCHED THEN UPDATE SET amount = s.amount
  WHEN NOT MATCHED THEN INSERT (order_id, amount) VALUES (s.order_id, s.amount);

-- [Snowflake] transient/temporary tables - cheaper, no fail-safe/time-travel retention
CREATE TRANSIENT TABLE staging.stg_scratch AS SELECT * FROM source.orders LIMIT 1000;

-- [Snowflake] Time Travel - query a table as of the past, without a separate backup
SELECT * FROM fact_sales AT (OFFSET => -3600);   -- table state 1 hour ago
```

Snowflake's **streams + tasks** combination gives you native, in-warehouse CDC and scheduling without an external tool - a stream tracks changes since it was last consumed, and a task runs SQL on a schedule, together forming a minimal built-in incremental pipeline. **Time travel** lets you query (or clone) a table's historical state directly, which is invaluable for "what did this table look like right before that bad deploy."

# PART 48 - REDSHIFT

```sql
-- [Redshift] distribution style controls which node stores each row
CREATE TABLE fact_sales (
    sale_id INT, customer_id INT, amount NUMERIC
)
DISTSTYLE KEY DISTKEY (customer_id)      -- co-locate rows sharing a customer_id on the same node
SORTKEY (order_date);                     -- physical row order for zone-map pruning

-- Bulk load from S3
COPY fact_sales FROM 's3://bucket/sales/' IAM_ROLE 'arn:aws:iam::123:role/redshift-role' FORMAT PARQUET;

-- Unload query results back to S3
UNLOAD ('SELECT * FROM mart.sales_mart')
TO 's3://bucket/exports/sales_mart_'
IAM_ROLE 'arn:aws:iam::123:role/redshift-role';

-- Maintenance: reclaim space from deleted rows and refresh statistics
VACUUM fact_sales;
ANALYZE fact_sales;
```

**Distribution styles**: `KEY` (rows with the same key land on the same node - minimizes cross-node shuffling for joins on that key), `ALL` (small dimension tables fully copied to every node - avoids shuffling entirely for tables that fit), `EVEN` (round-robin - simple, but every join needs a shuffle). **Sort keys** determine physical row order, similar in effect to partition pruning elsewhere - queries filtering on the sort key skip large blocks of unsorted-irrelevant data. `VACUUM`/`ANALYZE` matter more in Redshift than in fully-managed platforms because Redshift historically requires more manual maintenance to reclaim space after deletes and keep the query planner's statistics accurate. Redshift is a strong fit when a team is already deep in AWS and needs an MPP warehouse with tight S3/IAM integration.

# PART 49 - DATABRICKS SQL

```sql
-- [Databricks SQL] Delta Lake table - ACID transactions on top of Parquet files in object storage
CREATE TABLE gold.fact_sales
USING DELTA
PARTITIONED BY (order_month)
AS SELECT *, DATE_TRUNC('month', order_date) AS order_month FROM silver.sales_clean;

-- MERGE (upsert) directly against a Delta table
MERGE INTO gold.fact_sales AS t
USING silver.sales_clean AS s
ON t.sale_id = s.sale_id
WHEN MATCHED THEN UPDATE SET t.amount = s.amount
WHEN NOT MATCHED THEN INSERT (sale_id, amount, order_date) VALUES (s.sale_id, s.amount, s.order_date);

-- File/layout maintenance: compact small files, then physically sort by high-cardinality columns
OPTIMIZE gold.fact_sales ZORDER BY (customer_id);

-- Time travel, native on Delta tables
SELECT * FROM gold.fact_sales VERSION AS OF 12;
SELECT * FROM gold.fact_sales TIMESTAMP AS OF '2026-02-01';
```

**Medallion architecture** (Databricks' standard pattern, but applicable anywhere):

```
Bronze (raw)             -- exact copy of source, minimal/no transformation
   -> Silver (cleaned)     -- typed, deduplicated, validated, conformed
      -> Gold (business-ready)  -- aggregated, dimensional, ready for BI/ML consumption
```

Bronze maps directly onto this handbook's "Raw" layer, Silver onto "Staging/Intermediate/Core," and Gold onto "Dimension/Fact/Mart" - same underlying discipline, different naming convention. `OPTIMIZE ... ZORDER BY` is Databricks' clustering mechanism (Part 23): it physically co-locates rows by the Z-ordered columns so filters on them skip far more data files.

# PART 50 - MEDALLION ARCHITECTURE (worked example)

```sql
-- BRONZE: land raw, untouched, append-only
CREATE TABLE bronze.orders_raw USING DELTA AS
SELECT * FROM cloud_files('s3://landing/orders/', 'json');   -- [Databricks] Auto Loader concept

-- SILVER: typed, cleaned, deduplicated, validated
CREATE TABLE silver.orders_clean USING DELTA AS
SELECT
    CAST(order_id AS BIGINT) AS order_id,
    CAST(customer_id AS BIGINT) AS customer_id,
    CAST(amount AS DECIMAL(12,2)) AS amount,
    CAST(order_date AS DATE) AS order_date
FROM (
    SELECT *, ROW_NUMBER() OVER (PARTITION BY order_id ORDER BY _ingested_at DESC) AS rn
    FROM bronze.orders_raw
) WHERE rn = 1 AND amount >= 0;

-- GOLD: business-ready, joined, aggregated
CREATE TABLE gold.daily_sales_summary USING DELTA AS
SELECT order_date, COUNT(*) AS n_orders, SUM(amount) AS total_sales
FROM silver.orders_clean
GROUP BY order_date;
```

Each layer is independently re-runnable and independently testable; Bronze is never modified once written (source of truth for full reprocessing), Silver is the "trusted clean data" layer other teams should build on, and Gold is what BI tools/executives actually query.

# PART 51 - DATA LAKES AND SQL

SQL queries a data lake through **external tables** - a schema definition layered over files in object storage, without physically copying the data into the warehouse.

```sql
-- [PostgreSQL + foreign data wrapper, conceptually similar everywhere]
CREATE FOREIGN TABLE raw.events_csv (event_id INT, event_type TEXT, event_ts TIMESTAMP)
SERVER file_server OPTIONS (filename '/lake/events/2026/02/07/events.csv', format 'csv');

-- [BigQuery] external table directly over Parquet in cloud storage
CREATE EXTERNAL TABLE raw.events_parquet
OPTIONS (format = 'PARQUET', uris = ['gs://lake/events/*.parquet']);
```

**File formats**: **CSV/JSON** are row-oriented, human-readable, but slow for analytics (must read every column to get any column) and lack strong typing. **Parquet**/**ORC** are columnar formats: each column is stored contiguously, so a query touching 3 of 50 columns physically reads only those 3 columns' data, plus columnar formats compress far better (similar values stored together) and carry embedded schema/type information. This is why columnar formats are the default choice for analytics on a lake - the storage layout matches the analytical access pattern (wide tables, narrow column selection, heavy aggregation) far better than row-oriented formats do.

```python

# Python extracts -> SQL transforms -> Python validates -> SQL loads
import requests, pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine("postgresql+psycopg://etl:pw@host/warehouse")

# 1. Python extracts from an API (SQL cannot call an HTTP API)
resp = requests.get("https://api.example.com/orders?date=2026-02-07")
orders = pd.DataFrame(resp.json())
orders.to_sql("stg_orders_raw", engine, if_exists="append", index=False)

# 2. SQL transforms inside the warehouse (fastest place to do set-based work)
with engine.begin() as conn:
    conn.execute(text("""
        INSERT INTO warehouse.fact_orders (customer_id, amount, order_date)
        SELECT customer_id, amount, order_date FROM stg_orders_raw
        WHERE amount >= 0
    """))

# 3. Python validates the result (rich assertion/alerting logic is easier in Python)
row_count = pd.read_sql("SELECT COUNT(*) AS n FROM warehouse.fact_orders", engine)["n"][0]
assert row_count > 0, "Pipeline produced zero rows - alerting!"

# 4. SQL loads/refreshes the final reporting table
with engine.begin() as conn:
    conn.execute(text("REFRESH MATERIALIZED VIEW mart.daily_sales_summary"))   # [PostgreSQL]
```

# PART 54 - SQL AND ORCHESTRATION

Orchestration concepts:
- **DAG** (Directed Acyclic Graph): the pipeline's tasks and their dependencies, with no cycles - task B can depend on task A, but nothing can depend on itself transitively.
- **Task**: one unit of work (often, one SQL statement or one dbt model).
- **Dependency**: an edge in the DAG - "task B cannot start until task A succeeds."
- **Scheduling**: when the DAG runs (cron-like, or event-triggered).
- **Retries**: automatic re-attempt of a failed task, usually with backoff and a max attempt count.
- **Backfills**: re-running the DAG for a past date range (e.g. to fix a bug retroactively).
- **Monitoring**: tracking task success/failure/duration, alerting on SLA breach.

```python

# Apache Airflow - a DAG of SQL tasks with explicit dependencies
from airflow import DAG
from airflow.providers.postgres.operators.postgres import PostgresOperator
from datetime import datetime

with DAG("daily_sales_pipeline", schedule="0 6 * * *", start_date=datetime(2026,1,1)) as dag:
    stage = PostgresOperator(task_id="stage_orders", sql="sql/stage_orders.sql")
    validate = PostgresOperator(task_id="validate_orders", sql="sql/validate_orders.sql")
    load_fact = PostgresOperator(task_id="load_fact_sales", sql="sql/load_fact_sales.sql")
    refresh_mart = PostgresOperator(task_id="refresh_mart", sql="sql/refresh_mart.sql")

    stage >> validate >> load_fact >> refresh_mart   # explicit dependency chain
```

**Prefect** and **Dagster** solve the same problem with a more Python-native task-definition style (plain functions decorated as tasks/assets) and, in Dagster's case, an explicit focus on *data assets* (tables) rather than just tasks - conceptually still the same DAG-of-dependent-SQL-steps pattern. Every SQL step from Parts 1-53 of this handbook (stage, validate, transform, load, test) becomes one node in this DAG; the orchestrator's job is purely to run them in the right order, retry failures, and alert - the actual data logic still lives in the SQL.

# PART 56 - BUILD A COMPLETE DATA WAREHOUSE (e-commerce)

Source tables: `customers`, `products`, `orders`, `order_items`, `payments`, `shipments`.

```sql
-- DIMENSIONS
CREATE TABLE dim_customer (customer_key SERIAL PRIMARY KEY, customer_id INT UNIQUE, name TEXT, segment TEXT);
CREATE TABLE dim_product  (product_key  SERIAL PRIMARY KEY, product_id  INT UNIQUE, name TEXT, category TEXT);
CREATE TABLE dim_date     (date_key INT PRIMARY KEY, full_date DATE, month_name TEXT, year INT);
CREATE TABLE dim_location (location_key SERIAL PRIMARY KEY, city TEXT, country TEXT);

-- FACTS
CREATE TABLE fact_sales (
    sale_key BIGSERIAL PRIMARY KEY,
    order_id INT, customer_key INT REFERENCES dim_customer, product_key INT REFERENCES dim_product,
    date_key INT REFERENCES dim_date, quantity INT, sales_amount NUMERIC(12,2)
);
CREATE TABLE fact_payments (
    payment_key BIGSERIAL PRIMARY KEY,
    order_id INT, customer_key INT REFERENCES dim_customer, date_key INT REFERENCES dim_date,
    payment_amount NUMERIC(12,2), payment_method TEXT
);
CREATE TABLE fact_shipments (
    shipment_key BIGSERIAL PRIMARY KEY,
    order_id INT, ship_date_key INT REFERENCES dim_date, delivered_date_key INT REFERENCES dim_date,
    days_to_deliver INT
);

-- MARTS
CREATE TABLE mart.sales_mart AS
SELECT c.name AS customer_name, p.category, SUM(f.sales_amount) AS total_sales
FROM fact_sales f JOIN dim_customer c ON c.customer_key=f.customer_key
JOIN dim_product p ON p.product_key=f.product_key
GROUP BY c.name, p.category;

CREATE TABLE mart.customer_mart AS
SELECT c.customer_id, c.name, COUNT(DISTINCT f.order_id) AS n_orders, SUM(f.sales_amount) AS lifetime_value
FROM fact_sales f JOIN dim_customer c ON c.customer_key=f.customer_key
GROUP BY c.customer_id, c.name;

CREATE TABLE mart.product_mart AS
SELECT p.category, SUM(f.quantity) AS units_sold, SUM(f.sales_amount) AS revenue
FROM fact_sales f JOIN dim_product p ON p.product_key=f.product_key
GROUP BY p.category;
```

Key decisions: `order_items` (line-item grain) drives `fact_sales`' grain directly - one row per order item, so `sales_amount` sums correctly without inflation (Part 2); `fact_payments` and `fact_shipments` are kept separate from `fact_sales` rather than merged into one giant fact, because they have different grains and different update cadences (a shipment can update days after the sale is recorded); all three marts are conformed to the same `dim_customer`/`dim_product` so numbers are comparable across marts.

# PART 59 - SQL FOR MACHINE LEARNING DATA ENGINEERING

```sql
-- Point-in-time-correct customer features for a training dataset
-- (critical: every feature must only use data available BEFORE the label's timestamp,
--  otherwise the model trains on information it wouldn't have had in production - "leakage")

SELECT
    c.customer_id,
    COUNT(o.order_id)                                   AS transaction_count,
    SUM(o.amount)                                          AS total_spend,
    AVG(o.amount)                                             AS avg_transaction,
    MAX(o.order_date)                                            AS last_transaction_date,
    CURRENT_DATE - MAX(o.order_date)                                AS days_since_last_transaction
FROM dim_customer c
LEFT JOIN fact_sales o
       ON o.customer_key = c.customer_key
      AND o.order_date < '2026-01-01'          -- as-of cutoff: only history before the label date
GROUP BY c.customer_id;
```

**Point-in-time correctness**: a feature computed "as of" a given date must only aggregate events that happened strictly before that date - the `AND o.order_date < '2026-01-01'` clause above is what prevents leakage; without it, a model predicting January churn could accidentally see January's own orders as a feature, making it look far more accurate in testing than it will be in live production. **Rolling features** (e.g. "spend in the last 30 days") use a windowed date filter (`order_date BETWEEN cutoff - 30 AND cutoff`) instead of all-time aggregation. This SQL pattern - one row per entity, one column per feature, computed as-of a cutoff - is exactly what feature stores and ML training pipelines expect as input.

# PART 60 - SQL FOR AI ENGINEERING

SQL underpins most of the non-model parts of an AI system:

- **User data / event data**: who is using the system, what they asked, when (`chat_sessions`, `user_events` tables).
- **Feature data**: same discipline as Part 59, applied to AI product features (e.g. "user's message count today" for rate limiting).
- **Metadata**: document metadata (title, source, ingestion date), embeddings metadata (which model/version generated a vector, when it was generated), all stored relationally even though the vectors themselves usually aren't.
- **Retrieval systems / RAG pipelines**: SQL stores everything *around* the vectors - which documents exist, their chunks, their source, access permissions.
- **Evaluation data / feedback data**: `eval_results`, `user_feedback` tables tracking model output quality over time - a completely standard SQL analytics problem once captured.

```sql
CREATE TABLE ai.eval_results (
    eval_run_id TEXT, model_version TEXT, prompt_id TEXT,
    score NUMERIC, passed BOOLEAN, evaluated_at TIMESTAMP
);

SELECT model_version, AVG(score) AS avg_score, ROUND(100.0*SUM(CASE WHEN passed THEN 1 ELSE 0 END)/COUNT(*),1) AS pass_rate
FROM ai.eval_results
GROUP BY model_version
ORDER BY model_version;
```

# PART 61 - SQL FOR RAG PIPELINES

```
Documents -> ingestion -> cleaning -> chunking -> metadata -> embeddings -> storage -> retrieval -> evaluation
```

```sql
CREATE TABLE ai.documents (
    document_id   BIGSERIAL PRIMARY KEY,
    source        TEXT,
    title         TEXT,
    ingested_at   TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

CREATE TABLE ai.document_chunks (
    chunk_id      BIGSERIAL PRIMARY KEY,
    document_id   BIGINT REFERENCES ai.documents(document_id),
    chunk_index   INT,
    chunk_text    TEXT,
    token_count   INT
);

CREATE TABLE ai.embeddings_metadata (
    chunk_id       BIGINT REFERENCES ai.document_chunks(chunk_id),
    embedding_model TEXT,
    embedding_dim   INT,
    generated_at    TIMESTAMP,
    vector_store_id TEXT       -- pointer/ID into the actual vector database, not the vector itself
);

CREATE TABLE ai.retrieval_logs (
    query_id     BIGSERIAL PRIMARY KEY,
    query_text   TEXT,
    retrieved_chunk_ids BIGINT[],   -- [PostgreSQL] array of chunk_ids returned
    retrieved_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

CREATE TABLE ai.evaluation_results (
    query_id BIGINT REFERENCES ai.retrieval_logs(query_id),
    relevance_score NUMERIC,
    was_helpful BOOLEAN
);
```

Ordinary relational SQL is excellent for everything in these tables - documents, chunks, metadata, logs, evaluation - but it does **not** efficiently perform high-dimensional vector similarity search at scale on its own; that requires either a dedicated vector database (Pinecone, Weaviate, Qdrant) or a vector extension on top of a relational engine (e.g. `pgvector` for PostgreSQL). The right mental model: SQL owns the *system of record* around retrieval (what documents exist, their chunks, their permissions, what was retrieved and how well it performed), while a vector store or extension owns the actual nearest-neighbor search - and `vector_store_id` in the metadata table is the join key connecting the two systems.

# PART 62 - PROFESSIONAL SQL CODING STANDARDS

```sql
-- BAD: unclear names, SELECT *, no formatting, no comments
select * from t1 join t2 on t1.id=t2.tid where t1.x=1;

-- PROFESSIONAL: explicit columns, meaningful aliases, formatted, commented
SELECT
    c.customer_id,
    c.customer_name,
    o.order_id,
    o.order_amount
FROM dim_customer AS c
JOIN fact_orders  AS o
    ON o.customer_key = c.customer_key
WHERE c.is_active = TRUE   -- exclude churned customers from this report
ORDER BY o.order_amount DESC;
```

Standards: consistent naming (`snake_case`, singular vs plural chosen once and applied everywhere, `dim_`/`fact_`/`stg_` prefixes signal layer at a glance), never `SELECT *` in production code (breaks silently when the source adds a column, wastes I/O), explicit table aliases that read naturally, comments explaining *why* not *what*, CTEs organized top-to-bottom in the order they're used with descriptive names (`with_valid_orders`, not `cte1`), and modular SQL (break a huge query into named CTEs / staged views rather than one 300-line unreadable statement).

# PART 63 - SQL CODE REVIEW

**Example query under review:**

```sql
SELECT * FROM orders WHERE customer_id = '101' AND status <> 'cancelled'
```

| Issue type | Finding |
|---|---|
| Performance | `customer_id = '101'` compares an INT column to a string literal - forces an implicit cast on every row, defeats index use |
| Correctness | `status <> 'cancelled'` silently excludes NULL status rows too (NULL comparisons are never true) |
| Maintainability | `SELECT *` - unclear what the caller actually needs, brittle to schema changes |
| Data quality | No check that `customer_id` even exists in `dim_customer` |

**Rewritten:**

```sql
SELECT order_id, customer_id, order_amount, status
FROM orders
WHERE customer_id = 101                                  -- correct type, index-friendly
  AND (status <> 'cancelled' OR status IS NULL);          -- explicit NULL handling
```

Apply this same four-lens review (performance, correctness, maintainability, security/DQ) to every query before merging it into a production pipeline.

# PART 64 - COMMON DATA ENGINEERING SQL MISTAKES (curated 30, extensible pattern)

1. Full reload every run when an incremental watermark would work - wastes compute and time (Part 12).
2. Joining at the wrong grain, inflating downstream sums (Part 2).
3. Ignoring NULLs in filters (`<> 'x'` silently drops NULL rows).
4. No incremental strategy documented - nobody knows if a table is safe to re-run.
5. No audit logging - a silent failure goes unnoticed for days.
6. No data-quality tests before promoting a table to "trusted."
7. Implementing SCD Type 2 without closing the prior row's `effective_to` first (leaves two "current" rows).
8. Non-idempotent pipelines that double-count on retry (Part 28).
9. `SELECT *` in a production model - breaks silently on upstream schema change.
10. Missing indexes on foreign-key/join columns.
11. Poor partition key choice (low cardinality, or a column queries never filter on).
12. Unsafe `DELETE`/`UPDATE` with no `WHERE`, or a `WHERE` that doesn't match intent.
13. Missing transaction boundaries around multi-statement loads (partial failure leaves inconsistent state).
14. Hardcoded dates (`WHERE order_date = '2026-02-07'`) baked into a scheduled job instead of parameterized.
15. Ignoring late-arriving data - historical partitions never get corrected (Part 40).
16. No reconciliation against the source - silent data loss goes undetected (Part 38).
17. Incorrect watermark boundary logic (`>` vs `>=`) causing silent gaps or duplicates (Part 12).
18. Casting inside a `WHERE` clause on the indexed column, killing SARGability (Part 24).
19. Using `MERGE` with an `ON` clause that can match multiple source rows to one target row - undefined/incorrect behavior.
20. Comparing floating point values for exact equality instead of using a tolerance or `NUMERIC`.
21. Not handling timezone conversion consistently between source and warehouse timestamps.
22. Building a fact table on an unstable natural key instead of a stable surrogate key.
23. Forgetting `is_deleted`/soft-delete filters in downstream queries, double-counting "deleted" rows.
24. Letting a pipeline service account run with admin-level DB privileges (Part 44).
25. No schema-change alerting - a renamed source column silently nulls out a whole pipeline.
26. Over-normalizing a mart layer that BI tools then have to join expensively at query time.
27. Using `COUNT(column)` when `COUNT(*)` was intended (or vice versa) - `COUNT(column)` skips NULLs.
28. Aggregating before joining when the join could change the grain, double counting a measure.
29. Not testing referential integrity, allowing orphaned fact rows with no matching dimension.
30. Deploying a query straight to production without ever running `EXPLAIN ANALYZE` on realistic data volume.

Each entry follows the same corrective pattern used throughout this handbook: name the mistake, explain the mechanism, apply the fix from the relevant Part above - use this as a checklist template to keep growing past 30 as your own team encounters new failure modes.

# PART 65 - SQL DEBUGGING (representative set of 10 failures, extensible pattern)

**1. Problem**: Daily revenue dropped 40% overnight.
**Query**: `SELECT SUM(amount) FROM fact_sales WHERE order_date = CURRENT_DATE - 1`
**Diagnosis**: `EXPLAIN ANALYZE` shows the partition for yesterday only has half the expected rows.
**Root cause**: the incremental load's watermark used `>` instead of `>=`, skipping rows with a timestamp equal to the prior watermark.
**Corrected query**: use `>=` on the new watermark and deduplicate by business key downstream (Part 12).
**Prevention**: add a row-count reconciliation check (Part 38) that alerts when daily row counts fall outside a historical range.

**2. Problem**: A `MERGE` statement throws "cardinality violation."
**Diagnosis**: The `ON` clause matches multiple source rows to one target row.
**Root cause**: source staging table had undetected duplicate business keys.
**Corrected query**: deduplicate the source with `ROW_NUMBER()` (Part 11) before the `MERGE`.
**Prevention**: uniqueness test on the staging table before it's allowed to feed a `MERGE` (Part 35).

**3. Problem**: A dashboard total doesn't match finance's spreadsheet.
**Diagnosis**: `fact_sales` grain is per line item, but the dashboard query summed `order_total` (a per-order column) joined to line items.
**Root cause**: grain mismatch (Part 2).
**Corrected query**: sum `line_amount` instead of the repeated `order_total`.
**Prevention**: document grain explicitly in the table's data contract (Part 36).

**4. Problem**: A nightly job times out.
**Diagnosis**: `EXPLAIN` shows a sequential scan on a 200M-row table.
**Root cause**: the `WHERE` clause wraps the filter column in a function, disabling SARGability (Part 24).
**Corrected query**: rewrite as a raw-column range predicate.
**Prevention**: code review checklist item - "no functions on filtered/joined columns."

**5. Problem**: Duplicate rows appear after a pipeline retry.
**Diagnosis**: audit table shows the pipeline ran twice for the same `batch_id` after a transient failure.
**Root cause**: the load step was a blind `INSERT`, not idempotent (Part 28).
**Corrected query**: `DELETE WHERE batch_id = ...` before re-inserting, or switch to `MERGE` keyed on a stable ID.
**Prevention**: idempotency is a required design property before any pipeline ships.

**6. Problem**: A `NOT IN` subquery returns zero rows unexpectedly.
**Diagnosis**: the subquery's result set contains a `NULL`.
**Root cause**: `NOT IN (subquery containing NULL)` evaluates to unknown for every row, so nothing matches.
**Corrected query**: use `NOT EXISTS` instead of `NOT IN` whenever the subquery might contain NULLs.
**Prevention**: standardize on `NOT EXISTS` for anti-joins.

**7. Problem**: SCD Type 2 dimension has two "current" rows for the same customer.
**Diagnosis**: the load script inserted the new version without first closing the old one.
**Root cause**: missing/failed `UPDATE ... SET is_current = FALSE` step (Part 16).
**Corrected query**: wrap close+insert in one transaction so both succeed or neither does.
**Prevention**: a uniqueness test on `(customer_id) WHERE is_current = TRUE` (Part 35).

**8. Problem**: A report is missing an entire day of data.
**Diagnosis**: reconciliation shows the source has rows for that day, the warehouse doesn't.
**Root cause**: the source's daily export failed silently and the pipeline treated zero new rows as success.
**Corrected query**: none - this is an orchestration/monitoring gap.
**Prevention**: freshness check (Part 37) that alerts when expected row counts are zero.

**9. Problem**: Query returns different results each run with no data change.
**Diagnosis**: query relies on `ORDER BY` inside a subquery without a matching `LIMIT`/window function guarantee.
**Root cause**: SQL does not guarantee row order without an explicit `ORDER BY` at the outermost level tied to a deterministic key.
**Corrected query**: add a deterministic tiebreaker column to every `ORDER BY` used for row selection (e.g. `ORDER BY updated_at DESC, id DESC`).
**Prevention**: never assume implicit ordering; make every ordering explicit and deterministic.

**10. Problem**: A `JOIN` unexpectedly multiplies row counts.
**Diagnosis**: `EXPLAIN` shows far more output rows than either input table.
**Root cause**: the join column isn't unique on one side, creating a fan-out (classic grain problem, Part 2).
**Corrected query**: pre-aggregate one side to the join grain before joining.
**Prevention**: know and document the grain and uniqueness of every join key before writing the join.

# PART 66 - SQL INTERVIEW PREPARATION (curated set across levels, extensible pattern)

**Beginner**
1. What is the difference between `WHERE` and `HAVING`? -> `WHERE` filters rows before aggregation; `HAVING` filters groups after aggregation.
2. What is the difference between `DELETE`, `TRUNCATE`, and `DROP`? -> `DELETE` removes rows (can be filtered, logged, rolled back); `TRUNCATE` removes all rows fast (minimal logging, resets identity); `DROP` removes the table structure entirely.
3. What does a `LEFT JOIN` return that an `INNER JOIN` doesn't? -> all rows from the left table, with NULLs for unmatched right-table columns.
4. What is a primary key vs a foreign key? -> primary key uniquely identifies a row in its table; foreign key references a primary key in another table to enforce a relationship.
5. What does `GROUP BY` do? -> collapses rows sharing the same values in the grouped columns into one row per group, for use with aggregate functions.

**Intermediate**
6. Explain the difference between ETL and ELT. -> (Part 5) transform-before-load vs load-then-transform-in-warehouse.
7. What is a surrogate key and why use one? -> (Part 19) stable, warehouse-generated key decoupled from source system IDs, enables SCD Type 2.
8. How do you detect and remove duplicate rows keeping only the latest? -> `ROW_NUMBER() OVER (PARTITION BY key ORDER BY updated_at DESC)`, filter `rn = 1` (Part 11).
9. What is SARGability and why does it matter? -> (Part 24) whether a predicate can use an index/partition; wrapping a column in a function breaks it.
10. What's the difference between a star schema and a snowflake schema? -> (Part 17) denormalized dimensions vs further-normalized sub-dimensions.

**Advanced**
11. Design an incremental loading strategy for a 500M-row table with no reliable `updated_at` column. -> use CDC (log-based) if available; otherwise a full-table hash comparison against yesterday's snapshot, or a sequence/ID-based watermark if inserts are append-only.
12. How would you implement SCD Type 2 for a dimension with three tracked attributes changing independently? -> version the whole row on any tracked-attribute change; consider whether attributes should be tracked together or split into separate mini-dimensions if they change at very different rates.
13. Explain the `>` vs `>=` watermark boundary problem and how you'd solve it robustly. -> (Part 12) prefer a strictly increasing sequence key, or `>=` with downstream idempotent dedup.
14. How do you make a pipeline idempotent when the sink doesn't support `MERGE`? -> delete-then-insert scoped by `batch_id`, or write to a new partition/table version and atomically swap.
15. Walk through how you'd reconcile 150 missing rows between a source and a warehouse. -> (Part 38) row count -> key diff -> pattern analysis by date/batch -> root cause via audit logs.

**Senior / Lead / Architect**
16. How would you design a schema-evolution policy across 50 producer teams feeding one shared lakehouse? -> data contracts (Part 36) with mandatory backward-compatible-by-default changes, versioned breaking changes, automated contract validation at ingestion, and a deprecation timeline.
17. When would you choose ELT over ETL for a new cloud-native pipeline, and when would you still choose ETL? -> ELT when the warehouse has abundant elastic compute and you want re-derivable raw history; ETL when the target system is compute-constrained, when sensitive data must be filtered/masked before it ever lands in a shared warehouse, or when transformation logic needs a general-purpose language.
18. How do you decide between CDC and periodic snapshotting for a given source? -> (Part 15) access to transaction log, need for intermediate change history, source table size, and whether deletes must be captured.
19. Design a data-quality framework that scales to hundreds of tables without becoming unmaintainable. -> declarative tests colocated with model definitions (dbt-style), a shared `dq_results` table, severity tiers (hard-fail vs soft-warn), and ownership tied to each table's contract.
20. How would you architect a lakehouse migration from a legacy on-prem warehouse with minimal downtime? -> dual-write/parallel-run period, reconciliation between old and new systems (Part 38) until they agree consistently, then cut over consumers table by table with contracts.

## PART A - Theory (sample of 10, extend to 30)

1. Explain the difference between OLTP and OLAP. **Answer**: OLTP systems run the business with normalized schemas optimized for many small transactional writes; OLAP systems support analysis with denormalized/dimensional schemas optimized for large scans and aggregations.

4. What is a data contract? **Answer**: (Part 36) an explicit agreement on schema, ownership, freshness, quality, semantics, and SLAs between a data producer and its consumers.
5. Define idempotency in a pipeline context. **Answer**: (Part 28) re-running the same step with the same input leaves the system in the same end state as running it once.
6. What is the medallion architecture? **Answer**: (Part 50) Bronze (raw) -> Silver (cleaned) -> Gold (business-ready) layered lakehouse pattern.
7. Explain SCD Type 1 vs Type 2. **Answer**: (Part 16) Type 1 overwrites with no history; Type 2 inserts a new versioned row and preserves full history via effective dates and a current flag.
8. What is partition pruning? **Answer**: (Part 22) the query planner skips reading partitions that cannot contain matching rows based on the filter predicate.
9. What does SARGable mean? **Answer**: (Part 24) a predicate the engine can evaluate using an index/partition directly, without wrapping the column in a function.
