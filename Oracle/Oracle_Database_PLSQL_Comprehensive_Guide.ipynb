# 1. Oracle Database Fundamentals

## 1.1 What Oracle Database Is

Oracle Database is a commercial, enterprise-grade RDBMS developed by Oracle Corporation, widely used in large organizations (banking, telecom, government, insurance) for its maturity, scalability, and extensive feature set. Since you already know PostgreSQL, MySQL, and SQL Server, this handbook focuses on what makes Oracle distinct: its architecture, PL/SQL, and enterprise-grade tooling, rather than re-teaching relational fundamentals.

## 1.2 Oracle SQL vs PL/SQL

| | Oracle SQL | PL/SQL |
|---|---|---|
| Nature | Declarative - describes *what* data to retrieve/modify | Procedural - Oracle's proprietary procedural extension to SQL |
| Use | Queries, DML, DDL | Business logic, control flow, procedures, functions, packages, triggers |
| Comparable to | Standard SQL (ANSI), similar to PostgreSQL/MySQL/T-SQL | Comparable to PostgreSQL's PL/pgSQL, or SQL Server's T-SQL procedural extensions |

## 1.3 Oracle Database Architecture

Oracle's architecture is more explicitly layered than PostgreSQL or MySQL's, and understanding it is foundational to administration and performance tuning (Parts 22-25, 41-44).

```
Oracle Database
 └── Instance (memory + background processes)
      ├── SGA (System Global Area) - shared memory
      └── PGA (Program Global Area) - per-session memory
 └── Database (physical files on disk)
      ├── Datafiles
      ├── Control files
      ├── Online redo logs
      └── Undo tablespace
```

## 1.4 Database Instance

An **instance** is the combination of Oracle's memory structures (SGA, PGA) and background processes running in memory - it is the "live," running side of Oracle. A database can, in Oracle's advanced Real Application Clusters (RAC) configuration, be served by *multiple* instances simultaneously - a concept without a direct equivalent in PostgreSQL/MySQL's simpler single-process models.

## 1.5 SGA (System Global Area)

Shared memory used by all sessions connected to the instance.

| Component | Purpose |
|---|---|
| Buffer Cache | Caches data blocks read from datafiles, reducing disk I/O |
| Shared Pool | Caches parsed SQL/PL/SQL and execution plans (Part 23.8-23.9) |
| Redo Log Buffer | Buffers redo entries before they're written to the online redo logs (Part 43) |
| Large Pool / Java Pool | Specialized memory for backup operations, Java stored procedures |

## 1.6 PGA (Program Global Area)

Private memory allocated per session/process for sort operations, hash joins, and session-specific state - not shared across sessions, unlike the SGA.

## 1.7 Background Processes

| Process | Role |
|---|---|
| `PMON` | Process monitor - cleans up after failed processes |
| `SMON` | System monitor - performs instance recovery on startup |
| `DBWn` | Database writer - writes modified buffer cache blocks to datafiles |
| `LGWR` | Log writer - writes redo log buffer entries to online redo logs |
| `CKPT` | Checkpoint - signals DBWn and updates datafile headers |
| `ARCn` | Archiver - copies filled online redo logs to archived redo logs (Part 43) |

## 1.8 Datafiles, Control Files, and Online Redo Logs

| File Type | Purpose |
|---|---|
| Datafiles | Store actual table/index data, organized into tablespaces (Part 41) |
| Control files | Track the physical structure of the database (which datafiles, redo logs exist) |
| Online redo logs | Record every change for crash recovery, before being archived (Part 43) |

## 1.9 Undo

Oracle stores the "before" image of changed data in **undo tablespaces**, powering rollback and Oracle's distinctive read consistency model (Part 25, 42) - a mechanism notably different from PostgreSQL's MVCC (which keeps old row versions in the table itself) and from SQL Server's approach.

## 1.10 Tablespaces, Schemas, Users, and Database Objects

| Concept | Definition |
|---|---|
| Tablespace | A logical storage container mapping to one or more physical datafiles (Part 41) |
| Schema | The collection of objects (tables, views, procedures) owned by a specific user - in Oracle, schema and user are effectively the same thing |
| User | A database account with login credentials and privileges |
| Database object | Any named structure: table, view, index, sequence, procedure, etc. (Part 3) |

## 1.11 Oracle Database → Instance → Schema → Objects → Tables

```
Database (physical files)
  └── Instance (memory + processes, connects to the database)
       └── Schemas (one per user, e.g. HR, SCOTT)
            └── Objects (tables, views, indexes, procedures...)
                 └── Tables (rows and columns of data)
```

## 1.12 Key Distinctions

| Comparison | Explanation |
|---|---|
| Database vs Instance | The database is the physical files on disk; the instance is the memory and processes that access those files. In PostgreSQL/MySQL this distinction is far less pronounced - a "database" there is usually just a logical namespace inside one running server process. |
| User vs Schema | In Oracle these are effectively one and the same - creating a user automatically creates its schema. This differs from PostgreSQL, where a schema is a namespace that can be independent of any specific user, and multiple users can share schemas. |
| Tablespace vs Schema | A tablespace is *physical* storage; a schema is a *logical* ownership grouping. One schema's objects can span multiple tablespaces, and one tablespace can hold objects from multiple schemas. |
| Datafile vs Tablespace | A tablespace is the logical container; one or more datafiles are the actual physical OS files backing it. |

## 3.2 Synonyms

A **synonym** is an alias for another object, often used to simplify references to objects in another schema, or to provide location transparency.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE SYNONYM emp FOR hr.employees;
SELECT * FROM emp;   -- equivalent to SELECT * FROM hr.employees
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE PUBLIC SYNONYM emp FOR hr.employees;   -- available to all users
````

Synonyms have no direct equivalent in PostgreSQL (which uses search paths) or MySQL, though SQL Server's `SYNONYM` object is conceptually identical.

## 3.3 Database Links

A **database link** allows querying tables in a *remote* Oracle database as if they were local.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE DATABASE LINK sales_link
    CONNECT TO remote_user IDENTIFIED BY remote_password
    USING 'REMOTE_DB_SERVICE_NAME';

SELECT * FROM orders@sales_link;
````

This is roughly analogous to PostgreSQL's `postgres_fdw` foreign data wrapper or SQL Server's linked servers, but is a more deeply integrated, native Oracle feature.

## 3.4 How Oracle Organizes Objects

Every object belongs to exactly one schema (Part 1.10-1.11), referenced as `schema.object_name`. Within your own session, if you're connected as that schema's owner, the schema prefix can be omitted; accessing another user's objects requires either the full `schema.object` reference, an explicit `GRANT` (Part 45.4), or a synonym.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM hr.employees;   -- explicit schema reference
SELECT * FROM employees;      -- works only if connected as hr, or a synonym/grant exists
````

# 4. Oracle Data Types

## 4.1 Character Types

| Type | Description |
|---|---|
| `CHAR(n)` | Fixed-length character data, blank-padded to `n` bytes/chars. |
| `VARCHAR2(n)` | Variable-length character data, up to `n` bytes (or chars, if declared with `CHAR` semantics), no padding. |
| `NCHAR(n)` / `NVARCHAR2(n)` | Unicode (national character set) equivalents of `CHAR`/`VARCHAR2`. |

## 4.2 Why VARCHAR2 Is Preferred Over VARCHAR

`VARCHAR` currently behaves identically to `VARCHAR2` in Oracle, but Oracle has explicitly reserved `VARCHAR` for potential *future* redefinition with different semantics (e.g. different NULL/empty-string handling). Oracle's own documentation recommends always using `VARCHAR2`, never `VARCHAR`, to guarantee stable, predictable behavior across versions - unlike PostgreSQL, where `VARCHAR` is the standard, stable, ANSI-compliant type.

## 4.3 Numeric: NUMBER, Precision, and Scale

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
NUMBER            -- unconstrained: any precision/scale
NUMBER(10)        -- integer, up to 10 digits
NUMBER(10, 2)     -- 10 total digits, 2 after the decimal point
````

Unlike PostgreSQL (which has distinct `INTEGER`, `NUMERIC`, `REAL` types) or MySQL (`INT`, `DECIMAL`, `FLOAT`), Oracle's single flexible `NUMBER` type handles essentially all numeric needs, with precision and scale as optional constraints.

## 4.4 Date/Time Types

| Type | Stores |
|---|---|
| `DATE` | Date **and** time, to the second (no fractional seconds) - a common point of confusion for those coming from PostgreSQL/MySQL, where `DATE` means date-only. |
| `TIMESTAMP` | Date and time with fractional-second precision (up to 9 digits). |
| `TIMESTAMP WITH TIME ZONE` | Timestamp plus an explicit time zone offset stored with the value. |
| `TIMESTAMP WITH LOCAL TIME ZONE` | Stored normalized to the database's time zone; displayed converted to the session's time zone. |
| `INTERVAL YEAR TO MONTH` / `INTERVAL DAY TO SECOND` | Represents a duration, not a point in time. |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT SYSDATE, SYSTIMESTAMP FROM dual;
SELECT INTERVAL '2' DAY + INTERVAL '3' HOUR FROM dual;
````

## 4.5 Large Objects (LOBs)

| Type | Stores |
|---|---|
| `CLOB` | Large character data (up to 128 TB), for large text. |
| `NCLOB` | Unicode version of CLOB. |
| `BLOB` | Large binary data (images, files). |
| `BFILE` | A reference to a binary file stored *outside* the database, on the filesystem. |

Comparable to PostgreSQL's `TEXT`/`BYTEA` (though Oracle's LOB model is more explicit and configurable) and MySQL's `TEXT`/`BLOB` family.

## 4.6 Other Types

| Type | Purpose |
|---|---|
| `RAW(n)` | Raw binary data, up to 2000 bytes, unconverted by character set. |
| `ROWID` | The physical address of a row - Oracle's internal, extremely fast row identifier. |
| `UROWID` | A universal row identifier, supporting non-Oracle or index-organized tables. |
| `XMLType` | Native XML storage and querying (Part 39). |
| JSON | Oracle stores JSON in `VARCHAR2`, `CLOB`, or (in newer versions) a native `JSON` type, with dedicated functions and indexing (Part 38). |

## 4.7 Data-Type Selection by Domain

| Domain | Guidance |
|---|---|
| Finance/Banking | `NUMBER(p, s)` with an explicit precision/scale for currency (e.g. `NUMBER(15,2)`) - never floating-point types, to avoid rounding errors. |
| Healthcare | `VARCHAR2` for identifiers/codes; `TIMESTAMP WITH TIME ZONE` for events spanning multiple facilities/time zones; `CLOB` for clinical notes. |
| E-commerce | `NUMBER(12,2)` for prices; `VARCHAR2` for SKUs/descriptions; `TIMESTAMP` for order events. |
| Analytics | Favor `NUMBER` and `DATE`/`TIMESTAMP` types that aggregate and partition (Part 40) cleanly; avoid LOBs in fact tables. |


# 5. Table Creation and Design

## 5.1 CREATE TABLE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE employees (
    employee_id     NUMBER(6) PRIMARY KEY,
    first_name      VARCHAR2(50) NOT NULL,
    last_name       VARCHAR2(50) NOT NULL,
    email           VARCHAR2(100) UNIQUE NOT NULL,
    hire_date       DATE DEFAULT SYSDATE,
    salary          NUMBER(10,2) CHECK (salary > 0),
    department_id   NUMBER(4),
    CONSTRAINT fk_department FOREIGN KEY (department_id) REFERENCES departments(department_id)
);
````

Familiar territory if you know PostgreSQL/MySQL/SQL Server `CREATE TABLE` syntax - constraints and defaults behave the same conceptually, with Oracle-specific data types (Part 4).

## 5.2 ALTER TABLE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
ALTER TABLE employees ADD phone_number VARCHAR2(20);
ALTER TABLE employees MODIFY salary NUMBER(12,2);
ALTER TABLE employees DROP COLUMN phone_number;
ALTER TABLE employees ADD CONSTRAINT chk_salary CHECK (salary > 0);
````

**Oracle-specific note:** `MODIFY` (not `ALTER COLUMN`) is Oracle's keyword for changing a column's definition - a syntax difference worth remembering when translating from PostgreSQL/SQL Server (Part 55).

## 5.3 DROP TABLE and TRUNCATE TABLE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DROP TABLE employees;
DROP TABLE employees PURGE;   -- skip the recycle bin entirely
TRUNCATE TABLE employees;
````

**Oracle-specific:** by default, `DROP TABLE` moves the table to a **recycle bin** (like a trash folder) rather than permanently deleting it, allowing recovery with `FLASHBACK TABLE ... TO BEFORE DROP`. `PURGE` skips this safety net. PostgreSQL and MySQL have no equivalent recycle bin concept for dropped tables.

## 5.4 Constraints

|---|---|

| `FOREIGN KEY` | Enforces a valid relationship to another table |
| `UNIQUE` | Disallows duplicate values |

| `NOT NULL` | Disallows missing values |

## 5.5 Composite Keys

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE order_items (
    order_id    NUMBER,
    product_id  NUMBER,
    quantity    NUMBER,
    CONSTRAINT pk_order_items PRIMARY KEY (order_id, product_id)
);
````

## 5.6 Natural Keys vs Surrogate Keys vs Composite Keys

| Key Type | Definition | Example |
|---|---|---|
| Natural key | A real-world attribute used as the key | `email`, `national_id` |
| Surrogate key | An artificial identifier with no business meaning | An auto-generated `employee_id` |
| Composite key | A primary key made of multiple columns together | `(order_id, product_id)` |

**Professional guidance:** surrogate keys (typically generated via sequences or identity columns, Part 6) are the standard choice for primary keys in Oracle, exactly as in PostgreSQL/MySQL/SQL Server - natural keys remain valuable as `UNIQUE` constraints alongside a surrogate primary key.


# 6. Oracle Sequences and ID Generation

## 6.1 CREATE SEQUENCE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE SEQUENCE employees_seq
    START WITH 1
    INCREMENT BY 1
    MINVALUE 1
    MAXVALUE 999999999
    NOCYCLE
    CACHE 20;
````

## 6.2 NEXTVAL and CURRVAL

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employees_seq.NEXTVAL FROM dual;   -- generates and returns the next value
SELECT employees_seq.CURRVAL FROM dual;   -- returns the value most recently generated in this session

INSERT INTO employees (employee_id, first_name)
VALUES (employees_seq.NEXTVAL, 'Ada');
````

`CURRVAL` requires `NEXTVAL` to have already been called at least once in the current session - calling `CURRVAL` first raises an error.

## 6.3 Sequence Options Explained

| Option | Meaning |
|---|---|
| `START WITH` | The first value generated |
| `INCREMENT BY` | The step between values (can be negative, for descending sequences) |
| `MINVALUE` / `MAXVALUE` | The bounds a sequence will generate within |
| `CYCLE` / `NOCYCLE` | Whether the sequence wraps back to `MINVALUE` after reaching `MAXVALUE`, or raises an error instead |
| `CACHE n` | Pre-allocates `n` values in memory for performance, at the cost of potential gaps if the instance restarts |
| `ORDER` / `NOORDER` | In Oracle RAC (clustered) configurations, whether values are guaranteed to be generated in strict order across instances |

## 6.4 Oracle SEQUENCE vs PostgreSQL SEQUENCE vs MySQL AUTO_INCREMENT vs SQL Server IDENTITY

| | Oracle | PostgreSQL | MySQL | SQL Server |
|---|---|---|---|---|
| Mechanism | Standalone `SEQUENCE` object | `SEQUENCE` object (often via `SERIAL`/`IDENTITY`) | `AUTO_INCREMENT` column attribute | `IDENTITY` column property |
| Tied to a table? | No - independent, reusable across tables | Optionally independent, or tied via `SERIAL` | Tied directly to one column | Tied directly to one column |
| Manual value retrieval | `seq.NEXTVAL` / `seq.CURRVAL` | `nextval('seq')` / `currval('seq')` | `LAST_INSERT_ID()` | `SCOPE_IDENTITY()` |
| Explicit insert control | Full control - call `NEXTVAL` whenever needed | Full control, similar to Oracle | Automatic on insert only | Automatic on insert only |

## 6.5 Modern Identity Columns in Oracle

Since Oracle 12c, Oracle also supports `GENERATED ... AS IDENTITY`, offering syntax closer to PostgreSQL/SQL Server's identity columns while still being implemented internally via a sequence:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE employees (
    employee_id NUMBER GENERATED ALWAYS AS IDENTITY,
    first_name  VARCHAR2(50)
);

INSERT INTO employees (first_name) VALUES ('Ada');   -- employee_id is generated automatically
````

| Clause | Behavior |
|---|---|
| `GENERATED ALWAYS AS IDENTITY` | The database always generates the value; an explicit value in `INSERT` raises an error unless overridden. |
| `GENERATED BY DEFAULT AS IDENTITY` | The database generates a value only if none is explicitly provided. |

**Professional guidance:** for new Oracle development, `GENERATED ... AS IDENTITY` is generally preferred for simple auto-numbered primary keys, since it's more concise and closely matches conventions from other databases; standalone sequences remain essential when you need explicit, shared, or cross-table numbering control.

## 7.5 INSERT ALL and INSERT FIRST (Oracle-Specific)

`INSERT ALL` inserts into multiple tables in a single statement, based on one source query - useful for splitting incoming data across several destination tables at once.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
INSERT ALL
    INTO sales_domestic (order_id, amount) VALUES (order_id, amount)
    INTO sales_log (order_id, logged_at) VALUES (order_id, SYSDATE)
SELECT order_id, amount FROM staging_orders WHERE country = 'US';
````

`INSERT FIRST` evaluates conditions in order and inserts each source row into only the **first** matching branch - useful for conditional routing:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
INSERT FIRST
    WHEN amount > 10000 THEN INTO large_orders VALUES (order_id, amount)
    WHEN amount > 1000  THEN INTO medium_orders VALUES (order_id, amount)
    ELSE INTO small_orders VALUES (order_id, amount)
SELECT order_id, amount FROM staging_orders;
````

Neither `INSERT ALL` nor `INSERT FIRST` has a direct equivalent in PostgreSQL, MySQL, or SQL Server - both are genuinely Oracle-specific, powerful for ETL-style conditional data distribution (Part 50).

# 8. SELECT and Oracle SQL

## 8.1 Core Clauses (Recap)

`SELECT`, `DISTINCT`, `WHERE`, `ORDER BY`, `GROUP BY`, `HAVING` all work as you already know from PostgreSQL/MySQL/SQL Server - Oracle follows the same standard SQL semantics for these clauses.

## 8.2 Oracle's Modern Row-Limiting Syntax: FETCH FIRST / OFFSET

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, salary
FROM employees
ORDER BY salary DESC
OFFSET 10 ROWS FETCH NEXT 5 ROWS ONLY;
````

This is the modern, ANSI SQL:2008-standard row-limiting syntax, available since Oracle 12c.

## 8.3 Comparison: Row-Limiting Syntax Across Databases

| Database | Syntax |
|---|---|
| Oracle (12c+) | `OFFSET n ROWS FETCH NEXT m ROWS ONLY` |
| PostgreSQL | `LIMIT m OFFSET n` |
| MySQL | `LIMIT n, m` |
| SQL Server | `OFFSET n ROWS FETCH NEXT m ROWS ONLY` (SQL Server 2012+) or `TOP m` |

Oracle's modern syntax is identical to SQL Server's `OFFSET/FETCH`, since both adopted the same ANSI standard - the most portable choice going forward if you need to write similar logic across both.

## 8.4 Legacy Oracle Row Limiting: ROWNUM

Before 12c, Oracle developers used the pseudo-column `ROWNUM`:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM (
    SELECT employee_id, salary FROM employees ORDER BY salary DESC
) WHERE ROWNUM <= 5;
````

**Important trap:** `ROWNUM` is assigned *before* `ORDER BY` is applied within the same query - which is exactly why the query above wraps the ordered query in a subquery first, applying `ROWNUM` to the *already-sorted* result. Writing `WHERE ROWNUM <= 5 ORDER BY salary DESC` directly (without the subquery) gives the wrong 5 rows, since `ROWNUM` would be filtering *before* sorting happens. This remains one of the most common Oracle beginner mistakes (Part 16.1 covers `ROWNUM` further).

## 8.5 Logical Query Processing Order

```
FROM  ->  WHERE  ->  GROUP BY  ->  HAVING  ->  SELECT  ->  ORDER BY  ->  OFFSET/FETCH
```

This logical order (identical across Oracle, PostgreSQL, MySQL, and SQL Server) explains several common confusions: why a column alias defined in `SELECT` cannot be used in `WHERE` (since `WHERE` is logically evaluated first), but *can* be used in `ORDER BY` (evaluated last).

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT salary * 1.1 AS adjusted_salary
FROM employees
WHERE salary > 5000        -- cannot reference adjusted_salary here
ORDER BY adjusted_salary;  -- but CAN reference it here
````

# 9. Oracle NULL Behavior

## 9.1 The Critical Oracle-Specific Rule: Empty String Equals NULL

**This is one of the most important Oracle-specific behaviors to internalize:** Oracle treats an empty string (`''`) in a `VARCHAR2`/`CHAR` column as `NULL`, not as a distinct zero-length value.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
INSERT INTO employees (employee_id, middle_name) VALUES (1, '');
SELECT * FROM employees WHERE middle_name IS NULL;   -- the row above IS returned
````

This differs from PostgreSQL and MySQL, both of which treat `''` and `NULL` as genuinely distinct values. This has real practical consequences: application code written and tested against PostgreSQL/MySQL that relies on distinguishing `''` from `NULL` will behave incorrectly against Oracle without adjustment.

## 9.2 IS NULL and IS NOT NULL (Recap)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM employees WHERE commission_pct IS NULL;
SELECT * FROM employees WHERE commission_pct IS NOT NULL;
````

Standard behavior, identical across all four databases - `=` and `!=` never match `NULL`; only `IS [NOT] NULL` does.

## 9.3 NVL

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, NVL(commission_pct, 0) AS commission FROM employees;
````

`NVL(expr, replacement)` returns `replacement` if `expr` is `NULL`, otherwise returns `expr` - Oracle's original null-substitution function.

## 9.4 NVL2

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, NVL2(commission_pct, 'Has Commission', 'No Commission') AS status
FROM employees;
````

`NVL2(expr, value_if_not_null, value_if_null)` - a three-way conditional based specifically on nullness, with no direct single-function equivalent in PostgreSQL/MySQL/SQL Server (which would require a `CASE` expression instead).

## 9.5 COALESCE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT COALESCE(commission_pct, bonus_pct, 0) AS effective_rate FROM employees;
````

Standard ANSI SQL, available identically across Oracle, PostgreSQL, MySQL, and SQL Server - returns the first non-NULL value from a list of expressions.

## 9.6 NULLIF

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT NULLIF(department_id, 0) FROM employees;
````

Returns `NULL` if the two arguments are equal, otherwise returns the first argument - useful for converting a "sentinel" value (like `0` meaning "unassigned") into a true `NULL`.

## 9.7 NVL vs COALESCE

| | NVL | COALESCE |
|---|---|---|
| Arguments | Exactly 2 | 2 or more |
| Standard | Oracle-proprietary | ANSI SQL standard, portable |
| Short-circuit evaluation | No - both arguments are always evaluated | Yes - stops at the first non-NULL argument, which matters for expensive expressions or subqueries |
| Portability | Oracle-only | Works across PostgreSQL, MySQL, SQL Server, and Oracle |

**Professional guidance:** prefer `COALESCE` for portability and short-circuit evaluation, unless you're specifically working in a codebase with an established `NVL` convention.

## 9.8 NULL Handling Compared Across Databases

| Function/Behavior | Oracle | PostgreSQL | MySQL | SQL Server |
|---|---|---|---|---|
| Empty string vs NULL | Treated as the same value | Distinct values | Distinct values | Distinct values |
| Null-coalesce function | `NVL`, `COALESCE` | `COALESCE` | `IFNULL`, `COALESCE` | `ISNULL`, `COALESCE` |
| Three-way null conditional | `NVL2` | `CASE` expression | `CASE` expression | `CASE` expression |

# 12. Joins

## 12.1 Modern ANSI JOIN Syntax (Recap)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT e.first_name, d.department_name
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;

SELECT e.first_name, d.department_name
FROM employees e
LEFT JOIN departments d ON e.department_id = d.department_id;
````

`INNER`, `LEFT`, `RIGHT`, `FULL OUTER`, `CROSS`, and `SELF` joins all behave identically to PostgreSQL/MySQL/SQL Server - no Oracle-specific semantics here at the conceptual level.

## 12.2 Oracle's Legacy Outer-Join Syntax

Older Oracle code (and interview questions) frequently use Oracle's proprietary `(+)` operator for outer joins, predating ANSI join syntax support:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Legacy Oracle outer join syntax
SELECT e.first_name, d.department_name
FROM employees e, departments d
WHERE e.department_id = d.department_id(+);
````

The `(+)` marks the "optional" side of the join - here, it produces the same result as a `LEFT JOIN` from `employees` to `departments` (departments is optional/nullable).

## 12.3 Why Modern ANSI JOIN Syntax Should Be Preferred

| Legacy `(+)` Syntax | Modern ANSI `JOIN` Syntax |
|---|---|
| Oracle-only, not portable | Portable across PostgreSQL, MySQL, SQL Server, Oracle |
| Cannot express a full outer join at all | Supports `FULL OUTER JOIN` directly |
| Join conditions mixed with filter conditions in `WHERE`, easy to misplace | Join conditions cleanly separated in `ON`, filters in `WHERE` |
| Ambiguous/error-prone with more than two tables | Explicit, unambiguous with any number of tables |

**Professional guidance:** always use ANSI `JOIN` syntax in new code; recognize `(+)` only because you will likely encounter it in legacy Oracle systems.

## 12.4 Join Conditions, Cardinality, and Duplicate Rows

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT c.customer_name, o.order_id
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id;
````

If a customer has 3 orders, this join produces 3 rows for that customer - a direct consequence of relationship cardinality (Part 1.8 concepts you already know from other databases). Forgetting this is a common cause of accidentally inflated aggregate results (e.g. `SUM()` overcounting due to a fan-out join) - identical risk in Oracle as in any relational database.

## 12.5 NULLs in Joins

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT c.customer_name, o.order_id
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;   -- customers with no orders at all
````

## 12.6 Many-to-Many Relationships

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT s.student_name, c.course_name
FROM students s
JOIN enrollments e ON s.student_id = e.student_id
JOIN courses c ON e.course_id = c.course_id;
````

Standard junction-table pattern, identical in concept to what you already know from other databases.

## 12.7 SELF JOIN

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT e.first_name AS employee, m.first_name AS manager
FROM employees e
JOIN employees m ON e.manager_id = m.employee_id;
````

# 13. Subqueries

## 13.1 Scalar Subqueries

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, salary,
       (SELECT AVG(salary) FROM employees) AS company_avg
FROM employees;
````

A scalar subquery returns exactly one row, one column - usable anywhere a single value is expected.

## 13.2 Single-Row and Multi-Row Subqueries

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Single-row: comparison operators (=, >, <) work directly
SELECT * FROM employees WHERE salary > (SELECT AVG(salary) FROM employees);

-- Multi-row: requires IN, ANY, ALL
SELECT * FROM employees WHERE department_id IN (SELECT department_id FROM departments WHERE location_id = 1700);
````

## 13.3 Correlated Subqueries

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT e.employee_id, e.salary
FROM employees e
WHERE e.salary > (
    SELECT AVG(salary) FROM employees e2 WHERE e2.department_id = e.department_id
);
````

A **correlated** subquery references a column from the outer query (`e.department_id`), so it is logically re-evaluated for every outer row - conceptually identical to correlated subqueries in PostgreSQL/MySQL/SQL Server, though Oracle's optimizer (Part 22) may rewrite it internally for efficiency.

## 13.4 EXISTS and NOT EXISTS

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT c.customer_id FROM customers c
WHERE EXISTS (SELECT 1 FROM orders o WHERE o.customer_id = c.customer_id);

SELECT c.customer_id FROM customers c
WHERE NOT EXISTS (SELECT 1 FROM orders o WHERE o.customer_id = c.customer_id);
````

`EXISTS` stops as soon as it finds one matching row - often more efficient than `IN` for large subquery result sets, since it doesn't need to materialize the full list.

## 13.5 IN and NOT IN

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM employees WHERE department_id IN (10, 20, 30);
SELECT * FROM employees WHERE department_id NOT IN (10, 20, 30);
````

## 13.6 The NULL Trap With NOT IN

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM employees
WHERE department_id NOT IN (SELECT department_id FROM departments WHERE manager_id IS NULL);
````

**Critical, universal SQL trap (not Oracle-specific, but essential to restate):** if the subquery's result set contains even a single `NULL`, `NOT IN` returns **no rows at all** - because comparing anything to `NULL` yields `UNKNOWN`, not `TRUE` or `FALSE`, and `NOT IN` requires every comparison to be definitively `TRUE`.

**Safe alternative:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM employees e
WHERE NOT EXISTS (
    SELECT 1 FROM departments d WHERE d.manager_id IS NULL AND d.department_id = e.department_id
);
````

`NOT EXISTS` does not suffer from this trap, since it doesn't compare against a flat list of potentially-`NULL` values - this makes `NOT EXISTS` the generally safer default over `NOT IN` whenever the subquery's column might contain `NULL`.


# 14. Common Table Expressions

## 14.1 Basic CTE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
WITH high_earners AS (
    SELECT * FROM employees WHERE salary > 10000
)
SELECT department_id, COUNT(*) FROM high_earners GROUP BY department_id;
````

Behaves identically to CTEs in PostgreSQL/MySQL 8+/SQL Server.

## 14.2 Multiple CTEs

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
WITH dept_totals AS (
    SELECT department_id, SUM(salary) AS total_salary FROM employees GROUP BY department_id
),
dept_avg AS (
    SELECT AVG(total_salary) AS avg_total FROM dept_totals
)
SELECT dt.department_id, dt.total_salary
FROM dept_totals dt, dept_avg da
WHERE dt.total_salary > da.avg_total;
````

## 14.3 Recursive CTEs

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
WITH RECURSIVE employee_hierarchy (employee_id, first_name, manager_id, lvl) AS (
    SELECT employee_id, first_name, manager_id, 1 FROM employees WHERE manager_id IS NULL
    UNION ALL
    SELECT e.employee_id, e.first_name, e.manager_id, eh.lvl + 1
    FROM employees e
    JOIN employee_hierarchy eh ON e.manager_id = eh.employee_id
)
SELECT * FROM employee_hierarchy ORDER BY lvl;
````

Standard ANSI recursive CTE syntax, supported in modern Oracle versions - identical structure to PostgreSQL/SQL Server recursive CTEs.

## 14.4 Oracle's Hierarchical Queries: CONNECT BY

Oracle has its own, older, and still very widely used hierarchical query syntax that predates recursive CTEs:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, first_name, manager_id, LEVEL
FROM employees
START WITH manager_id IS NULL
CONNECT BY PRIOR employee_id = manager_id
ORDER SIBLINGS BY first_name;
````

| Clause | Meaning |
|---|---|
| `START WITH` | Defines the root row(s) of the hierarchy |
| `CONNECT BY PRIOR` | Defines the parent-child relationship; `PRIOR` refers to the *parent* row's value |
| `LEVEL` | A pseudo-column giving the depth in the hierarchy (root = 1) |
| `ORDER SIBLINGS BY` | Sorts children within each parent, preserving the hierarchy's shape (unlike a plain `ORDER BY`, which would flatten it) |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Useful string-formatting trick for visualizing hierarchy depth
SELECT LPAD(' ', 2 * (LEVEL - 1)) || first_name AS org_chart
FROM employees
START WITH manager_id IS NULL
CONNECT BY PRIOR employee_id = manager_id;
````

## 14.5 CONNECT BY vs Recursive CTE

| | `CONNECT BY` | Recursive CTE |
|---|---|---|
| Portability | Oracle-only | Standard ANSI, works across PostgreSQL/SQL Server/modern MySQL |
| Conciseness | Very concise for simple hierarchies | More verbose, but more explicit/flexible |
| Depth tracking | Built-in `LEVEL` pseudo-column | Must be tracked manually as a computed column |
| Cycle detection | Built-in `CONNECT_BY_ISCYCLE`, `NOCYCLE` clause | Must be handled manually |
| Sibling ordering | Native `ORDER SIBLINGS BY` | Requires manual path-tracking to replicate |

**Professional guidance:** `CONNECT BY` remains extremely common in existing Oracle codebases and is often more concise for straightforward hierarchies; recursive CTEs are preferable when portability across databases matters, or when the recursive logic is more complex than a simple parent-child walk.


# 15. Analytic/Window Functions

## 15.1 The OVER Clause

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, department_id, salary,
       AVG(salary) OVER (PARTITION BY department_id) AS dept_avg_salary
FROM employees;
````

`OVER (...)` turns an ordinary aggregate into a **window function**: it computes across a set of related rows (the "window") without collapsing them into fewer output rows - every original row is preserved, exactly as covered for standard SQL in your Database Programming handbook (Part 4.10 there), now with Oracle-specific depth.

## 15.2 PARTITION BY and ORDER BY Within OVER

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, department_id, salary,
       RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) AS dept_rank
FROM employees;
````

`PARTITION BY` resets the window's calculation for each group; `ORDER BY` (inside `OVER`) defines the row ordering for ranking, running totals, and offset functions like `LAG`/`LEAD`.

## 15.3 ROW_NUMBER, RANK, DENSE_RANK

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, salary,
       ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num,
       RANK()       OVER (ORDER BY salary DESC) AS rnk,
       DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rnk
FROM employees;
````

| Function | Behavior on Ties |
|---|---|
| `ROW_NUMBER()` | Always assigns unique, sequential numbers, even for tied values (arbitrary tie-breaking) |
| `RANK()` | Tied rows get the same rank; the next rank *skips* accordingly (1, 1, 3) |
| `DENSE_RANK()` | Tied rows get the same rank; the next rank does *not* skip (1, 1, 2) |

## 15.4 NTILE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, salary,
       NTILE(4) OVER (ORDER BY salary) AS salary_quartile
FROM employees;
````

Divides rows into `n` roughly equal-sized buckets - commonly used for quartiles/deciles in analytics.

## 15.5 LAG and LEAD

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT order_date, revenue,
       LAG(revenue, 1) OVER (ORDER BY order_date) AS prev_day_revenue,
       LEAD(revenue, 1) OVER (ORDER BY order_date) AS next_day_revenue
FROM daily_revenue;
````

`LAG` looks backward, `LEAD` looks forward, by a specified number of rows - foundational for period-over-period comparisons (Part 15.9).

## 15.6 FIRST_VALUE and LAST_VALUE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, department_id, salary,
       FIRST_VALUE(salary) OVER (PARTITION BY department_id ORDER BY salary DESC) AS highest_in_dept,
       LAST_VALUE(salary) OVER (
           PARTITION BY department_id ORDER BY salary DESC
           ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING
       ) AS lowest_in_dept
FROM employees;
````

**Common mistake:** `LAST_VALUE` without an explicit `ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING` frame only sees rows up to the *current* row by default, producing a misleading result equal to the current row's own value rather than the true group maximum/minimum - always specify the frame explicitly when using `LAST_VALUE`.

## 15.7 Running Totals

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT order_date, revenue,
       SUM(revenue) OVER (ORDER BY order_date ROWS UNBOUNDED PRECEDING) AS running_total
FROM daily_revenue;
````

## 15.8 Moving Averages

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT order_date, revenue,
       AVG(revenue) OVER (ORDER BY order_date ROWS BETWEEN 6 PRECEDING AND CURRENT ROW) AS moving_avg_7day
FROM daily_revenue;
````

## 15.9 Month-over-Month and Year-over-Year Growth

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT month, revenue,
       LAG(revenue, 1) OVER (ORDER BY month) AS prev_month_revenue,
       ROUND(
           (revenue - LAG(revenue, 1) OVER (ORDER BY month))
           / LAG(revenue, 1) OVER (ORDER BY month) * 100, 2
       ) AS mom_growth_pct
FROM monthly_revenue;
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT year, revenue,
       LAG(revenue, 1) OVER (ORDER BY year) AS prev_year_revenue,
       ROUND(
           (revenue - LAG(revenue, 1) OVER (ORDER BY year))
           / LAG(revenue, 1) OVER (ORDER BY year) * 100, 2
       ) AS yoy_growth_pct
FROM yearly_revenue;
````

## 15.10 Cumulative Percentages

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT product_id, revenue,
       SUM(revenue) OVER (ORDER BY revenue DESC ROWS UNBOUNDED PRECEDING)
           / SUM(revenue) OVER () * 100 AS cumulative_pct_of_total
FROM product_revenue;
````

## 15.11 Top-N Analysis

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM (
    SELECT employee_id, department_id, salary,
           RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) AS rnk
    FROM employees
)
WHERE rnk <= 3;   -- top 3 earners per department
````

## 15.12 Real-World Examples

**Banking - detecting large day-over-day balance swings:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT account_id, txn_date, balance,
       balance - LAG(balance) OVER (PARTITION BY account_id ORDER BY txn_date) AS balance_change
FROM account_balances;
````

**HR - ranking employees within department by tenure:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT employee_id, department_id, hire_date,
       ROW_NUMBER() OVER (PARTITION BY department_id ORDER BY hire_date) AS tenure_rank
FROM employees;
````

## 16.4 KEEP (DENSE_RANK ...)

## 16.7 MODEL Clause Concepts

The `MODEL` clause is an advanced, Oracle-exclusive SQL extension that lets you define spreadsheet-like formulas and cell references directly within a SQL query (e.g. `cell[year, product] = cell[year-1, product] * 1.1`), useful for complex financial forecasting scenarios that would otherwise require pulling data out into an external tool.

# 19. Views

## 19.1 CREATE VIEW

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE VIEW active_employees AS
SELECT employee_id, first_name, last_name, department_id
FROM employees
WHERE termination_date IS NULL;

SELECT * FROM active_employees WHERE department_id = 90;
````

A view is a named, stored query - identical concept to PostgreSQL/MySQL/SQL Server views.

## 19.2 Updatable Views

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
UPDATE active_employees SET department_id = 100 WHERE employee_id = 101;
````

A view is updatable (its underlying base table can be modified through it) only if it meets certain conditions: it selects from a single base table (or, in limited cases, a simple join), includes no `DISTINCT`, `GROUP BY`, aggregate functions, or set operators, and every `NOT NULL` column without a default is included.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE VIEW active_employees_secure AS
SELECT employee_id, first_name, last_name, department_id
FROM employees
WHERE termination_date IS NULL
WITH CHECK OPTION;
````

`WITH CHECK OPTION` prevents an `UPDATE`/`INSERT` through the view from creating a row that would no longer satisfy the view's own `WHERE` clause (e.g. accidentally "terminating" an employee via this view, which would make them disappear from it).

## 19.3 Complex Views

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE VIEW department_summary AS
SELECT d.department_name, COUNT(e.employee_id) AS headcount, AVG(e.salary) AS avg_salary
FROM departments d
LEFT JOIN employees e ON d.department_id = e.department_id
GROUP BY d.department_name;
````

A view involving joins and aggregation - not updatable, but extremely valuable for encapsulating complex reporting logic behind a simple, reusable name.

## 19.4 Security Through Views

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE VIEW public_employee_directory AS
SELECT employee_id, first_name, last_name, department_id
FROM employees;   -- deliberately excludes salary, ssn, etc.

GRANT SELECT ON public_employee_directory TO junior_staff_role;
````

Granting access to a view (rather than the underlying table directly) lets you expose only specific columns/rows to a given role - a common, professional pattern for enforcing column-level and row-level security (Part 45).

## 19.5 Abstraction and Reporting

Views decouple the *logical* shape reporting tools and applications rely on from the *physical* structure of underlying tables - if a table is later restructured, a view can often be redefined to preserve the same external interface, avoiding breaking every downstream consumer.


# 20. Materialized Views

## 20.1 Why Materialized Views Exist

Unlike a regular view (which re-executes its query every time it's accessed), a **materialized view** physically stores the query's result set on disk, like a table - dramatically speeding up repeated access to expensive aggregations or joins, at the cost of the data potentially being slightly out of date until refreshed.

## 20.2 Creating a Materialized View

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE MATERIALIZED VIEW mv_department_summary
BUILD IMMEDIATE
REFRESH COMPLETE ON DEMAND
AS
SELECT department_id, COUNT(*) AS headcount, SUM(salary) AS total_salary
FROM employees
GROUP BY department_id;
````

## 20.3 Query Performance and Data Warehousing

Materialized views are foundational to data warehousing (Part 51): a nightly ETL job can populate detailed fact tables, and materialized views can pre-aggregate them into summary tables that dashboards query directly, avoiding re-running expensive joins/aggregations on every dashboard refresh.

## 20.4 Refresh Strategies

| Strategy | Behavior |
|---|---|
| `REFRESH COMPLETE` | Fully re-executes the defining query, replacing all data - simple but potentially expensive for large datasets. |
| `REFRESH FAST` | Applies only the *changes* since the last refresh, using a materialized view log on the base table(s) - much more efficient, but requires the query to meet specific eligibility rules. |
| `REFRESH ON DEMAND` | Refreshed only when explicitly requested (`DBMS_MVIEW.REFRESH`) or on a schedule. |
| `REFRESH ON COMMIT` | Automatically refreshed whenever a transaction modifying the base table commits - keeps the materialized view always current, at the cost of extra overhead on every relevant commit. |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Required for FAST refresh to be possible
CREATE MATERIALIZED VIEW LOG ON employees WITH ROWID, SEQUENCE (department_id, salary) INCLUDING NEW VALUES;

CREATE MATERIALIZED VIEW mv_department_summary
REFRESH FAST ON COMMIT
AS
SELECT department_id, COUNT(*) AS headcount, SUM(salary) AS total_salary
FROM employees
GROUP BY department_id;
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Manually triggering a refresh
BEGIN
    DBMS_MVIEW.REFRESH('MV_DEPARTMENT_SUMMARY', 'C');   -- 'C' = complete refresh
END;
/
````

## 20.5 Materialized Views vs Normal Views

| | View | Materialized View |
|---|---|---|
| Storage | None - re-executes the query every time | Physically stores the result set |
| Freshness | Always current (reflects live data) | Current only as of the last refresh |
| Query performance | Same cost as running the underlying query | Fast - reading pre-computed, stored data |
| Best for | Simple abstraction, security, always-current logic | Expensive aggregations/joins queried repeatedly, data warehousing |

**Note:** PostgreSQL has its own `MATERIALIZED VIEW` with similar concepts but a simpler refresh model (only `REFRESH MATERIALIZED VIEW`, manually or scheduled externally - no built-in `ON COMMIT` option). MySQL has no native materialized view feature at all (commonly emulated via a real table plus a scheduled event or trigger). SQL Server's closest equivalent is an "indexed view."


# 21. Indexes

## 21.1 B-tree Indexes (Default)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_employees_dept ON employees(department_id);
````

Oracle's default index type - a balanced tree structure, ideal for high-selectivity columns (Part 21.9) and range queries. Conceptually identical to the default B-tree index in PostgreSQL, MySQL (InnoDB), and SQL Server.

## 21.2 Bitmap Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE BITMAP INDEX idx_employees_gender ON employees(gender);
````

Stores a bitmap (a string of bits) per distinct value, marking which rows contain it - extremely efficient for **low-cardinality** columns (few distinct values, like gender or status flags) in read-heavy, analytical workloads (data warehouses), but poorly suited to tables with frequent concurrent writes, since updating a bitmap index can lock a large range of rows. **Bitmap indexes are a distinctly Oracle Enterprise Edition feature** - PostgreSQL builds bitmap index *scans* dynamically at query time rather than storing a persistent bitmap index, and MySQL/SQL Server have no direct equivalent at all.

## 21.3 Unique Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE UNIQUE INDEX idx_employees_email ON employees(email);
````

## 21.4 Composite Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_orders_cust_date ON orders(customer_id, order_date);
````

As with any database, most effective when queries filter on the leading column(s), in order (same principle as PostgreSQL/MySQL/SQL Server).

## 21.5 Function-Based Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_employees_upper_email ON employees(UPPER(email));

SELECT * FROM employees WHERE UPPER(email) = 'ADA@EXAMPLE.COM';   -- can now use the index
````

A function-based index indexes the *result* of an expression, not the raw column - solving the classic problem where wrapping an indexed column in a function (Part 14.6 of your Database Programming handbook) would otherwise prevent index usage entirely.

## 21.6 Invisible Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
ALTER INDEX idx_employees_dept INVISIBLE;
````

An **invisible index** is maintained (updated on writes) but ignored by the optimizer by default - a uniquely Oracle feature allowing you to safely test the impact of *removing* an index (by making it invisible) without actually dropping it, with an instant rollback path (`ALTER INDEX ... VISIBLE`) if something goes wrong.

## 21.7 Reverse Key Indexes

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_orders_id_rev ON orders(order_id) REVERSE;
````

Reverses the bytes of the indexed value before storing it - spreading sequential values (like an ascending sequence-generated ID) across the index structure more evenly, reducing a specific type of contention ("right-growing index" hot-block contention) in high-concurrency insert workloads, particularly relevant in Oracle RAC clustered environments. No direct equivalent in PostgreSQL/MySQL/SQL Server.

## 21.8 Partitioned Indexes

Covered together with table partitioning in Part 40 - indexes on a partitioned table can be **local** (one index segment per partition, aligned with the table's partitioning) or **global** (a single index structure spanning all partitions).

## 21.9 Index Selectivity and Cardinality (Recap, Oracle-Specific Depth)

**Cardinality** is the number of distinct values in a column; **selectivity** is cardinality relative to total row count. High-selectivity columns (like `email`) benefit from B-tree indexes; low-selectivity columns (like `is_active`) are often poor B-tree candidates but can be excellent bitmap index candidates in read-heavy contexts (Part 21.2).

## 21.10 Index Scans vs Full Table Scans

| Access Path | When Chosen |
|---|---|
| Full table scan | No useful index exists, or the optimizer estimates most/all rows must be read anyway |
| Index unique scan | A unique/primary key index, expecting at most one matching row |
| Index range scan | A non-unique index, expecting a range or multiple matching rows |
| Index full scan | Reading the entire index (still often faster than a table scan if the index is much smaller than the table) |

Fully explored with `EXPLAIN PLAN` in Part 22.

## 21.11 Over-Indexing

Every additional index slows down `INSERT`/`UPDATE`/`DELETE` (each index must be maintained) and consumes storage - the same fundamental tradeoff you already know from PostgreSQL/MySQL/SQL Server applies identically in Oracle; Oracle's `INVISIBLE` index feature (Part 21.6) specifically makes it easier to safely evaluate whether a given index is still earning its keep.

## 21.12 Oracle Indexing Compared

| Feature | Oracle | PostgreSQL | MySQL | SQL Server |
|---|---|---|---|---|
| Default index type | B-tree | B-tree | B-tree (InnoDB) | B-tree |
| Persistent bitmap index | Yes (Enterprise Edition) | No (dynamic bitmap scans only) | No | No |
| Function-based index | Yes | Yes (expression indexes) | Yes (functional key parts, 8.0.13+) | Yes (computed column + index) |
| Invisible index | Yes | No | Yes (MySQL 8.0+, similar concept) | Disable, not true invisibility |
| Reverse key index | Yes | No native equivalent | No native equivalent | No native equivalent |


# 22. Oracle Optimizer

## 22.1 The Cost-Based Optimizer (CBO)

Oracle's optimizer chooses an execution plan by estimating the relative **cost** (a unitless measure combining estimated I/O, CPU, and memory usage) of each possible way to execute a query, and selecting the lowest-cost plan - conceptually similar to PostgreSQL's and SQL Server's own cost-based optimizers, but with Oracle-specific statistics, hints, and plan stability features.

## 22.2 Statistics

The optimizer's cost estimates are only as good as the statistics it has about the data - row counts, distinct value counts, data distribution histograms.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    DBMS_STATS.GATHER_TABLE_STATS('HR', 'EMPLOYEES');
END;
/
````

Stale statistics (e.g. after a large bulk load without a subsequent stats refresh) are one of the most common real-world causes of a suddenly poor execution plan.

## 22.3 Cardinality and Selectivity (Recap, Applied to the Optimizer)

The optimizer estimates how many rows each step of a plan will produce (**cardinality estimates**), directly driving which access paths and join methods it chooses - accurate statistics are what make these estimates trustworthy.

## 22.4 EXPLAIN PLAN

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
EXPLAIN PLAN FOR
SELECT * FROM employees WHERE department_id = 90;

SELECT * FROM TABLE(DBMS_XPLAN.DISPLAY);
````

```
--------------------------------------------------------------------------
| Id  | Operation                   | Name                | Rows | Cost |
--------------------------------------------------------------------------
|   0 | SELECT STATEMENT            |                      |    5 |    2 |
|   1 |  TABLE ACCESS BY INDEX ROWID| EMPLOYEES            |    5 |    2 |
|   2 |   INDEX RANGE SCAN          | IDX_EMPLOYEES_DEPT   |    5 |    1 |
--------------------------------------------------------------------------
```

`DBMS_XPLAN.DISPLAY` is the standard, readable way to view a plan generated by `EXPLAIN PLAN` - roughly analogous to PostgreSQL's `EXPLAIN`, MySQL's `EXPLAIN`, or SQL Server's graphical/text execution plan.

## 22.5 Access Paths

| Access Path | Meaning |
|---|---|
| Full table scan | Reads every block of the table |
| Index unique scan | At most one row expected, via a unique index |
| Index range scan | A range of rows via a non-unique (or partially matched unique) index |
| Index full scan | Reads the entire index structure in order |
| Index fast full scan | Reads the entire index, unordered, treating it like a smaller table scan |

## 22.6 Join Methods

| Method | Best Suited For |
|---|---|
| Nested Loops | Small outer row sets joined to an indexed inner table - efficient when one side is small |
| Hash Join | Large, unsorted row sets with no useful index - builds an in-memory hash table on the smaller set |
| Sort Merge Join | Both inputs are already sorted (or an index provides that order) on the join key |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT /*+ USE_NL(e d) */ e.first_name, d.department_name
FROM employees e JOIN departments d ON e.department_id = d.department_id;
````

`/*+ ... */` is an Oracle **hint** - a directive suggesting (not guaranteeing) a specific access path or join method to the optimizer, used sparingly to correct rare cases where the optimizer's estimate is wrong despite good statistics.

## 22.7 Tools for Viewing Execution Plans

| Tool | Use |
|---|---|
| `EXPLAIN PLAN` + `DBMS_XPLAN.DISPLAY` | Shows the *estimated* plan without executing the query |
| `DBMS_XPLAN.DISPLAY_CURSOR` | Shows the *actual* plan used by a query that has already run, including real row counts |
| SQL Developer's "Explain Plan" and "Autotrace" | Graphical plan viewing |
| `V$SQL_PLAN` | Data dictionary view exposing cached plans directly (Part 46) |


# 23. Oracle Query Performance

## 23.1 The General Tuning Workflow

Identify the slow query -> examine its execution plan (Part 22.4) -> check statistics freshness (Part 22.2) -> check for missing/unused indexes (Part 21) -> rewrite the SQL if needed -> re-measure.

## 23.2 SARGability Concepts

A predicate is **SARGable** ("Search ARGument-able") if the optimizer can use an index to satisfy it directly. Wrapping an indexed column in a function or performing implicit type conversion breaks SARGability - a concept identical across Oracle, PostgreSQL, MySQL, and SQL Server.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- NOT SARGable: function wraps the indexed column
WHERE TRUNC(hire_date) = DATE '2026-01-01'

-- SARGable: range comparison directly on the raw column
WHERE hire_date >= DATE '2026-01-01' AND hire_date < DATE '2026-01-02'
````

## 23.3 Bind Variables, Hard Parse, and Soft Parse

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Without bind variables: a NEW literal each time forces re-parsing
SELECT * FROM employees WHERE department_id = 90;
SELECT * FROM employees WHERE department_id = 100;

-- With bind variables: the SAME parsed plan is reused
SELECT * FROM employees WHERE department_id = :dept_id;
````

| Concept | Meaning |
|---|---|
| Hard parse | The optimizer fully analyzes and builds a new execution plan from scratch - expensive. |
| Soft parse | Oracle finds an already-parsed, matching statement in the shared pool (Part 1.5) and reuses its plan - cheap. |
| Bind variables | Placeholders (`:dept_id`) that let structurally identical statements with different literal values share the same parsed plan, avoiding repeated hard parses. |

This is precisely why parameterized queries (already familiar to you as a SQL-injection defense, from your Database Programming handbook Part 7) are *also* a significant Oracle performance best practice - every unique literal value in an unparameterized query forces Oracle to hard-parse it separately, potentially causing severe shared-pool contention ("bind variable starvation") under high load.

## 23.4 SQL Plan Management Concepts

Oracle's **SQL Plan Management (SPM)** lets the database "freeze" a known-good execution plan for a specific SQL statement, preventing the optimizer from switching to a new (potentially worse) plan after a statistics change or Oracle version upgrade, without a controlled evaluation step first - a level of plan stability control not directly matched by PostgreSQL/MySQL, and only partially by SQL Server's Query Store.

## 23.5 A Before-and-After Optimization Example

**Before (slow, full table scan, non-SARGable predicate):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM orders WHERE TO_CHAR(order_date, 'YYYY') = '2026';
````

**After (SARGable range predicate, index-friendly):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM orders
WHERE order_date >= DATE '2026-01-01' AND order_date < DATE '2027-01-01';
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_orders_date ON orders(order_date);
````

**Before (unnecessary columns, expensive correlated subquery per row):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT e.*, (SELECT department_name FROM departments d WHERE d.department_id = e.department_id) AS dept_name
FROM employees e;
````

**After (explicit JOIN, only needed columns, allows a hash/nested-loop join instead of row-by-row subquery execution):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT e.employee_id, e.first_name, d.department_name
FROM employees e
JOIN departments d ON e.department_id = d.department_id;
````

# 24. Transactions and ACID

## 24.1 COMMIT, ROLLBACK, SAVEPOINT (Recap)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
UPDATE accounts SET balance = balance - 100 WHERE account_id = 1;
UPDATE accounts SET balance = balance + 100 WHERE account_id = 2;
COMMIT;
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SAVEPOINT before_transfer;
UPDATE accounts SET balance = balance - 100 WHERE account_id = 1;
-- something goes wrong
ROLLBACK TO SAVEPOINT before_transfer;
````

ACID properties (Atomicity, Consistency, Isolation, Durability) are identical concepts to what you already know from PostgreSQL/MySQL/SQL Server (see also your Database Programming handbook, Part 15).

## 24.2 Oracle's Distinctive Transaction Behavior

| Behavior | Detail |
|---|---|
| No autocommit by default | Most Oracle client tools (SQL\*Plus, SQLcl) require an explicit `COMMIT` - unlike MySQL, which autocommits by default. DDL statements (`CREATE`, `ALTER`, `DROP`), however, **always** implicitly commit in Oracle, even mid-transaction. |
| One transaction per session at a time | Oracle does not support nested/independent transactions within a single session the way some databases' savepoint-heavy patterns might suggest - `SAVEPOINT` provides partial rollback within a single ongoing transaction, not a separate nested transaction. |
| Read consistency via undo | Oracle uses undo data (Part 42) to give every query a consistent, point-in-time view of the data, rather than row-versioning stored directly in the table (PostgreSQL's MVCC approach) or locking-based consistency (older SQL Server behavior without snapshot isolation enabled). |

## 24.3 DDL and Implicit Commits

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
UPDATE employees SET salary = salary * 1.1 WHERE department_id = 90;
CREATE TABLE temp_log (msg VARCHAR2(100));   -- implicitly commits the UPDATE above!
````

**Common mistake:** assuming a `CREATE TABLE`, `ALTER TABLE`, or similar DDL statement inside a larger transactional script is "just definitional" and won't affect pending DML - in Oracle, it silently commits everything before it. This is a genuine, practical trap when mixing DDL and DML in the same script or PL/SQL block, and differs from some other databases where DDL can participate in an explicit transaction (notably, PostgreSQL supports transactional DDL).

## 24.4 A Practical Transfer Example

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    UPDATE accounts SET balance = balance - 100 WHERE account_id = 1;
    UPDATE accounts SET balance = balance + 100 WHERE account_id = 2;
    COMMIT;
EXCEPTION
    WHEN OTHERS THEN
        ROLLBACK;
        RAISE;
END;
/
````

# 25. Oracle Concurrency and Consistency

## 25.1 Read Consistency

Every Oracle query sees a consistent snapshot of the data as of the moment the query (or, in serializable mode, the transaction) began - achieved using undo data (Part 42) to reconstruct "before" images of rows changed by other, concurrently running transactions. Readers never block writers, and writers never block readers in Oracle - a deliberate, foundational design decision.

## 25.2 Multiversion Concurrency, Oracle-Style

Oracle's approach is often grouped under the general term "MVCC" (multiversion concurrency control) alongside PostgreSQL, but the *mechanism* differs meaningfully:

| | Oracle | PostgreSQL |
|---|---|---|
| Old row versions stored | In the separate undo tablespace (Part 42) | Directly within the table's own storage (old tuple versions) |
| Cleanup | Undo segments are reused/expired based on retention settings | `VACUUM` must periodically reclaim dead tuple versions |
| Long-running query risk | Can encounter `ORA-01555: snapshot too old` if needed undo has been overwritten (Part 42.5) | Long-running transactions can cause table bloat until vacuumed |

## 25.3 Locks and Row-Level Locking

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM employees WHERE employee_id = 100 FOR UPDATE;
````

Oracle locks at the **row level** by default for DML - never escalating to a full table lock automatically the way some other databases historically have (SQL Server's older lock escalation behavior, for example). `FOR UPDATE` explicitly locks selected rows for the duration of the transaction, preventing other sessions from modifying (or also locking) them until commit/rollback.

## 25.4 Blocking and Deadlocks

**Blocking** occurs when one session holds a lock another session needs, and the second session simply waits. A **deadlock** occurs when two sessions each hold a lock the other needs - Oracle automatically detects this cycle and terminates one of the transactions with `ORA-00060: deadlock detected`, forcing that session to roll back and retry.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM v$lock WHERE block = 1;   -- inspecting current blocking locks
````

## 25.5 Isolation Levels in Oracle

| Isolation Level | Oracle Support |
|---|---|
| Read Committed | Oracle's default - every statement sees data committed as of the statement's own start time |
| Serializable | Available; the entire transaction sees a consistent snapshot as of its *first* query, and will raise `ORA-08177` if a serialization conflict is detected on write |
| Read Uncommitted | **Not supported** - Oracle has no "dirty read" isolation level at all, unlike some other databases that technically permit it |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SET TRANSACTION ISOLATION LEVEL SERIALIZABLE;
````

## 25.6 Oracle vs PostgreSQL vs MySQL/InnoDB vs SQL Server: Concurrency Model Comparison

| | Oracle | PostgreSQL | MySQL/InnoDB | SQL Server |
|---|---|---|---|---|
| Default isolation | Read Committed | Read Committed | Repeatable Read (MySQL's own default differs from the SQL standard's usual default) | Read Committed |
| Readers block writers? | Never | Never | Never (with MVCC) | By default, yes, unless snapshot isolation/RCSI is enabled |
| Old-version storage | Undo tablespace | In-table (old tuples) | Undo logs (rollback segments), similar in spirit to Oracle | Version store in tempdb (when snapshot isolation is enabled) |
| Dirty reads possible? | Never | Never | Only if explicitly set to Read Uncommitted | Only if explicitly set to Read Uncommitted, or via `NOLOCK` hints |


# 26. PL/SQL Fundamentals

## 26.1 What PL/SQL Is

PL/SQL (Procedural Language/SQL) is Oracle's proprietary procedural extension to SQL, embedding control flow, variables, and error handling directly around SQL statements - comparable in role to PostgreSQL's PL/pgSQL or SQL Server's T-SQL, but with its own distinct syntax and semantics.

## 26.2 The Anonymous Block Structure

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    -- variable declarations (optional)
    v_message VARCHAR2(50);
BEGIN
    -- executable statements (required)
    v_message := 'Hello, PL/SQL';
    DBMS_OUTPUT.PUT_LINE(v_message);
EXCEPTION
    -- exception handlers (optional)
    WHEN OTHERS THEN
        DBMS_OUTPUT.PUT_LINE('An error occurred');
END;
/
````

| Section | Purpose |
|---|---|
| `DECLARE` | Declares variables, constants, cursors, types - optional if the block needs none. |
| `BEGIN ... END` | The required executable body - always present. |
| `EXCEPTION` | Optional error-handling section (Part 30). |
| `/` | Tells the client tool (SQL\*Plus/SQLcl) to execute the block. |

This is called an **anonymous block** because it has no name and cannot be called from elsewhere - useful for quick scripts and testing, as opposed to named, reusable procedures/functions (Parts 31-32).

## 26.3 Variables

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_first_name VARCHAR2(50);
    v_salary     NUMBER(10,2) := 5000;
    v_is_active  BOOLEAN := TRUE;
BEGIN
    v_first_name := 'Ada';
    DBMS_OUTPUT.PUT_LINE(v_first_name || ' earns ' || v_salary);
END;
/
````

Variables use standard Oracle data types (Part 4), plus PL/SQL-only types like `BOOLEAN` (which has no direct SQL column equivalent - SQL tables cannot store a `BOOLEAN` column directly in Oracle, a notable difference from PostgreSQL).

## 26.4 Constants

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    c_tax_rate CONSTANT NUMBER(4,3) := 0.075;
BEGIN
    DBMS_OUTPUT.PUT_LINE('Tax rate: ' || c_tax_rate);
END;
/
````

## 26.5 Assignment and Expressions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
v_total := v_price * v_quantity;
v_full_name := v_first_name || ' ' || v_last_name;
````

`:=` is PL/SQL's assignment operator (distinct from `=`, which is strictly comparison) - the same convention used throughout Oracle's procedural code.

## 26.6 Scope and Nested Blocks

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_outer VARCHAR2(20) := 'outer value';
BEGIN
    DECLARE
        v_inner VARCHAR2(20) := 'inner value';
    BEGIN
        DBMS_OUTPUT.PUT_LINE(v_outer);   -- accessible: outer scope is visible to inner blocks
        DBMS_OUTPUT.PUT_LINE(v_inner);
    END;
    -- DBMS_OUTPUT.PUT_LINE(v_inner);   -- would FAIL here: v_inner is out of scope
END;
/
````

A nested block can access variables from any enclosing block, but variables declared *inside* a nested block cease to exist once that block ends - directly analogous to lexical scoping concepts from general programming (closures, enclosing scope) you already know.


# 27. PL/SQL Control Flow

## 27.1 IF / ELSIF / ELSE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_salary NUMBER := 8500;
BEGIN
    IF v_salary >= 10000 THEN
        DBMS_OUTPUT.PUT_LINE('High earner');
    ELSIF v_salary >= 5000 THEN
        DBMS_OUTPUT.PUT_LINE('Mid earner');
    ELSE
        DBMS_OUTPUT.PUT_LINE('Entry level');
    END IF;
END;
/
````

## 27.2 CASE (PL/SQL Statement Form)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_grade CHAR(1) := 'B';
BEGIN
    CASE v_grade
        WHEN 'A' THEN DBMS_OUTPUT.PUT_LINE('Excellent');
        WHEN 'B' THEN DBMS_OUTPUT.PUT_LINE('Good');
        ELSE DBMS_OUTPUT.PUT_LINE('Needs improvement');
    END CASE;
END;
/
````

## 27.3 Basic LOOP

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_counter NUMBER := 1;
BEGIN
    LOOP
        DBMS_OUTPUT.PUT_LINE('Iteration ' || v_counter);
        v_counter := v_counter + 1;
        EXIT WHEN v_counter > 5;
    END LOOP;
END;
/
````

## 27.4 WHILE LOOP

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_counter NUMBER := 1;
BEGIN
    WHILE v_counter <= 5 LOOP
        DBMS_OUTPUT.PUT_LINE('Count: ' || v_counter);
        v_counter := v_counter + 1;
    END LOOP;
END;
/
````

## 27.5 FOR LOOP

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    FOR i IN 1..5 LOOP
        DBMS_OUTPUT.PUT_LINE('i = ' || i);
    END LOOP;

    FOR i IN REVERSE 1..5 LOOP
        DBMS_OUTPUT.PUT_LINE('reversed i = ' || i);
    END LOOP;
END;
/
````

The loop variable (`i`) is implicitly declared, scoped only to the loop, and requires no `DECLARE` section entry.

## 27.6 Nested Loops, EXIT, and CONTINUE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    FOR i IN 1..3 LOOP
        FOR j IN 1..3 LOOP
            IF j = 2 THEN
                CONTINUE;   -- skip to the next iteration of the inner loop
            END IF;
            IF i = 3 AND j = 3 THEN
                EXIT;       -- exits only the INNER loop
            END IF;
            DBMS_OUTPUT.PUT_LINE('i=' || i || ', j=' || j);
        END LOOP;
    END LOOP;
END;
/
````

**Common mistake:** assuming `EXIT` inside a nested loop exits *all* loops - by default it only exits the innermost one. To exit an outer loop directly, label it:

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    <<outer_loop>>
    FOR i IN 1..3 LOOP
        FOR j IN 1..3 LOOP
            IF i = 2 AND j = 2 THEN
                EXIT outer_loop;   -- exits the labeled OUTER loop directly
            END IF;
            DBMS_OUTPUT.PUT_LINE('i=' || i || ', j=' || j);
        END LOOP;
    END LOOP;
END;
/
````

## 27.7 Realistic Business Example: Applying a Bonus by Tier

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    CURSOR emp_cursor IS SELECT employee_id, salary FROM employees;
BEGIN
    FOR emp_rec IN emp_cursor LOOP
        IF emp_rec.salary >= 15000 THEN
            UPDATE employees SET salary = salary * 1.03 WHERE employee_id = emp_rec.employee_id;
        ELSIF emp_rec.salary >= 8000 THEN
            UPDATE employees SET salary = salary * 1.05 WHERE employee_id = emp_rec.employee_id;
        ELSE
            UPDATE employees SET salary = salary * 1.08 WHERE employee_id = emp_rec.employee_id;
        END IF;
    END LOOP;
    COMMIT;
END;
/
````

## 28.2 %TYPE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_salary employees.salary%TYPE;
BEGIN
    SELECT salary INTO v_salary FROM employees WHERE employee_id = 100;
    DBMS_OUTPUT.PUT_LINE(v_salary);
END;
/
````

`%TYPE` declares a variable with the **same data type as a specific column** (or another variable) - if `employees.salary` is later changed from `NUMBER(8,2)` to `NUMBER(10,2)`, this variable's declaration automatically stays correct with zero code changes.

## 28.3 %ROWTYPE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_emp employees%ROWTYPE;
BEGIN
    SELECT * INTO v_emp FROM employees WHERE employee_id = 100;
    DBMS_OUTPUT.PUT_LINE(v_emp.first_name || ' ' || v_emp.last_name);
END;
/
````

`%ROWTYPE` declares a **record** variable with one field per column in the referenced table (or cursor) - avoiding having to declare a separate scalar variable for every single column.

## 28.4 Why %TYPE and %ROWTYPE Matter for Maintainability

| Without %TYPE/%ROWTYPE | With %TYPE/%ROWTYPE |
|---|---|
| A column's data type change requires manually updating every PL/SQL variable declared to match it | The variable's type updates automatically, since it's derived directly from the table's current definition |
| Adding/removing table columns requires updating every `%ROWTYPE`-equivalent hand-written record | `%ROWTYPE` automatically reflects the table's current column list |
| Type mismatches between PL/SQL variables and table columns are a real, recurring source of bugs | Type mismatches with the source table become structurally impossible |

This is the PL/SQL analog of writing code that references a schema *dynamically* rather than hardcoding assumptions about it - directly connecting to good maintainability practices you already value from general software engineering.

# 29. PL/SQL Cursors

## 29.1 Implicit Cursors

Every SQL statement executed inside PL/SQL (even a plain `UPDATE` or `SELECT INTO`) uses an **implicit cursor** behind the scenes, automatically managed by Oracle - you never declare or open it yourself.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    UPDATE employees SET salary = salary * 1.1 WHERE department_id = 90;
    DBMS_OUTPUT.PUT_LINE(SQL%ROWCOUNT || ' rows updated');
END;
/
````

`SQL%ROWCOUNT`, `SQL%FOUND`, and `SQL%NOTFOUND` are implicit cursor attributes reporting on the most recently executed implicit SQL statement.

## 29.2 Explicit Cursors: Declaration, Opening, Fetching, Closing

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    CURSOR emp_cursor IS
        SELECT employee_id, first_name, salary FROM employees WHERE department_id = 90;
    v_employee_id employees.employee_id%TYPE;
    v_first_name  employees.first_name%TYPE;
    v_salary      employees.salary%TYPE;
BEGIN
    OPEN emp_cursor;
    LOOP
        FETCH emp_cursor INTO v_employee_id, v_first_name, v_salary;
        EXIT WHEN emp_cursor%NOTFOUND;
        DBMS_OUTPUT.PUT_LINE(v_first_name || ': ' || v_salary);
    END LOOP;
    CLOSE emp_cursor;
END;
/
````

| Step | Purpose |
|---|---|
| `CURSOR ... IS` | Declares the query the cursor will execute. |
| `OPEN` | Executes the query and positions the cursor before the first row. |
| `FETCH ... INTO` | Retrieves the next row into variables. |
| `%NOTFOUND` | `TRUE` once `FETCH` has gone past the last row. |
| `CLOSE` | Releases the cursor's resources - always required for explicit cursors. |

## 29.3 Cursor FOR Loops

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    FOR emp_rec IN (SELECT employee_id, first_name, salary FROM employees WHERE department_id = 90) LOOP
        DBMS_OUTPUT.PUT_LINE(emp_rec.first_name || ': ' || emp_rec.salary);
    END LOOP;
END;
/
````

A cursor `FOR` loop implicitly opens the cursor, fetches every row, and closes the cursor automatically - the standard, preferred style for straightforward row-by-row iteration, since it eliminates the boilerplate of Part 29.2's manual `OPEN`/`FETCH`/`EXIT WHEN`/`CLOSE`.

## 29.4 Parameterized Cursors

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    CURSOR emp_cursor(p_dept_id NUMBER) IS
        SELECT employee_id, first_name FROM employees WHERE department_id = p_dept_id;
BEGIN
    FOR emp_rec IN emp_cursor(90) LOOP
        DBMS_OUTPUT.PUT_LINE(emp_rec.first_name);
    END LOOP;
    FOR emp_rec IN emp_cursor(60) LOOP   -- same cursor, different parameter
        DBMS_OUTPUT.PUT_LINE(emp_rec.first_name);
    END LOOP;
END;
/
````

## 29.5 When Cursors Are Appropriate, and Why Set-Based SQL Is Usually Preferable

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Row-by-row (cursor-based) - generally SLOWER, but sometimes necessary
FOR emp_rec IN (SELECT employee_id, salary FROM employees) LOOP
    UPDATE employees SET salary = salary * 1.1 WHERE employee_id = emp_rec.employee_id;
END LOOP;

-- Set-based - almost always FASTER for pure data transformation
UPDATE employees SET salary = salary * 1.1;
````

| Use a Cursor When | Prefer Set-Based SQL When |
|---|---|
| Each row requires complex, row-specific procedural logic (calling other procedures, external validation) that cannot be expressed as a single SQL statement | The logic is fundamentally a data transformation expressible directly in `UPDATE`/`INSERT`/`DELETE` |
| Building a report or export that genuinely needs row-by-row formatting/output | Aggregating, filtering, or joining data - SQL's engine is fundamentally optimized for set operations |

**Professional guidance:** the single most common PL/SQL performance mistake is reaching for a cursor loop out of habit (or from experience in a general-purpose programming language) when the same result could be achieved with one plain SQL statement. Always ask "could this be one `UPDATE`/`INSERT ... SELECT` instead?" before writing a cursor loop - and when a cursor loop genuinely is necessary at scale, prefer `BULK COLLECT`/`FORALL` (Part 36) over plain row-by-row fetching.


# 30. PL/SQL Exception Handling

## 30.1 The EXCEPTION Section

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    -- executable code
    NULL;
EXCEPTION
    WHEN exception_name THEN
        -- handling code
        NULL;
END;
/
````

Conceptually parallel to `try`/`except` in Python, or `try`/`catch` elsewhere - PL/SQL's block-structured error handling.

## 30.2 Predefined Exceptions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_salary NUMBER;
BEGIN
    SELECT salary INTO v_salary FROM employees WHERE employee_id = 999999;
EXCEPTION
    WHEN NO_DATA_FOUND THEN
        DBMS_OUTPUT.PUT_LINE('No employee with that ID exists');
    WHEN TOO_MANY_ROWS THEN
        DBMS_OUTPUT.PUT_LINE('Query unexpectedly matched more than one row');
    WHEN ZERO_DIVIDE THEN
        DBMS_OUTPUT.PUT_LINE('Division by zero');
END;
/
````

| Predefined Exception | Raised When |
|---|---|
| `NO_DATA_FOUND` | A `SELECT INTO` matches zero rows |
| `TOO_MANY_ROWS` | A `SELECT INTO` matches more than one row |
| `ZERO_DIVIDE` | Division by zero |
| `DUP_VAL_ON_INDEX` | A unique constraint/index is violated |
| `VALUE_ERROR` | A conversion, truncation, or numeric/value error |
| `INVALID_NUMBER` | A string fails to convert to a number |

## 30.3 User-Defined Exceptions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    e_insufficient_funds EXCEPTION;
    v_balance NUMBER := 50;
    v_withdrawal NUMBER := 100;
BEGIN
    IF v_withdrawal > v_balance THEN
        RAISE e_insufficient_funds;
    END IF;
EXCEPTION
    WHEN e_insufficient_funds THEN
        DBMS_OUTPUT.PUT_LINE('Insufficient funds for this withdrawal');
END;
/
````

## 30.4 RAISE and RAISE_APPLICATION_ERROR

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    RAISE_APPLICATION_ERROR(-20001, 'Withdrawal amount exceeds available balance');
END;
/
````

`RAISE_APPLICATION_ERROR` raises a custom error with a specific error number (must be between -20000 and -20999, reserved for application use) and a descriptive message - the standard, professional way to surface a business-rule violation back to the calling application with a clear, meaningful error rather than a generic Oracle error code.

## 30.5 Exception Propagation

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE withdraw_funds(p_account_id NUMBER, p_amount NUMBER) IS
    v_balance NUMBER;
BEGIN
    SELECT balance INTO v_balance FROM accounts WHERE account_id = p_account_id;
    IF p_amount > v_balance THEN
        RAISE_APPLICATION_ERROR(-20001, 'Insufficient funds');
    END IF;
    UPDATE accounts SET balance = balance - p_amount WHERE account_id = p_account_id;
END;
/
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    withdraw_funds(1, 100000);   -- if this raises, control passes to the CALLER's exception handler
EXCEPTION
    WHEN OTHERS THEN
        DBMS_OUTPUT.PUT_LINE('Transaction failed: ' || SQLERRM);
END;
/
````

An unhandled exception inside a procedure **propagates** up to its caller's own exception handler (or further, if that caller also doesn't handle it) - identical in spirit to exception propagation in general-purpose languages.

## 30.6 Logging Errors: A Production-Quality Pattern

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE process_payment(p_account_id NUMBER, p_amount NUMBER) IS
BEGIN
    UPDATE accounts SET balance = balance - p_amount WHERE account_id = p_account_id;
    COMMIT;
EXCEPTION
    WHEN OTHERS THEN
        ROLLBACK;
        INSERT INTO error_log (error_code, error_message, occurred_at, procedure_name)
        VALUES (SQLCODE, SQLERRM, SYSTIMESTAMP, 'PROCESS_PAYMENT');
        COMMIT;   -- commit the error log entry itself, in its own autonomous scope (see note below)
        RAISE;    -- re-raise so the caller still knows the operation failed
END;
/
````

**Professional note:** committing a log entry from inside a handler that also needs to `ROLLBACK` the failed business transaction requires care - in real production code, this logging insert is typically done inside an **autonomous transaction** (`PRAGMA AUTONOMOUS_TRANSACTION`) so the log entry survives independently of the outer rollback. `SQLCODE` and `SQLERRM` return the current error's numeric code and message text respectively, and are the standard way to capture diagnostic details for logging.

## 31.2 IN, OUT, and IN OUT Parameters

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE get_account_balance(
    p_account_id IN  NUMBER,
    p_balance    OUT NUMBER
) IS
BEGIN
    SELECT balance INTO p_balance FROM accounts WHERE account_id = p_account_id;
END;
/
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_balance NUMBER;
BEGIN
    get_account_balance(1, v_balance);
    DBMS_OUTPUT.PUT_LINE('Balance: ' || v_balance);
END;
/
````

| Mode | Direction | Analogy |
|---|---|---|
| `IN` (default) | Caller passes a value in; the procedure cannot modify the caller's variable | A regular function argument |
| `OUT` | The procedure sets a value that flows back to the caller; incoming value is ignored | A "return value" delivered through a parameter |
| `IN OUT` | The procedure receives a value AND can modify it, reflected back to the caller | Pass-by-reference |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE apply_discount(p_amount IN OUT NUMBER, p_pct IN NUMBER) IS
BEGIN
    p_amount := p_amount - (p_amount * p_pct / 100);
END;
/
````

## 31.3 Validation, Transactions, and Exception Handling Inside Procedures

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE transfer_funds(
    p_from_account IN NUMBER,
    p_to_account   IN NUMBER,
    p_amount       IN NUMBER
) IS
    v_from_balance NUMBER;
BEGIN
    IF p_amount <= 0 THEN
        RAISE_APPLICATION_ERROR(-20001, 'Transfer amount must be positive');
    END IF;

    SELECT balance INTO v_from_balance FROM accounts WHERE account_id = p_from_account FOR UPDATE;

    IF v_from_balance < p_amount THEN
        RAISE_APPLICATION_ERROR(-20002, 'Insufficient funds');
    END IF;

    UPDATE accounts SET balance = balance - p_amount WHERE account_id = p_from_account;
    UPDATE accounts SET balance = balance + p_amount WHERE account_id = p_to_account;

    COMMIT;
EXCEPTION
    WHEN OTHERS THEN
        ROLLBACK;
        RAISE;
END;
/
````

`FOR UPDATE` (Part 25.3) locks the source account's row for the duration of the transaction, preventing a race condition where two concurrent transfers could both read the same starting balance before either commits.

## 31.7 Business Example: Sales Processing

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE process_sale(
    p_customer_id IN NUMBER,
    p_product_id  IN NUMBER,
    p_quantity    IN NUMBER
) IS
    v_unit_price NUMBER;
    v_stock      NUMBER;
BEGIN
    SELECT price, stock_quantity INTO v_unit_price, v_stock
    FROM products WHERE product_id = p_product_id FOR UPDATE;

    IF v_stock < p_quantity THEN
        RAISE_APPLICATION_ERROR(-20003, 'Insufficient stock');
    END IF;

    INSERT INTO orders (order_id, customer_id, order_date)
    VALUES (orders_seq.NEXTVAL, p_customer_id, SYSDATE);

    UPDATE products SET stock_quantity = stock_quantity - p_quantity WHERE product_id = p_product_id;

    COMMIT;
END;
/
````

## 31.8 Business Example: Fraud Detection Flagging

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE flag_suspicious_transactions IS
BEGIN
    INSERT INTO fraud_flags (transaction_id, flagged_at, reason)
    SELECT transaction_id, SYSTIMESTAMP, 'Amount exceeds 3x account average'
    FROM transactions t
    WHERE amount > (
        SELECT AVG(amount) * 3 FROM transactions t2 WHERE t2.account_id = t.account_id
    );
    COMMIT;
END;
/
````

# 32. PL/SQL Functions

## 32.1 CREATE OR REPLACE FUNCTION

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE FUNCTION calculate_annual_salary(p_monthly_salary NUMBER)
RETURN NUMBER IS
BEGIN
    RETURN p_monthly_salary * 12;
END;
/

SELECT employee_id, calculate_annual_salary(salary) AS annual_salary FROM employees;
````

Unlike a procedure, a function **must** return exactly one value via `RETURN`, and - as shown here - can be called directly inside SQL statements, not only from PL/SQL blocks.

## 32.2 Parameters

Functions support the same `IN`/`OUT`/`IN OUT` parameter modes as procedures (Part 31.2), though functions called from SQL (as opposed to PL/SQL-only) are restricted to `IN` parameters only.

## 32.3 SQL-Callable Functions vs PL/SQL-Only Functions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE FUNCTION get_full_name(p_first VARCHAR2, p_last VARCHAR2)
RETURN VARCHAR2 IS
BEGIN
    RETURN p_first || ' ' || p_last;
END;
/

SELECT get_full_name(first_name, last_name) FROM employees;   -- SQL-callable
````

A function is **SQL-callable** if it uses only data types and features valid in SQL context (no `BOOLEAN` return type, no `OUT`/`IN OUT` parameters, no direct DML side effects when called from a query in most contexts). A function using `BOOLEAN`, `OUT` parameters, or performing DML can still be a perfectly valid PL/SQL function - it simply can't be called directly from a `SELECT` statement.

## 32.4 Deterministic Functions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE FUNCTION celsius_to_fahrenheit(p_celsius NUMBER)
RETURN NUMBER
DETERMINISTIC IS
BEGIN
    RETURN p_celsius * 9/5 + 32;
END;
/
````

`DETERMINISTIC` tells Oracle the function always returns the same output for the same input, with no dependency on external state - enabling the optimizer to cache results and use the function in function-based indexes (Part 21.5) or materialized views with query rewrite. Declaring a genuinely non-deterministic function as `DETERMINISTIC` is a real, subtle bug source, since Oracle may incorrectly reuse a stale cached result.

## 32.5 Comparing Oracle Functions With PostgreSQL, MySQL, and SQL Server

| | Oracle | PostgreSQL | MySQL | SQL Server |
|---|---|---|---|---|
| Definition keyword | `CREATE OR REPLACE FUNCTION ... RETURN type IS` | `CREATE OR REPLACE FUNCTION ... RETURNS type AS $$ ... $$` | `CREATE FUNCTION ... RETURNS type` | `CREATE FUNCTION ... RETURNS type` |
| Called from SQL directly | Yes (if SQL-callable) | Yes | Yes | Yes (scalar/table-valued) |
| Deterministic hint | `DETERMINISTIC` | `IMMUTABLE` | `DETERMINISTIC` | `SCHEMABINDING` + deterministic rules |
| Can perform DML | Yes (with restrictions when called from SQL) | Yes | Limited | Limited (scalar functions generally cannot) |
| Grouped into packages | Yes, natively (Part 33) | No native packages (schemas serve a similar organizational role) | No native packages | No native packages |


# 33. Packages

## 33.1 What a Package Is

A **package** groups related procedures, functions, variables, and types under one named unit, split into a **specification** (the public interface) and a **body** (the implementation) - a distinctly Oracle feature with no direct equivalent in PostgreSQL, MySQL, or SQL Server, all of which organize procedures/functions as flat, individually-named objects.

## 33.2 Package Specification

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PACKAGE payroll_pkg IS
    PROCEDURE run_payroll(p_pay_period_end IN DATE);
    FUNCTION calculate_bonus(p_employee_id IN NUMBER) RETURN NUMBER;
END payroll_pkg;
/
````

The specification declares *what* is publicly available - other code can only call procedures/functions listed here.

## 33.3 Package Body

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PACKAGE BODY payroll_pkg IS

    -- private helper, NOT declared in the specification, so it's invisible outside this package
    FUNCTION get_tenure_years(p_hire_date DATE) RETURN NUMBER IS
    BEGIN
        RETURN MONTHS_BETWEEN(SYSDATE, p_hire_date) / 12;
    END;

    FUNCTION calculate_bonus(p_employee_id IN NUMBER) RETURN NUMBER IS
        v_salary NUMBER;
        v_hire_date DATE;
    BEGIN
        SELECT salary, hire_date INTO v_salary, v_hire_date
        FROM employees WHERE employee_id = p_employee_id;

        IF get_tenure_years(v_hire_date) >= 5 THEN
            RETURN v_salary * 0.10;
        ELSE
            RETURN v_salary * 0.05;
        END IF;
    END;

    PROCEDURE run_payroll(p_pay_period_end IN DATE) IS
    BEGIN
        FOR emp_rec IN (SELECT employee_id, salary FROM employees) LOOP
            INSERT INTO payroll_runs (employee_id, pay_period_end, gross_pay, bonus)
            VALUES (emp_rec.employee_id, p_pay_period_end, emp_rec.salary,
                    calculate_bonus(emp_rec.employee_id));
        END LOOP;
        COMMIT;
    END;

END payroll_pkg;
/
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    payroll_pkg.run_payroll(DATE '2026-01-31');
END;
/
````

## 33.4 Public vs Private Procedures

| | Public | Private |
|---|---|---|
| Declared in specification? | Yes | No |
| Callable from outside the package? | Yes, as `package_name.procedure_name` | No - only from other code within the same package body |
| Purpose | The package's intended external API | Internal implementation detail, hidden from consumers |

`get_tenure_years` above is private - a clean, direct application of encapsulation, the same principle behind leading-underscore "internal" functions in Python modules you already know from module design best practices.

## 33.5 Package Variables and Package State

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PACKAGE session_cache_pkg IS
    PROCEDURE set_current_user(p_user_id NUMBER);
    FUNCTION get_current_user RETURN NUMBER;
END session_cache_pkg;
/

CREATE OR REPLACE PACKAGE BODY session_cache_pkg IS
    g_current_user_id NUMBER;   -- package-level (global to the package) variable

    PROCEDURE set_current_user(p_user_id NUMBER) IS
    BEGIN
        g_current_user_id := p_user_id;
    END;

    FUNCTION get_current_user RETURN NUMBER IS
    BEGIN
        RETURN g_current_user_id;
    END;
END session_cache_pkg;
/
````

A package-level variable (`g_current_user_id`) persists for the **duration of the database session**, shared across every call to that package within the same session - a powerful but easy-to-misuse form of state, since it is invisible from the package's calling code and can introduce subtle bugs if a session unexpectedly reuses stale state.

## 33.6 Package Initialization

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PACKAGE BODY config_pkg IS
    g_tax_rate NUMBER;
BEGIN
    -- this initialization block runs ONCE, the first time the package is referenced in a session
    SELECT rate INTO g_tax_rate FROM tax_config WHERE is_current = 'Y';
END config_pkg;
/
````

An optional `BEGIN...END` block at the very end of a package body runs exactly once per session, the first time any part of the package is accessed - useful for one-time setup like loading configuration into package variables.

## 33.7 Why Packages Matter in Enterprise Oracle Applications

| Benefit | Explanation |
|---|---|
| Encapsulation | Private helper logic (Part 33.4) stays hidden, exposing only a clean public API. |
| Organization | Related business logic (e.g. everything payroll-related) lives together under one named unit, rather than scattered flat procedures. |
| Performance | The entire package body is loaded into the shared pool (Part 1.5) as a unit on first use, reducing repeated parsing overhead for frequently-called related procedures. |
| State management | Package variables provide session-scoped state without needing a separate table or external cache for simple cases. |
| Overloading | Multiple procedures/functions with the same name but different parameter signatures can coexist within one package - not possible with standalone procedures. |


# 34. Triggers

## 34.1 BEFORE and AFTER Triggers

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE TRIGGER trg_employees_before_insert
BEFORE INSERT ON employees
FOR EACH ROW
BEGIN
    IF :NEW.hire_date IS NULL THEN
        :NEW.hire_date := SYSDATE;
    END IF;
END;
/
````

`BEFORE` triggers run before the triggering DML statement takes effect - commonly used for validation or defaulting values, since `:NEW` column values can still be modified. `AFTER` triggers run after the change has been applied - commonly used for auditing or cascading side effects, since the change is already final.

## 34.2 Row-Level vs Statement-Level Triggers

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Row-level: runs once PER ROW affected
CREATE OR REPLACE TRIGGER trg_audit_row
AFTER UPDATE ON employees
FOR EACH ROW
BEGIN
    INSERT INTO employee_audit (employee_id, old_salary, new_salary, changed_at)
    VALUES (:OLD.employee_id, :OLD.salary, :NEW.salary, SYSTIMESTAMP);
END;
/

-- Statement-level: runs ONCE per triggering statement, regardless of row count
CREATE OR REPLACE TRIGGER trg_audit_statement
AFTER UPDATE ON employees
BEGIN
    INSERT INTO statement_log (table_name, action, occurred_at)
    VALUES ('EMPLOYEES', 'UPDATE', SYSTIMESTAMP);
END;
/
````

`FOR EACH ROW` is what makes a trigger row-level; without it, a trigger is statement-level and cannot reference `:NEW`/`:OLD` at all.

## 34.3 :NEW and :OLD

| Reference | Available In | Meaning |
|---|---|---|
| `:NEW` | `INSERT`, `UPDATE` triggers | The row's value *after* the change |
| `:OLD` | `UPDATE`, `DELETE` triggers | The row's value *before* the change |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE TRIGGER trg_prevent_salary_decrease
BEFORE UPDATE OF salary ON employees
FOR EACH ROW
BEGIN
    IF :NEW.salary < :OLD.salary THEN
        RAISE_APPLICATION_ERROR(-20010, 'Salary cannot be decreased');
    END IF;
END;
/
````

## 34.4 Compound Trigger Concepts

A **compound trigger** combines multiple timing points (`BEFORE STATEMENT`, `BEFORE EACH ROW`, `AFTER EACH ROW`, `AFTER STATEMENT`) into a single trigger object, sharing state across all of them - solving the "mutating table" error (`ORA-04091`, Part 60) that arises when a row-level trigger tries to query the very table it's currently modifying.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE TRIGGER trg_department_total_compound
FOR UPDATE OF salary ON employees
COMPOUND TRIGGER

    TYPE t_dept_list IS TABLE OF NUMBER INDEX BY PLS_INTEGER;
    g_affected_depts t_dept_list;

    AFTER EACH ROW IS
    BEGIN
        g_affected_depts(g_affected_depts.COUNT + 1) := :NEW.department_id;
    END AFTER EACH ROW;

    AFTER STATEMENT IS
    BEGIN
        FOR i IN 1..g_affected_depts.COUNT LOOP
            UPDATE department_totals
            SET total_salary = (SELECT SUM(salary) FROM employees WHERE department_id = g_affected_depts(i))
            WHERE department_id = g_affected_depts(i);
        END LOOP;
    END AFTER STATEMENT;

END trg_department_total_compound;
/
````

## 34.5 Use Cases: Audit Trails, Validation, Automatic Updates, Security

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Audit trail (recap of Part 34.2)
-- Validation (recap of Part 34.3)

-- Automatic timestamp update
CREATE OR REPLACE TRIGGER trg_update_modified_at
BEFORE UPDATE ON customers
FOR EACH ROW
BEGIN
    :NEW.modified_at := SYSTIMESTAMP;
END;
/

-- Security: preventing changes outside business hours
CREATE OR REPLACE TRIGGER trg_restrict_hours
BEFORE INSERT OR UPDATE OR DELETE ON payroll_runs
BEGIN
    IF TO_CHAR(SYSDATE, 'HH24') NOT BETWEEN '08' AND '18' THEN
        RAISE_APPLICATION_ERROR(-20020, 'Payroll changes are restricted to business hours');
    END IF;
END;
/
````

## 34.6 Risks of Excessive Trigger Usage

| Risk | Explanation |
|---|---|
| Hidden behavior | Logic buried in triggers is invisible to anyone reading the calling application code, making systems harder to understand and debug. |
| Cascading trigger chains | A trigger firing another trigger (on a different table) can create hard-to-trace execution chains and even infinite loops if not carefully designed. |
| Performance overhead | Every row-level trigger adds per-row overhead to DML, which compounds under `BULK COLLECT`/`FORALL` (Part 36) or large batch operations. |
| Mutating table errors | Row-level triggers querying their own table mid-statement fail with `ORA-04091` unless restructured (e.g. via compound triggers, Part 34.4). |
| Testing complexity | Triggers fire implicitly, making them easy to forget about when writing unit tests for the "obvious" application code path. |

**Professional guidance:** reserve triggers for genuinely cross-cutting concerns (auditing, enforcing invariants that must hold regardless of *how* data is changed) rather than routine business logic, which is generally clearer and more testable living explicitly in application code or stored procedures.


# 35. Dynamic SQL

## 35.1 What Native Dynamic SQL Is

**Dynamic SQL** builds and executes a SQL/PL-SQL statement whose exact text is only known at runtime - necessary when table names, column names, or the overall query structure cannot be hardcoded in advance.

## 35.2 EXECUTE IMMEDIATE

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_table_name VARCHAR2(30) := 'EMPLOYEES';
    v_count NUMBER;
BEGIN
    EXECUTE IMMEDIATE 'SELECT COUNT(*) FROM ' || v_table_name INTO v_count;
    DBMS_OUTPUT.PUT_LINE(v_count);
END;
/
````

## 35.3 Bind Variables in Dynamic SQL

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_dept_id NUMBER := 90;
    v_count NUMBER;
BEGIN
    EXECUTE IMMEDIATE
        'SELECT COUNT(*) FROM employees WHERE department_id = :dept'
        INTO v_count
        USING v_dept_id;
    DBMS_OUTPUT.PUT_LINE(v_count);
END;
/
````

Just as in ordinary SQL (Part 23.3), using `USING` bind variables inside dynamic SQL allows the same statement *shape* to be reused across different parameter values, avoiding excessive hard parsing.

## 35.4 Dynamic DDL

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    EXECUTE IMMEDIATE 'CREATE TABLE temp_report AS SELECT * FROM employees WHERE 1=0';
END;
/
````

DDL statements (`CREATE`, `ALTER`, `DROP`) can only be executed dynamically in PL/SQL - they cannot appear as static PL/SQL statements at all.

## 35.5 Dynamic DML

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    v_sql VARCHAR2(200);
BEGIN
    v_sql := 'UPDATE employees SET salary = salary * :pct WHERE department_id = :dept';
    EXECUTE IMMEDIATE v_sql USING 1.05, 90;
    COMMIT;
END;
/
````

## 35.6 SQL Injection Risks and Safe Dynamic SQL

**Unsafe - concatenating raw, untrusted input directly into dynamic SQL:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
EXECUTE IMMEDIATE 'SELECT * FROM employees WHERE last_name = ''' || p_user_input || '''';
````

Exactly the same fundamental vulnerability you already know from your Database Programming handbook (Part 7 there) - a malicious `p_user_input` value like `x' OR '1'='1` alters the query's actual structure.

**Safe - using bind variables even within dynamic SQL:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
EXECUTE IMMEDIATE 'SELECT * FROM employees WHERE last_name = :ln' USING p_user_input;
````

**Especially dangerous case - dynamically building identifiers (table/column names), which cannot be bound as parameters at all:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Table/column names CANNOT use bind variables - they must be concatenated
-- Validate against a strict allow-list before ever concatenating an identifier
IF v_table_name NOT IN ('EMPLOYEES', 'DEPARTMENTS', 'CUSTOMERS') THEN
    RAISE_APPLICATION_ERROR(-20030, 'Invalid table name');
END IF;
EXECUTE IMMEDIATE 'SELECT COUNT(*) FROM ' || v_table_name INTO v_count;
````

**Rule of thumb:** bind variables handle *values* safely; identifiers (table/column names) can never be bound and must instead be validated against a strict, known allow-list before being concatenated into dynamic SQL.


# 36. Bulk Processing

## 36.1 The Problem: Row-by-Row Context Switching

Every time a PL/SQL loop executes an individual SQL statement (an `INSERT`, `UPDATE`, or `FETCH`), Oracle must **context switch** between the PL/SQL engine and the SQL engine - a real, measurable cost that becomes significant across thousands or millions of iterations.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- SLOW: one context switch per row
BEGIN
    FOR i IN 1..100000 LOOP
        INSERT INTO staging_numbers VALUES (i);
    END LOOP;
    COMMIT;
END;
/
````

## 36.2 BULK COLLECT

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    TYPE t_salary_list IS TABLE OF employees.salary%TYPE;
    v_salaries t_salary_list;
BEGIN
    SELECT salary BULK COLLECT INTO v_salaries FROM employees;
    DBMS_OUTPUT.PUT_LINE('Fetched ' || v_salaries.COUNT || ' salaries in one round trip');
END;
/
````

`BULK COLLECT` fetches an entire result set into a collection (Part 37) in a single context switch, instead of one switch per row.

## 36.3 FORALL

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    TYPE t_id_list IS TABLE OF employees.employee_id%TYPE;
    v_ids t_id_list;
BEGIN
    SELECT employee_id BULK COLLECT INTO v_ids FROM employees WHERE department_id = 90;

    FORALL i IN 1..v_ids.COUNT
        UPDATE employees SET salary = salary * 1.1 WHERE employee_id = v_ids(i);

    COMMIT;
END;
/
````

`FORALL` sends an entire batch of `INSERT`/`UPDATE`/`DELETE` statements to the SQL engine in a single context switch, rather than executing them one at a time in a normal loop - functionally similar to what `executemany()` does in Python's DB-API (already familiar to you from your Database Programming handbook, Part 3.8), applied specifically to PL/SQL.

## 36.4 LIMIT (Bounded BULK COLLECT)

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    CURSOR emp_cursor IS SELECT employee_id FROM employees;
    TYPE t_id_list IS TABLE OF employees.employee_id%TYPE;
    v_ids t_id_list;
BEGIN
    OPEN emp_cursor;
    LOOP
        FETCH emp_cursor BULK COLLECT INTO v_ids LIMIT 1000;   -- process 1000 rows at a time
        FORALL i IN 1..v_ids.COUNT
            UPDATE employees SET salary = salary * 1.05 WHERE employee_id = v_ids(i);
        EXIT WHEN emp_cursor%NOTFOUND;
    END LOOP;
    CLOSE emp_cursor;
    COMMIT;
END;
/
````

`LIMIT` bounds how many rows `BULK COLLECT` fetches per iteration, avoiding excessive PGA memory consumption (Part 1.6) on very large result sets - the PL/SQL equivalent of `chunksize` in `pandas.read_sql_query()`, already familiar from your Database Programming handbook (Part 11.11 there).

## 36.5 A Before-and-After Performance Example

**Before - row-by-row (slow):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
BEGIN
    FOR emp_rec IN (SELECT employee_id FROM employees) LOOP
        UPDATE employees SET bonus = 500 WHERE employee_id = emp_rec.employee_id;
    END LOOP;
    COMMIT;
END;
/
````

**After - BULK COLLECT + FORALL (fast):**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    TYPE t_id_list IS TABLE OF employees.employee_id%TYPE;
    v_ids t_id_list;
BEGIN
    SELECT employee_id BULK COLLECT INTO v_ids FROM employees;
    FORALL i IN 1..v_ids.COUNT
        UPDATE employees SET bonus = 500 WHERE employee_id = v_ids(i);
    COMMIT;
END;
/
````

**Best - avoid PL/SQL entirely for a pure set-based operation:**

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
UPDATE employees SET bonus = 500;
````

For large tables, `BULK COLLECT`/`FORALL` can be an order of magnitude faster than row-by-row processing; a single plain SQL statement (when the logic genuinely permits it, per Part 29.5) is typically faster still.


# 37. Collections

## 37.1 Associative Arrays

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    TYPE t_salary_map IS TABLE OF NUMBER INDEX BY VARCHAR2(50);
    v_salaries t_salary_map;
BEGIN
    v_salaries('Ada') := 9500;
    v_salaries('Alan') := 8700;

    DBMS_OUTPUT.PUT_LINE(v_salaries('Ada'));

    -- iteration
    DECLARE
        v_key VARCHAR2(50) := v_salaries.FIRST;
    BEGIN
        WHILE v_key IS NOT NULL LOOP
            DBMS_OUTPUT.PUT_LINE(v_key || ': ' || v_salaries(v_key));
            v_key := v_salaries.NEXT(v_key);
        END LOOP;
    END;
END;
/
````

An **associative array** (also called an "index-by table") is a key-value structure, conceptually similar to a Python dictionary - indexed by either a string or an integer, sparse (indexes need not be contiguous), and existing only in PL/SQL memory (not storable directly in a database column).

## 37.2 Nested Tables

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TYPE t_number_list AS TABLE OF NUMBER;
/

DECLARE
    v_numbers t_number_list := t_number_list(10, 20, 30);
BEGIN
    v_numbers.EXTEND;
    v_numbers(4) := 40;
    FOR i IN 1..v_numbers.COUNT LOOP
        DBMS_OUTPUT.PUT_LINE(v_numbers(i));
    END LOOP;
END;
/
````

Unlike associative arrays, a **nested table** type can be defined at the schema level (`CREATE TYPE`) and used as a column type in an actual table - a genuinely Oracle-distinctive capability for storing a variable-length collection directly within a row.

## 37.3 VARRAYs

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TYPE t_top_scores AS VARRAY(5) OF NUMBER;
/

DECLARE
    v_scores t_top_scores := t_top_scores(95, 88, 92);
BEGIN
    v_scores.EXTEND;
    v_scores(4) := 79;
    DBMS_OUTPUT.PUT_LINE(v_scores.COUNT);
END;
/
````

A **VARRAY** (variable-size array) has a fixed maximum size declared upfront (`VARRAY(5)` here), and - unlike a nested table - preserves element order and dense indexing, making it suited to genuinely array-like, bounded collections (e.g. "the top 5 scores").

## 37.4 Declaration, Population, Iteration: Summary Comparison

| | Associative Array | Nested Table | VARRAY |
|---|---|---|---|
| Storable as a table column? | No | Yes | Yes |
| Maximum size | Unbounded | Unbounded | Fixed, declared upfront |
| Indexing | Sparse (string or integer keys) | Dense, but elements can be individually deleted (sparse) | Dense, fixed order |
| Typical use | PL/SQL-only lookup tables/caches | Storing a variable-length collection in a column, or use with `BULK COLLECT` | Small, fixed-bound ordered lists |

## 37.5 Collection Methods

| Method | Purpose |
|---|---|
| `.COUNT` | Number of elements currently in the collection |
| `.FIRST` / `.LAST` | Lowest / highest index in use |
| `.NEXT(n)` / `.PRIOR(n)` | The next / previous index after/before `n` |
| `.EXISTS(n)` | Whether an element exists at index `n` |
| `.EXTEND` | Adds a new element (nested tables/VARRAYs) |
| `.DELETE` | Removes an element (or all elements) |
| `.TRIM` | Removes elements from the end (VARRAYs/nested tables) |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
IF v_numbers.EXISTS(4) THEN
    DBMS_OUTPUT.PUT_LINE('Element 4 exists: ' || v_numbers(4));
END IF;
````

## 37.6 Searching a Collection

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
DECLARE
    TYPE t_id_list IS TABLE OF NUMBER;
    v_ids t_id_list := t_id_list(10, 25, 40);
    v_found BOOLEAN := FALSE;
BEGIN
    FOR i IN 1..v_ids.COUNT LOOP
        IF v_ids(i) = 25 THEN
            v_found := TRUE;
            EXIT;
        END IF;
    END LOOP;
    DBMS_OUTPUT.PUT_LINE(CASE WHEN v_found THEN 'Found' ELSE 'Not found' END);
END;
/
````

PL/SQL collections have no built-in "contains" search method - searching requires an explicit loop (or, for large collections, using `MEMBER OF` against a SQL-comparable nested table type within a SQL context).


# 38. JSON in Oracle

## 38.1 JSON Storage

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE api_events (
    id NUMBER GENERATED ALWAYS AS IDENTITY,
    payload VARCHAR2(4000) CHECK (payload IS JSON),   -- pre-21c style, using a CHECK constraint
    created_at TIMESTAMP DEFAULT SYSTIMESTAMP
);
````

Older Oracle versions store JSON in a `VARCHAR2`/`CLOB` column with an `IS JSON` check constraint enforcing validity. Oracle 21c+ introduces a native `JSON` data type with more efficient binary storage. **Version-dependent note:** always confirm your specific Oracle version's JSON capabilities before relying on native-type-specific features.

## 38.2 JSON Validation

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
INSERT INTO api_events (payload) VALUES ('{"user_id": 1, "action": "login"}');
INSERT INTO api_events (payload) VALUES ('not valid json');   -- fails the IS JSON check constraint
````

## 38.3 JSON Extraction

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT id, JSON_VALUE(payload, '$.user_id') AS user_id
FROM api_events;
````

`JSON_VALUE` extracts a single scalar value from a JSON document using a JSON path expression (`$.user_id`).

## 38.4 JSON Querying

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT id, payload
FROM api_events
WHERE JSON_VALUE(payload, '$.action') = 'login';

SELECT jt.*
FROM api_events,
     JSON_TABLE(payload, '$' COLUMNS (
         user_id NUMBER PATH '$.user_id',
         action  VARCHAR2(50) PATH '$.action'
     )) jt;
````

# 40. Oracle Partitioning

## 40.1 Why Partitioning Helps Very Large Tables

**Partitioning** splits one logical table into multiple physical segments, based on a chosen key - allowing queries, maintenance operations, and index scans to work against only the relevant subset of data ("partition pruning"), dramatically improving performance and manageability for very large tables (commonly billions of rows in enterprise data warehouses, Part 51).

## 40.2 Range Partitioning

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE sales (
    sale_id   NUMBER,
    sale_date DATE,
    amount    NUMBER
)
PARTITION BY RANGE (sale_date) (
    PARTITION p_2024 VALUES LESS THAN (DATE '2025-01-01'),
    PARTITION p_2025 VALUES LESS THAN (DATE '2026-01-01'),
    PARTITION p_2026 VALUES LESS THAN (DATE '2027-01-01')
);
````

The most common partitioning strategy for time-series/transactional data - a query filtering on `sale_date` can skip entire partitions outside the relevant range entirely.

## 40.3 List Partitioning

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE customers (
    customer_id NUMBER,
    region      VARCHAR2(20)
)
PARTITION BY LIST (region) (
    PARTITION p_north VALUES ('North', 'Northeast'),
    PARTITION p_south VALUES ('South', 'Southeast'),
    PARTITION p_other VALUES (DEFAULT)
);
````

Partitions by explicit, discrete values - suited to categorical columns with a known, manageable set of distinct values.

## 40.4 Hash Partitioning

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE transactions (
    transaction_id NUMBER,
    account_id     NUMBER
)
PARTITION BY HASH (account_id) PARTITIONS 8;
````

Distributes rows evenly across a fixed number of partitions using an internal hash function - useful when there's no natural range or list to partition by, but even data distribution (for parallelism and I/O balancing) is still valuable.

## 40.5 Composite Partitioning

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE sales (
    sale_id   NUMBER,
    sale_date DATE,
    region    VARCHAR2(20),
    amount    NUMBER
)
PARTITION BY RANGE (sale_date)
SUBPARTITION BY LIST (region) (
    PARTITION p_2026 VALUES LESS THAN (DATE '2027-01-01') (
        SUBPARTITION p_2026_north VALUES ('North'),
        SUBPARTITION p_2026_south VALUES ('South')
    )
);
````

Combines two strategies - here, partitioned by year and further subpartitioned by region - for very large, multi-dimensional access patterns.

## 40.6 Partition Pruning

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
SELECT * FROM sales WHERE sale_date BETWEEN DATE '2026-01-01' AND DATE '2026-03-31';
````

The optimizer (Part 22) automatically recognizes that only the `p_2026` partition needs scanning to satisfy this filter - visible in the execution plan as `PARTITION RANGE SINGLE` (or a limited range) rather than scanning every partition.

## 40.7 Local vs Global Indexes on Partitioned Tables

| | Local Index | Global Index |
|---|---|---|
| Structure | One index segment per table partition, automatically aligned | A single index structure spanning the whole table |
| Maintenance | Simple - dropping/adding a partition only affects its own local index segment | More complex - partition maintenance operations can invalidate the entire global index, requiring a rebuild |
| Best for | Most common Oracle partitioning use cases, especially with matching partition keys | Queries needing efficient lookups on a column *not* aligned with the partitioning key |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE INDEX idx_sales_date_local ON sales(sale_date) LOCAL;
CREATE INDEX idx_sales_id_global ON sales(sale_id) GLOBAL;
````

**Note:** PostgreSQL's declarative partitioning (since PG 10) and SQL Server's partitioned tables/indexes offer conceptually similar capabilities; MySQL also supports partitioning (range, list, hash) though historically with more limitations than Oracle's mature implementation, particularly around composite partitioning and global indexes.


# 41. Oracle Tablespaces

## 41.1 The Storage Hierarchy

```
Database
  └── Tablespace (logical container, e.g. USERS, DATA_TS)
       └── Datafile (physical OS file(s) backing the tablespace)
            └── Segment (the storage for one specific object, e.g. a table or index)
                 └── Extent (a contiguous chunk of space allocated to a segment)
                      └── Block (the smallest unit of I/O, typically 8KB by default)
```

## 41.2 Tablespaces and Datafiles

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLESPACE sales_data
DATAFILE '/u01/oradata/orcl/sales_data01.dbf' SIZE 500M
AUTOEXTEND ON NEXT 100M MAXSIZE UNLIMITED;
````

A tablespace is where tables, indexes, and other segments are ultimately stored - assigning a table to a specific tablespace is how DBAs control physical storage placement, backup grouping, and I/O distribution across disks.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE sales (sale_id NUMBER, amount NUMBER) TABLESPACE sales_data;
````

## 41.3 Locally Managed Tablespaces

Modern Oracle tablespaces are **locally managed** (space usage tracked via bitmaps within the tablespace's own datafiles), as opposed to the older, now-deprecated "dictionary managed" approach (space tracked in central data dictionary tables) - locally managed tablespaces are simpler and more performant, and are the default/only recommended choice in current Oracle versions.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLESPACE sales_data
DATAFILE '/u01/oradata/orcl/sales_data01.dbf' SIZE 500M
EXTENT MANAGEMENT LOCAL
UNIFORM SIZE 1M;
````

## 41.4 Extents and Segments

A **segment** is the storage allocated to one specific database object (a table, an index, an undo segment). A segment grows by acquiring additional **extents** - contiguous groups of blocks - as more space is needed.

## 41.5 Blocks

The smallest unit Oracle reads/writes at once, typically 8KB by default (configurable per tablespace) - every row ultimately lives within some block, and Oracle's buffer cache (Part 1.5) caches data at the block level.

## 41.6 Comparison Note

PostgreSQL has a broadly analogous "tablespace" concept (mapping to a filesystem directory), though its storage model is otherwise significantly simpler (no explicit extent/segment tuning exposed to typical users). MySQL/InnoDB has tablespaces too (a single shared tablespace by default, or "file per table"). SQL Server uses "filegroups" as its closest analogous concept. Oracle's storage hierarchy is generally the most explicit and tunable of the four, reflecting its enterprise/DBA-oriented heritage.


# 42. Oracle Undo

## 42.1 What Undo Data Is

**Undo data** stores the "before" image of every row changed by a transaction, held in a dedicated **undo tablespace** - the mechanism behind `ROLLBACK` (Part 24) and Oracle's read consistency model (Part 25.1).

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE UNDO TABLESPACE undotbs2
DATAFILE '/u01/oradata/orcl/undotbs2.dbf' SIZE 200M AUTOEXTEND ON;
````

## 42.2 Read Consistency Through Undo

When a long-running query starts, Oracle guarantees it sees the data exactly as it existed at that starting moment - even if other transactions commit changes while the query is still running, by transparently reading the *undo* version of any blocks that have since changed.

## 42.3 Rollback

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
UPDATE accounts SET balance = balance - 100 WHERE account_id = 1;
ROLLBACK;   -- Oracle restores the pre-update value using undo data
````

## 42.4 Undo and Long-Running Queries

A long-running report query, running concurrently with an OLTP workload constantly updating the same tables, relies on undo remaining available for the *entire* duration of the query - if the undo data it needs has since been overwritten (because the undo tablespace's retention period was too short relative to the query's runtime), the query fails.

## 42.5 ORA-01555: Snapshot Too Old

```
ORA-01555: snapshot too old: rollback segment number ... with name "..." too small
```

| Stage | Detail |
|---|---|
| Cause | A long-running query needed an undo (before-image) version of a block that has since been overwritten by newer transactions, because the undo tablespace didn't retain it long enough. |
| Common triggers | Undo tablespace too small; `UNDO_RETENTION` too low relative to query duration; a very long-running report competing with heavy concurrent write activity. |
| Diagnosis | Check `UNDO_RETENTION` and undo tablespace sizing; examine whether the report query's runtime is unusually long, or write activity unusually heavy, relative to normal conditions. |
| Solution | Increase undo tablespace size and/or `UNDO_RETENTION`; consider `GUARANTEE` retention if this must never happen; optimize the long-running query itself. |

## 42.6 Why Undo Is Critical to Oracle Transaction Processing

Without sufficient undo, Oracle cannot provide `ROLLBACK`, cannot guarantee read consistency for concurrent readers/writers (Part 25.1-25.2), and cannot support Flashback features (querying data as of a past point in time) - undo is not an optional performance feature, but a foundational requirement for correct transactional behavior in Oracle's architecture.


# 43. Redo and Recovery

## 43.1 Redo Logs

**Redo** records every change made to the database, written first to the **redo log buffer** in the SGA (Part 1.5), then flushed by `LGWR` to the **online redo logs** on disk - this is what allows Oracle to recover committed transactions after an unexpected crash (a durability guarantee, the "D" in ACID, Part 24).

## 43.2 Online Redo Logs

Oracle maintains multiple online redo log groups, writing to them in a circular fashion - once one group fills, Oracle switches to the next ("log switch"), eventually cycling back to overwrite the first, unless it has been archived first (Part 43.3).

## 43.3 Archived Redo Logs

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
ALTER DATABASE ARCHIVELOG;
````

When a database runs in `ARCHIVELOG` mode, every online redo log is copied to a permanent **archived redo log** file before being overwritten - preserving a complete history of every change, essential for point-in-time recovery (Part 44.7) and for taking hot (online) backups.

## 43.4 Redo Generation

Every DML operation, and even most DDL, generates redo - this is why bulk operations (Part 36) that touch millions of rows generate proportionally large amounts of redo, a real, practical consideration when planning large batch loads (sometimes mitigated with `NOLOGGING` for specific, recoverable-by-other-means bulk load operations).

## 43.5 Recovery Concepts

| Recovery Type | Uses |
|---|---|
| Instance recovery (automatic, on startup after a crash) | Online redo logs only - replays committed changes since the last checkpoint, then rolls back any uncommitted transactions using undo (Part 42) |
| Media recovery (after losing a datafile) | A backup of the datafile, plus archived and online redo logs, to roll it forward to the point of failure |

## 43.6 Redo vs Undo vs Archive Logs: The Key Distinction

| | Redo | Undo | Archived Redo Logs |
|---|---|---|---|
| Purpose | Reapply changes during recovery (roll *forward*) | Reverse changes for `ROLLBACK` and read consistency (roll *back*) | A permanent, disk-based history of redo, preserved before online redo logs are overwritten |
| Used for | Crash/media recovery, ensuring durability | Transaction rollback, read consistency, Flashback queries | Point-in-time recovery, restoring from an old backup forward to a specific moment |
| Where it lives | Online redo log files (then archived) | The undo tablespace | Separate archive destination (disk, tape, or a backup system) |

This redo/undo distinction is one of the most conceptually important - and most commonly confused, especially by those newer to Oracle specifically - architectural ideas in the entire platform: redo protects *durability going forward*, undo protects *consistency and enables reversal*.


# 44. Oracle Backup and Recovery

## 44.1 RMAN Concepts

**RMAN** (Recovery Manager) is Oracle's built-in backup and recovery tool, aware of the database's physical structure at the block level - able to perform block-level incremental backups, corruption detection, and automated recovery orchestration, well beyond a simple file copy.

```
rman target /
```

## 44.2 Full Backup

```
RMAN> BACKUP DATABASE;
```

A complete copy of every datafile in the database.

## 44.3 Incremental Backup

```
RMAN> BACKUP INCREMENTAL LEVEL 1 DATABASE;
```

Backs up only the blocks changed since the last backup at a lower (or equal) level - significantly reducing backup time/size for large databases compared to always taking a full backup.

## 44.4 Archived Logs in Backup Strategy

```
RMAN> BACKUP ARCHIVELOG ALL;
```

Backing up archived redo logs (Part 43.3) alongside datafile backups is what enables recovery *forward* to any point between backups, not just to the moment of the last full backup.

## 44.5 Restore and Recovery

```
RMAN> RESTORE DATABASE;
RMAN> RECOVER DATABASE;
```

`RESTORE` copies backup files back into place; `RECOVER` then applies archived and online redo logs to bring the restored files forward to a consistent, up-to-date (or specifically targeted) state.

## 44.6 Point-in-Time Recovery

```
RMAN> RUN {
    SET UNTIL TIME "TO_DATE('2026-01-15 09:00:00','YYYY-MM-DD HH24:MI:SS')";
    RESTORE DATABASE;
    RECOVER DATABASE;
};
```

Restores and recovers the database to its exact state at a specific past moment - essential for recovering from an accidental mass data change (e.g. an erroneous bulk `UPDATE`/`DELETE`) by rolling back to just before it happened.

## 44.7 Disaster Recovery

Beyond RMAN backups, enterprise Oracle deployments commonly use **Oracle Data Guard** to maintain a continuously synchronized standby database (often in a physically separate data center), ready to take over with minimal data loss if the primary database becomes entirely unavailable.

## 44.8 RPO and RTO

| Term | Meaning |
|---|---|
| **RPO** (Recovery Point Objective) | The maximum acceptable amount of *data loss*, measured in time (e.g. "we can afford to lose at most 5 minutes of transactions"). |
| **RTO** (Recovery Time Objective) | The maximum acceptable amount of *downtime* before the system must be back online. |

These two targets directly drive backup/DR architecture decisions: a near-zero RPO typically requires synchronous standby replication (Data Guard); a longer acceptable RPO may be satisfied by frequent incremental RMAN backups and archived-log shipping alone.

# 46. Oracle Data Dictionary

## 46.1 USER_\*, ALL_\*, DBA_\*, V$, and GV$

Oracle exposes its own internal metadata through a rich set of built-in views, collectively called the **data dictionary**.

| Prefix | Scope |
|---|---|
| `USER_*` | Objects owned by the *currently connected* schema only |
| `ALL_*` | Objects the current user owns, **plus** objects owned by others that the user has been granted access to |
| `DBA_*` | Every object in the entire database (requires DBA-level privileges) |
| `V$*` | Dynamic performance views - real-time instance/session/memory statistics, not persistent metadata |

| View | Reveals |
|---|---|
| `USER_TABLES` | Tables owned by the current schema |

# 50. Oracle for Data Engineering

## 50.1 ETL vs ELT

| | ETL | ELT |
|---|---|---|
| Order | Extract -> Transform (outside the target) -> Load | Extract -> Load (raw) -> Transform (inside the target database) |
| Where transformation happens | An external tool/application | Inside Oracle itself, using SQL/PL/SQL |
| When favored | Complex transformations, or limited target database compute | Oracle's own processing power is leveraged directly, common in modern data warehouse patterns |

## 50.2 Batch Processing and Incremental Loading

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Full batch load
TRUNCATE TABLE staging_orders;
INSERT INTO staging_orders SELECT * FROM external_orders_feed;

-- Incremental load: only new/changed rows since the last run
INSERT INTO staging_orders
SELECT * FROM external_orders_feed
WHERE last_modified > (SELECT MAX(last_loaded_at) FROM etl_watermark);

UPDATE etl_watermark SET last_loaded_at = SYSTIMESTAMP;
````

Incremental loading avoids reprocessing the entire source dataset on every run - essential once a source table grows beyond what a full reload can process within an acceptable time window.

## 50.3 Change Data Capture (CDC) Concepts

**CDC** captures and propagates only the specific rows that changed in a source system, typically via database triggers, log-mining (Oracle GoldenGate, a dedicated Oracle product for this), or periodic timestamp-based comparison (the simpler approach shown in Part 50.2) - avoiding full-table comparisons or reloads to detect what's new.

## 50.4 Slowly Changing Dimensions (SCD)

| Type | Behavior |
|---|---|
| Type 1 | Overwrite the old value - no history kept |
| Type 2 | Insert a new row for the changed value, preserving history with effective date ranges |
| Type 3 | Add a new column to track a limited, specific history (e.g. "previous region") |

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
-- Type 2 SCD: closing out the old record, inserting a new current one
UPDATE dim_customer SET is_current = 'N', end_date = SYSDATE
WHERE customer_id = 123 AND is_current = 'Y';

INSERT INTO dim_customer (customer_id, region, is_current, start_date, end_date)
VALUES (123, 'West', 'Y', SYSDATE, DATE '9999-12-31');
````

## 50.5 Star Schemas: Fact Tables and Dimension Tables

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE fact_sales (
    sale_id       NUMBER,
    date_key      NUMBER REFERENCES dim_date(date_key),
    customer_key  NUMBER REFERENCES dim_customer(customer_key),
    product_key   NUMBER REFERENCES dim_product(product_key),
    amount        NUMBER
);
````

Fact tables hold measurable, numeric events (sales, transactions); dimension tables hold descriptive context (customers, products, dates) - fully explored architecturally in Part 51.

## 50.6 A Complete Oracle ETL Example

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE run_daily_sales_etl IS
BEGIN
    -- EXTRACT + LOAD (raw)
    INSERT INTO staging_sales
    SELECT * FROM external_pos_feed WHERE feed_date = TRUNC(SYSDATE) - 1;

    -- TRANSFORM (inside Oracle, ELT-style)
    INSERT INTO fact_sales (sale_id, date_key, customer_key, product_key, amount)
    SELECT s.sale_id,
           TO_NUMBER(TO_CHAR(s.sale_date, 'YYYYMMDD')),
           c.customer_key,
           p.product_key,
           s.amount
    FROM staging_sales s
    JOIN dim_customer c ON s.customer_id = c.customer_id AND c.is_current = 'Y'
    JOIN dim_product p ON s.product_id = p.product_id;

    DELETE FROM staging_sales;
    COMMIT;
EXCEPTION
    WHEN OTHERS THEN
        ROLLBACK;
        INSERT INTO etl_error_log (procedure_name, error_message, occurred_at)
        VALUES ('RUN_DAILY_SALES_ETL', SQLERRM, SYSTIMESTAMP);
        COMMIT;
        RAISE;
END;
/
````

# 51. Oracle Data Warehousing

## 51.1 OLTP vs OLAP

| | OLTP | OLAP |
|---|---|---|
| Purpose | Day-to-day transactional operations | Analytical reporting and business intelligence |
| Schema shape | Highly normalized (3NF, Part 5 concepts) | Denormalized, star/snowflake schemas |
| Typical query | Fetch/update a single record | Aggregate across millions of rows |
| Example | The `orders` table an application writes to constantly | A `fact_sales` table a BI dashboard queries |

## 51.2 Star Schema

```
                dim_date
                    |
dim_customer -- fact_sales -- dim_product
                    |
                dim_store
```

A central fact table connected directly to several dimension tables - simple, fast for typical BI queries, and the most common data warehouse design pattern.

## 51.3 Snowflake Schema

A star schema where dimension tables are further normalized into sub-dimensions (e.g. `dim_product` referencing a separate `dim_category` table) - trades some query simplicity/performance for reduced redundancy, generally favored only when dimension tables are very large or slowly changing dimensions (Part 50.4) require it.

## 51.4 Surrogate Keys in Dimensions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE dim_customer (
    customer_key  NUMBER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,   -- surrogate key
    customer_id   NUMBER,    -- the natural/business key from the source system
    name          VARCHAR2(100),
    region        VARCHAR2(50),
    is_current    CHAR(1),
    start_date    DATE,
    end_date      DATE
);
````

Dimension tables use a surrogate key (Part 5.6) specifically so that Type 2 SCD history (multiple rows per business `customer_id`, Part 50.4) can be modeled cleanly - the fact table always references the surrogate `customer_key`, correctly pointing to the customer's attributes *as they were* at the time of the sale.

## 51.5 Date Dimensions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE dim_date (
    date_key    NUMBER PRIMARY KEY,   -- e.g. 20260615
    full_date   DATE,
    year        NUMBER,
    quarter     NUMBER,
    month       NUMBER,
    day_of_week VARCHAR2(10),
    is_weekend  CHAR(1)
);
````

A pre-populated date dimension avoids repeatedly computing date parts (`EXTRACT`, `TO_CHAR`) in every analytical query, and allows straightforward filtering/grouping by fiscal periods, holidays, or other business-specific calendar attributes not derivable from a plain date alone.

## 51.6 A Complete Retail Data Warehouse Example

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE dim_date (date_key NUMBER PRIMARY KEY, full_date DATE, year NUMBER, month NUMBER);
CREATE TABLE dim_customer (customer_key NUMBER GENERATED ALWAYS AS IDENTITY PRIMARY KEY, customer_id NUMBER, region VARCHAR2(50));
CREATE TABLE dim_product (product_key NUMBER GENERATED ALWAYS AS IDENTITY PRIMARY KEY, product_id NUMBER, category VARCHAR2(50));

CREATE TABLE fact_sales (
    sale_id      NUMBER PRIMARY KEY,
    date_key     NUMBER REFERENCES dim_date(date_key),
    customer_key NUMBER REFERENCES dim_customer(customer_key),
    product_key  NUMBER REFERENCES dim_product(product_key),
    quantity     NUMBER,
    amount       NUMBER
)
PARTITION BY RANGE (date_key) (
    PARTITION p_2025 VALUES LESS THAN (20260101),
    PARTITION p_2026 VALUES LESS THAN (20270101)
);

-- A typical BI query against this warehouse
SELECT d.year, p.category, SUM(f.amount) AS total_revenue
FROM fact_sales f
JOIN dim_date d ON f.date_key = d.date_key
JOIN dim_product p ON f.product_key = p.product_key
GROUP BY d.year, p.category
ORDER BY d.year, total_revenue DESC;
````

Combining partitioning (Part 40) directly on the fact table's date key with a clean star schema is a standard, professional design for a large-scale Oracle data warehouse.

## 52.3 A Small CRUD API With FastAPI

````python
from fastapi import FastAPI, HTTPException
import oracledb

app = FastAPI()
pool = oracledb.create_pool(user="app_user", password="password", dsn="localhost:1521/ORCLPDB1", min=2, max=10)

@app.get("/employees/{employee_id}")
def get_employee(employee_id: int):
    with pool.acquire() as connection:
        cursor = connection.cursor()
        cursor.execute(
            "SELECT employee_id, first_name, last_name FROM employees WHERE employee_id = :id",
            id=employee_id
        )
        row = cursor.fetchone()
        if row is None:
            raise HTTPException(status_code=404, detail="Employee not found")
        return {"id": row[0], "first_name": row[1], "last_name": row[2]}

@app.post("/employees")
def create_employee(first_name: str, last_name: str, email: str):
    with pool.acquire() as connection:
        cursor = connection.cursor()
        cursor.execute(
            "INSERT INTO employees (employee_id, first_name, last_name, email) "
            "VALUES (employees_seq.NEXTVAL, :fn, :ln, :em)",
            fn=first_name, ln=last_name, em=email
        )
        connection.commit()
        return {"status": "created"}
````

## 52.4 Transactions in a Backend Context

````python
@app.post("/transfer")
def transfer_funds(from_account: int, to_account: int, amount: float):
    with pool.acquire() as connection:
        cursor = connection.cursor()
        try:
            cursor.execute("UPDATE accounts SET balance = balance - :amt WHERE account_id = :id",
                            amt=amount, id=from_account)
            cursor.execute("UPDATE accounts SET balance = balance + :amt WHERE account_id = :id",
                            amt=amount, id=to_account)
            connection.commit()
            return {"status": "success"}
        except oracledb.DatabaseError as e:
            connection.rollback()
            raise HTTPException(status_code=500, detail=str(e))
````

## 56.1 Project 1 - Retail Management System

**Business problem:** Track products, inventory, and sales across multiple stores.

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE TABLE stores (store_id NUMBER PRIMARY KEY, store_name VARCHAR2(100), region VARCHAR2(50));
CREATE TABLE products (product_id NUMBER PRIMARY KEY, name VARCHAR2(100), price NUMBER(10,2));
CREATE TABLE inventory (store_id NUMBER REFERENCES stores, product_id NUMBER REFERENCES products,
    quantity NUMBER, PRIMARY KEY (store_id, product_id));
CREATE TABLE sales (sale_id NUMBER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    store_id NUMBER REFERENCES stores, product_id NUMBER REFERENCES products,
    quantity NUMBER, sale_date DATE DEFAULT SYSDATE);
````

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
CREATE OR REPLACE PROCEDURE record_sale(p_store_id NUMBER, p_product_id NUMBER, p_qty NUMBER) IS
    v_stock NUMBER;
BEGIN
    SELECT quantity INTO v_stock FROM inventory WHERE store_id = p_store_id AND product_id = p_product_id FOR UPDATE;
    IF v_stock < p_qty THEN
        RAISE_APPLICATION_ERROR(-20001, 'Insufficient inventory');
    END IF;
    UPDATE inventory SET quantity = quantity - p_qty WHERE store_id = p_store_id AND product_id = p_product_id;
    INSERT INTO sales (store_id, product_id, quantity) VALUES (p_store_id, p_product_id, p_qty);
    COMMIT;
END;
/
````

**Analytical query:** top-selling products per region using `RANK()` (Part 15.3); **Optimization:** composite index on `sales(store_id, sale_date)`; **Security:** a read-only `reporting_role` for BI tools accessing only aggregated views (Part 19.4).

# 60. Common Oracle Errors

## 60.1 ORA-00904: invalid identifier

| | Detail |
|---|---|
| Meaning | A referenced column or identifier doesn't exist, or is misspelled |
| Cause | Typo in a column name, or referencing a column not in the current query's tables |
| Example | `SELECT salry FROM employees;` |
| Diagnosis | Check `USER_TAB_COLUMNS` (Part 46.2) for the actual column name |
| Solution | Correct the spelling: `SELECT salary FROM employees;` |
| Prevention | Use SQL Developer's auto-complete; validate column names against the data dictionary |

## 60.2 ORA-00933: SQL command not properly ended

| | Detail |
|---|---|
| Meaning | Extra or misplaced tokens after what Oracle expected to be the end of the statement |
| Cause | Often a trailing clause in the wrong position, or a MySQL/PostgreSQL-specific syntax Oracle doesn't recognize (Part 55) |
| Example | `SELECT * FROM employees LIMIT 10;` (MySQL/PostgreSQL syntax, invalid in Oracle) |
| Solution | `SELECT * FROM employees FETCH FIRST 10 ROWS ONLY;` (Part 8.2) |
| Prevention | Remember Oracle's specific row-limiting and clause syntax when translating from other databases |

## 60.3 ORA-00942: table or view does not exist

| | Detail |
|---|---|
| Meaning | The referenced table/view either doesn't exist, or you lack privileges to see it |
| Cause | Typo, wrong schema, or a missing `GRANT` |
| Diagnosis | Check `SELECT * FROM all_tables WHERE table_name = 'YOUR_TABLE';` (Part 46.3) |
| Solution | Correct the name, use `schema.table_name`, or request the appropriate `GRANT` (Part 45.3) |

## 60.4 ORA-01400: cannot insert NULL into (...)

| | Detail |
|---|---|
| Meaning | An `INSERT`/`UPDATE` attempted to set a `NOT NULL` column to `NULL` |
| Cause | Missing a required value, or - per Part 9.1 - inserting an empty string into a character column |
| Solution | Supply a valid, non-null value for the column |
| Prevention | Validate application input before submitting DML, especially remembering the empty-string-as-NULL rule |

## 60.5 ORA-01438: value larger than specified precision allows for this column

| | Detail |
|---|---|
| Meaning | A numeric value exceeds the column's defined precision (Part 4.3) |
| Example | Inserting `123456` into a `NUMBER(5)` column |
| Solution | Widen the column's precision, or validate/round the value before insert |

## 60.6 ORA-01722: invalid number

| | Detail |
|---|---|
| Meaning | A string could not be converted to a number |
| Cause | Implicit or explicit conversion (`TO_NUMBER`) on a non-numeric string |
| Example | `SELECT TO_NUMBER('abc') FROM dual;` |
| Solution | Validate the string represents a genuine number first, or use `TO_NUMBER(x DEFAULT NULL ON CONVERSION ERROR)` (12c+) |

## 60.7 ORA-01861: literal does not match format string

| | Detail |
|---|---|
| Meaning | A date/timestamp literal doesn't match the format mask supplied to `TO_DATE`/`TO_TIMESTAMP` |
| Example | `TO_DATE('2026/06/15', 'YYYY-MM-DD')` (mismatched separators) |
| Solution | Ensure the format mask (Part 10.5) exactly matches the literal's actual structure |

## 60.8 ORA-00001: unique constraint violated

| | Detail |
|---|---|
| Meaning | An `INSERT`/`UPDATE` would create a duplicate value in a column protected by a `UNIQUE` or primary key constraint |
| Diagnosis | Check the constraint name in the error message against `USER_CONSTRAINTS` (Part 46.2) |
| Solution | Handle via `DUP_VAL_ON_INDEX` (Part 30.2) in PL/SQL, or resolve the underlying duplicate data issue |

## 60.9 ORA-01403: no data found

| | Detail |
|---|---|
| Meaning | A `SELECT INTO` matched zero rows |
| Solution | Handle with `WHEN NO_DATA_FOUND` (Part 30.2), or verify the row genuinely exists before querying |

## 60.10 ORA-06502: PL/SQL: numeric or value error

| | Detail |
|---|---|
| Meaning | A value doesn't fit its target variable (e.g. a string too long for a `VARCHAR2` variable, or a failed numeric conversion) |
| Solution | Widen the variable's declared size, or validate/truncate the source value first; catch with `WHEN VALUE_ERROR` |

## 60.11 ORA-04091: table is mutating, trigger/function may not see it

| | Detail |
|---|---|
| Meaning | A row-level trigger attempted to query the very table currently being modified by the triggering statement |
| Solution | Restructure using a compound trigger (Part 34.4), or move the logic to an `AFTER STATEMENT` trigger with a package-level collection |

Fully covered in Part 42.5 - typically resolved by increasing undo retention/size, or optimizing the long-running query.


# 61. Oracle Best Practices

## 61.1 SQL

- Always use `VARCHAR2`, never `VARCHAR` (Part 4.2).
- Use ANSI `JOIN` syntax, never the legacy `(+)` operator, in new code (Part 12.3).
- Prefer `EXISTS`/`NOT EXISTS` over `IN`/`NOT IN` when the subquery result could contain `NULL` (Part 13.6).
- Prefer `COALESCE` over `NVL` for portability, unless matching an existing codebase convention (Part 9.7).
- Always use bind variables for literals that vary between executions (Part 23.3).

## 61.2 PL/SQL

- Prefer set-based SQL over cursor loops whenever the logic permits it (Part 29.5).
- Use `BULK COLLECT`/`FORALL` for any necessary large-scale row-by-row processing (Part 36).
- Always use `%TYPE`/`%ROWTYPE` instead of hardcoding data types that mirror table columns (Part 28.4).
- Keep exception handling specific (`WHEN NO_DATA_FOUND`, etc.) rather than defaulting to a catch-all `WHEN OTHERS` that silently hides real bugs.

## 61.3 Database Design

- Use surrogate keys as primary keys; natural keys as `UNIQUE` constraints (Part 5.6).
- Normalize OLTP schemas to at least 3NF; deliberately denormalize only for analytical/warehouse tables (Part 17 of your Database Programming handbook, Part 51 here).
- Design fact/dimension tables with surrogate keys to support historical tracking (Part 51.4).

## 61.4 Indexing

- Index columns used in frequent filters/joins; verify actual usage with `EXPLAIN PLAN` (Part 22.4), don't assume.
- Reserve bitmap indexes for low-cardinality columns in read-heavy analytical contexts, never high-concurrency OLTP writes (Part 21.2).
- Use invisible indexes to safely evaluate removing an index before dropping it outright (Part 21.6).

## 61.5 Transactions

- Never rely on implicit autocommit behavior - commit explicitly, and be aware that DDL always implicitly commits (Part 24.2-24.3).
- Use `FOR UPDATE` to lock rows explicitly when a read-then-write sequence must be atomic against concurrent modification (Part 25.3, 31.3).

## 61.6 Security

- Grant privileges to roles, not directly to individual users (Part 45.4).
- Always parameterize dynamic SQL values; validate identifiers against an allow-list (Part 35.6).
- Never hardcode credentials in application code; use a secrets manager or environment variables, exactly as covered in your Database Programming handbook (Part 7.7, 28.2 there).

## 61.7 Exception Handling

- Log errors with enough context (`SQLCODE`, `SQLERRM`) to diagnose without exposing sensitive data externally (Part 30.6).
- Use autonomous transactions for error logging that must survive a surrounding rollback (Part 30.6).

## 61.8 Packages and Procedures

- Group related business logic into packages; keep implementation details private (Part 33.4).
- Keep package-level state minimal and well-documented, since it persists for the session and can be a source of subtle bugs (Part 33.5).

## 61.9 Performance

- Gather statistics regularly, especially after large bulk loads (Part 22.2).
- Avoid wrapping indexed columns in functions unless a matching function-based index exists (Part 21.5, 23.2).
- Partition very large tables aligned with common query filters (Part 40).

## 61.10 Backup

- Test restore procedures regularly, not just backup jobs themselves (Part 44, echoing the same principle from your Database Programming handbook, Part 28.5 there).
- Define RPO/RTO targets explicitly before choosing a backup/DR architecture (Part 44.8).

## 61.11 Data Quality

- Validate incoming data before loading into production tables (Part 59.5, Q69).
- Build automated data quality checks into ETL pipelines, not just ad hoc manual review (Part 50.6).

## 61.12 Documentation and Deployment

- Document package specifications thoroughly, since they are the contract other developers rely on (Part 33.2).
- Use versioned migration scripts for schema changes, mirroring the migration discipline from your Database Programming handbook (Part 25 there), adapted to Oracle's DDL and RMAN/backup practices.
- Test schema changes on a staging environment before applying to production, especially for large tables (Part 58.4 Q88).

## 62.8 CTEs

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
WITH cte_name AS (SELECT ...) SELECT * FROM cte_name;
WITH RECURSIVE cte (...) AS (... UNION ALL ...) SELECT * FROM cte;
-- Oracle hierarchical alternative:
SELECT ..., LEVEL FROM t START WITH cond CONNECT BY PRIOR parent_id = id;
````

## 62.9 Window Functions

````python
-- %%sql (Oracle SQL / PL/SQL - not executable in a Python kernel)
ROW_NUMBER() OVER (PARTITION BY x ORDER BY y)
RANK() OVER (ORDER BY y DESC)
DENSE_RANK() OVER (ORDER BY y DESC)
LAG(col, 1) OVER (ORDER BY y)
LEAD(col, 1) OVER (ORDER BY y)
SUM(col) OVER (ORDER BY y ROWS UNBOUNDED PRECEDING)   -- running total
AVG(col) OVER (ORDER BY y ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)  -- moving avg
````
