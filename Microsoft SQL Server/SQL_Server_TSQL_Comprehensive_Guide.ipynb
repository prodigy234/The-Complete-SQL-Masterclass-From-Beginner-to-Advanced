## 1. SQL Server Fundamentals

### 1.1 What SQL Server Is

Microsoft SQL Server is a relational database management system (RDBMS) developed by Microsoft, running primarily on Windows and (since SQL Server 2017) Linux and Docker containers. Like PostgreSQL and MySQL, it stores data in tables, enforces relationships and constraints, and is queried with SQL — but it is built around Microsoft's own database engine, tooling ecosystem, and a SQL dialect called **T-SQL (Transact-SQL)**.

### 1.2 SQL Server vs SQL vs T-SQL

| Term | What It Is |
|---|---|
| **SQL** | The ANSI/ISO-standardized query language itself — `SELECT`, `INSERT`, `JOIN`, and so on, common in some form across virtually every RDBMS. |
| **T-SQL (Transact-SQL)** | Microsoft's proprietary extension of SQL, adding procedural constructs (`BEGIN`/`END`, `WHILE`, variables, `TRY`/`CATCH`), SQL Server-specific functions, and syntax found nowhere else (`TOP`, `OUTPUT`, `APPLY`, `MERGE`). |
| **SQL Server** | The actual database engine/product that executes T-SQL. |

Throughout this handbook, every syntax block is labeled **[Standard SQL]**, **[T-SQL]**, or **[SQL Server Feature]** so you always know whether something is portable or Microsoft-specific — directly relevant since you already know PostgreSQL and MySQL.

### 1.3 SQL Server Architecture

At a high level, SQL Server consists of the **Database Engine** (the core service that stores, processes, and secures data — handling query execution, transactions, and storage), the **SQL Server Agent** (a job scheduler, Part 38), and various services (Analysis Services, Integration Services, Reporting Services) not covered in depth here since this handbook focuses on the relational engine and T-SQL.

### 1.4 SQL Server Instances

An **instance** is a single running installation of the SQL Server Database Engine — one physical or virtual machine can host multiple independent instances (a **default instance** and any number of **named instances**), each with its own set of databases, logins, and configuration, fully isolated from one another. This has no direct equivalent in a typical PostgreSQL/MySQL setup, where "one server process, one port" is more common; SQL Server instances are more like several independent PostgreSQL clusters coexisting on one machine, each addressable as `ServerName\InstanceName`.

### 1.5 Core Object Hierarchy

```
SQL Server Instance
    └── Database
            └── Schema
                    └── Objects (Tables, Views, Procedures, Functions, Triggers)
                            └── Rows / Columns
```

This is the single most important structural concept to internalize coming from PostgreSQL (which has an identical `database → schema → object` hierarchy) or MySQL (where, notably, a MySQL "database" and "schema" are actually the same thing — there is no separate schema layer). In SQL Server, exactly as in PostgreSQL, a schema is a genuine namespace *within* a database — `Sales.Orders` and `HR.Orders` can coexist as two entirely different tables in the same database.

### 1.6 Server vs Instance vs Database vs Schema vs Table vs User vs Login

| Term | Meaning |
|---|---|
| **Server** | The physical or virtual machine (or cloud resource) hosting SQL Server. |
| **Instance** | One installation of the Database Engine on that server (Section 1.4). |
| **Database** | A container for schemas, tables, and other objects — the unit typically backed up, restored, and secured independently (Part 3). |
| **Schema** | A namespace within a database grouping related objects (e.g. `Sales`, `HR`). |
| **Table** | A structured collection of rows and columns within a schema. |
| **Login** | A server-level security principal — an identity that can connect to the *instance* (Part 35). |
| **User** | A database-level security principal, typically mapped to a login, granting access *within a specific database*. |

This login/user distinction is a genuinely SQL Server-specific (and SQL Server-important) concept — PostgreSQL uses a single unified "role" for both server- and database-level access, and MySQL ties privileges directly to a `user@host` combination without this separate two-tier model. Covered fully in Part 35.

### 1.7 Databases, Tables, Views, Stored Procedures, Functions, Triggers, Indexes, Constraints, Transactions

These core concepts — tables, views, stored procedures, functions, triggers, indexes, constraints, and transactions — are already familiar from PostgreSQL/MySQL and are not re-taught from scratch here; each gets its own dedicated, SQL Server-focused part later in this handbook (Parts 5, 11-25) where the specific T-SQL syntax and SQL Server behavior are covered in depth.

### 1.8 System Databases

Every SQL Server instance automatically includes four system databases, distinct from any user-created database:

| Database | Purpose |
|---|---|
| **master** | Records instance-level metadata: logins, configuration, and the existence of every other database. Never store application data here. |
| **model** | A template database — every new database is created as a copy of `model`'s initial structure and settings. |
| **msdb** | Used by SQL Server Agent (Part 38) to store jobs, schedules, alerts, and backup/restore history. |
| **tempdb** | A shared scratch database for temporary tables, sorting, hashing, and other intermediate work — recreated fresh on every server restart (covered fully in Part 40). |

Neither PostgreSQL nor MySQL has a direct equivalent to this specific four-database system layer — PostgreSQL has a `postgres` maintenance database and internal catalogs, MySQL has `information_schema`/`mysql` system databases, but SQL Server's `model`/`msdb`/`tempdb` split, and especially `tempdb`'s central role in query execution (Part 40), is distinctly its own architecture worth learning deliberately.

### 1.9 User Databases

A **user database** is any database you create to hold actual application data — `CREATE DATABASE Sales;` (Part 3) — logically and physically separate from the four system databases above.

## 2. SQL Server Tools

### 2.1 SQL Server Management Studio (SSMS)

SSMS is Microsoft's full-featured, Windows-only graphical client for SQL Server — the closest equivalent to pgAdmin (PostgreSQL) or MySQL Workbench (MySQL), providing Object Explorer (a tree view of every database, table, and object), a Query Editor, execution-plan visualization (Part 29), and administrative tooling (backups, security, jobs) in one application.

### 2.2 Azure Data Studio

Azure Data Studio is a modern, cross-platform (Windows/macOS/Linux), lightweight alternative to SSMS, built on the same editor foundation as VS Code — notebooks, Git integration, and extensions are first-class, at the cost of some of SSMS's deeper administrative GUI tooling. A reasonable default choice if you're not on Windows or prefer a VS Code-like experience.

### 2.3 VS Code SQL Tooling

VS Code, with the "SQL Server (mssql)" extension, provides connection management, query execution, and IntelliSense directly inside the editor many developers already use daily for application code — a practical choice when SQL Server work is one part of a broader development workflow rather than the primary focus.

### 2.4 sqlcmd

```bash
sqlcmd -S localhost -d Sales -U sa -P YourPassword -Q "SELECT COUNT(*) FROM dbo.Orders;"```


`sqlcmd` is SQL Server's command-line query tool — the direct equivalent of `psql` (PostgreSQL) or the `mysql` CLI client — essential for scripting, automation, and any environment without a GUI (containers, CI pipelines, remote servers).

| Flag | Meaning |
|---|---|
| `-S` | Server/instance to connect to. |
| `-d` | Target database. |
| `-U` / `-P` | SQL Server Authentication username/password (Section 2.6). |
| `-E` | Use Windows Authentication instead of `-U`/`-P`. |
| `-Q` | Run a single query and exit (vs `-i` for a script file). |

### 2.5 SQL Server Configuration Manager

A Windows-specific administrative tool for managing SQL Server *services* (starting/stopping the engine and Agent), configuring network protocols (TCP/IP, named pipes), and setting the port an instance listens on — administrative, instance-level configuration rather than day-to-day query work.

### 2.6 Authentication: Windows vs SQL Server Authentication

| Mode | How It Works |
|---|---|
| **Windows Authentication** | Uses the currently logged-in Windows/Active Directory identity; no separate SQL Server password to manage. The default and generally recommended mode on Windows-based enterprise networks. |
| **SQL Server Authentication** | A traditional username/password stored and verified by SQL Server itself, independent of the OS — required for non-Windows clients, cross-platform apps, and cloud scenarios (Azure SQL). |

This dual-authentication-mode design is distinctly SQL Server's own; PostgreSQL and MySQL each have their own separate authentication method configuration (`pg_hba.conf`, MySQL's plugin-based auth) but neither maps cleanly onto "Windows identity vs database-native credential" as a first-class, built-in choice.

### 2.7 Connecting, Creating Databases and Users, Running and Saving Queries

`GO` is a **batch separator** recognized by SSMS/`sqlcmd` (not a real T-SQL statement, and not sent to the server itself) — it tells the client tool "send everything above this point as one batch, then start a new batch." This matters because certain statements (like `CREATE DATABASE` followed immediately by `USE`) must be in separate batches. Query scripts are saved as `.sql` files, versioned in source control exactly like application code — a professional habit worth establishing from the start (Part 62).

## 3. Database and Schema Management

### 3.1 CREATE DATABASE / ALTER DATABASE / DROP DATABASE

The base `CREATE DATABASE` syntax is portable across PostgreSQL/MySQL/SQL Server; SQL Server's `ALTER DATABASE` gains many database-wide options (recovery model, compatibility level, file management) that PostgreSQL and MySQL either don't expose the same way or handle via separate configuration.

### 3.2 CREATE SCHEMA / ALTER SCHEMA / DROP SCHEMA

**[T-SQL note for MySQL users]:** MySQL has no independent schema layer — `CREATE SCHEMA` in MySQL is just a synonym for `CREATE DATABASE`. In SQL Server (like PostgreSQL), a schema is a true sub-container inside a database; a single database routinely holds many schemas (`Sales`, `HR`, `dbo` the default schema) each organizing a related set of objects.

### 3.3 Database Files — MDF, NDF, LDF

| File Type | Extension | Purpose |
|---|---|---|
| **Primary data file** | `.mdf` | Holds the database's core data and system tables; every database has exactly one. |
| **Secondary data file(s)** | `.ndf` | Optional additional data files, used to spread data across multiple disks/filegroups for very large databases. |
| **Log file** | `.ldf` | The transaction log — a sequential record of every change, used for crash recovery, transaction rollback, and point-in-time restore (Part 37). |

This is a genuinely SQL Server-specific storage architecture. PostgreSQL organizes storage as a directory of files per table/index inside a "cluster" data directory; MySQL (InnoDB) uses tablespace files (`.ibd`) per table by default. SQL Server's explicit MDF/NDF/LDF file model, with the transaction log as a distinctly separate physical file, directly shapes both backup strategy (Part 37) and performance tuning (`tempdb`'s own files, Part 40).

### 3.4 Filegroups

A **filegroup** is a named collection of one or more data files that tables/indexes can be explicitly placed onto — the **PRIMARY** filegroup (holding the `.mdf` and always present) is the default; additional, secondary filegroups let large tables, indexes, or partitions (Part 44) be physically placed on specific disks for performance or archival separation. Neither PostgreSQL's tablespaces nor MySQL's tablespace files map exactly onto SQL Server's filegroup concept, though tablespaces serve a broadly similar physical-placement role in PostgreSQL.

### 3.5 Recovery Models

| Model | Behavior |
|---|---|
| **Simple** | The transaction log is automatically truncated after each checkpoint; no transaction log backups possible. Simplest to manage, but only point-in-time recovery to the last full/differential backup. |
| **Full** | The log retains every transaction until explicitly backed up, enabling point-in-time recovery (Part 37) and transaction log backups — the standard choice for production databases. |
| **Bulk-logged** | A middle ground, minimally logging certain bulk operations (`BULK INSERT`, index rebuilds) for performance, while still supporting log backups. |

Recovery models are a distinctly SQL Server administrative concept without a direct one-to-one equivalent in PostgreSQL (which uses WAL — Write-Ahead Logging — with continuous archiving configured separately) or MySQL (binary logging, configured independently of a single "recovery model" setting). Choosing the recovery model deliberately (Full for anything requiring point-in-time recovery) is a foundational production decision covered further in Part 37.

### 3.6 Professional Database Organization Strategy

- One database per logical application/bounded context, rather than one giant shared database for unrelated systems.
- Schemas used deliberately to group related objects (`Sales`, `HR`, `Audit`) rather than dumping everything into the default `dbo` schema.
- `tempdb`-heavy workloads (Part 40) considered when sizing and placing files/filegroups.
- Recovery model chosen explicitly per database based on actual recovery requirements (Part 37), never left as an accidental default.

## 4. SQL Server Data Types

### 4.1 Numeric Types

| Type | Storage | Range/Notes |
|---|---|---|
| `tinyint` | 1 byte | 0 to 255 (unsigned only — no negative values, unlike most other integer types) |
| `smallint` | 2 bytes | -32,768 to 32,767 |
| `int` | 4 bytes | ~-2.1 billion to 2.1 billion |
| `bigint` | 8 bytes | Very large range; use for surrogate keys expected to exceed `int`'s range |
| `decimal(p,s)` / `numeric(p,s)` | Varies by precision | Exact, fixed-point — the standard choice for money and precise calculations; `decimal` and `numeric` are fully synonymous in SQL Server |
| `money` | 8 bytes | Fixed 4 decimal places; **generally avoided professionally** in favor of `decimal(19,4)` for clearer precision control and portability |
| `smallmoney` | 4 bytes | Smaller-range version of `money`, same caveats |
| `float` | 4 or 8 bytes | Approximate, floating-point — not appropriate for exact financial values |
| `real` | 4 bytes | Approximate, single-precision floating-point |

**[SQL Server vs PostgreSQL/MySQL]:** `tinyint` being unsigned-only (0-255) is a genuine SQL Server quirk — PostgreSQL has no `tinyint` at all (its smallest integer is `smallint`), and MySQL's `tinyint` supports an explicit signed range (-128 to 127) by default. Always double-check `tinyint`'s range assumption when porting schemas.

### 4.2 Character Types

| Type | Storage | Notes |
|---|---|---|
| `char(n)` | Fixed, n bytes | Fixed-length, space-padded; rarely the right default choice. |
| `varchar(n)` | Variable, up to n bytes | Variable-length, non-Unicode (ASCII/extended ASCII per collation). |
| `nchar(n)` | Fixed, 2n bytes | Fixed-length Unicode. |
| `nvarchar(n)` | Variable, up to 2n bytes | Variable-length Unicode — the standard professional default for any text that might contain non-English characters. |
| `text` | Deprecated | Legacy large-text type; **avoid** in new development — use `varchar(max)`/`nvarchar(max)` instead (Section 4.2.1). |

#### 4.2.1 Why text Is Avoided

`text` (and its counterparts `ntext`, `image`) are deprecated legacy types predating `varchar(max)`/`nvarchar(max)`/`varbinary(max)`, which were introduced specifically to replace them with full support for standard string functions, indexing flexibility, and modern T-SQL features that `text` lacks or handles awkwardly. **Always use `varchar(max)`/`nvarchar(max)` for large text in new SQL Server development.**

**[SQL Server vs PostgreSQL/MySQL]:** PostgreSQL's `text` type is the modern, fully-supported, unlimited-length standard — completely unrelated to SQL Server's deprecated `text`; don't let the shared name cause confusion. MySQL's `TEXT` (uppercase-conventional) is also actively used and not deprecated. This is purely a SQL Server-specific naming trap.

### 4.3 Date/Time Types

| Type | Precision | Storage | Notes |
|---|---|---|---|
| `date` | Day | 3 bytes | Date only, no time component. |
| `time` | Up to 100ns | 3-5 bytes | Time only, no date. |
| `datetime` | ~3.33ms | 8 bytes | Legacy type with an odd rounding precision and a limited range (1753-9999); prefer `datetime2` in new development. |
| `datetime2` | Up to 100ns | 6-8 bytes | Modern, higher-precision, wider-range replacement for `datetime` — the standard professional default. |
| `smalldatetime` | 1 minute | 4 bytes | Compact, low-precision legacy type. |
| `datetimeoffset` | Up to 100ns | 8-10 bytes | `datetime2` plus a UTC offset — the correct choice for any application spanning multiple time zones. |

**[SQL Server vs PostgreSQL/MySQL]:** PostgreSQL's `timestamp`/`timestamptz` map roughly onto SQL Server's `datetime2`/`datetimeoffset` respectively; MySQL's `DATETIME`/`TIMESTAMP` distinction is different again (MySQL's `TIMESTAMP` is UTC-converting and range-limited to 2038, unlike SQL Server's `datetimeoffset`). Always re-verify these mappings explicitly when porting schemas rather than assuming a name match implies identical behavior.

### 4.4 Other Types

| Type | Purpose |
|---|---|
| `bit` | Boolean-equivalent: 0, 1, or NULL. SQL Server has no native `boolean` type — `bit` is the standard substitute. |
| `uniqueidentifier` | A 128-bit GUID, SQL Server's native globally-unique identifier type — see `NEWID()`/`NEWSEQUENTIALID()` (Section 5.4). |
| `binary(n)` / `varbinary(n)` | Fixed/variable-length raw binary data; `varbinary(max)` for large binary objects (files, images). |
| `xml` | Native XML storage and querying (Part 32). |
| JSON | SQL Server has **no dedicated JSON data type** — JSON is stored as `nvarchar(max)` and manipulated with JSON *functions* (Part 31), a notable difference from PostgreSQL's native `json`/`jsonb` types. |
| Spatial types (`geometry`, `geography`) | Native spatial/GIS data types, broadly analogous to PostgreSQL's PostGIS extension, but built into the engine directly. |

**[SQL Server vs PostgreSQL]:** The absence of a true JSON data type is one of the most consequential differences from PostgreSQL for application developers used to `jsonb`'s native indexing and operators — SQL Server's JSON support (Part 31) is function-based over plain `nvarchar(max)`, which changes both the syntax and the indexing strategy (typically via computed columns, Section 31.6) needed for performant JSON querying.

### 4.5 SQL Server Data-Type Selection Guide

| Business Need | Recommended Type |
|---|---|
| Primary/surrogate key | `int IDENTITY` or `bigint IDENTITY` (Section 5.3); `uniqueidentifier` when values must be generated outside the database or merged across systems |
| Money/financial amounts | `decimal(19,4)`, not `money` or `float` |
| Names, emails, short text | `nvarchar(n)` sized to a sensible realistic maximum |
| Large free text | `nvarchar(max)` |
| Yes/No flags | `bit` |
| Timestamps (single time zone) | `datetime2` |
| Timestamps (multi-time-zone) | `datetimeoffset` |
| Date only (birthdate, order date) | `date` |
| JSON payloads | `nvarchar(max)` with `ISJSON()` constraint (Part 31) |

## 5. Table Design

### 5.1 CREATE TABLE / ALTER TABLE / DROP TABLE / TRUNCATE TABLE

`TRUNCATE TABLE` behaves the same conceptually as in PostgreSQL/MySQL — dramatically faster than `DELETE` for clearing an entire table since it deallocates data pages rather than logging row-by-row deletes, but it cannot be filtered with `WHERE` and resets `IDENTITY` (Section 5.3) back to its seed value.

### 5.2 Keys — Primary, Foreign, Composite, Candidate, Surrogate, Natural

| Key Type | Meaning |
|---|---|
| **Primary key** | The column(s) uniquely identifying each row; enforced with a unique index automatically. |
| **Foreign key** | A column referencing another table's primary key, enforcing referential integrity. |
| **Composite key** | A primary/unique key made of more than one column. |
| **Candidate key** | Any column (or set of columns) that *could* validly serve as the primary key. |
| **Surrogate key** | An artificial, meaningless identifier (typically `IDENTITY` or `uniqueidentifier`) with no business meaning — the standard professional choice for primary keys. |
| **Natural key** | A real-world, business-meaningful column (an email, a national ID) used as a key — appropriate as a `UNIQUE` constraint, but generally avoided as the *primary* key since business values can change. |

### 5.3 IDENTITY

`IDENTITY(seed, increment)` is SQL Server's **[T-SQL]** auto-incrementing column feature — conceptually equivalent to PostgreSQL's `SERIAL`/`GENERATED ALWAYS AS IDENTITY` and MySQL's `AUTO_INCREMENT`, but with its own retrieval function: `SCOPE_IDENTITY()` returns the last identity value generated in the *current session and scope*, deliberately safer than the older `@@IDENTITY` (which can return a value from an unrelated trigger-fired insert on a different table).

### 5.4 SQL Server SEQUENCE

A `SEQUENCE` is a standalone, schema-scoped number generator, independent of any single table — usable across multiple tables, reservable in advance, and controllable (`CACHE`, `CYCLE`, `MINVALUE`/`MAXVALUE`) with more flexibility than `IDENTITY`. This maps closely onto PostgreSQL's `SEQUENCE` object (which `SERIAL` is itself built on internally); MySQL has no native sequence object at all, historically relying purely on `AUTO_INCREMENT`.

### 5.5 IDENTITY vs SEQUENCE — When to Use Each

| Scenario | Prefer |
|---|---|
| Simple auto-incrementing single-table primary key | `IDENTITY` |
| A shared number series needed across multiple tables | `SEQUENCE` |
| Values needed *before* the `INSERT` executes (e.g. to use in related records first) | `SEQUENCE` (`NEXT VALUE FOR` can be called independently) |
| Need to control caching/cycling behavior explicitly | `SEQUENCE` |

### 5.6 DEFAULT, NOT NULL, UNIQUE, CHECK

These constraint types behave essentially identically to PostgreSQL/MySQL and are not re-taught from first principles here; the syntax shown is standard T-SQL, directly portable in concept from either database you already know.

### 5.7 A Professional Table-Design Example

Every business rule expressible as a constraint (valid status values, non-negative totals) is enforced directly at the schema level — a professional habit that prevents invalid data regardless of which application, script, or careless ad-hoc query ultimately writes to the table.

## 6. CRUD Operations

### 6.1 INSERT — Single Row, Multiple Rows, From Another Table

### 6.2 UPDATE — Multiple Records, Conditional

`UPDATE ... FROM` is **[T-SQL-specific]** syntax for updating one table based on a join to another; PostgreSQL uses `UPDATE ... SET ... FROM ...` similarly, while MySQL uses `UPDATE t1 JOIN t2 ON ... SET ...` — same capability, genuinely different syntax across all three.

### 6.3 DELETE — Safe Deletion, Related Records

**Common Mistake:** Running an `UPDATE`/`DELETE` without a `WHERE` clause, updating/deleting every row in the table. Professional practice: always run the equivalent `SELECT` with the same `WHERE` clause first to confirm the affected row count before executing the write, and wrap ad-hoc production changes in an explicit transaction (Part 23) that can be rolled back if the row count looks wrong.

### 6.4 OUTPUT — Capturing Inserted, Updated, or Deleted Rows

`OUTPUT` exposes two special pseudo-tables during a DML statement: `inserted` (the new row values, for `INSERT`/`UPDATE`) and `deleted` (the old row values, for `UPDATE`/`DELETE`) — letting a single statement both modify data *and* return exactly what changed, without a separate follow-up `SELECT`. This is directly comparable in purpose to PostgreSQL's `RETURNING` clause, though the `inserted`/`deleted` pseudo-table syntax, and the ability to reference *both* old and new values simultaneously in an `UPDATE`, is distinctly T-SQL. MySQL has no equivalent feature at all.

**Business use case:** capturing exactly which rows were archived/cancelled/updated for audit logging, without a race-condition-prone separate `SELECT` before or after the write.

## 7. SELECT and T-SQL Querying

### 7.1 Core Clauses

`SELECT`, `DISTINCT`, `WHERE`, `ORDER BY`, `GROUP BY`, `HAVING` behave the same as in PostgreSQL/MySQL and are not re-taught here.

### 7.2 TOP — Deep Dive and Comparison

| Feature | SQL Server | PostgreSQL | MySQL |
|---|---|---|---|
| Syntax | `SELECT TOP (n) ... ` (before column list) | `SELECT ... LIMIT n` (after everything) | `SELECT ... LIMIT n` (after everything) |
| Percentage-based limiting | `TOP n PERCENT` | Not built in (compute manually) | Not built in |
| Include ties at the boundary | `TOP n WITH TIES` | Not built in directly (`FETCH ... WITH TIES` in modern PostgreSQL) | Not built in |
| Offset/pagination | `OFFSET ... FETCH` (Section 7.3) | `LIMIT n OFFSET m` | `LIMIT m, n` |

`TOP` must always be paired with `ORDER BY` to be meaningful — without it, "top" rows are returned in whatever arbitrary order the engine happens to produce them, which is a common, subtle correctness bug when porting `LIMIT`-based PostgreSQL/MySQL queries that happened to omit `ORDER BY`.

### 7.3 OFFSET / FETCH — Standard SQL Pagination

Unlike `TOP`, `OFFSET...FETCH` is standard SQL and directly portable — the recommended choice for pagination logic in code intended to work across databases, or simply as the more modern, standards-based idiom even within a SQL-Server-only codebase.

### 7.4 Logical Query Processing Order

```
1. FROM        (including JOINs)
2. WHERE
3. GROUP BY
4. HAVING
5. SELECT      (including DISTINCT's column list is evaluated here)
6. DISTINCT
7. ORDER BY
8. TOP / OFFSET-FETCH
```

This logical order — identical in concept across PostgreSQL, MySQL, and SQL Server, since it derives from the SQL standard, not any vendor — explains many common beginner confusions: a column alias defined in `SELECT` cannot be referenced in `WHERE` (since `WHERE` is logically evaluated *before* `SELECT`), but *can* be referenced in `ORDER BY` (evaluated *after* `SELECT`). **Physical execution** (Part 28-29) does not necessarily follow this order literally — SQL Server's optimizer is free to reorder operations for performance as long as the *logical result* is identical — but reasoning about correctness should always follow the logical order above.

## 8. Operators and Expressions

### 8.1 Arithmetic, Comparison, and Logical Operators

`+ - * / %`, `= <> != < > <= >=`, and `AND OR NOT` all behave as expected from PostgreSQL/MySQL. Note: `<>` and `!=` are both valid "not equal" operators in T-SQL; `<>` is the ANSI-standard form and generally preferred for portability.

### 8.2 IN, BETWEEN, LIKE, IS NULL

### 8.3 Wildcard Patterns

| Pattern | Meaning |
|---|---|
| `%` | Any sequence of zero or more characters. |
| `_` | Any single character. |
| `[abc]` | Any single character within the listed set — **[T-SQL-specific]**; PostgreSQL/MySQL `LIKE` has no bracket-set syntax at all (PostgreSQL uses `SIMILAR TO` or regex `~` for this). |
| `[^abc]` | Any single character **not** in the listed set — also **[T-SQL-specific]**. |
| `[a-f]` | Any single character in the given range. |

T-SQL's bracket-set wildcards inside `LIKE` are a genuinely distinctive feature — a middle ground between plain `%`/`_` wildcards and full regular expressions, useful for simple structured-pattern matching without needing `PATINDEX` (Part 9) or a full regex engine (which T-SQL's native `LIKE` does not support at all, unlike PostgreSQL's `~`).

## 9. T-SQL Functions

### 9.1 String Functions

| Function | Notes |
|---|---|
| `LEN` | Character count, trailing spaces ignored — differs subtly from PostgreSQL's `LENGTH()`, which does not trim. |
| `CHARINDEX(substr, str)` | Like PostgreSQL's `POSITION(substr IN str)` / MySQL's `LOCATE`; finds a literal substring's position. |
| `PATINDEX('%pattern%', str)` | **[T-SQL-specific]** — finds a `LIKE`-style *pattern* (with wildcards) position, not just a literal substring. |
| `TRANSLATE(str, from, to)` | **[T-SQL, SQL Server 2017+]** — character-by-character substitution, like a compact multi-character `REPLACE` in one call. |
| `STRING_AGG(expr, separator)` | **[T-SQL, SQL Server 2017+]** — aggregates values across rows into one delimited string, directly equivalent to PostgreSQL's `STRING_AGG` (same name!) and MySQL's `GROUP_CONCAT`. |
| `STRING_SPLIT(str, separator)` | **[T-SQL, SQL Server 2016+]** — splits a delimited string into rows (a table-valued function); comparable to PostgreSQL's `unnest(string_to_array(...))` or MySQL's lack of a direct built-in equivalent. |

### 9.2 Numeric Functions

Standard across all three databases in concept; `CEILING` (SQL Server's spelling) vs `CEIL` (PostgreSQL/MySQL both accept `CEIL`, and PostgreSQL also accepts `CEILING`) is a minor naming difference worth remembering when porting.

### 9.3 Date Functions

| Function | Notes |
|---|---|
| `DATEADD(unit, n, date)` | Comparable to PostgreSQL's `date + interval '30 days'` or MySQL's `DATE_ADD`; T-SQL's unit-keyword-first syntax is its own. |
| `DATEDIFF(unit, start, end)` | Returns an *integer count* of unit boundaries crossed — a common source of off-by-one surprises (e.g. `DATEDIFF(YEAR, '2025-12-31', '2026-01-01')` returns `1`, despite only one day apart, since it counts calendar-year boundaries crossed, not elapsed time). |
| `DATEDIFF_BIG` | Same as `DATEDIFF` but returns `bigint`, avoiding overflow when counting small units (seconds, milliseconds) across a large date range. |
| `EOMONTH` | **[T-SQL, SQL Server 2012+]** — end-of-month date, directly useful for monthly reporting boundaries without manual date arithmetic. |
| `DATEFROMPARTS`/`DATETIMEFROMPARTS` | **[T-SQL]** — construct a date/datetime from separate integer components, safer and clearer than string concatenation + `CAST`. |

### 9.4 Conversion Functions — CAST, CONVERT, TRY_CAST, TRY_CONVERT

`CAST` is standard SQL; `CONVERT` is **[T-SQL-specific]**, functionally similar but additionally accepting a numeric *style* code (like `120` above) controlling exact output formatting — particularly useful for date/time-to-string conversions, an area PostgreSQL handles instead with `TO_CHAR()`'s format-string approach.

**Why TRY_CAST/TRY_CONVERT matter for data cleaning:** a plain `CAST`/`CONVERT` on invalid input **raises an error and aborts the batch**; `TRY_CAST`/`TRY_CONVERT` instead return `NULL` on failure, letting a data-cleaning query (Part 33) process an entire messy dataset in one pass — flagging or filtering unconvertible rows via `WHERE TRY_CAST(...) IS NULL` — rather than crashing on the first bad row. PostgreSQL has no exact equivalent (community extensions aside); MySQL's implicit, often-silent type coercion behaves quite differently again, making `TRY_CAST`/`TRY_CONVERT` a distinctly valuable, explicit T-SQL tool.

### 9.5 NULL Handling — ISNULL, COALESCE, NULLIF

| Aspect | `ISNULL` | `COALESCE` |
|---|---|---|
| Standard? | **[T-SQL-specific]** | **[Standard SQL]**, portable to PostgreSQL/MySQL |
| Number of arguments | Exactly 2 | 2 or more |
| Return type | Determined by the *first* argument's type | Determined by the *highest-precedence* type among all arguments |
| Performance | Occasionally marginally faster (simpler internally) | Negligible difference in practice |

**Professional recommendation:** prefer `COALESCE` for portability and its more predictable type-resolution rules, reserving `ISNULL` for cases specifically needing its particular type-preservation behavior — a subtlety that occasionally matters when the fallback value's implied type differs from the primary expression's.

## 11. Joins

### 11.1 Join Types Overview

Join mechanics (`INNER`, `LEFT`, `RIGHT`, `FULL OUTER`, `CROSS`, self-joins) are conceptually identical across PostgreSQL, MySQL, and SQL Server — the relational algebra doesn't change — so this part focuses on T-SQL syntax specifics and behavior worth confirming explicitly, rather than re-deriving joins from scratch.

**MySQL note:** MySQL did not support `FULL OUTER JOIN` directly until relatively recent behavior workarounds (traditionally simulated via `UNION` of `LEFT JOIN` and `RIGHT JOIN`); SQL Server has always supported it natively, same as PostgreSQL.

### 11.2 Relationship Cardinality and Duplicate Rows

A **one-to-many** join (one customer, many orders) naturally produces one output row per matching order — entirely expected. A **many-to-many** join (through an intermediate table, like `OrderItems` linking `Orders` and `Products`) or an unintentionally imprecise join condition can silently multiply rows in ways that inflate `SUM()`/`COUNT()` aggregates — one of the most common real-world query correctness bugs, entirely independent of which SQL dialect is used. Always sanity-check aggregate results against an independent row count when joins are involved.

### 11.3 NULL Behavior in Joins

`LEFT`/`RIGHT`/`FULL OUTER JOIN` populate unmatched columns with `NULL` — and critically, filtering on those `NULL`-able columns in `WHERE` (rather than in the `ON` clause) can silently convert an outer join back into an inner join, since `WHERE SomeOuterColumn = 'X'` is never true when `SomeOuterColumn IS NULL`. Move such filters into the `ON` clause when the intent is genuinely to preserve unmatched rows.

## 13. CTEs

### 13.1 Basic CTE — WITH

A CTE (Common Table Expression) is standard SQL, identical in syntax and purpose to PostgreSQL/MySQL 8+ — a named, temporary result set scoped to the single statement that follows it, primarily used for readability by breaking a complex query into named logical steps.

### 13.2 Multiple CTEs and CTE Chains

Each CTE can reference any CTE defined *earlier* in the same `WITH` clause, chaining transformations into clearly named stages — dramatically more readable than deeply nested subqueries for multi-step analytical logic.

### 13.3 Recursive CTEs

| Part | Explanation |
|---|---|
| **Anchor member** | The starting point — rows with no manager (the top of the hierarchy). |
| `UNION ALL` | Combines the anchor with each successive recursive pass; must be `UNION ALL`, not `UNION`, in SQL Server's recursive CTE syntax. |
| **Recursive member** | References `OrgChart` itself, joining each pass's newly found rows to find the *next* level down; SQL Server automatically repeats this until no new rows are found. |

Recursive CTEs are standard SQL, working nearly identically in PostgreSQL (`WITH RECURSIVE`, note the required `RECURSIVE` keyword PostgreSQL needs that SQL Server does not) — a genuinely important syntax difference to remember when porting. MySQL added recursive CTE support only in version 8.0+.

**Business use cases:** organizational charts (above), bill-of-materials explosion (a product made of sub-parts, themselves made of sub-parts), date/number sequence generation, and graph-like relationship traversal (finding all paths between connected nodes).

`OPTION (MAXRECURSION n)` is **[T-SQL-specific]** — SQL Server defaults to a maximum of 100 recursion levels as a safety limit against runaway/infinite recursive CTEs; this must be raised explicitly (or set to `0` for unlimited, with caution) for a recursion depth genuinely expected to exceed 100.

## 15. Temporary Objects

### 15.1 Local Temporary Tables — #Table

A `#Table` is **[T-SQL-specific]** syntax (the `#` prefix) — visible only within the creating session (or, if created inside a stored procedure, that procedure's scope and anything it calls), physically stored in `tempdb` (Part 40), and automatically dropped when the session/scope ends. PostgreSQL uses `CREATE TEMP TABLE`/`CREATE TEMPORARY TABLE` (no special prefix, session-scoped); MySQL uses `CREATE TEMPORARY TABLE` similarly.

### 15.2 Global Temporary Tables — ##Table

A `##Table` (double-hash prefix) is visible to **every** session on the instance, and persists until the creating session ends *and* no other session is still referencing it — a genuinely SQL Server-specific concept with no direct PostgreSQL/MySQL equivalent; used rarely, mainly for cross-session coordination in specific administrative scenarios.

### 15.3 Table Variables

A table variable (`@Name TABLE (...)`) is scoped to the current batch/procedure, cannot have most explicit indexes added after creation (only inline `PRIMARY KEY`/`UNIQUE` constraints, which do create supporting indexes), and — importantly — has historically been assumed by the optimizer to hold exactly one row for cardinality estimation purposes in older SQL Server versions (improved by "table variable deferred compilation" in SQL Server 2019+, but still generally less statistics-aware than a temp table).

### 15.4 Temporary Table Indexing

Temp tables support full indexing, statistics, and constraints just like permanent tables — a key practical advantage over table variables for larger intermediate result sets where the optimizer's query plan quality genuinely benefits from real statistics.

### 15.5 Temporary Table vs Table Variable vs CTE — When Professionals Use Each

| Situation | Prefer |
|---|---|
| Small, simple intermediate set, single statement | CTE |
| Intermediate set reused across multiple statements, small-to-medium size | Table variable |
| Large intermediate set, needs indexing, benefits from real statistics for the optimizer | Temp table (`#Table`) |
| Result set needs to survive and be queried across multiple batches within a stored procedure with complex logic | Temp table |
| Rollback behavior matters: table variables are **not** rolled back by an enclosing transaction rollback (a genuinely distinctive, sometimes-useful SQL Server behavior), while temp tables *are* rolled back | Table variable, when audit/logging data must survive a rolled-back transaction |

## 16. APPLY Operators

### 16.1 CROSS APPLY and OUTER APPLY — SQL Server-Specific

### 16.2 How APPLY Differs From Ordinary JOINs

A standard `JOIN`'s `ON` condition cannot reference a *derived, per-row computed table* on its right side — `APPLY` specifically allows the right-hand table expression to be a subquery (or table-valued function) that is **re-evaluated once per row of the left side**, with full access to that row's columns. This is precisely what makes the "latest order per customer" pattern above possible in a single query — a plain `JOIN` to a `TOP 1 ... ORDER BY` subquery cannot correlate that `TOP 1` to each specific customer the way `CROSS APPLY` can. PostgreSQL's closest equivalent is a `LATERAL JOIN` (`JOIN LATERAL (...)`), which serves the same purpose with different syntax; MySQL added `LATERAL` derived table support only in version 8.0.14+, and it remains less commonly used there than `CROSS APPLY` is in SQL Server code.

### 16.3 Advanced Examples

`CROSS APPLY`/`OUTER APPLY` combined with `OPENJSON` (Part 31) or a table-valued function are the standard, idiomatic T-SQL way to "unnest" or "explode" one row into several related rows computed per-row — a pattern PostgreSQL developers would reach for `LATERAL` to accomplish, and MySQL developers would often resort to a correlated subquery or application-side logic for, historically.

## 17. PIVOT and UNPIVOT

### 17.1 PIVOT — Rows Into Columns

`PIVOT` transforms distinct values from one column (`OrderYear`) into separate output columns, aggregating (`SUM`) the measure for each — turning "tall" data (one row per order) into a "wide" report layout (one row per category, one column per year), directly useful for financial/BI reporting tables. PostgreSQL has no native `PIVOT` keyword at all (achieved instead via `CASE`-based conditional aggregation, or the `crosstab()` function from the `tablefunc` extension); MySQL likewise has no native `PIVOT`, relying on the same `CASE`-based manual approach.

### 17.2 UNPIVOT — Columns Into Rows

`UNPIVOT` performs the reverse transformation — converting several columns back into rows, useful for normalizing a wide reporting table back into a "tall", analysis-friendly shape.

### 17.3 Dynamic PIVOT Concepts

Because `PIVOT`'s column list must be known at parse time, a **dynamic** pivot (where the actual set of column values isn't known until runtime — e.g. "however many distinct years exist in the data") requires building and executing the query as a string via dynamic SQL (Part 36) — `QUOTENAME()` here safely brackets each generated column name, and `sp_executesql` is the safe way to execute the constructed statement.

**Business use cases:** monthly/yearly financial summary reports, sales-by-region-by-quarter dashboards, survey-response cross-tabulations — any report whose natural presentation is a matrix/grid rather than a tall table.

### 18.2 Duplicate Handling, Column Compatibility, Performance

## 19. Views

### 19.1 CREATE / ALTER / DROP VIEW

### 19.2 Why Views Exist

Views provide **security** (granting access to a view without exposing underlying base tables' full column set or row set — see Part 35's row-level security for a more granular alternative), **abstraction** (hiding complex join/aggregation logic behind a simple `SELECT * FROM view`), **reusability** (one definition, queried by many reports/applications), and simplified **reporting** access for consumers (like Power BI, Part 52) who should query a stable, well-named interface rather than raw, potentially-changing base tables.

### 19.3 Standard Views vs Indexed Views

A standard view is just stored SQL text, re-executed against current data every time it's queried. An **indexed view** (created with `WITH SCHEMABINDING`, then given a `UNIQUE CLUSTERED INDEX`) physically materializes and persists its result set on disk, automatically kept in sync as underlying data changes — SQL Server's equivalent of PostgreSQL's `MATERIALIZED VIEW`, with one crucial difference: **a SQL Server indexed view updates automatically and incrementally with every underlying data change**, while a PostgreSQL materialized view requires an explicit `REFRESH MATERIALIZED VIEW` command and is stale until then. MySQL has no materialized/indexed view feature at all.

### 19.4 Requirements and Limitations of Indexed Views

`SCHEMABINDING` locks the view's definition to the exact current structure of its underlying tables (preventing those tables from being altered in incompatible ways without first dropping the view); the view's `SELECT` cannot use `SELECT *`, non-deterministic functions, outer joins, or several other constructs; `COUNT(*)` must be written as `COUNT_BIG(*)`; and the first index created must be a `UNIQUE CLUSTERED` index specifically. These restrictions exist because the engine must be able to maintain the materialized result incrementally and deterministically on every underlying write — a genuine engineering trade-off between query-time speed and both write-time overhead and definitional flexibility.

## 22. Triggers

### 22.1 DML Triggers — AFTER and INSTEAD OF

Triggers use the same `inserted`/`deleted` pseudo-tables introduced with `OUTPUT` in Part 6 — a nice syntactic consistency within T-SQL. `AFTER` triggers (the more common type) run once the triggering statement's changes are already applied; `INSTEAD OF` triggers intercept the action entirely, running their own logic in its place — most commonly used to make an otherwise non-updatable view (one involving joins or aggregation) support `INSERT`/`UPDATE`/`DELETE`.

### 22.2 DDL Triggers

A DDL trigger fires on schema-changing statements (`CREATE`/`ALTER`/`DROP`) rather than data-changing ones — used for governance: preventing accidental structural changes, or logging every schema modification for compliance auditing.

### 22.3 Practical Use Cases

- **Audit trails:** recording who changed what and when (Section 22.1's example), often superseded in modern SQL Server by Temporal Tables (Part 42) or Change Data Capture (Part 43) for this specific purpose.
- **Change tracking:** propagating a change in one table to a derived/denormalized value in another.
- **Security:** DDL triggers restricting schema changes; DML triggers enforcing row-level business rules beyond what a `CHECK` constraint can express.
- **Business rules:** enforcing cross-table invariants a single-table `CHECK` constraint cannot see (e.g. "an order's total must equal the sum of its line items").

### 22.4 Why Excessive Trigger Usage Can Make Systems Difficult to Maintain

Triggers execute *implicitly* — invisible in the calling application code or ad-hoc query that triggered them — which means a simple-looking `UPDATE` statement can silently cascade into arbitrarily complex hidden logic, additional writes, or even other triggers firing recursively. This makes debugging unexpected behavior and reasoning about performance significantly harder than equivalent logic made *explicit* in a stored procedure or application layer. **Professional guidance:** prefer explicit logic in stored procedures/application code where practical; reserve triggers for genuine cross-cutting concerns (audit, enforced invariants) that must apply unconditionally regardless of which code path performs the write, and keep trigger logic itself as simple and fast as possible.

## 24. Transaction Isolation

### 24.1 Isolation Levels

| Level | Prevents Dirty Reads? | Prevents Non-Repeatable Reads? | Prevents Phantom Reads? | Mechanism |
|---|---|---|---|---|
| Read Uncommitted | No | No | No | No locking on reads at all |
| Read Committed (SQL Server default) | Yes | No | No | Shared locks held only during the individual read |
| Repeatable Read | Yes | Yes | No | Shared locks held until end of transaction |
| Serializable | Yes | Yes | Yes | Range locks preventing any concurrent modification of the read range |
| Snapshot | Yes | Yes | Yes | Row versioning (Section 24.4), not locking |
| Read Committed Snapshot Isolation (RCSI) | Yes | No | No | Row versioning at the statement level |

### 24.2 Dirty Reads, Non-Repeatable Reads, Phantom Reads

| Anomaly | Description |
|---|---|
| **Dirty read** | Reading another transaction's *uncommitted* changes, which might later be rolled back. |
| **Non-repeatable read** | Re-reading the same row within one transaction returns *different* data because another transaction committed a change in between. |
| **Phantom read** | Re-running the same query within one transaction returns *additional or fewer rows* because another transaction inserted/deleted matching rows in between. |

### 24.3 Blocking, Lock Escalation, Deadlocks

**Blocking** occurs when one transaction holds a lock another transaction needs, forcing it to wait. **Lock escalation** is SQL Server automatically converting many fine-grained row/page locks into a single coarser table lock once a threshold is exceeded, trading concurrency for lock-management overhead. A **deadlock** occurs when two transactions each hold a lock the other needs, forming a cycle with no possible resolution — SQL Server automatically detects deadlocks and terminates one transaction (the "deadlock victim", chosen by estimated rollback cost) to break the cycle, returning error 1205 to the terminated transaction's caller, whose application code should be prepared to retry.

### 24.4 SQL Server Locking and Row-Versioning — In Depth

By default (`Read Committed`), SQL Server uses **pessimistic, lock-based** concurrency control: readers take shared locks, writers take exclusive locks, and readers can block behind writers. Enabling **Read Committed Snapshot Isolation (RCSI)** at the database level, or explicitly using the `SNAPSHOT` isolation level, switches to **optimistic, row-versioning-based** concurrency: readers see a consistent, appropriately-timestamped *version* of each row from `tempdb`'s version store (Part 40) instead of waiting on a writer's lock at all — dramatically reducing reader/writer blocking, at the cost of `tempdb` version-store overhead and the possibility of write-write conflicts under `SNAPSHOT` isolation specifically (raising error 3960, requiring a retry).

**[SQL Server vs PostgreSQL/MySQL]:** PostgreSQL uses MVCC (Multi-Version Concurrency Control) — row versioning — as its *default*, always-on concurrency model, meaning PostgreSQL readers essentially never block behind writers out of the box; SQL Server's traditional default (`Read Committed` without RCSI) is lock-based and *can* experience reader/writer blocking unless RCSI is explicitly enabled — a genuinely important operational difference to know when moving between the two systems. MySQL/InnoDB also defaults to MVCC-based `REPEATABLE READ` isolation, again differing from SQL Server's out-of-the-box locking behavior.

## 25. Error Handling

### 25.1 TRY / CATCH

`BEGIN TRY`/`BEGIN CATCH` (**[T-SQL-specific]**) is structurally similar to PostgreSQL's `EXCEPTION` block within a `DO`/function body, and conceptually parallel to application-level `try`/`except` (as covered for Python in the Testing and Debugging handbook's Part 9) — but scoped specifically to catchable T-SQL errors, and requiring explicit `@@TRANCOUNT` checking and rollback, unlike PostgreSQL's automatic transaction rollback on an unhandled exception within a block.

### 25.2 Error Information Functions

| Function | Returns |
|---|---|
| `ERROR_NUMBER()` | The numeric error code. |
| `ERROR_MESSAGE()` | The full error message text. |
| `ERROR_LINE()` | The line number within the batch/procedure where the error occurred. |
| `ERROR_PROCEDURE()` | The name of the stored procedure/function where the error occurred (`NULL` if not inside one). |
| `ERROR_SEVERITY()` | The error's severity level (16 is a common "user-correctable" default). |
| `ERROR_STATE()` | A caller-defined state value, useful for distinguishing multiple raise points sharing the same error number. |

These are only valid **inside a `CATCH` block** — calling them elsewhere returns `NULL`.

### 25.3 THROW vs Older Patterns

**Why `THROW` is generally preferred:** a bare `THROW` (no arguments) inside a `CATCH` block re-raises the *original* error with its original line number, procedure name, and severity fully intact — `RAISERROR` cannot do this at all, only generate a *new* error referencing the `CATCH` block's own location, losing the original context. `THROW` with arguments also always terminates the batch (severity is effectively always 16, "error"), giving simpler, more predictable behavior than `RAISERROR`'s configurable severity levels, most of which are rarely appropriate for application error signaling. Use `RAISERROR` only when its specific legacy features (like `WITH NOWAIT` for immediate message flushing, useful in some long-running-job logging scenarios) are genuinely required.

## 26. Indexes

### 26.1 Heap

A table with no clustered index is called a **heap** — rows are stored in no particular physical order, referenced by other indexes via a physical row ID. Generally, professional guidance recommends every table have a clustered index (Section 26.2) rather than remaining a heap, since a heap can suffer from page fragmentation and lacks the ordered-access benefits a clustered index provides.

### 26.2 Clustered Index

A clustered index determines the **physical storage order** of the table's data itself — the table's rows *are* the index's leaf level; there can be at most **one** clustered index per table (since data can only be physically sorted one way). By default, `PRIMARY KEY` creates a clustered index automatically unless `NONCLUSTERED` is specified.

### 26.3 Nonclustered Index

A nonclustered index is a separate structure, holding a copy of the indexed column(s) plus a pointer back to the full row (the clustering key, if the table has a clustered index, or a physical row locator if it's a heap) — a table can have many nonclustered indexes (up to 999), each optimized for a different query pattern.

### 26.4 Clustered vs Nonclustered — And Comparison With PostgreSQL/MySQL

| Aspect | SQL Server | PostgreSQL | MySQL (InnoDB) |
|---|---|---|---|
| Default table storage | Heap, unless a clustered index is created | Heap-like (unless `CLUSTER` is run once, non-maintained afterward) | Always clustered by primary key (InnoDB's fundamental storage model) |
| Clustered index maintenance | Automatically, continuously maintained | `CLUSTER` command is a one-time physical reorder, not continuously maintained | Automatic and continuous (built into InnoDB) |
| Max clustered indexes per table | 1 | N/A (no true continuously-maintained clustered index concept) | Always exactly 1 (the primary key) |

This is a genuinely important conceptual difference: MySQL/InnoDB tables are *always* clustered by primary key with no choice in the matter; SQL Server gives you the explicit choice of clustering key (commonly, but not necessarily, the primary key) and continuously maintains it; PostgreSQL's `CLUSTER` is a one-time physical reorganization that immediately begins drifting out of sync as new rows are inserted, unlike SQL Server's always-current clustered index.

### 26.5 Unique, Composite, Included Columns, Filtered, Covering Indexes

**Included columns** let a nonclustered index become a **covering index** for a specific query — one where every column the query needs (in `SELECT`, `WHERE`, `JOIN`) is present directly in the index itself, letting the engine satisfy the entire query from the (typically much smaller) index alone, without a **key lookup** (Section 26.6) back to the full table row. PostgreSQL added a comparable `INCLUDE` clause for indexes in version 11+; MySQL has no direct equivalent concept, though a composite index can serve a similar covering purpose if all needed columns happen to be part of the key itself.

**Filtered indexes** index only rows matching a `WHERE` predicate — smaller, faster to maintain, and more selective for queries that always filter on that same condition (like "only pending orders"). PostgreSQL's `CREATE INDEX ... WHERE ...` (a "partial index") is functionally identical; MySQL has no equivalent feature at all.

### 26.6 Index Seek vs Index Scan vs Key Lookup

| Operation | Meaning |
|---|---|
| **Index seek** | The engine navigates the index's B-tree structure directly to the needed row(s) — efficient, typically `O(log n)`. |
| **Index scan** | The engine reads the *entire* index (or table) sequentially — can still be efficient for large result-set queries, but expensive when only a few rows are actually needed. |
| **Key lookup** | After a nonclustered index seek finds matching rows, the engine must go back to the clustered index (or heap) to retrieve any *additional* columns not included in the nonclustered index — an extra, potentially costly random-access step per row, exactly what covering indexes (Section 26.5) eliminate. |

Seeing many key lookups in an execution plan (Part 29) for a query touching many rows is a strong, actionable signal that adding `INCLUDE` columns to make the index "covering" would likely improve performance meaningfully.

### 26.7 B-Tree Structure and Index Selectivity

Both clustered and nonclustered indexes in SQL Server are implemented as **B-trees** — balanced tree structures allowing `O(log n)` seeks — conceptually identical to PostgreSQL's default B-tree index type and MySQL/InnoDB's index structure. **Selectivity** measures what fraction of a table's rows a given predicate typically matches; a highly selective predicate (`WHERE Email = 'specific@example.com'`, matching ~1 row) benefits enormously from an index seek, while a low-selectivity predicate (`WHERE IsActive = 1`, matching 95% of rows) often performs *better* with a table scan than an index seek plus enormous numbers of key lookups — the optimizer's cost-based decision here is exactly why "just add an index" is not always the right answer (Part 30).

### 26.8 SARGability and Composite Index Column Ordering

Covered fully in Part 30; briefly: a predicate is **SARGable** (Search ARGument-able) when the engine can use an index to seek directly to matching rows, rather than needing to evaluate a function/expression against every row. Composite index column order should generally place the most selective and/or most commonly-filtered-with-equality column first, since a composite index can only be seek-efficient on a *left-to-right prefix* of its columns — an index on `(CustomerID, OrderDate)` seeks efficiently for `WHERE CustomerID = 5`, and for `WHERE CustomerID = 5 AND OrderDate > '2026-01-01'`, but **not** efficiently for `WHERE OrderDate > '2026-01-01'` alone.

### 26.9 Over-Indexing and Missing Indexes

Every additional index speeds up *some* reads but slows down **every** `INSERT`/`UPDATE`/`DELETE` that touches indexed columns (since each index must also be maintained on every write) and consumes additional storage — over-indexing a heavily-written table is a genuine, common performance mistake (Part 61). SQL Server's Dynamic Management Views (`sys.dm_db_missing_index_details` and related DMVs) surface indexes the optimizer *would* have used had they existed, based on actual query history — a valuable, data-driven starting point for index design, though missing-index suggestions should always be evaluated critically (Section 30) rather than applied blindly, since they don't account for the write-cost trade-off above.

## 27. Columnstore

### 27.1 Rowstore vs Columnstore

Traditional indexes (Part 26) are **rowstore**: all of a row's column values are stored physically together, efficient for retrieving/modifying whole rows (typical OLTP access patterns). A **columnstore** index instead stores each *column's* values together, contiguously, across all rows — dramatically more efficient for analytical queries that aggregate over millions of rows but touch only a handful of columns, since only the needed columns' data need be read from disk at all, and highly repetitive column data compresses far better stored this way.

### 27.2 Clustered and Nonclustered Columnstore Indexes

A **clustered columnstore index** replaces the table's entire storage with columnar format — the standard choice for dedicated fact tables (Part 51) in a data warehouse. A **nonclustered columnstore index** adds a columnar copy of specified columns *alongside* a table's normal rowstore storage, useful for enabling fast analytical queries against an otherwise transactional (rowstore) table without migrating its primary storage format.

### 27.3 Compression and Why Columnstore Matters for Analytics

Columnstore indexes use aggressive compression (often 5-10x smaller than equivalent rowstore data) since column-aligned data tends to be highly repetitive (a `Status` column with only 4 distinct values compresses far better stored column-wise than interleaved with unrelated row data), and SQL Server executes analytical operations against columnstore data using **batch mode** processing (operating on ~900-row batches at a time, vectorized, rather than row-by-row) — the combination of far less I/O (only needed columns read, heavily compressed) and vectorized execution is what makes columnstore indexes routinely deliver order-of-magnitude speedups for `GROUP BY`/`SUM`/`AVG`-heavy analytical workloads compared to rowstore, at the cost of being comparatively inefficient for single-row lookups and frequent small updates — precisely why columnstore is recommended for data-warehouse fact tables (Part 51) and analytical workloads, not OLTP tables with heavy transactional row-level activity.

**[SQL Server vs PostgreSQL/MySQL]:** Neither PostgreSQL nor MySQL includes a native columnstore index type in their core engine (PostgreSQL has third-party columnar extensions like `cstore_fdw`/Citus; MySQL has no first-party equivalent at all) — this is one of SQL Server's most significant built-in analytical-performance advantages over the two, and a major reason SQL Server is frequently chosen as the engine underneath an in-house data warehouse (Part 51).

## 28. Query Performance

### 28.1 Execution Plans — Estimated vs Actual

An **estimated execution plan** shows what the optimizer *plans* to do, based on statistics, without actually running the query; an **actual execution plan** runs the query and additionally reports real row counts and timings alongside the same plan shape — always prefer the actual plan when diagnosing a genuinely slow query, since large gaps between *estimated* and *actual* row counts (Section 29's "poor cardinality estimates") are themselves one of the most common root causes of bad plans.

```
Table 'Orders'. Scan count 1, logical reads 4521, physical reads 0, read-ahead reads 0.
SQL Server Execution Times: CPU time = 156 ms, elapsed time = 203 ms.
```

`SET STATISTICS IO ON` reports **logical reads** (pages read from the buffer cache/memory — the most important number to watch, since it's a stable, hardware-independent measure of query cost) and **physical reads** (pages actually read from disk, indicating the buffer cache didn't already have the data). `SET STATISTICS TIME ON` reports actual CPU and elapsed time. Together, these are the standard first diagnostic step for any "why is this query slow" investigation, cheaper and often more immediately actionable than opening a full graphical execution plan.

### 28.2 Query Cost and Cardinality Estimation

The optimizer assigns every candidate execution plan an estimated **cost** (a unitless number combining estimated I/O and CPU work) and selects the cheapest plan it can find within a bounded search — it does not guarantee the *globally* optimal plan, only a "good enough, found quickly" one. **Cardinality estimation** — predicting how many rows each operation will produce — drives nearly every downstream decision (which join algorithm, whether to use an index, how much memory to grant); accurate cardinality estimation depends entirely on up-to-date **statistics** (Section 28.3).

### 28.3 Statistics

Statistics are compact summaries (histograms) of a column's data distribution, automatically created and maintained (`AUTO_CREATE_STATISTICS`/`AUTO_UPDATE_STATISTICS`, on by default) as data changes — stale statistics (common after a large bulk load without a subsequent update) are one of the single most common root causes of a previously-fast query suddenly performing poorly, since the optimizer's cardinality estimates become wrong.

### 28.4 Join Algorithms

| Algorithm | Best Suited For |
|---|---|
| **Nested Loops** | Small outer input joined to an index-seekable inner input — very efficient when applicable. |
| **Hash Join** | Large, unsorted inputs with no useful index — builds an in-memory hash table from the smaller input, probes it with the larger. |
| **Merge Join** | Both inputs already sorted on the join key (e.g. via an index) — very efficient, avoids building a hash table entirely. |

The optimizer chooses among these automatically based on estimated input sizes and available indexes/sort order; seeing an unexpected Hash Join where a Nested Loops join was expected (or vice versa) in an execution plan is often a symptom, not a cause — usually traceable back to missing indexes or poor cardinality estimates (Part 29) rather than something to "fix" by forcing a different join type directly.

### 28.5 Sort Operations, Memory Grants, Spills, Parallelism

A **sort** operation (for `ORDER BY`, some `GROUP BY` strategies, or merge joins needing sorted input) requires memory, requested in advance as a **memory grant** based on cardinality estimates; when the actual data is larger than the granted memory (often because the underlying cardinality estimate was wrong), the sort **spills** to `tempdb` (Part 40) — visible in an actual execution plan as a "Sort Warning" and a dramatic performance cliff, since disk-based spilling is vastly slower than an in-memory sort. **Parallelism** (multiple CPU cores cooperating on one query) can speed up large analytical queries substantially but carries its own coordination overhead, making it inappropriate (and, by default configuration, avoided by the optimizer) for small, fast OLTP-style queries.

### 28.6 Diagnosing Slow Queries — A Practical Methodology

1. Confirm the query is actually slow in isolation (rule out application-side or network factors first).
2. Capture the **actual** execution plan (Section 28.1), not just the estimated one.
3. Compare estimated vs actual row counts at each operator — large discrepancies point to stale statistics or non-SARGable predicates (Part 30).
4. Check `SET STATISTICS IO ON` output for unexpectedly high logical reads on any one table.
5. Look for key lookups (Section 26.6), sort spills (Section 28.5), or unexpected scans where a seek would be expected.
6. Only then consider specific fixes: new/adjusted indexes, updated statistics, or query rewriting (Part 30) — in that order of investigation, never index changes as a first, unverified guess.

## 29. Explaining Execution Plans

### 29.1 Reading a Plan From Right to Left, Top to Bottom

SQL Server execution plans are read **right to left**, with data logically "flowing" from the rightmost (typically a table/index access) operators toward the leftmost (final `SELECT`) operator — each operator's **relative cost percentage** (shown as a percentage of the total batch cost) is the first place to look for the actual bottleneck within a larger plan.

### 29.2 A Good Execution Plan Example

A "good" plan for a highly selective, point-lookup-style query shows an **Index Seek** with a small estimated row count matching the actual row count closely, minimal or no key lookups, and low relative cost.

### 29.3 A Bad Execution Plan Example — Missing Index

Without a usable index, the engine must scan the entire table (or clustered index), evaluating the `WHERE` predicate against every single row — the relative cost is overwhelmingly dominated by this one operator, and the fix (adding an index on `CustomerID`) is directly suggested by SQL Server's "Missing Index" recommendation, often shown directly in the graphical plan.

### 29.4 Non-SARGable Predicates and Implicit Conversions

`YEAR(OrderDate) = 2026` forces the engine to evaluate `YEAR()` against *every row* before it can compare — the column being wrapped in a function makes the predicate **non-SARGable**, and the plan shows a scan instead of a seek even with a perfectly good index on `OrderDate` present. This is directly visible in the execution plan as a scan operator with a `Predicate` (post-filter) rather than a `Seek Predicate`. Implicit conversions (comparing mismatched types, especially `varchar` vs `nvarchar`, or numeric vs string) cause the same problem and are frequently *invisible* unless specifically looked for — SQL Server's plan XML/warnings do flag them, and it's a standard, high-value item to check when a seemingly well-indexed query still scans.

### 29.5 Parameter-Sensitive Behavior ("Parameter Sniffing")

A stored procedure's execution plan is typically compiled once and cached based on the **first** parameter value(s) it happened to be called with — if that first call was highly atypical (e.g. a customer with an unusually huge order history), the resulting cached plan can perform very poorly for subsequent, more typical calls (or vice versa) — a genuinely SQL-Server-specific operational concern known as "parameter sniffing." Mitigations include `OPTION (RECOMPILE)` (accepting recompilation cost for a plan tailored to each call's actual parameters), `OPTIMIZE FOR` hints, or (in modern SQL Server) Parameter-Sensitive Plan (PSP) optimization, which can automatically maintain multiple cached plan variants for different parameter value ranges.

### 29.6 Practical Optimization Methodology, Not Just "Add an Index"

Genuine query optimization methodology, in order: (1) confirm the query is logically correct and returns the minimum necessary data (Part 30); (2) verify statistics are current; (3) identify non-SARGable predicates and implicit conversions and rewrite them; (4) *then* evaluate whether an index change would help, informed by the actual plan's bottleneck, not a reflexive guess; (5) re-measure with `SET STATISTICS IO/TIME ON` after each change to confirm genuine improvement rather than assuming it.

## 30. Query Optimization

### 30.1 SARGability — Practical Rewrites

The general rule: **never wrap an indexed column in a function or arithmetic expression** if the predicate needs to remain seekable; instead, transform the constant/comparison side of the expression to isolate the raw column.

### 30.2 Selecting Only Required Columns; Avoiding Unnecessary DISTINCT

`DISTINCT` used to paper over row duplication caused by a join (Section 11.2) rather than addressing the join itself both hides the real issue and forces an expensive deduplication pass (Section 18.2) that `EXISTS` avoids entirely.

### 30.3 Join Optimization, Statistics Maintenance, Parameterization

Ensuring join columns are indexed appropriately (Part 26), keeping statistics current (`UPDATE STATISTICS`, or relying on `AUTO_UPDATE_STATISTICS`), and using **parameterized queries** (both for the SQL-injection-prevention reasons covered in Part 36, and because parameterized queries allow the same cached execution plan to be reused across calls instead of triggering a fresh, costly compilation for every literal-value variation) are all foundational, broadly-applicable optimization practices.

### 30.4 Temp Table Strategies and Set-Based Operations vs Cursors

SQL Server (like PostgreSQL and MySQL) is fundamentally optimized for **set-based** operations — processing many rows at once as a single relational operation — not row-by-row procedural iteration. A cursor forces the engine to process one row at a time, with per-row overhead, and is almost always dramatically slower than an equivalent set-based rewrite for anything that *can* be expressed set-based (which is the large majority of business logic, once the habit of thinking in sets rather than loops is developed). Cursors remain occasionally appropriate for genuinely sequential, stateful administrative tasks (iterating databases for a maintenance script) but should be treated as a last resort for ordinary business-logic queries.

### 30.5 Before/After Optimization Example

Measuring `SET STATISTICS IO` logical reads before and after such a rewrite is the concrete, quantifiable way to confirm and communicate the actual improvement achieved — "logical reads dropped from 4,521 to 8" is a far more convincing (and diagnosable) claim than "it feels faster."

## 34. Database Design and Normalization

### 34.1 Entities, Attributes, Relationships, Keys, ER Diagrams

These foundational relational-modeling concepts (identical across any RDBMS, including PostgreSQL and MySQL) are assumed already familiar; this part focuses on how they inform practical SQL Server table design decisions rather than re-deriving ER modeling from first principles.

### 34.2 Normal Forms — 1NF, 2NF, 3NF, BCNF

| Form | Requirement |
|---|---|
| **1NF** | Every column holds a single, atomic value; no repeating groups or arrays in one column. |
| **2NF** | 1NF, plus every non-key column depends on the **whole** primary key (relevant only for composite keys). |
| **3NF** | 2NF, plus no non-key column depends on *another non-key column* (no "transitive" dependency). |
| **BCNF** | A stricter version of 3NF, addressing certain edge cases involving multiple overlapping candidate keys. |

### 34.3 Normalization vs Denormalization

**Normalization** minimizes data redundancy and update anomalies by splitting data into well-structured related tables — the standard approach for OLTP systems (Section 34.4) where data integrity and efficient, small transactional writes matter most. **Denormalization** deliberately reintroduces redundancy (duplicating data across tables, or pre-aggregating values) to reduce the number of joins a *read*-heavy analytical query needs — the standard approach for OLAP/data-warehouse design (Part 51), where read performance for large aggregate queries outweighs the write-side and storage costs of redundancy.

### 34.4 OLTP Design vs OLAP Design

| Aspect | OLTP (Transactional) | OLAP (Analytical) |
|---|---|---|
| Primary workload | Many small, fast reads/writes | Fewer, large, complex aggregate reads |
| Schema shape | Normalized (3NF+) | Denormalized (star/snowflake schema, Part 51) |
| Typical index strategy | Many targeted nonclustered indexes for specific lookups | Fewer, wider indexes; heavy columnstore use (Part 27) |
| Example | An e-commerce checkout system | A sales data warehouse used for BI reporting |

SQL Server is routinely used for **both** roles — often literally the same product, configured and modeled differently — a genuinely practical reason to understand this distinction deeply as a SQL Server professional working across both backend development (Part 53) and analytics/warehousing (Parts 47, 51) contexts.

## 35. SQL Server Security

### 35.1 Logins, Users, Roles, Schemas, Permissions

### 35.2 GRANT, DENY, REVOKE

`DENY` is worth calling out specifically: it **always wins** over any `GRANT`, from any role or source — a genuinely important behavior for enforcing hard security boundaries that cannot be accidentally overridden by adding a user to some other permissive role later. `REVOKE` merely removes a specific grant/denial, returning to a neutral "no explicit permission" state, which is different from actively denying.

### 35.3 Server-Level vs Database-Level vs Object-Level Security, Least Privilege

| Level | Example |
|---|---|
| **Server-level** | `CREATE LOGIN`, server roles (`sysadmin`, `dbcreator`) — controls what a login can do across the *entire instance*. |
| **Database-level** | Database roles (`db_datareader`, `db_datawriter`, custom roles) — controls access *within one specific database*. |
| **Object-level** | `GRANT SELECT ON Sales.Orders` — the finest-grained permission, on a specific table/view/procedure. |

**Least privilege** — granting a login/user only the exact permissions its role genuinely requires, nothing more — is the foundational security principle underlying all of Part 35; an application's service account, for instance, should almost never hold `sysadmin` or `db_owner`, even though it would technically "work" with those broad permissions.

### 35.4 Row-Level Security

Row-Level Security (**[SQL Server feature]**, introduced 2016) transparently filters (or blocks writes to) rows based on a predicate function evaluated per query — a salesperson querying `Sales.Customers` automatically sees only their own region's rows, with **no application code changes required at all**, since the filtering happens inside the engine itself for every query, regardless of which application or ad-hoc tool issues it. PostgreSQL has a very similar native `ROW LEVEL SECURITY` feature (`CREATE POLICY`); MySQL has no native equivalent, typically requiring views or application-level filtering instead.

### 35.5 Dynamic Data Masking

Dynamic Data Masking (**[SQL Server feature]**) obscures sensitive column values in query *results* for users without an explicit `UNMASK` permission, while leaving the actual stored data untouched — useful for letting support staff or reporting tools see "enough" of a value (like the last 4 digits of an SSN) without exposing the full sensitive value, without needing separate masked copies of the data maintained elsewhere.

### 35.6 Encryption Concepts

| Feature | What It Protects Against |
|---|---|
| **Transparent Data Encryption (TDE)** | Encrypts the database's physical files at rest — protects against someone stealing the physical `.mdf`/`.bak` files; the server itself, and any authorized query, sees plain data transparently. |
| **Always Encrypted** | Encrypts specific column *values* client-side, before they ever reach the server in plaintext — even a DBA with full server access cannot see the plaintext value; decryption happens only in an authorized client application holding the encryption key. |
| **Column-level (cell) encryption** | Older, more manual T-SQL function-based (`ENCRYPTBYKEY`, etc.) column encryption, largely superseded in new development by Always Encrypted. |

**Banking/healthcare example:** TDE is a standard, comparatively low-effort baseline (protecting backup files if physically stolen); Always Encrypted is appropriate for the most sensitive fields specifically (SSNs, full card numbers) where even privileged database administrators should never be able to view plaintext values — a materially stronger, "zero server-side trust" security posture increasingly required by financial and healthcare compliance regimes.

## 36. Dynamic SQL

### 36.1 sp_executesql and Parameterized Dynamic SQL

`sp_executesql` executes a dynamically-built SQL string **with proper parameterization** — the parameter values are passed separately from the SQL text itself, exactly like a parameterized query in any application-level driver (Part 49), meaning the values can never be misinterpreted as SQL syntax.

### 36.2 SQL Injection — Unsafe vs Safe Dynamic SQL

This is the exact T-SQL equivalent of the SQL-injection defense pattern covered for Python in the Regular Expressions and Testing handbooks: **never build SQL by string-concatenating untrusted input; always pass untrusted values as genuine parameters**, whether via `sp_executesql` in T-SQL or a parameterized query from an application driver (Part 49).

### 36.3 Dynamic Filtering, Reporting, and PIVOT

Dynamic reporting and dynamic `PIVOT` (Section 17.3) follow the same fundamental pattern: build the SQL text programmatically to accommodate a structure not known until runtime (which filters were supplied; which columns should appear), but **always** route the actual data values through `sp_executesql`'s parameter mechanism, never string concatenation, regardless of how dynamic the surrounding query structure needs to be.

## 37. SQL Server Backup and Restore

### 37.1 Backup Types

| Type | Captures |
|---|---|
| **Full backup** | The entire database at a point in time — the foundation every other backup type builds on. |
| **Differential backup** | Everything changed *since the last full backup* — smaller and faster than another full backup, but grows over time until the next full backup resets the baseline. |
| **Transaction log backup** | Every transaction recorded in the log since the last log backup — enables point-in-time recovery (Section 37.3); only possible under the Full (or Bulk-logged) recovery model (Section 3.5). |

### 37.2 BACKUP DATABASE, BACKUP LOG, RESTORE DATABASE, RESTORE LOG

`WITH NORECOVERY` leaves the database in a "restoring" state, ready to accept the *next* backup file in sequence (differential, then log); the final restore step in the chain uses `WITH RECOVERY` to bring the database fully online. Restoring the wrong order, or forgetting `NORECOVERY` on an intermediate step, is a common practical mistake worth rehearsing in a non-production environment before ever relying on it during a real incident.

### 37.3 Point-in-Time Recovery

`STOPAT` restores the transaction log only up to a specific moment — the standard recovery technique for "someone ran a bad `UPDATE` at 2:35pm, restore to just before it" scenarios, only possible when the Full recovery model (Section 3.5) has been in effect with regular log backups being taken.

### 37.4 RPO, RTO, and Disaster Recovery

| Term | Meaning |
|---|---|
| **RPO (Recovery Point Objective)** | The maximum acceptable amount of *data loss*, measured in time — "we can tolerate losing up to 15 minutes of data" directly determines how frequently log backups must run. |
| **RTO (Recovery Time Objective)** | The maximum acceptable *downtime* before service is restored — directly informs whether a slower full-restore-from-backup strategy is adequate, or a High Availability solution (Part 45) with near-instant failover is required instead. |

A backup/restore strategy should always be derived explicitly from a stated RPO/RTO, not chosen arbitrarily — these two numbers are a business decision (how much data loss and downtime is genuinely acceptable), which the technical backup schedule and architecture then serve.

## 38. SQL Server Agent

### 38.1 Jobs, Job Steps, Schedules, Alerts, Operators

SQL Server Agent is the built-in job scheduler running as a Windows service alongside the Database Engine — a **job** is a named unit of scheduled work made of one or more **job steps** (each a T-SQL script, SSIS package, PowerShell script, or other supported step type), triggered by a **schedule** (time-based) or in response to an **alert** (a specific error/performance condition), with **operators** (people/distribution lists) notified of success/failure via email or other configured methods.

SQL Server Agent has no direct built-in equivalent in either PostgreSQL (typically relying on the OS's own scheduler, `cron`, or the `pg_cron` extension) or MySQL (the `EVENT` scheduler, a comparatively lighter-weight built-in feature) — Agent's integration with `msdb` (Section 1.8), its rich job-step type system, and its native alerting are a genuinely more mature, SQL-Server-native automation platform.

### 38.2 Automating Data Imports, ETL, Backups, Reports, Maintenance

Common professional Agent job categories: nightly ETL/data-warehouse loading (Part 50), scheduled full/differential/log backups (Part 37) run on an automated cadence matching the stated RPO, index maintenance (rebuilding/reorganizing fragmented indexes on a regular schedule), automated report generation and distribution, and scheduled data-quality checks (re-running validation queries like Section 33's and alerting an operator if quality thresholds are violated).

## 40. tempdb

### 40.1 What tempdb Is

Introduced in Section 1.8: `tempdb` is a system database, shared by every database and session on the instance, recreated fresh on every SQL Server restart, used as scratch space for a wide range of internal operations — not merely explicit `#temp` tables.

### 40.2 What Uses tempdb

| Use | Example |
|---|---|
| Temporary tables/table variables | `#Table`, `@TableVar` (Part 15) |
| Sorting | `ORDER BY` operations exceeding available memory grant (Section 28.5) |
| Hashing | Hash joins and hash aggregates needing a working area |
| Version store | Row versioning for Snapshot isolation/RCSI (Section 24.4) |
| Intermediate operations | Query plan operators like spools, and some index-rebuild operations |

Because so many unrelated internal operations all compete for the same shared `tempdb`, it is frequently one of the most contended resources on a busy SQL Server instance — **tempdb contention** manifests as blocking specifically on `tempdb`'s internal system pages (visible via `PAGELATCH` waits in monitoring DMVs), a genuinely SQL-Server-specific operational concern with no precise PostgreSQL/MySQL parallel, since neither uses one single shared scratch database this centrally for such a broad range of engine-internal purposes.

### 40.3 Monitoring and Configuration Concepts

Professional guidance for busy instances: configure **multiple tempdb data files** of equal size (commonly one per CPU core, up to a reasonable cap) specifically to reduce internal allocation-page contention, monitor tempdb space usage and growth (an out-of-space `tempdb` can halt query processing instance-wide, affecting *every* database, not just one), and be alert to any single long-running transaction under Snapshot/RCSI isolation, since it can prevent the version store from being cleaned up, causing `tempdb` to grow unexpectedly large.

## 43. Change Data Capture and Change Tracking

### 43.1 Change Data Capture (CDC)

CDC captures the **full detail** of every INSERT/UPDATE/DELETE against a tracked table — including old and new column values — into dedicated change tables, by reading the transaction log asynchronously (minimal impact on the source table's transactional workload). This is purpose-built for feeding **ETL pipelines** (Part 50): a downstream data warehouse load process can pull only what changed since its last run, rather than re-scanning the entire source table every time.

### 43.2 Change Tracking

Change Tracking is a lighter-weight alternative to CDC: it records **that** a row changed (and its primary key) but not the actual old/new column values — sufficient for synchronization scenarios (like "which rows need to be re-fetched by a downstream cache or search index") that don't need full change history, at lower storage and overhead cost than full CDC.

### 43.3 CDC vs Change Tracking — Choosing

| Need | Choose |
|---|---|
| Full audit trail of exactly what changed, old and new values | CDC |
| Just need to know *which* rows changed, to re-sync/re-fetch them | Change Tracking |
| Feeding an incremental ETL/data-warehouse load (Part 50) | CDC (needs the actual changed values) |
| Keeping a search index or cache in sync with a source table | Change Tracking (usually sufficient, lighter weight) |

Neither PostgreSQL nor MySQL bundles a directly equivalent first-party feature at this level of built-in integration (PostgreSQL's logical replication and various CDC tools like Debezium serve an analogous purpose but as separate components, not a single built-in engine feature pair like SQL Server's).

## 44. Partitioning

### 44.1 Partition Functions and Schemes

A **partition function** defines the logical boundaries (here, by year) dividing a table's rows; a **partition scheme** maps each resulting partition to a physical filegroup (Section 3.4) — together, they let one logical table be physically stored (and, importantly, *managed*) as several separate underlying partitions, transparent to ordinary querying.

### 44.2 Range Partitioning and Partition Switching

**Partition switching** is the standout capability: moving a whole partition's worth of data into or out of a partitioned table is a near-instantaneous, purely metadata-level operation (no actual data is physically copied or moved), regardless of how many millions of rows the partition contains — the standard technique for efficiently archiving old data out of an active table, or rapidly loading a new bulk data batch in, without the locking/logging cost of an equivalent row-by-row `DELETE`/`INSERT`.

### 44.3 How Partitioning Helps Large Enterprise Datasets

Partitioning improves **manageability** (index maintenance, backups, and archival can target individual partitions rather than an entire massive table), **query performance** for queries that filter on the partitioning column (**partition elimination** lets the engine skip scanning irrelevant partitions entirely), and enables the near-instant archival pattern above — all without changing how the table is queried from the application's perspective. PostgreSQL's native declarative partitioning (`PARTITION BY RANGE`) and MySQL's `PARTITION BY RANGE` serve a broadly similar purpose, though SQL Server's partition-switching mechanism and its tight integration with filegroups reflect a somewhat more mature, enterprise-oriented feature set historically.

## 45. SQL Server High Availability

### 45.1 High Availability vs Backup/Disaster Recovery — A Critical Distinction

**High Availability (HA)** minimizes *downtime* by maintaining one or more synchronized standby copies of a database ready for near-instant failover if the primary fails. **Backup and Disaster Recovery (Part 37)** protects against *data loss* and provides a recovery path after a catastrophic failure (including scenarios HA itself cannot address, like widespread corruption replicated to every HA copy, or an entire data center being destroyed). **These are complementary, not substitutes for each other** — a robust production architecture requires both: HA for routine failures (a single server crashing) and backups for catastrophic or logical failures (corruption, accidental mass deletion, a data-center-level disaster) that HA alone cannot protect against.

### 45.2 Always On Availability Groups

Always On Availability Groups (**[SQL Server feature]**, 2012+) maintain one or more synchronized secondary replicas of a set of databases, automatically (or manually) failing over to a secondary if the primary becomes unavailable — replicas can also serve **read-only** query workloads (offloading reporting queries from the primary), a commonly-used side benefit beyond pure HA.

### 45.3 Failover, Log Shipping, Database Mirroring

| Mechanism | Summary |
|---|---|
| **Always On Availability Groups** | Modern, flexible, multi-replica HA/DR solution; the current recommended approach. |
| **Log Shipping** | Periodically ships and restores transaction log backups to one or more secondary servers — simple, reliable, but with higher latency (minutes, not seconds) than Availability Groups. |
| **Database Mirroring** | An older (pre-2012) single-secondary HA feature, now deprecated in favor of Availability Groups; understood today mainly for historical/legacy-system context. |

### 45.4 Replication's Role

Replication (Part 46) is not primarily an HA mechanism at all — it is oriented toward *distributing* data to other locations/systems for different purposes (reporting offload, integration with other systems), though it is sometimes loosely associated with availability due to maintaining additional data copies elsewhere.

## 46. Replication

### 46.1 Publisher, Distributor, Subscriber

| Role | Function |
|---|---|
| **Publisher** | The source database whose changes are being replicated out. |
| **Distributor** | Coordinates the movement of changes from publisher to subscriber(s); can be the same server as the publisher or a dedicated separate server. |
| **Subscriber** | The destination database receiving replicated changes. |

### 46.2 Transactional, Snapshot, and Merge Replication

| Type | Behavior | Typical Use |
|---|---|---|
| **Transactional replication** | Near-real-time, continuous propagation of individual transactions from publisher to subscriber(s). | Feeding a reporting server or another system with low-latency, up-to-date copies of operational data. |
| **Snapshot replication** | Periodically copies the *entire* current dataset, rather than incremental changes. | Smaller, infrequently-changing datasets, or as the initial seeding step before transactional/merge replication begins. |
| **Merge replication** | Allows changes at *both* publisher and subscriber(s), with conflict-detection/resolution logic reconciling divergent changes. | Distributed/offline scenarios where subscribers (e.g. field sales laptops) must be able to make local changes while disconnected, later synchronized bidirectionally. |

### 46.3 Where Replication Fits Into Enterprise Data Architectures

Replication is most often used to distribute operational data to a **reporting server** (isolating heavy analytical query load away from the transactional primary, a distinct concern from — and often complementary to — Always On's read-replica capability), to feed data into **separate downstream systems** (a data warehouse's staging layer, Part 50), or to support **distributed/branch-office** scenarios via merge replication. In modern architectures, CDC (Part 43) combined with a dedicated ETL/ELT pipeline has increasingly displaced transactional replication for pure data-warehouse-feeding use cases, while replication remains relevant for genuine multi-master or distributed-write scenarios.

### 49.5 Transactions, Connection Management, Connection Pooling, Error Handling

````sql
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

engine = create_engine(
    'mssql+pyodbc://app_user:YourPassword@localhost/Sales?driver=ODBC+Driver+18+for+SQL+Server',
    pool_size=5, max_overflow=10, pool_pre_ping=True
)

try:
    with engine.begin() as conn:   # engine.begin() commits automatically on success, rolls back on exception
        conn.execute(text("UPDATE Banking.Accounts SET Balance = Balance - :amt WHERE AccountID = :id"),
                     {'amt': 500, 'id': 1})
        conn.execute(text("UPDATE Banking.Accounts SET Balance = Balance + :amt WHERE AccountID = :id"),
                     {'amt': 500, 'id': 2})
except SQLAlchemyError as e:
    print(f"Transaction failed and was rolled back: {e}")
````

| Concept | Explanation |
|---|---|
| `pool_size` / `max_overflow` | Connection pooling — reusing a bounded set of already-open connections rather than opening/closing a new physical connection per query, avoiding significant per-connection overhead under load. |
| `pool_pre_ping=True` | Tests each pooled connection's liveness before reuse, gracefully handling connections SQL Server or a network intermediary silently closed. |
| `engine.begin()` | A context manager that wraps the block in an explicit transaction, auto-committing on success and auto-rolling-back on any exception — the Python-side analogue of T-SQL's `TRY`/`CATCH` + `COMMIT`/`ROLLBACK` pattern (Section 25.1). |

### 50.1 ETL vs ELT

**ETL** (Extract, Transform, Load) transforms data *before* loading it into the target warehouse — transformation logic typically lives in a separate tool/service. **ELT** (Extract, Load, Transform) loads raw data into the target first, then transforms it *using the target database's own processing power* (T-SQL, in SQL Server's case) — increasingly the more common modern pattern, leveraging SQL Server's own set-based processing strength (Part 30) for the transformation step itself rather than an external engine.

## 51. SQL Server Data Warehousing

### 51.1 Star Schema — Fact and Dimension Tables

A **star schema** places a central **fact table** (numeric measures — `Amount`) surrounded by **dimension tables** (descriptive context — customer, date, product), joined via surrogate keys — deliberately denormalized (Section 34.3) compared to an OLTP schema, optimized for the large aggregate `GROUP BY` queries typical of BI reporting rather than for minimizing storage redundancy.

### 51.2 Snowflake Schema

A **snowflake schema** further normalizes dimension tables (e.g. splitting `DimCustomer` into `DimCustomer` + a separate `DimRegion`) — reduces some redundancy at the cost of additional joins for queries, a genuine trade-off against the star schema's simplicity; star schemas are more common in practice for their query simplicity and (especially combined with columnstore indexes, Part 27) strong performance.

### 51.3 Date Dimensions and Surrogate Keys

A dedicated `DimDate` table (pre-populated with one row per calendar date, often years into the future) is a near-universal data-warehouse convention — it lets date-based grouping/filtering (`Quarter`, `DayName`, fiscal periods) happen via simple joins and `WHERE` clauses against pre-computed columns, rather than repeatedly recomputing `DATEPART`-style expressions (which, per Part 30, are also often non-SARGable) across every fact-table query.

### 51.4 Building a Complete Sales Data Warehouse

The complete pattern combines Sections 51.1-51.3 with Part 50's ETL/SCD Type 2 loading pattern and Part 27's columnstore indexing on the fact table — together forming the standard architecture of a SQL-Server-based sales data warehouse: normalized OLTP source → staged, cleaned, and conformed via ETL → loaded into a star-schema warehouse with surrogate keys and SCD-tracked dimensions → queried via columnstore-accelerated analytical SQL (Part 47) → surfaced to business users through Power BI (Part 52).

## 52. SQL Server + Power BI

### 52.1 Import Mode vs DirectQuery

| Mode | Behavior |
|---|---|
| **Import** | Power BI copies a snapshot of the data into its own compressed in-memory model — very fast query/interaction performance, but data is only as fresh as the last scheduled refresh. |
| **DirectQuery** | Power BI sends live queries back to SQL Server for every interaction — always current data, but performance depends entirely on the underlying SQL Server query performance (Parts 28-30), and the query patterns Power BI itself generates. |

### 52.2 SQL Views for Power BI

Exposing a stable, well-named, pre-joined **view** (Part 19) — rather than having report authors connect directly to raw warehouse tables — is standard professional practice: it decouples the underlying schema from report definitions (a table rename/restructure doesn't break every downstream report), lets a DBA optimize the view definition centrally, and can itself be an indexed view (Section 19.3) for DirectQuery performance.

### 52.3 Query Optimization Before Power BI, Star Schema, Aggregations

DirectQuery performance is directly determined by the same optimization principles from Parts 28-30 applied to whatever queries Power BI generates — a star schema (Section 51.1) with appropriate indexes (including columnstore, Part 27) on the fact table is the standard foundation; Power BI's own **aggregation tables** feature (pre-summarized, smaller tables Power BI automatically queries instead of the full fact table when a report doesn't need row-level detail) further reduces the query burden reaching SQL Server.

### 52.4 Data Refresh and Gateway Concepts

For Import-mode reports, a scheduled **refresh** re-pulls data from SQL Server on a defined cadence; when Power BI Service (cloud) needs to reach an on-premises SQL Server, an **On-premises Data Gateway** — a locally installed service bridging the cloud service and the internal network — is required, since Power BI Service cannot directly reach a database behind a private network's firewall otherwise.

### 53.2 Flask Example

````sql
from flask import Flask, jsonify, request
from sqlalchemy import create_engine, text

app = Flask(__name__)
engine = create_engine('mssql+pyodbc://app_user:pass@localhost/Sales?driver=ODBC+Driver+18+for+SQL+Server')

@app.route('/customers/<int:customer_id>/orders')
def get_orders(customer_id):
    with engine.connect() as conn:
        result = conn.execute(
            text("SELECT OrderID, OrderDate, TotalAmount FROM Sales.Orders WHERE CustomerID = :cid"),
            {'cid': customer_id}
        )
        orders = [dict(row._mapping) for row in result]
    return jsonify(orders)
````

| Task | PostgreSQL | MySQL | SQL Server |
|---|---|---|---|
| Limit to 10 rows | `SELECT * FROM t LIMIT 10;` | `SELECT * FROM t LIMIT 10;` | `SELECT TOP 10 * FROM t;` |

## 56. SQL Portability

### 56.1 Identifying ANSI SQL vs Vendor-Specific Syntax

| Category | Examples |
|---|---|
| **ANSI SQL (portable)** | `SELECT`, `WHERE`, `JOIN` types, `GROUP BY`/`HAVING`, `CASE`, standard CTEs, `OFFSET...FETCH`, window functions, `UNION`/`INTERSECT`/`EXCEPT` |
| **PostgreSQL-specific** | `SERIAL`, `RETURNING`, `ILIKE`, array types, `jsonb` operators, `LATERAL` |
| **MySQL-specific** | `AUTO_INCREMENT`, `LIMIT m, n`, `ON DUPLICATE KEY UPDATE`, backtick identifiers |
| **SQL Server-specific** | `TOP`, `IDENTITY`, `OUTPUT`, `CROSS APPLY`/`OUTER APPLY`, `PIVOT`/`UNPIVOT`, square-bracket identifiers, `[T-SQL]`-labeled items throughout this handbook |

### 56.2 Writing Portable SQL

Favoring `OFFSET...FETCH` over `TOP`, standard CTEs over vendor-specific procedural extensions, and `CASE` over any single database's proprietary conditional shortcuts are all small, deliberate, and cumulative choices that keep application-level SQL portable when portability is a genuine project requirement (e.g. a product intended to support multiple database backends, or an organization anticipating a future migration).

### 56.3 Why Understanding Both Portable and Vendor-Specific SQL Matters

Writing exclusively portable SQL leaves substantial performance and functionality on the table — `MERGE`, `CROSS APPLY`, columnstore indexes, and Row-Level Security (all covered in this handbook) have no portable equivalent, and avoiding them entirely for the sake of theoretical portability is frequently the wrong trade-off for a SQL-Server-only production system. The professional standard: **default to portable SQL when a project's requirements genuinely call for cross-database compatibility; deliberately and knowingly use vendor-specific features when their concrete benefit (performance, functionality, or genuine developer productivity) outweighs portability for that specific project** — and always know, explicitly, which category any given piece of SQL falls into, rather than discovering it painfully during a future migration attempt.

## 58. Complete SQL Server Capstone

### 58.1 Capstone Architecture

```
Raw Data (CSV exports from a retail POS system)
    ↓
SQL Server (Staging schema - Part 39's BULK INSERT)
    ↓
Data Cleaning (Part 33 - TRY_CAST, duplicate removal, outlier flagging)
    ↓
Data Warehouse (Part 51 - star schema, SCD Type 2, columnstore-indexed FactSales)
    ↓
Analytical SQL (Part 47 - revenue, cohort, retention, Pareto queries)
    ↓
Python (Part 49 - pandas pulls aggregated features via SQLAlchemy)
    ↓
Machine Learning (scikit-learn - churn/CLV model trained on SQL-prepared features)
    ↓
Predictions (Part 54 - written back to SQL Server's ML.ChurnPredictions table)
    ↓
SQL Server (predictions joined with warehouse dimensions for reporting)
    ↓
Power BI (Part 52 - executive dashboard combining actuals and predictions)
```

### 58.2 Stage-by-Stage Explanation

**Raw Data → SQL Server:** Nightly POS export files land via `BULK INSERT` (Part 39) into a `Staging` schema, all-text columns, no constraints — capturing data exactly as received before any assumptions are applied.

**Data Cleaning:** `TRY_CAST`/`TRY_CONVERT` (Section 9.4) validate and convert staged data; malformed rows are quarantined (Section 33.6) rather than silently dropped or allowed to break the pipeline; duplicate transactions are detected and removed deterministically (Section 33.2).

**Data Warehouse:** Cleaned data is loaded into a star schema (Part 51) — `FactSales` (columnstore-indexed, Part 27, for analytical performance) joined to `DimCustomer` (SCD Type 2, tracking customer attribute changes over time), `DimProduct`, and `DimDate`.

**Analytical SQL:** A library of views (Section 52.2) and parameterized stored procedures compute revenue trends, customer cohorts (Section 47.2), and Pareto/CLV metrics (Section 47.4) directly against the warehouse.

**Python:** A scheduled Python job connects via SQLAlchemy (Part 49), pulling a **pre-aggregated, feature-engineered** dataset (customer-level RFM — recency, frequency, monetary — features, Section 48.2) rather than raw transaction rows, keeping the expensive aggregation work inside SQL Server where it's most efficient.

**Machine Learning:** scikit-learn trains a churn-classification model (or a CLV regression model) on the pulled features, following the ML-testing discipline from the Testing and Debugging handbook's Part 22 (train/test split with no leakage, reproducible with a fixed seed).

**Predictions → SQL Server:** The trained model scores the full customer base; predictions, along with model version and scoring timestamp (Section 54.4), are written back via `to_sql()` (Section 49.4) into a dedicated predictions table.

**Power BI:** A final reporting view (Section 52.2) joins warehouse actuals with the predictions table, feeding an executive dashboard showing both historical performance and forward-looking churn/CLV risk — combining every layer of this capstone into one decision-support artifact.

### 58.3 Why This Architecture Reflects Professional Practice

Every stage does the work it is genuinely best suited for: SQL Server handles bulk data movement, cleaning, and set-based aggregation (its core strength, per Part 30); Python/scikit-learn handles the actual statistical modeling (its core strength); and the round-trip back into SQL Server keeps predictions alongside the operational and warehouse data business users already query through Power BI — no separate, disconnected "data science silo" that the rest of the organization can't easily consume.

## 59. Interview Preparation

### 59.1 Beginner (Q1-Q10)

**Q1. What is the difference between SQL Server, T-SQL, and SQL?** SQL is the standard query language; T-SQL is Microsoft's extended dialect adding procedural constructs and SQL-Server-specific syntax; SQL Server is the database engine that executes T-SQL (Section 1.2).

**Q2. What is the object hierarchy in SQL Server?** Instance → Database → Schema → Objects → Rows/Columns (Section 1.5).

**Q3. What is the difference between a login and a user?** A login is a server-level identity; a user is a database-level identity mapped to a login (Section 1.6).

**Q4. What are the four system databases?** `master`, `model`, `msdb`, `tempdb` (Section 1.8).

**Q5. What is the difference between `CHAR` and `VARCHAR`?** `CHAR` is fixed-length, space-padded; `VARCHAR` is variable-length, storing only the actual content length.

**Q6. What does `IDENTITY(1,1)` mean?** An auto-incrementing column starting at 1, incrementing by 1 per new row (Section 5.3).

**Q7. What is the difference between `DELETE` and `TRUNCATE`?** `DELETE` can be filtered with `WHERE` and is fully logged row-by-row; `TRUNCATE` removes all rows, cannot be filtered, and is minimally logged (Section 5.1).

**Q8. What does `TOP` do, and how does it differ from `LIMIT`?** Restricts the number of returned rows; SQL Server's `TOP` syntax precedes the column list, unlike PostgreSQL/MySQL's trailing `LIMIT` (Section 7.2).

**Q9. What is a primary key, and can a table have more than one?** A column (or set of columns) uniquely identifying each row; a table can have only one primary key, though it can be composite (multiple columns, Section 5.2).

**Q10. What is `NULL`, and how does it differ from an empty string or zero?** `NULL` represents an unknown/missing value; it is never equal to an empty string or zero, and any comparison with `NULL` yields `UNKNOWN`, not `TRUE`/`FALSE`.

### 59.2 Intermediate (Q11-Q20)

**Q11. What is the difference between `INNER JOIN` and `LEFT JOIN`?** `INNER JOIN` returns only matching rows; `LEFT JOIN` returns all rows from the left table, with `NULL`s for unmatched right-side columns (Section 11.1).

**Q12. Why can `NOT IN` behave unexpectedly with `NULL` values?** If the subquery's result set contains any `NULL`, the entire `NOT IN` comparison evaluates to `UNKNOWN` for every row, returning zero rows (Section 12.4).

**Q13. What is a CTE, and when would you use one?** A named temporary result set (`WITH ... AS (...)`), scoped to one statement, used to improve readability of multi-step queries (Part 13).

**Q14. What is the difference between `RANK()` and `DENSE_RANK()`?** `RANK()` skips subsequent rank values after a tie; `DENSE_RANK()` does not (Section 14.2).

**Q15. What is the difference between a temp table and a table variable?** A temp table supports full indexing and better statistics; a table variable has more limited indexing and historically weaker cardinality estimation, but is unaffected by transaction rollback (Part 15).

**Q16. What does `CROSS APPLY` do that a regular `JOIN` cannot?** Allows the right-hand table expression to be correlated to each row of the left side, such as a per-row `TOP N` subquery (Part 16).

**Q17. What is the difference between a view and an indexed view?** A standard view is just stored SQL, re-executed on each query; an indexed view physically materializes and automatically maintains its result set (Section 19.3).

**Q18. What does `SCOPE_IDENTITY()` return, and why is it preferred over `@@IDENTITY`?** The last identity value generated in the current session and scope; `@@IDENTITY` can incorrectly return a value from an unrelated trigger-fired insert (Section 5.3).

**Q19. What is the purpose of `TRY_CAST`/`TRY_CONVERT`?** They return `NULL` on a failed conversion instead of raising an error, valuable for cleaning messy data in one pass (Section 9.4).

**Q20. What is the difference between `ISNULL` and `COALESCE`?** `ISNULL` is T-SQL-specific, takes exactly 2 arguments; `COALESCE` is standard SQL, takes 2 or more, with more predictable type resolution (Section 9.5).

### 59.3 Advanced (Q21-Q30)

**Q21. What is a non-SARGable predicate, and how do you fix one?** A predicate that prevents index seek usage, typically by wrapping an indexed column in a function; fix by rewriting to isolate the raw column (Section 29.4, Section 30.1).

**Q22. What is parameter sniffing?** A cached execution plan compiled for one parameter value performing poorly for other, differently-shaped parameter values (Section 29.5).

**Q23. What is the difference between a clustered and nonclustered index?** A clustered index determines physical row storage order (max one per table); a nonclustered index is a separate structure pointing back to the table (Sections 26.2-26.3).

**Q24. What is a covering index?** A nonclustered index containing every column a specific query needs, avoiding a key lookup (Section 26.5).

**Q25. What is the difference between `RCSI` and `SNAPSHOT` isolation?** RCSI provides statement-level row versioning as a drop-in replacement for `Read Committed`'s locking behavior; `SNAPSHOT` provides transaction-level consistency and can raise write-write conflict errors (Section 24.4).

**Q26. What causes a deadlock, and how does SQL Server resolve one?** Two transactions holding locks the other needs, forming a cycle; SQL Server detects it and terminates the lower-cost "victim" transaction (Section 24.3).

**Q27. Why is `THROW` generally preferred over `RAISERROR`?** A bare `THROW` in a `CATCH` block re-raises the original error with its original context intact; `RAISERROR` always generates a new error at the `CATCH` block's location (Section 25.3).

**Q28. What is the difference between CDC and Change Tracking?** CDC captures full old/new column values; Change Tracking only records that a row changed, at lower overhead (Part 43).

**Q29. What is partition switching, and why is it fast?** Moving a partition's data in/out of a table via a metadata-only operation, with no physical data movement, regardless of row count (Section 44.2).

**Q30. What is the difference between High Availability and Disaster Recovery?** HA minimizes downtime via synchronized standby replicas; DR protects against data loss and provides recovery after catastrophic failure — complementary, not substitutes (Section 45.1).

### 59.4 Senior (Q31-Q40)

**Q31. How would you diagnose a query that suddenly became slow after months of good performance?** Check for stale statistics first, then compare estimated vs actual row counts in the execution plan, then look for non-SARGable predicates or implicit conversions before considering index changes (Sections 28.6, 29.6).

**Q32. When would you choose a nonclustered columnstore index over a clustered rowstore index?** When a transactional table needs to also support fast, large-scale analytical aggregation without migrating its primary storage format (Section 27.2).

**Q33. How do you design a schema to support both OLTP and OLAP workloads on the same SQL Server instance?** Keep the OLTP schema normalized for transactional integrity; feed a separately-modeled, denormalized star-schema warehouse via ETL/CDC for analytical workloads, isolating the two access patterns (Sections 34.4, 50, 51).

**Q34. What's the risk of over-relying on missing-index DMV suggestions?** They don't account for the write-side maintenance cost of additional indexes; blindly applying every suggestion can degrade write performance without properly evaluating the trade-off (Section 26.9).

**Q35. How would you implement point-in-time recovery for a critical production database?** Use the Full recovery model with regular transaction log backups, then `RESTORE DATABASE ... WITH NORECOVERY` followed by `RESTORE LOG ... WITH RECOVERY, STOPAT` (Section 37.3).

**Q36. How does Always On Availability Groups differ architecturally from Log Shipping?** Availability Groups provide near-real-time, often automatic failover with optional readable secondaries; Log Shipping is simpler but has higher latency and typically requires manual failover (Section 45.3).

**Q37. What are the trade-offs of Always Encrypted vs Transparent Data Encryption?** TDE protects data at rest transparently but is fully visible to authorized server-side queries/DBAs; Always Encrypted protects specific column values even from privileged server-side access, at the cost of requiring client-side key management (Section 35.6).

**Q38. How would you safely perform a large-scale data archival without locking a live production table?** Use partition switching (Section 44.2) to move an entire partition's data out as a near-instant metadata operation, rather than a row-by-row `DELETE`.

**Q39. What's the professional argument against overusing triggers for business logic?** Triggers execute implicitly and invisibly to calling code, making systems harder to debug and reason about; explicit logic in stored procedures or application code is generally preferred except for genuine cross-cutting concerns (Section 22.4).

**Q40. How would you evaluate whether a query should be rewritten as set-based instead of using a cursor?** Nearly any row-by-row procedural loop performing a per-row DML operation can typically be rewritten as a single set-based statement; cursors should be reserved for genuinely sequential administrative tasks, not ordinary business logic (Section 30.4).

### 59.5 Data Analyst (Q41-Q50)

**Q41. How do you calculate month-over-month growth in T-SQL?** Using `LAG()` over a monthly-aggregated CTE, comparing each month's value to the previous month's (Section 14.5).

**Q42. How do you find the top 3 products per category?** `ROW_NUMBER() OVER (PARTITION BY Category ORDER BY Revenue DESC)` filtered to `<= 3`, or `CROSS APPLY` with `TOP 3` per category (Sections 14.2, 16.3).

**Q43. How would you calculate a running total of sales?** `SUM(Amount) OVER (ORDER BY Date ROWS UNBOUNDED PRECEDING)` (Section 14.4).

**Q44. How do you identify customers with no transactions?** `LEFT JOIN` with a `WHERE ... IS NULL` filter, or `NOT EXISTS` (Sections 11.1, 12.3).

**Q45. What is cohort analysis, and how would you implement it in SQL?** Grouping customers by their acquisition period and tracking their activity in subsequent periods, typically via a CTE joining signup month to activity month (Section 47.2).

**Q46. How do you calculate customer retention rate?** Comparing the set of customers active in one period against those still active in a subsequent period, typically via `INTERSECT` or a self-join on customer ID across periods.

**Q47. What is Pareto analysis, and how is it computed in SQL?** Identifying what proportion of entities account for a given proportion of a total (the "80/20 rule"), computed via a running-total window function divided by the grand total (Section 47.4).

**Q48. How would you find the second-highest salary in a table?** `SELECT MAX(Salary) FROM Employees WHERE Salary < (SELECT MAX(Salary) FROM Employees)`, or `DENSE_RANK()` filtered to rank 2 (Section 60's coding challenges).

**Q49. What's the difference between `PIVOT` and conditional aggregation with `CASE`?** `PIVOT` is a dedicated T-SQL operator for a known, fixed set of columns; conditional aggregation is more flexible and portable but slightly more verbose (Part 17).

**Q50. How would you detect gaps in a sequence of dates?** Compare each date to the previous date via `LAG()`, flagging where the difference exceeds one day (Section 60's coding challenges).

### 59.6 Data Scientist (Q51-Q58)

**Q51. What kinds of feature engineering are best done in SQL vs Python?** Aggregations, joins, and simple derived columns are efficient in SQL; complex nonlinear transforms, embeddings, and text processing are better suited to Python (Section 48.4).

**Q52. How would you construct a training label in SQL for a churn model?** A `CASE`/`NOT EXISTS` expression checking whether a customer has any activity within a defined future window after a reference date (Section 48.3).

**Q53. How do you prevent data leakage when preparing features in SQL for a time-series-sensitive model?** Ensure every feature is computed using only data available *as of* the reference/prediction date, never including future information relative to that point.

**Q54. Why might you prefer `TABLESAMPLE` over `ORDER BY NEWID()` for large-table sampling?** `TABLESAMPLE` samples at the physical page level without a full table sort, making it far faster on large tables, at some cost to sampling uniformity (Section 48.1).

**Q55. How would you version and monitor ML predictions stored in SQL Server?** Store `ModelVersion` and `ScoredAt` alongside each prediction, enabling drift monitoring queries over time (Section 54.4).

**Q56. What's the professional division of labor between SQL Server and a Python ML framework?** SQL Server handles data preparation, aggregation, and serving; Python/ML frameworks handle model training and complex feature transforms (Section 54.1).

**Q57. How would you join SQL-extracted features to a pandas DataFrame for modeling?** Query the aggregated feature set directly into a DataFrame via `pandas.read_sql()` over a SQLAlchemy engine (Section 49.3).

**Q58. Why is it important to push aggregation into SQL rather than pulling raw rows into pandas?** SQL Server's set-based engine performs large-scale aggregation far more efficiently than row-by-row processing in Python, and reduces the data volume transferred (Section 48.4).

### 59.7 Data Engineer (Q59-Q68)

**Q59. What's the difference between ETL and ELT?** ETL transforms data before loading into the target; ELT loads raw data first, then transforms using the target system's own processing power (Section 50.1).

**Q60. How would you implement incremental loading?** A high-water-mark pattern tracking the last successfully loaded timestamp, or CDC/Change Tracking for more granular incremental capture (Sections 43, 50.2).

**Q61. What is a Slowly Changing Dimension, and when would you use Type 2?** A dimension table design pattern preserving history; Type 2 is used when historical accuracy of a changing attribute (e.g. a customer's region at the time of a past order) must be preserved (Section 50.2).

**Q62. What is the difference between CDC and transactional replication for ETL purposes?** CDC captures changes for consumption by a separate ETL process on a schedule; transactional replication continuously propagates changes in near real time to another database (Sections 43, 46.2).

**Q63. Why use a star schema rather than a fully normalized schema for a data warehouse?** Denormalization reduces the number of joins needed for large aggregate queries, which matters more for OLAP read performance than storage redundancy does (Section 34.3, Part 51).

**Q64. What is partition elimination, and why does it matter for large fact tables?** The optimizer's ability to skip scanning irrelevant partitions entirely when a query filters on the partitioning column, improving performance on very large tables (Section 44.3).

**Q65. How would you handle a malformed row during a bulk import without aborting the entire load?** Use `BULK INSERT`'s `MAXERRORS` and error-file options, or land all data into an all-text staging table and validate/convert afterward (Sections 33.6, 39.2).

**Q66. What automation tool would you use to schedule a nightly ETL job in SQL Server?** SQL Server Agent, defining a job with appropriate steps and a schedule (Part 38).

**Q67. How would you design a conformed dimension shared across multiple fact tables?** Build one authoritative `DimDate`/`DimCustomer` table referenced by every relevant fact table, ensuring consistent granularity and keys across subject areas (Section 57.10).

**Q68. What role does `OUTPUT` play in an ETL procedure?** Capturing exactly which rows were inserted/updated/deleted in a single statement, useful for logging and auditing pipeline behavior (Part 6).

### 59.8 Backend Developer (Q69-Q78)

**Q69. How do you prevent SQL injection when building dynamic queries?** Always pass untrusted values as parameters via `sp_executesql` (T-SQL) or a driver's parameterized query mechanism, never string concatenation (Sections 36.2, 49.1).

**Q70. What is connection pooling, and why does it matter for a web API?** Reusing a bounded set of already-open database connections rather than opening a new one per request, avoiding significant per-connection overhead under load (Section 49.5).

**Q71. How would you wrap a multi-step database operation in a transaction from Python?** Use `engine.begin()` as a context manager in SQLAlchemy, which commits on success and rolls back on any exception (Section 49.5).

**Q72. What's the benefit of `OUTPUT` for an API's create-resource endpoint?** Returning the newly created row's generated ID directly from the `INSERT` statement, without a separate follow-up query (Section 6.4).

**Q73. How does an ORM like Django's typically prevent SQL injection?** It automatically parameterizes generated queries rather than concatenating values into SQL text (Section 53.4).

**Q74. What SQL Server security principle should a backend service account follow?** Least privilege — granting only the exact permissions the service genuinely needs, never broad roles like `db_owner` by default (Section 35.3).

**Q75. How would you design a CRUD API's database access layer for testability?** Separate query-building logic from execution, allowing the database connection/session to be injected and mocked in tests (a direct application of the Testing handbook's Part 38 dependency-injection principle).

**Q76. What's a common mistake when building dynamic filtering for a search endpoint?** Concatenating filter values directly into SQL text instead of building the WHERE clause dynamically but passing values as parameters (Section 36.3).

**Q77. Why might a backend prefer stored procedures over inline application-level SQL for certain operations?** Centralizing complex business logic and transaction handling in the database, providing a stable interface less coupled to application code changes, and enabling database-level permission grants scoped to specific procedures rather than raw table access.

**Q78. How would you handle a transient connection failure to SQL Server from a backend service?** Implement retry logic with backoff around the connection/query call, similar in principle to the API retry patterns covered in the Testing handbook's Part 13.

### 59.9 Database Developer (Q79-Q88)

**Q79. When would you choose a scalar function vs an inline table-valued function?** Scalar functions for a single computed value; inline TVFs when the reusable logic naturally produces a set of rows, since they inline into the query plan more efficiently than multi-statement TVFs (Part 21).

**Q80. What is the difference between a standard view and an indexed view?** An indexed view physically materializes and automatically maintains its result set, requiring `SCHEMABINDING` and a unique clustered index (Section 19.3).

**Q81. How would you implement a recursive hierarchy query?** A recursive CTE, with an anchor member selecting the top-level rows and a recursive member joining back to the CTE itself (Section 13.3).

**Q82. What is the purpose of `SET NOCOUNT ON` in a stored procedure?** Suppresses row-count messages, reducing unnecessary network chatter — a standard professional habit (Section 20.1).

**Q83. How do you safely build a dynamic PIVOT query?** Construct the column list dynamically, then execute the assembled query via `sp_executesql`, keeping any actual data values parameterized (Section 17.3).

**Q84. What's the danger of using `MERGE` in SQL Server, and what's a safer alternative?** `MERGE` has had historical concurrency-related bugs and subtle correctness issues under concurrent access; a safer alternative is separate, explicit `UPDATE` and `INSERT` statements wrapped in a transaction (Part 61).

**Q85. How would you design a table to support both current and historical queries without manual audit triggers?** Use a system-versioned temporal table, which automatically maintains history with no additional application logic (Part 42).

**Q86. What's the difference between a DML trigger and a DDL trigger?** A DML trigger fires on data changes (`INSERT`/`UPDATE`/`DELETE`); a DDL trigger fires on schema changes (`CREATE`/`ALTER`/`DROP`) (Part 22).

**Q87. How would you enforce a business rule spanning multiple tables that a `CHECK` constraint cannot express?** A trigger, though used sparingly given Section 22.4's maintainability caveat; alternatively, enforce it in a stored procedure that is the sole sanctioned write path.

**Q88. What's the professional guidance on function usage in `WHERE` clauses against indexed columns?** Avoid wrapping the indexed column itself in a function, since it defeats SARGability; transform the comparison side of the expression instead (Section 30.1).

### 59.10 Database Administrator (Q89-Q100)

**Q89. What's the difference between the Simple and Full recovery models?** Simple truncates the log automatically and supports no log backups; Full retains the log for point-in-time recovery via log backups (Section 3.5).

**Q90. How would you restore a database to a specific point in time?** Restore the last full backup `WITH NORECOVERY`, any differential `WITH NORECOVERY`, then the log backup(s) `WITH RECOVERY, STOPAT` (Section 37.3).

**Q91. What is lock escalation, and why might it be undesirable?** SQL Server automatically converting many fine-grained locks into a coarser table lock, which can reduce concurrency for other transactions needing access to the same table (Section 24.3).

**Q92. How would you diagnose tempdb contention?** Monitor `PAGELATCH` waits via DMVs, and consider configuring multiple equally-sized tempdb data files (Sections 40.2-40.3).

**Q93. What's the difference between High Availability and a backup strategy, and why do you need both?** HA minimizes downtime via standby replicas; backups protect against data loss and catastrophic scenarios HA cannot address, like widespread corruption (Section 45.1).

**Q94. How would you find and address missing indexes on a production system?** Query `sys.dm_db_missing_index_details` and related DMVs, but evaluate each suggestion critically against the write-cost trade-off before implementing (Section 26.9).

**Q95. What is RPO, and how does it inform backup scheduling?** Recovery Point Objective — the maximum acceptable data loss — directly determines how frequently transaction log backups must run (Section 37.4).

**Q96. How would you audit who has access to sensitive tables in a database?** Query `sys.database_principals`, `sys.database_permissions`, and related system catalog views (Part 41).

**Q97. What is the purpose of Dynamic Data Masking, and what does it not protect against?** Obscures sensitive values in query results for unauthorized users; it does not encrypt the underlying stored data, and users with sufficient permission (or via certain workarounds) can potentially infer masked values, so it should not be relied upon as the sole protection for highly sensitive data (Section 35.5).

**Q98. How would you plan for a planned failover using Always On Availability Groups?** Verify replica synchronization state, perform a manual (planned) failover during a maintenance window, and confirm application connection strings correctly redirect to the new primary (Section 45.2).

**Q99. What is the risk of over-indexing a heavily-written OLTP table?** Every additional index adds write overhead to every `INSERT`/`UPDATE`/`DELETE` touching indexed columns, potentially degrading transactional throughput (Section 26.9).

**Q100. How would you approach investigating a sudden spike in blocking on a production system?** Identify the head blocker via blocking-chain DMVs, examine what it's waiting on and why, and determine whether it's a long-running transaction, a missing index causing lock escalation, or an isolation-level mismatch (Section 24.3-24.4).

## 61. Common SQL Server Mistakes

| # | Mistake | Explanation / Fix |
|---|---|---|
| 1 | **Incorrect joins** | Joining on the wrong column, or omitting a join condition entirely (accidental cross join) — always verify row counts against expectations after writing a new join (Section 11.2). |
| 2 | **Duplicate rows from joins** | A one-to-many or many-to-many relationship silently inflating aggregate results (Section 11.2) — verify with an independent `COUNT(DISTINCT ...)` check. |
| 3 | **NULL mistakes** | Assuming `= NULL` works (it doesn't — use `IS NULL`); forgetting `NULL` breaks simple equality/inequality comparisons in unexpected ways (Section 12.4). |
| 4 | **Incorrect GROUP BY** | Including a non-aggregated column not in the `GROUP BY` list conceptually, or grouping by too many/few columns, silently changing result granularity. |
| 5 | **Incorrect HAVING** | Using `HAVING` for a row-level filter that belongs in `WHERE` (evaluated too late, after unnecessary aggregation work) — `HAVING` is for filtering *aggregated* results only. |
| 6 | **NOT IN + NULL** | Returns zero rows unexpectedly if the subquery result contains any `NULL` (Section 12.4) — use `NOT EXISTS` instead. |
| 7 | **Implicit conversion** | Comparing mismatched types (`varchar` vs `nvarchar`, string vs numeric) silently defeats index seeks (Section 29.4). |
| 8 | **Non-SARGable queries** | Wrapping indexed columns in functions in `WHERE` clauses (Section 29.4, 30.1). |
| 9 | **Poor indexing** | Missing indexes on frequently-filtered/joined columns, verified via execution plans (Part 29) rather than guessed. |
| 10 | **Over-indexing** | Too many indexes on a heavily-written table, degrading `INSERT`/`UPDATE`/`DELETE` performance (Section 26.9). |
| 11 | **Cursor overuse** | Row-by-row procedural logic where a set-based statement would work — almost always slower (Section 30.4). |
| 12 | **Dynamic SQL injection** | String-concatenating untrusted input into dynamic SQL instead of using `sp_executesql` parameters (Section 36.2). |
| 13 | **Transaction mistakes** | Forgetting to check `@@TRANCOUNT` before `ROLLBACK`, or leaving a transaction open (forgotten `COMMIT`) holding locks indefinitely. |
| 14 | **Blocking** | Long-running transactions holding locks longer than necessary, blocking unrelated queries (Section 24.3) — keep transactions as short as possible. |
| 15 | **Deadlocks** | Two transactions acquiring the same resources in a different order — standardize resource-acquisition order across the codebase to reduce risk (Section 24.3). |
| 16 | **Poor data types** | Using `money`/`float` for financial values instead of `decimal`; oversized `nvarchar` columns with no realistic bound; `datetime` instead of `datetime2` in new development (Part 4). |
| 17 | **Missing constraints** | Relying on application code alone to enforce data integrity that a `CHECK`/`FOREIGN KEY` constraint should guarantee at the database level (Section 5.6). |
| 18 | **Bad normalization** | Either under-normalizing (redundant, inconsistent data) or over-normalizing an OLTP schema into excessive joins for simple queries (Part 34). |
| 19 | **MERGE misuse** | See Section 61.1 below — a dedicated, detailed treatment given the specific historical concerns. |

### 61.1 MERGE — Historical Concerns and Safer Alternatives

`MERGE` has had well-documented issues (particularly under `Read Committed` isolation without RCSI, and around trigger-firing behavior and unique-constraint violations under concurrent load) that have caused genuine, hard-to-reproduce production bugs — Microsoft's own documentation has historically carried explicit cautionary notes about `MERGE`'s concurrency behavior. **Professional guidance:** for simple upsert logic, prefer explicit separate `UPDATE`/`INSERT` statements wrapped in a transaction (as shown), which is more verbose but has fully predictable, well-understood behavior; reserve `MERGE` for scenarios where its specific capabilities (matching on multiple conditions, `WHEN NOT MATCHED BY SOURCE` for deletions) are genuinely needed and the concurrency profile of the operation is well understood and low-risk (e.g. a single-threaded nightly batch job, not a highly concurrent OLTP path).

## 62. Professional SQL Server Best Practices

| Category | Checklist |
|---|---|
| **Naming conventions** | Consistent, descriptive schema/table/column names; a documented convention for constraint names (`PK_`, `FK_`, `CK_`, `IX_` prefixes as used throughout this handbook). |
| **Data types** | `decimal` for money, never `float`/`money`; `datetime2` over `datetime`; `nvarchar` sized realistically, never blanket `nvarchar(max)` by default (Part 4). |
| **Constraints** | Every business rule expressible as a constraint enforced at the database level, not application code alone (Section 5.6). |
| **Indexing** | Deliberate clustered index choice; covering indexes for known hot query paths; regular review against actual query patterns, not guesswork (Part 26). |
| **Query design** | SARGable predicates; select only needed columns; set-based over cursors (Part 30). |
| **Transactions** | Kept as short as possible; explicit `TRY`/`CATCH` with proper `@@TRANCOUNT` checking (Parts 23, 25). |
| **Error handling** | `THROW` preferred over `RAISERROR`; meaningful custom error messages/numbers (Part 25). |
| **Security** | Least privilege; parameterized/dynamic SQL done safely; sensitive data masked/encrypted appropriately (Parts 35-36). |
| **Backups** | Recovery model matched to actual RPO/RTO requirements; backups tested via actual restores, not just assumed to work (Part 37). |
| **Documentation** | Schema and complex procedures documented; system-catalog-driven auditing queries maintained (Part 41). |
| **Performance** | Baseline metrics captured (`SET STATISTICS IO/TIME`); execution plans reviewed for genuinely slow queries before guessing at fixes (Parts 28-30). |
| **Testing** | Database changes tested in a non-production environment first, including restore drills; stored procedures covered by automated tests where practical (see the Testing and Debugging handbook for the application-layer testing discipline this pairs with). |
| **Deployment** | Schema changes version-controlled and deployed via repeatable scripts/migrations, not ad-hoc manual changes directly against production. |
| **Monitoring** | Blocking, deadlocks, tempdb contention, and long-running queries monitored proactively, not only investigated reactively after a user complaint. |
| **Disaster recovery** | HA and backup strategy both explicitly planned and periodically tested — never assumed to work untested (Part 45). |

## Conclusion

SQL Server, at its foundation, solves the exact same relational problems PostgreSQL and MySQL solve — and the deep conceptual fluency you already bring from those systems transfers directly: relational modeling, joins, transactions, indexing theory, and SQL's declarative query logic are not being re-learned here, only re-expressed through T-SQL's particular syntax and SQL Server's particular engine behavior. What this handbook has tried to make explicit, at every turn, is precisely where that transfer holds cleanly (`CASE`, window functions, CTEs — nearly identical everywhere) and where it does not (`TOP` vs `LIMIT`, lock-based vs MVCC-based concurrency by default, `IDENTITY`/`SEQUENCE`, the complete absence of native JSON/regex types, and SQL Server's genuinely distinctive built-in capabilities like columnstore indexes, temporal tables, and Always On Availability Groups).

Professional SQL Server fluency is ultimately less about memorizing every function in Part 9 or every DMV in Part 41, and more about the same disciplined thinking that likely already serves you well in PostgreSQL and MySQL: understand what a query logically needs to compute (Part 7's logical processing order), verify how the engine actually executes it (Parts 28-29's execution-plan literacy), and choose the right tool for the right layer of the stack — SQL Server for set-based data work and transactional integrity, Python for statistical modeling, Power BI for the final presentation layer (Parts 47-54's integration guidance) — rather than forcing any one layer to do work another is better suited for.

Returning to the best-practices checklist in Part 62 and the cheat sheet in Part 63 as ongoing references, working consistently through the projects in Part 57 and the capstone in Part 58, and revisiting the PostgreSQL/MySQL comparison in Part 55 whenever a new T-SQL feature's "does this exist elsewhere, and how" question arises, is the most direct path from your existing multi-database foundation to genuine SQL Server and T-SQL professional mastery.
